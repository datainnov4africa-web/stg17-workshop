## À quoi sert ce notebook

Les lumières nocturnes sont l'une des rares sources de données **gratuites, mondiales, quotidiennes et
politiquement neutres** dont dispose un institut national de statistique. Mais entre le fichier que la NASA
met en ligne et un indicateur publiable, il y a une étape que presque personne n'explique correctement :
**comprendre ce qu'il y a dans le fichier**.

Un fichier Black Marble n'est pas une image. C'est un **classeur** contenant une vingtaine de tableaux de
nombres entiers, chacun avec sa propre signification, sa propre échelle et sa propre valeur « vide ».
Lire ce classeur de travers — oublier un facteur d'échelle, confondre une valeur de remplissage avec un
zéro, ignorer un drapeau qualité — produit des chiffres qui *semblent* corrects et qui sont **faux**.

Ce notebook démonte le fichier pièce par pièce. À la fin, vous saurez répondre, preuve à l'appui, à
sept questions :

**Ce que fait cette étape.** Le notebook identifie d’abord l’environnement d’exécution
(Colab, Kaggle, Jupyter local, VS&nbsp;Code), puis vérifie les bibliothèques.

La distinction importante est celle entre **noyau requis** et **extensions optionnelles** :

- Le **noyau** — `numpy`, `pandas`, `matplotlib`, `h5py` — suffit à faire tourner **la totalité** de ce
  notebook. Ces quatre bibliothèques sont déjà présentes sur Colab et Kaggle.
- Les **extensions** — `geopandas`, `folium`, `rasterio`, `requests` — ajoutent du confort (frontières
  administratives, carte interactive, export GeoTIFF, téléchargement). Si l’une manque ou refuse de

````python
# =============================================================================
#  ÉTAPE 01 — ENVIRONNEMENT ET DÉPENDANCES
# =============================================================================
import importlib.util
import os
import platform
import subprocess
import sys
from pathlib import Path

# --- Réglage : mettre à False pour interdire toute installation automatique ---
INSTALLER_AUTOMATIQUEMENT = True


def detecter_environnement() -> str:
    """Identifie la plateforme d'exécution du notebook."""
    if "google.colab" in sys.modules:
        return "Google Colab"
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE") or Path("/kaggle/input").exists():
        return "Kaggle"
    if os.environ.get("VSCODE_PID") or os.environ.get("VSCODE_CWD"):
        return "VS Code"
    if "ipykernel" in sys.modules:
        return "Jupyter local"
    return "Python (hors notebook)"


ENVIRONNEMENT = detecter_environnement()

# Nom d'import  ->  nom du paquet pip
NOYAU = {"numpy": "numpy", "pandas": "pandas", "matplotlib": "matplotlib", "h5py": "h5py"}
EXTENSIONS = {"requests": "requests", "shapely": "shapely",
              "geopandas": "geopandas", "folium": "folium", "rasterio": "rasterio"}


def installer(paquets) -> bool:
    """Installe une liste de paquets via pip. Renvoie True si tout s'est bien passé."""
    try:
        subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", *paquets],
                              stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
        return True
    except Exception:
        return False


def verifier(table, etiquette, obligatoire):
    """Vérifie la présence des modules, tente l'installation des manquants."""
    manquants = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]
    if manquants and INSTALLER_AUTOMATIQUEMENT:
        print(f"  installation de {', '.join(manquants)} ...")
        installer(manquants)
        manquants = [pip for imp, pip in table.items() if importlib.util.find_spec(imp) is None]

    disponibles = {}
    for imp in table:
        present = importlib.util.find_spec(imp) is not None
        disponibles[imp] = present
        if present:
            try:
                version = getattr(__import__(imp), "__version__", "?")
            except Exception:
                version = "?"
            print(f"  [ok]      {imp:<12} {version}")
        else:
            print(f"  [{'MANQUE' if obligatoire else 'absent'}]  {imp:<12} ({etiquette})")
    return disponibles


print("=" * 74)
print(f"  ENVIRONNEMENT   : {ENVIRONNEMENT}")
print(f"  PYTHON          : {platform.python_version()}  ({platform.system()} {platform.machine()})")
print(f"  DOSSIER COURANT : {Path.cwd()}")
print("=" * 74)

print("\nNoyau requis")
etat_noyau = verifier(NOYAU, "indispensable", obligatoire=True)

print("\nExtensions optionnelles")
DISPO = verifier(EXTENSIONS, "confort", obligatoire=False)

if not all(etat_noyau.values()):
    raise SystemExit(
        "\nIl manque une bibliothèque du noyau. Installez-la à la main puis relancez :\n"
        "    pip install numpy pandas matplotlib h5py")

print("\n" + "=" * 74)
print("  Noyau complet — le notebook peut s'exécuter intégralement.")
if not all(DISPO.values()):
    absentes = [k for k, v in DISPO.items() if not v]
    print(f"  Extensions absentes : {', '.join(absentes)} — les étapes concernées seront")
    print("  simplement signalées puis contournées, sans interrompre le notebook.")
print("=" * 74)
````

````python
# =============================================================================
#  ÉTAPE 03 a — BOÎTE À OUTILS
# =============================================================================
import json
import re
import warnings

import h5py
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from matplotlib.colors import LinearSegmentedColormap, ListedColormap, LogNorm

warnings.filterwarnings("ignore")

# -----------------------------------------------------------------------------
#  1. CHARTE GRAPHIQUE  (palette du guide de style de la Banque)
# -----------------------------------------------------------------------------
COULEURS = dict(
    vert="#00A86A", vert_fonce="#00704A", foret="#00553A", or_="#F5C242",
    ocre="#D49A00", sarcelle="#0E7C86", terre="#C4621D", brique="#B83B2E",
    encre="#231F20", ardoise="#5E6964", brume="#F4F7F5", menthe="#E8F5EF", sauge="#D5DED9",
)

# Palette « vue de nuit depuis l'orbite » : noir -> vert -> or -> crème
CMAP_NUIT = LinearSegmentedColormap.from_list(
    "bad_nuit", ["#04120C", "#00553A", "#00704A", "#00A86A", "#F5C242", "#FFF6DF"])
# Palette séquentielle claire, pour les cartes sur fond blanc
CMAP_VERT = LinearSegmentedColormap.from_list(
    "bad_vert", ["#E8F5EF", "#9ED9C0", "#57BD92", "#10A06A", "#00704A", "#00553A"])

_polices = {f.name for f in mpl.font_manager.fontManager.ttflist}
POLICE = "Calibri" if "Calibri" in _polices else "DejaVu Sans"

mpl.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "figure.dpi": FIG_DPI, "font.family": POLICE, "font.size": 9.5,
    "axes.titlesize": 12, "axes.titleweight": "bold", "axes.titlecolor": COULEURS["foret"],
    "axes.labelcolor": COULEURS["ardoise"], "axes.edgecolor": COULEURS["sauge"],
    "xtick.color": COULEURS["ardoise"], "ytick.color": COULEURS["ardoise"],
    "axes.grid": False, "axes.spines.top": False, "axes.spines.right": False,
    "legend.frameon": False, "figure.autolayout": False,
})


def titre_figure(fig, titre, sous_titre="", source=""):
    """Applique l'habillage institutionnel standard à une figure."""
    fig.text(0.012, 0.985, titre, ha="left", va="top", fontsize=13.5,
             fontweight="bold", color=COULEURS["foret"])
    if sous_titre:
        fig.text(0.012, 0.938, sous_titre, ha="left", va="top", fontsize=10,
                 color=COULEURS["ardoise"])
    if source:
        fig.text(0.012, 0.004, source, ha="left", va="bottom", fontsize=7.2,
                 style="italic", color=COULEURS["ardoise"])


# -----------------------------------------------------------------------------
#  2. LA GRILLE BLACK MARBLE
# -----------------------------------------------------------------------------
#  Le globe est découpé en tuiles de 10° x 10°, indexées h00..h35 (ouest -> est)
#  et v00..v17 (nord -> sud). Chaque tuile contient 2400 x 2400 pixels, soit une
#  résolution de 15 secondes d'arc (~ 463 m à l'équateur).
# -----------------------------------------------------------------------------
TUILE_DEGRES = 10.0
TUILE_PIXELS = 2400
RESOLUTION_ARCSEC = 15.0

# Le nom du groupe HDF5 dépend du produit
GROUPE_PAR_PRODUIT = {
    "VNP46A1": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A2": "HDFEOS/GRIDS/VNP_Grid_DNB/Data Fields",
    "VNP46A3": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
    "VNP46A4": "HDFEOS/GRIDS/VIIRS_Grid_DNB_2d/Data Fields",
}
FAMILLE = {"VNP46A1": "journalier (brut)", "VNP46A2": "journalier",
           "VNP46A3": "mensuel", "VNP46A4": "annuel"}


def numeros_tuile(tuile: str):
    """'h17v08' -> (17, 8)"""
    m = re.search(r"h(\d{2})v(\d{2})", tuile)
    if not m:
        raise ValueError(f"Identifiant de tuile illisible : {tuile!r}")
    return int(m.group(1)), int(m.group(2))


def emprise_tuile(tuile: str):
    """'h17v08' -> (ouest, est, sud, nord) en degrés décimaux."""
    h, v = numeros_tuile(tuile)
    ouest = -180.0 + h * TUILE_DEGRES
    nord = 90.0 - v * TUILE_DEGRES
    return (ouest, ouest + TUILE_DEGRES, nord - TUILE_DEGRES, nord)


def tuiles_couvrant(ouest, est, sud, nord):
    """Liste les tuiles nécessaires pour couvrir une emprise géographique."""
    h0, h1 = int((ouest + 180) // 10), int((est + 180) // 10)
    v0, v1 = int((90 - nord) // 10), int((90 - sud) // 10)
    return [f"h{h:02d}v{v:02d}" for v in range(v0, v1 + 1) for h in range(h0, h1 + 1)]


# -----------------------------------------------------------------------------
#  3. LIRE UN FICHIER HDF5 SANS SE TROMPER
# -----------------------------------------------------------------------------
def trouver_groupe_donnees(fichier: h5py.File) -> str:
    """Retrouve le groupe 'Data Fields', quel que soit le produit.

    On ne code jamais le chemin en dur : on parcourt l'arborescence et on
    retient le premier groupe dont le nom se termine par 'Data Fields'.
    """
    trouves = []
    fichier.visititems(
        lambda nom, obj: trouves.append(nom)
        if isinstance(obj, h5py.Group) and nom.endswith("Data Fields") else None)
    if not trouves:
        raise KeyError("Aucun groupe 'Data Fields' : ce fichier n'est pas un produit VNP46.")
    return trouves[0]


def lire_attribut(objet, nom, defaut=None):
    """Lit un attribut HDF5 et le convertit en type Python simple.

    Les attributs HDF5 arrivent sous des formes très variées : octets, tableau
    numpy à un élément, tableau à plusieurs éléments... Cette fonction ramène
    tout à une chaîne, un nombre, ou une liste.
    """
    if nom not in objet.attrs:
        return defaut
    valeur = objet.attrs[nom]
    if isinstance(valeur, (bytes, bytearray, np.bytes_)):
        return valeur.decode("utf-8", "replace").strip()
    if isinstance(valeur, str):
        return valeur.strip()
    plat = np.ravel(valeur)
    if plat.size == 0:
        return defaut
    if plat.size == 1:
        element = plat[0]
        if isinstance(element, (bytes, bytearray, np.bytes_)):
            return element.decode("utf-8", "replace").strip()
        return element.item()
    return plat.tolist()


def tous_les_attributs(objet) -> dict:
    """Renvoie tous les attributs d'un objet HDF5, proprement convertis."""
    return {nom: lire_attribut(objet, nom) for nom in objet.attrs}


# Ordre de préférence pour la couche de radiance principale
COUCHES_RADIANCE = [
    "Gap_Filled_DNB_BRDF-Corrected_NTL",   # A2 : version comblée
    "DNB_BRDF-Corrected_NTL",              # A2 : version brute corrigée BRDF
    "AllAngle_Composite_Snow_Free",        # A3 / A4 : la référence usuelle
    "NearNadir_Composite_Snow_Free",       # A3 / A4 : vue proche du nadir
    "OffNadir_Composite_Snow_Free",
    "AllAngle_Composite_Snow_Covered",
]


def choisir_couche_radiance(groupe: h5py.Group) -> str:
    """Choisit la couche de radiance principale présente dans le fichier."""
    present = set(groupe.keys())
    for nom in COUCHES_RADIANCE:
        if nom in present:
            return nom
    # Repli : n'importe quelle couche 2D dont l'unité est une radiance
    for nom, obj in groupe.items():
        if isinstance(obj, h5py.Dataset) and obj.ndim == 2:
            if "nW" in str(lire_attribut(obj, "units", "")):
                return nom
    raise KeyError("Aucune couche de radiance identifiable dans ce fichier.")


# -----------------------------------------------------------------------------
#  4. DÉCODER UN NOM DE FICHIER
# -----------------------------------------------------------------------------
MOTIF_NOM = re.compile(
    r"^(?P<produit>VNP46A[1-4])"      # produit
    r"\.A(?P<annee>\d{4})(?P<jour>\d{3})"   # A + année + jour julien
    r"\.(?P<tuile>h\d{2}v\d{2})"      # tuile
    r"\.(?P<version>\d{3})"           # version de la collection
    r"\.(?P<horodatage>[^.]+)"        # horodatage de production
    r"\.h5$", re.IGNORECASE)


def decoder_nom_fichier(nom: str):
    """Décompose un nom de fichier VNP46 en ses éléments constitutifs."""
    m = MOTIF_NOM.match(str(nom))
    if not m:
        return None
    d = m.groupdict()
    d["annee"] = int(d["annee"])
    d["jour"] = int(d["jour"])
    d["date"] = pd.Timestamp(d["annee"], 1, 1) + pd.Timedelta(days=d["jour"] - 1)
    d["famille"] = FAMILLE.get(d["produit"].upper(), "?")
    d["emprise"] = emprise_tuile(d["tuile"])
    return d


print(f"Boîte à outils chargée  ·  police des figures : {POLICE}")
print(f"Grille Black Marble     : tuiles de {TUILE_DEGRES:.0f}° × {TUILE_DEGRES:.0f}°, "
      f"{TUILE_PIXELS} × {TUILE_PIXELS} pixels, {RESOLUTION_ARCSEC:.0f} secondes d'arc")
````

````python
# =============================================================================
#  ÉTAPE 03 b — FILET DE SÉCURITÉ : GÉNÉRATEUR DE FICHIERS DE DÉMONSTRATION
#  (défini maintenant, appelé seulement si aucune donnée réelle n'est trouvée)
# =============================================================================

# --- Spécification des couches, calquée sur la collection 2 (version 002) ----
SPEC_COMPOSITE = {   # produits VNP46A3 (mensuel) et VNP46A4 (annuel)
    "AllAngle_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="All angle composite, snow-free, BRDF-corrected NTL"),
    "AllAngle_Composite_Snow_Free_Num":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Number of Observations",
             titre="Number of observations, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Std":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Standard deviation, all-angle snow-free composite"),
    "AllAngle_Composite_Snow_Free_Quality":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Quality Flag",
             titre="Quality flag, all-angle snow-free composite"),
    "NearNadir_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Near-nadir composite, snow-free, BRDF-corrected NTL"),
    "NearNadir_Composite_Snow_Free_Num":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Number of Observations",
             titre="Number of observations, near-nadir snow-free composite"),
    "OffNadir_Composite_Snow_Free":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Off-nadir composite, snow-free, BRDF-corrected NTL"),
    "Land_Water_Mask":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Land/Water background mask"),
    "DNB_Platform":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Source satellite platform"),
}

SPEC_JOURNALIER = {   # produit VNP46A2
    "DNB_BRDF-Corrected_NTL":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="BRDF-corrected nighttime lights radiance"),
    "Gap_Filled_DNB_BRDF-Corrected_NTL":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Gap-filled BRDF-corrected nighttime lights radiance"),
    "DNB_Lunar_Irradiance":
        dict(type="uint16", echelle=0.1, remplissage=65535, unite="nW_cm-2_sr-1",
             titre="Lunar irradiance at observation time"),
    "Latest_High_Quality_Retrieval":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="Days",
             titre="Days since latest high-quality retrieval"),
    "Mandatory_Quality_Flag":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Mandatory quality flag"),
    "QF_Cloud_Mask":
        dict(type="uint16", echelle=1.0, remplissage=65535, unite="none",
             titre="VIIRS cloud mask quality flags (bit field)"),
    "Snow_Flag":
        dict(type="uint8", echelle=1.0, remplissage=255, unite="none",
             titre="Snow / no-snow flag"),
}

# --- Descriptions officielles des drapeaux (reprises des guides NASA) --------
DESC_QUALITE_COMPOSITE = "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
DESC_QUALITE_JOURNALIER = ("0 = High-quality, persistent nighttime lights "
                           "1 = High-quality, ephemeral nighttime lights "
                           "2 = Poor-quality, outlier corrected by VIIRS 255 = No retrieval")
DESC_TERRE_EAU = ("0 = Shallow ocean 1 = Land 2 = Coastline 3 = Shallow inland water "
                  "4 = Ephemeral water 5 = Deep inland water 6 = Continental 7 = Deep ocean")

# --- Foyers lumineux utilisés pour simuler une scène plausible ---------------
FOYERS = {
    "h17v08": [("Abidjan", -4.02, 5.35, 1.00), ("Yamoussoukro", -5.28, 6.82, 0.34),
               ("San-Pédro", -6.63, 4.75, 0.22), ("Daloa", -6.45, 6.88, 0.20),
               ("Bouaké", -5.03, 7.69, 0.30), ("Accra", -0.19, 5.60, 0.75)],
    "h17v07": [("Korhogo", -5.63, 9.46, 0.22), ("Odienné", -7.56, 9.51, 0.10),
               ("Bobo-Dioulasso", -4.30, 11.18, 0.28), ("Ouagadougou", -1.53, 12.37, 0.55)],
    "h20v07": [("Nyala", 24.88, 12.05, 0.30), ("El Obeid", 30.22, 13.18, 0.30),
               ("Khartoum", 32.53, 15.55, 1.00), ("El Fasher", 25.35, 13.63, 0.22)],
    "h20v06": [("Dongola", 30.48, 19.17, 0.15), ("Atbara", 33.99, 17.70, 0.20)],
}


def _champ_radiance(tuile, cote, graine=0, intensite=1.0, bruit=0.05):
    """Fabrique un champ de radiance plausible : fond rural + halos urbains."""
    ouest, est, sud, nord = emprise_tuile(tuile)
    LON, LAT = np.meshgrid(np.linspace(ouest, est, cote), np.linspace(nord, sud, cote))
    rng = np.random.default_rng(graine + numeros_tuile(tuile)[0] * 101 + numeros_tuile(tuile)[1])
    champ = 0.15 + rng.gamma(1.2, 0.18, LON.shape)                 # bruit de fond rural
    for _nom, clon, clat, poids in FOYERS.get(tuile, []):
        sigma = 0.10 + 0.22 * poids
        champ += 260.0 * poids * intensite * np.exp(
            -(((LON - clon) / sigma) ** 2 + ((LAT - clat) / sigma) ** 2) / 2)
        champ += 18.0 * poids * intensite * np.exp(          # halo périurbain (blooming)
            -(((LON - clon) / (sigma * 3.2)) ** 2 + ((LAT - clat) / (sigma * 3.2)) ** 2) / 2)
    if bruit:
        champ *= (1 + rng.normal(0, bruit, LON.shape)).clip(0.4, 2.0)
    return champ.clip(0, 6000), LON, LAT


def _encoder(valeurs, spec):
    """Réel -> entier stocké, en respectant le facteur d'échelle."""
    brut = np.round(np.nan_to_num(valeurs) / spec["echelle"])
    return np.clip(brut, 0, spec["remplissage"] - 1).astype(spec["type"])


def _octets(texte: str):
    """Encode une chaîne en octets UTF-8 pour un attribut HDF5.

    np.bytes_() encode en ASCII et échoue sur le moindre accent : on passe
    donc explicitement par .encode("utf-8").
    """
    return np.frombuffer(str(texte).encode("utf-8"), dtype="S1").tobytes()


def _ecrire_couche(groupe, nom, tableau, spec, extra=None):
    """Écrit un jeu de données avec la panoplie complète d'attributs NASA."""
    jeu = groupe.create_dataset(nom, data=tableau, compression="gzip", compression_opts=4)
    jeu.attrs["long_name"] = _octets(spec["titre"])
    jeu.attrs["units"] = _octets(spec["unite"])
    jeu.attrs["scale_factor"] = np.array([spec["echelle"]], dtype="float64")
    jeu.attrs["add_offset"] = np.array([0.0], dtype="float64")
    jeu.attrs["_FillValue"] = np.array([spec["remplissage"]], dtype=tableau.dtype)
    jeu.attrs["valid_min"] = np.array([0], dtype=tableau.dtype)
    jeu.attrs["valid_max"] = np.array([spec["remplissage"] - 1], dtype=tableau.dtype)
    for cle, val in (extra or {}).items():
        jeu.attrs[cle] = _octets(val) if isinstance(val, str) else val


def _struct_metadata(tuile, cote, grille):
    """Reproduit le bloc StructMetadata.0 propre au format HDF-EOS5."""
    ouest, est, sud, nord = emprise_tuile(tuile)
    return (f"GROUP=SwathStructure\nEND_GROUP=SwathStructure\n"
            f"GROUP=GridStructure\n\tGROUP=GRID_1\n"
            f"\t\tGridName=\"{grille}\"\n\t\tXDim={cote}\n\t\tYDim={cote}\n"
            f"\t\tUpperLeftPointMtrs=({ouest * 1e6:.6f},{nord * 1e6:.6f})\n"
            f"\t\tLowerRightMtrs=({est * 1e6:.6f},{sud * 1e6:.6f})\n"
            f"\t\tProjection=HE5_GCTP_GEO\n\t\tGridOrigin=HE5_HDFE_GD_UL\n"
            f"\tEND_GROUP=GRID_1\nEND_GROUP=GridStructure\nEND\n")


def fabriquer_fichier_demo(chemin, produit, tuile, annee, jour, cote=240,
                           graine=0, intensite=1.0):
    """Écrit un fichier .h5 de démonstration structurellement conforme."""
    chemin = Path(chemin)
    chemin.parent.mkdir(parents=True, exist_ok=True)
    journalier = produit == "VNP46A2"
    grille = "VNP_Grid_DNB" if journalier else "VIIRS_Grid_DNB_2d"
    spec_produit = SPEC_JOURNALIER if journalier else SPEC_COMPOSITE

    radiance, LON, LAT = _champ_radiance(tuile, cote, graine, intensite,
                                         bruit=0.25 if journalier else 0.05)
    rng = np.random.default_rng(graine + 7)

    # Masque terre/eau : océan et trait de côte pour la tuile littorale.
    # Le littoral suit une courbe et non une droite, pour rester plausible.
    terre_eau = np.ones(radiance.shape, dtype="uint8")
    if tuile == "h17v08":
        latitude_cote = (4.75 + 0.55 * np.sin((LON + 7.0) * 0.55)
                         + 0.18 * np.sin(LON * 2.1))
        terre_eau[LAT < latitude_cote] = 0                              # océan
        terre_eau[(LAT >= latitude_cote)
                  & (LAT < latitude_cote + 0.18)] = 2                   # trait de côte

    with h5py.File(chemin, "w") as f:
        # --- attributs globaux ------------------------------------------------
        ouest, est, sud, nord = emprise_tuile(tuile)
        h, v = numeros_tuile(tuile)
        for cle, val in [("ShortName", produit), ("PlatformShortName", "Suomi-NPP"),
                         ("InstrumentShortName", "VIIRS"), ("ProcessingLevel", "Level 3"),
                         ("ProductionTime", f"{annee}-06-01 00:00:00.000"),
                         ("HorizontalTileNumber", f"{h:02d}"),
                         ("VerticalTileNumber", f"{v:02d}"),
                         ("DEMO_SYNTHETIC", "TRUE - structure conforme, valeurs simulees")]:
            f.attrs[cle] = _octets(val)
        for cle, val in [("WestBoundingCoord", ouest), ("EastBoundingCoord", est),
                         ("SouthBoundingCoord", sud), ("NorthBoundingCoord", nord)]:
            f.attrs[cle] = np.array([val], dtype="float64")

        f.create_group("HDFEOS INFORMATION").create_dataset(
            "StructMetadata.0", data=_octets(_struct_metadata(tuile, cote, grille)))

        groupe = f.create_group(f"HDFEOS/GRIDS/{grille}/Data Fields")
        for nom, spec in spec_produit.items():
            if nom == "Land_Water_Mask":
                _ecrire_couche(groupe, nom, terre_eau, spec, {"description": DESC_TERRE_EAU})
            elif nom == "DNB_Platform":
                _ecrire_couche(groupe, nom, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = Suomi-NPP 1 = NOAA-20"})
            elif nom.endswith("_Quality") or nom == "Mandatory_Quality_Flag":
                q = np.zeros(radiance.shape, "uint8")
                q[rng.random(radiance.shape) < (0.22 if journalier else 0.06)] = 1
                q[rng.random(radiance.shape) < (0.10 if journalier else 0.02)] = 2
                q[terre_eau == 0] = 255
                _ecrire_couche(groupe, nom, q, spec, {
                    "description": DESC_QUALITE_JOURNALIER if journalier else DESC_QUALITE_COMPOSITE})
            elif nom.endswith("_Num"):
                _ecrire_couche(groupe, nom,
                               np.clip(rng.poisson(14, radiance.shape), 0, 254).astype("uint8"), spec)
            elif nom.endswith("_Std"):
                _ecrire_couche(groupe, nom, _encoder(radiance * 0.18, spec), spec)
            elif nom == "Snow_Flag":
                _ecrire_couche(groupe, nom, np.zeros(radiance.shape, "uint8"), spec,
                               {"description": "0 = No snow 1 = Snow"})
            elif nom == "QF_Cloud_Mask":
                qf = np.full(radiance.shape, 0b10, dtype="uint16")   # bit 1 : masque exploitable
                qf[rng.random(radiance.shape) < 0.18] |= (0b10 << 6)  # bits 6-7 : nuage détecté
                _ecrire_couche(groupe, nom, qf, spec,
                               {"description": "VIIRS cloud mask, bit field"})
            elif nom == "DNB_Lunar_Irradiance":
                _ecrire_couche(groupe, nom, _encoder(np.full(radiance.shape, 12.5), spec), spec)
            elif nom == "Latest_High_Quality_Retrieval":
                _ecrire_couche(groupe, nom,
                               np.clip(rng.poisson(2, radiance.shape), 0, 254).astype("uint8"), spec)
            elif "NearNadir" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 0.97, spec), spec)
            elif "OffNadir" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 1.04, spec), spec)
            elif "Gap_Filled" in nom:
                _ecrire_couche(groupe, nom, _encoder(radiance * 1.01, spec), spec)
            else:
                # Couche de radiance principale : l'eau reçoit la valeur de remplissage
                encode = _encoder(radiance, spec)
                encode[terre_eau == 0] = spec["remplissage"]
                _ecrire_couche(groupe, nom, encode, spec)

        # Les produits composites embarquent aussi les coordonnées explicites
        if not journalier:
            resolution = TUILE_DEGRES / cote
            axes = {"lon": (ouest + (np.arange(cote) + 0.5) * resolution,
                            "Longitude", "degrees_east"),
                    "lat": (nord - (np.arange(cote) + 0.5) * resolution,
                            "Latitude", "degrees_north")}
            for nom_axe, (valeurs, titre_axe, unite) in axes.items():
                jeu = groupe.create_dataset(nom_axe, data=valeurs.astype("float64"))
                jeu.attrs["long_name"] = _octets(titre_axe)
                jeu.attrs["units"] = _octets(unite)
                jeu.attrs["_FillValue"] = np.array([-999.9], dtype="float64")
    return chemin


def fabriquer_jeu_demo(dossier, cote=240):
    """Fabrique un petit jeu complet : 3 années, 4 mois, 2 nuits."""
    dossier = Path(dossier)
    produits = []
    for annee in (2021, 2022, 2023):
        for tuile in ("h17v07", "h17v08"):
            produits.append(("VNP46A4", tuile, annee, 1,
                             f"annuel/{annee}", 1 + 0.07 * (annee - 2021)))
    for jour, _mois in [(1, "jan"), (91, "avr"), (182, "jui"), (274, "oct")]:
        produits.append(("VNP46A3", "h17v08", 2023, jour, "mensuel/2023", 1.0))
    for jour in (12, 200):
        produits.append(("VNP46A2", "h17v08", 2023, jour, "journalier/2023", 1.0))

    ecrits = []
    for i, (produit, tuile, annee, jour, sous_dossier, intensite) in enumerate(produits):
        nom = f"{produit}.A{annee}{jour:03d}.{tuile}.002.DEMO{i:09d}.h5"
        ecrits.append(fabriquer_fichier_demo(dossier / sous_dossier / nom, produit, tuile,
                                             annee, jour, cote=cote, graine=annee + jour,
                                             intensite=intensite))
    return ecrits


print("Générateur de démonstration prêt (il ne sera appelé qu'en dernier recours).")
````

````python
# =============================================================================
#  ÉTAPE 03 c — LOCALISER LES DONNÉES, PUIS DRESSER L'INVENTAIRE
# =============================================================================

NOMS_DOSSIERS_USUELS = ["blackmarble_data", "blackmarble", "black_marble", "BlackMarble",
                        "ntl_data", "NTL", "ntl", "data", "donnees", "Collecte",
                        "Collecte/blackmarble_data", "VNP46"]


def _contient_vnp46(dossier: Path, limite=4000) -> bool:
    """Le dossier contient-il au moins un fichier VNP46*.h5 ? (arrêt au premier)"""
    try:
        for i, fichier in enumerate(dossier.rglob("*.h5")):
            if fichier.name.upper().startswith("VNP46"):
                return True
            if i > limite:
                break
    except (PermissionError, OSError):
        pass
    return False


def localiser_donnees(chemin_explicite=None):
    """Applique l'échelle de priorité et renvoie (dossier, origine)."""
    # 1. chemin explicite
    if chemin_explicite:
        d = Path(chemin_explicite).expanduser()
        if d.exists():
            return d, ("chemin explicite (DATA_DIR)" if _contient_vnp46(d)
                       else "chemin explicite (DATA_DIR) — aucun VNP46 détecté")
        print(f"  ! DATA_DIR pointe vers un dossier inexistant : {d}")

    # 2. Kaggle
    racine_kaggle = Path("/kaggle/input")
    if racine_kaggle.exists():
        for d in sorted(racine_kaggle.iterdir()):
            if d.is_dir() and _contient_vnp46(d):
                return d, "jeu de données Kaggle"

    # 3. Colab
    for d in [Path("/content/drive/MyDrive"), Path("/content")]:
        if d.exists() and _contient_vnp46(d):
            return d, "Google Colab"

    # 4. dossiers usuels, en remontant l'arborescence
    ici = Path.cwd()
    for base in [ici, *list(ici.parents)[:4]]:
        for nom in NOMS_DOSSIERS_USUELS:
            d = base / nom
            if d.exists() and _contient_vnp46(d):
                return d, "dossier local usuel"

    # 5. recherche large sous le dossier courant
    if _contient_vnp46(ici):
        return ici, "dossier courant"

    return None, "aucune donnée trouvée"


# --- Application de l'échelle -------------------------------------------------
print("Recherche des données ...")
DOSSIER_DONNEES, ORIGINE = localiser_donnees(DATA_DIR)
MODE_DEMO = False

if DOSSIER_DONNEES is None:
    if AUTORISER_DEMO:
        DOSSIER_DONNEES = Path.cwd() / "ntl_demo_data"
        if not _contient_vnp46(DOSSIER_DONNEES):
            print("  aucune donnée réelle — fabrication du jeu de démonstration ...")
            fabriquer_jeu_demo(DOSSIER_DONNEES, cote=DEMO_PX)
        ORIGINE, MODE_DEMO = "jeu de démonstration fabriqué localement", True
    else:
        raise SystemExit("Aucune donnée trouvée et AUTORISER_DEMO = False. "
                         "Renseignez DATA_DIR à l'étape 02.")


# --- Inventaire ---------------------------------------------------------------
def dresser_inventaire(dossier: Path) -> pd.DataFrame:
    """Parcourt le dossier et catalogue tous les fichiers VNP46 exploitables."""
    lignes = []
    for fichier in sorted(Path(dossier).rglob("*.h5")):
        infos = decoder_nom_fichier(fichier.name)
        if infos is None:
            continue
        lignes.append({
            "produit": infos["produit"].upper(),
            "famille": infos["famille"],
            "annee": infos["annee"],
            "jour": infos["jour"],
            "date": infos["date"].date(),
            "tuile": infos["tuile"],
            "taille_Mo": round(fichier.stat().st_size / 1e6, 2),
            "chemin": str(fichier),
        })
    colonnes = ["produit", "famille", "annee", "jour", "date", "tuile", "taille_Mo", "chemin"]
    if not lignes:
        return pd.DataFrame(columns=colonnes)
    return (pd.DataFrame(lignes)
            .sort_values(["produit", "annee", "jour", "tuile"])
            .reset_index(drop=True))


INVENTAIRE = dresser_inventaire(DOSSIER_DONNEES)

if INVENTAIRE.empty:
    raise SystemExit(
        f"Des fichiers .h5 existent sous {DOSSIER_DONNEES} mais aucun ne suit la convention\n"
        f"de nommage VNP46. Vérifiez qu'ils n'ont pas été renommés au téléchargement.")


def fichier_exemple(produit=None):
    """Renvoie un fichier représentatif du produit demandé (ou du plus riche)."""
    if produit:
        sous = INVENTAIRE[INVENTAIRE["produit"] == produit]
        if not sous.empty:
            return Path(sous.iloc[-1]["chemin"])
    for repli in ("VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"):
        sous = INVENTAIRE[INVENTAIRE["produit"] == repli]
        if not sous.empty:
            return Path(sous.iloc[-1]["chemin"])
    return Path(INVENTAIRE.iloc[-1]["chemin"])


FICHIER_EXEMPLE = fichier_exemple(PRODUIT_PRINCIPAL)

# --- Restitution --------------------------------------------------------------
bandeau = "  MODE DÉMONSTRATION — valeurs simulées, structure conforme  " if MODE_DEMO \
          else "  DONNÉES RÉELLES  "
print("\n" + "=" * 74)
print(bandeau.center(74, "*" if MODE_DEMO else "="))
print("=" * 74)
print(f"  Origine           : {ORIGINE}")
print(f"  Dossier           : {DOSSIER_DONNEES}")
print(f"  Fichiers VNP46    : {len(INVENTAIRE)}   ({INVENTAIRE['taille_Mo'].sum():.1f} Mo)")
print(f"  Produits présents : {', '.join(sorted(INVENTAIRE['produit'].unique()))}")
print(f"  Tuiles présentes  : {', '.join(sorted(INVENTAIRE['tuile'].unique()))}")
print(f"  Années couvertes  : {INVENTAIRE['annee'].min()} – {INVENTAIRE['annee'].max()}")
print(f"  Fichier d'exemple : {FICHIER_EXEMPLE.name}")
print("=" * 74)

# Synthèse par produit
synthese = (INVENTAIRE.groupby(["produit", "famille"])
            .agg(fichiers=("chemin", "size"),
                 annees=("annee", lambda s: f"{s.min()}–{s.max()}"),
                 tuiles=("tuile", lambda s: s.nunique()),
                 volume_Mo=("taille_Mo", "sum"))
            .reset_index())
display(synthese)
display(INVENTAIRE.drop(columns=["chemin"]).head(12))
````

**Le problème que résout la grille.** Une image mondiale à 500 m de résolution pèserait des
téraoctets. La NASA découpe donc le globe en **carrés de 10° × 10°** : 36 colonnes (`h00` à `h35`, d’ouest
en est) par 18 lignes (`v00` à `v17`, du nord au sud), soit 648 tuiles. Chaque tuile fait 2400 × 2400 pixels.

Le calcul du coin supérieur gauche d’une tuile tient en deux lignes, et c’est tout ce dont vous avez besoin
pour géoréférencer un fichier :

```
longitude_ouest = -180 + 10 × h
latitude_nord   =  +90 - 10 × v
```

La conséquence pratique : **un pays chevauche presque toujours plusieurs tuiles**, et il vous faudra les

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Un fichier n’est pas une image : c’est un classeur</div>

**L’image mentale.** Ouvrez un fichier `.h5` en pensant à un **classeur à intercalaires**,
pas à une photo. Trois notions suffisent :

- un **groupe** est un intercalaire — il ne contient pas de données, seulement d’autres groupes ou des jeux
  de données ; c’est l’équivalent d’un dossier ;
- un **jeu de données** (*dataset*) est une feuille de chiffres — un tableau, ici presque toujours
  bidimensionnel, de type entier ;
- un **attribut** est une annotation collée sur un groupe ou sur une feuille : unité, facteur d’échelle,
  valeur de remplissage, description des codes.

Dans un fichier Black Marble, le chemin utile est toujours de la forme
`HDFEOS / GRIDS / <nom de grille> / Data Fields`. Le nom de grille, lui, **change selon le produit** :
`VIIRS_Grid_DNB_2d` pour les composites A3 et A4, `VNP_Grid_DNB` pour le journalier A2. C’est précisément
pour cela que la fonction `trouver_groupe_donnees()` cherche le groupe au lieu de le supposer.

````python
# =============================================================================
#  ÉTAPE 07 — L'ARBORESCENCE DU FICHIER
# =============================================================================
print(f"FICHIER : {FICHIER_EXEMPLE.name}")
print(f"TAILLE  : {FICHIER_EXEMPLE.stat().st_size / 1e6:.2f} Mo")
print("=" * 78)

compteurs = {"groupes": 0, "jeux": 0, "octets": 0}


def afficher_noeud(nom, objet):
    """Affiche un nœud de l'arborescence avec une indentation proportionnelle."""
    profondeur = nom.count("/")
    feuille = nom.split("/")[-1]
    marge = "   " * profondeur
    if isinstance(objet, h5py.Group):
        compteurs["groupes"] += 1
        n_attrs = len(objet.attrs)
        suffixe = f"   [{n_attrs} attribut(s)]" if n_attrs else ""
        print(f"{marge}├─ ▸ {feuille}/{suffixe}")
    else:
        compteurs["jeux"] += 1
        compteurs["octets"] += objet.size * objet.dtype.itemsize
        forme = "×".join(str(d) for d in objet.shape) if objet.shape else "scalaire"
        print(f"{marge}├─ ▪ {feuille:<42} {forme:>12}  {str(objet.dtype):<8} "
              f"{len(objet.attrs):>2} attr.")


with h5py.File(FICHIER_EXEMPLE, "r") as f:
    # 1. Les attributs globaux du fichier
    globaux = tous_les_attributs(f)
    if globaux:
        print("\nATTRIBUTS GLOBAUX DU FICHIER")
        for cle, valeur in list(globaux.items())[:14]:
            texte = str(valeur)
            print(f"   {cle:<26} = {texte[:46]}{'…' if len(texte) > 46 else ''}")

    # 2. L'arborescence complète
    print("\nARBORESCENCE")
    print("   ▸ groupe (intercalaire)   ▪ jeu de données (feuille de chiffres)\n")
    f.visititems(afficher_noeud)

    chemin_groupe = trouver_groupe_donnees(f)

print("\n" + "=" * 78)
print(f"  {compteurs['groupes']} groupe(s)  ·  {compteurs['jeux']} jeu(x) de données  ·  "
      f"{compteurs['octets'] / 1e6:.1f} Mo une fois décompressé en mémoire")
print(f"  Groupe utile détecté automatiquement :\n     {chemin_groupe}")
print("=" * 78)
````

````python
# =============================================================================
#  ÉTAPE 08 — LE CATALOGUE DES COUCHES
# =============================================================================

def classer_couche(nom: str) -> str:
    """Traduit le nom technique d'une couche en un rôle compréhensible."""
    if nom in ("lat", "lon", "latitude", "longitude"):
        return "5 · Coordonnées"
    if nom == "Land_Water_Mask":
        return "4 · Contexte — terre / eau"
    if nom == "DNB_Platform":
        return "4 · Contexte — satellite"
    if nom.endswith("_Quality") or "Quality_Flag" in nom:
        return "3 · Qualité"
    if "Cloud" in nom or nom == "Snow_Flag":
        return "3 · Qualité — nuage / neige"
    if nom.endswith("_Num"):
        return "2 · Incertitude — nb d'observations"
    if nom.endswith("_Std"):
        return "2 · Incertitude — écart-type"
    if "Lunar" in nom:
        return "4 · Contexte — éclairement lunaire"
    if "Retrieval" in nom:
        return "3 · Qualité — fraîcheur"
    if "Composite" in nom or "NTL" in nom:
        return "1 · RADIANCE"
    return "6 · Autre"


lignes = []
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    couche_principale = choisir_couche_radiance(groupe)
    for nom, objet in groupe.items():
        if not isinstance(objet, h5py.Dataset):
            continue
        lignes.append({
            "couche": nom,
            "rôle": classer_couche(nom),
            "forme": "×".join(str(d) for d in objet.shape) if objet.shape else "scalaire",
            "type": str(objet.dtype),
            "unité": lire_attribut(objet, "units", ""),
            "échelle": lire_attribut(objet, "scale_factor", ""),
            "décalage": lire_attribut(objet, "add_offset", lire_attribut(objet, "offset", "")),
            "_FillValue": lire_attribut(objet, "_FillValue", ""),
            "Mo": round(objet.size * objet.dtype.itemsize / 1e6, 2),
        })

CATALOGUE = pd.DataFrame(lignes).sort_values(["rôle", "couche"]).reset_index(drop=True)

print(f"{len(CATALOGUE)} couche(s) dans {FICHIER_EXEMPLE.name}")
print(f"Couche de radiance retenue automatiquement : {couche_principale}\n")
display(CATALOGUE)

print("\nPOIDS MÉMOIRE PAR RÔLE")
display(CATALOGUE.groupby("rôle").agg(couches=("couche", "size"), Mo=("Mo", "sum")))
````

 <div style="color:#00553A;font-size:21px;font-weight:800;margin-top:4px;line-height:1.25;">Décoder les nombres</div>
 <div style="color:#231F20;font-size:14px;margin-top:4px;">Trois attributs qui séparent le vrai du faux</div>
</div>

**Le problème.** Pour économiser de la place, la NASA ne stocke pas des nombres décimaux mais
des **entiers**. La radiance réelle s’obtient par une conversion, et les paramètres de cette conversion
sont écrits **dans le fichier lui-même**, sous forme d’attributs :

$$\text{radiance} = \text{valeur stockée} \times \texttt{scale\_factor} + \texttt{add\_offset}$$

à ceci près qu’une valeur particulière — le `_FillValue` — ne signifie pas « zéro » mais
**« aucune donnée ici »**. Elle doit être convertie en `NaN` **avant** toute opération.

Concrètement, pour `AllAngle_Composite_Snow_Free` : `scale_factor = 0,1`, `add_offset = 0`,
`_FillValue = 65535`. Une valeur stockée de `1247` vaut donc **124,7 nW·cm⁻²·sr⁻¹**, et une valeur
stockée de `65535` ne vaut **rien du tout**.

<div style="background:#FBEDEA;border:1px solid #B83B2E44;border-left:4px solid #B83B2E;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#B83B2E;font-weight:700;font-size:13.5px;">&#9940;&nbsp; L’erreur qui invalide silencieusement une analyse entière</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Traiter le <code>_FillValue</code> comme un nombre. Avec <code>_FillValue = 65535</code> et <code>scale_factor = 0,1</code>, chaque pixel vide devient <b>6 553,5 nW·cm⁻²·sr⁻¹</b> — une valeur supérieure à celle du centre de Lagos. Une somme des lumières calculée ainsi peut être fausse d’un facteur dix, sans qu’aucun message d’erreur n’apparaisse. La carte produite reste jolie. <b>C’est la première chose à vérifier</b> quand un chiffre semble trop beau.</div>
</div>

````python
# =============================================================================
#  ÉTAPE 09 — DÉCODER LES NOMBRES : scale_factor, add_offset, _FillValue
# =============================================================================
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    jeu = groupe[couche_principale]

    attributs = tous_les_attributs(jeu)
    echelle = lire_attribut(jeu, "scale_factor", 1.0)
    decalage = lire_attribut(jeu, "add_offset", lire_attribut(jeu, "offset", 0.0))
    remplissage = lire_attribut(jeu, "_FillValue", None)

    # Un petit échantillon carré, pris là où il y a de la lumière
    cote = jeu.shape[0]
    apercu = jeu[::max(cote // 300, 1), ::max(cote // 300, 1)].astype("float64")
    masque_valide = apercu != remplissage if remplissage is not None else np.ones_like(apercu, bool)
    if masque_valide.any():
        i_max, j_max = np.unravel_index(np.argmax(np.where(masque_valide, apercu, -1)),
                                        apercu.shape)
        pas = max(cote // 300, 1)
        li = int(np.clip(i_max * pas - 2, 0, cote - 5))
        co = int(np.clip(j_max * pas - 2, 0, cote - 5))
    else:
        li = co = 0
    bloc_brut = jeu[li:li + 5, co:co + 5].astype("float64")

print(f"COUCHE : {couche_principale}\n")
print("ATTRIBUTS LUS DANS LE FICHIER (jamais codés en dur)")
for cle, valeur in attributs.items():
    texte = str(valeur)
    print(f"   {cle:<16} = {texte[:62]}{'…' if len(texte) > 62 else ''}")

print(f"\nL'ÉQUATION DE DÉCODAGE")
print(f"   radiance = valeur_stockée × {echelle} + {decalage}")
print(f"   puis     : toute valeur égale à {remplissage} devient NaN\n")

# --- Application sur un bloc 5 × 5 -------------------------------------------
bloc_decode = np.where(bloc_brut == remplissage, np.nan, bloc_brut * echelle + decalage)

print(f"BLOC 5 × 5 EXTRAIT À LA POSITION (ligne {li}, colonne {co})\n")
print("   valeurs stockées (entiers bruts)")
print(pd.DataFrame(bloc_brut.astype("int64")).to_string(
    header=[f"c{co + k}" for k in range(5)], index=[f"l{li + k}" for k in range(5)]))
print(f"\n   radiances décodées (nW·cm⁻²·sr⁻¹)")
print(pd.DataFrame(np.round(bloc_decode, 2)).to_string(
    header=[f"c{co + k}" for k in range(5)], index=[f"l{li + k}" for k in range(5)]))

# --- Démonstration du danger --------------------------------------------------
tableau_complet = None
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    tableau_complet = f[trouver_groupe_donnees(f)][couche_principale][:].astype("float64")

faux = (tableau_complet * echelle + decalage)                    # sans traiter le _FillValue
juste = np.where(tableau_complet == remplissage, np.nan, faux)   # correctement traité
n_vides = int(np.sum(tableau_complet == remplissage))

print("\n" + "=" * 78)
print("  CE QUE COÛTE L'OUBLI DU _FillValue")
print("=" * 78)
print(f"  Pixels de remplissage  : {n_vides:,} sur {tableau_complet.size:,} "
      f"({100 * n_vides / tableau_complet.size:.1f} %)".replace(",", " "))
print(f"  Somme des lumières — traitement correct : {np.nansum(juste):>18,.0f}".replace(",", " "))
print(f"  Somme des lumières — _FillValue ignoré  : {np.nansum(faux):>18,.0f}".replace(",", " "))
if np.nansum(juste) > 0:
    print(f"  Facteur d'erreur                       : "
          f"× {np.nansum(faux) / np.nansum(juste):.1f}")
print("=" * 78)
````

**Ce qu’il faut savoir.** Les produits Black Marble sont sur une **grille géographique
régulière** (EPSG:4326) : pas de projection compliquée, pas de rééchantillonnage. Le pixel (0, 0) est
le **coin nord-ouest** de la tuile, et la taille d’un pixel est constante en degrés :

```
résolution = 10° / 2400 = 0,00416667°  (soit 15 secondes d'arc)

longitude(colonne) = ouest + (colonne + 0,5) × résolution
latitude(ligne)    = nord  - (ligne    + 0,5) × résolution
```

Le `+ 0,5` place la coordonnée au **centre** du pixel, et non à son coin — une convention discrète mais
qui, oubliée, décale toute votre grille d’un demi-pixel, soit environ 230 mètres.

Les composites A3 et A4 embarquent en outre des jeux `lat` et `lon` explicites. La cellule ci-dessous
calcule les coordonnées, puis **les confronte** aux valeurs stockées : c’est une vérification que tout
notebook sérieux devrait faire une fois.

````python
# =============================================================================
#  ÉTAPE 10 — GÉORÉFÉRENCEMENT : DU PIXEL AUX COORDONNÉES
# =============================================================================
tuile_exemple = infos["tuile"]
ouest, est, sud, nord = emprise_tuile(tuile_exemple)
n_lignes, n_colonnes = tableau_complet.shape
resolution = TUILE_DEGRES / n_lignes


def coordonnees_pixel(ligne, colonne):
    """Centre du pixel (ligne, colonne) en degrés décimaux."""
    return (ouest + (colonne + 0.5) * resolution,
            nord - (ligne + 0.5) * resolution)


def pixel_depuis_coordonnees(longitude, latitude):
    """Opération inverse : quelle cellule contient ce point ?"""
    colonne = int((longitude - ouest) / resolution)
    ligne = int((nord - latitude) / resolution)
    if not (0 <= ligne < n_lignes and 0 <= colonne < n_colonnes):
        return None
    return ligne, colonne


# Axes complets (utiles pour l'affichage et l'export)
LONGITUDES = ouest + (np.arange(n_colonnes) + 0.5) * resolution
LATITUDES = nord - (np.arange(n_lignes) + 0.5) * resolution

# Emprise au format attendu par imshow : (gauche, droite, bas, haut)
EMPRISE_AFFICHAGE = (ouest, est, sud, nord)

# Transformation affine, au format GDAL / rasterio
TRANSFORMATION_AFFINE = (resolution, 0.0, ouest, 0.0, -resolution, nord)

print(f"TUILE {tuile_exemple}   ·   {n_lignes} × {n_colonnes} pixels\n")
print(f"  Coin nord-ouest   : longitude {ouest:+10.5f}°   latitude {nord:+9.5f}°")
print(f"  Coin sud-est      : longitude {est:+10.5f}°   latitude {sud:+9.5f}°")
print(f"  Résolution        : {resolution:.8f}°  =  {resolution * 3600:.2f}\" d'arc")
print(f"  Transformation    : (a={resolution:.8f}, b=0, c={ouest:.5f}, "
      f"d=0, e={-resolution:.8f}, f={nord:.5f})")

print("\nQUELQUES PIXELS DE CONTRÔLE")
controles = [(0, 0, "coin nord-ouest"),
             (0, n_colonnes - 1, "coin nord-est"),
             (n_lignes - 1, 0, "coin sud-ouest"),
             (n_lignes // 2, n_colonnes // 2, "centre de la tuile")]
table_controle = pd.DataFrame(
    [{"ligne": li_, "colonne": co_, "repère": nom_,
      "longitude": round(coordonnees_pixel(li_, co_)[0], 5),
      "latitude": round(coordonnees_pixel(li_, co_)[1], 5)}
     for li_, co_, nom_ in controles])
display(table_controle.set_index("repère"))

# --- Vérification contre les jeux lat / lon du fichier ------------------------
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    a_coordonnees = "lat" in groupe and "lon" in groupe
    if a_coordonnees:
        lat_fichier = groupe["lat"][:]
        lon_fichier = groupe["lon"][:]

if a_coordonnees:
    ecart_lat = float(np.max(np.abs(lat_fichier - LATITUDES)))
    ecart_lon = float(np.max(np.abs(lon_fichier - LONGITUDES)))
    print(f"\nVÉRIFICATION contre les jeux « lat » et « lon » du fichier")
    print(f"  écart maximal en latitude  : {ecart_lat:.3e}°  "
          f"({ecart_lat * 111320:.4f} m)")
    print(f"  écart maximal en longitude : {ecart_lon:.3e}°  "
          f"({ecart_lon * 111320:.4f} m)")
    verdict = "les deux méthodes concordent" if max(ecart_lat, ecart_lon) < 1e-6 \
              else "ÉCART SIGNIFICATIF — vérifiez la convention de centrage du pixel"
    print(f"  → {verdict}")
else:
    print("\n  Ce produit n'embarque pas de jeux « lat » / « lon » : "
          "le calcul ci-dessus est la seule source de géoréférencement.")

# --- Localiser une ville dans la grille ---------------------------------------
villes_tuile = FOYERS.get(tuile_exemple, [])
if villes_tuile:
    print("\nLOCALISATION DE QUELQUES VILLES DANS LA GRILLE")
    lignes_villes = []
    for nom_ville, vlon, vlat, _poids in villes_tuile:
        position = pixel_depuis_coordonnees(vlon, vlat)
        if position is None:
            continue
        li_v, co_v = position
        valeur = tableau_complet[li_v, co_v]
        radiance_ville = (np.nan if valeur == remplissage
                          else valeur * echelle + decalage)
        lignes_villes.append({"ville": nom_ville, "longitude": vlon, "latitude": vlat,
                              "ligne": li_v, "colonne": co_v,
                              "radiance": round(radiance_ville, 1)})
    display(pd.DataFrame(lignes_villes).set_index("ville"))
````

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Savoir à quels pixels faire confiance</div>

**Le principe.** La NASA attache à chaque pixel un petit code entier qui résume la confiance
qu’on peut lui accorder. Sa signification **dépend du produit**, et la bonne pratique consiste à lire le
codage dans l’attribut `description` du fichier plutôt que de le supposer.

Pour les composites (A3, A4) — couche `…_Quality` :

| code | signification | faut-il le garder ? |
|:---:|---|---|
| **0** | bonne qualité | oui, toujours |
| **1** | qualité médiocre | selon l’usage |
| **2** | valeur comblée (*gap-filled*) | à éviter pour un indicateur publié |
| **255** | valeur de remplissage | jamais |

Pour le journalier (A2) — couche `Mandatory_Quality_Flag` :

| code | signification | commentaire |
|:---:|---|---|
| **0** | haute qualité, lumières **persistantes** | le cœur du signal : villes, infrastructures |
| **1** | haute qualité, lumières **éphémères** | feux, torchères, pêche, activité saisonnière |
| **2** | qualité médiocre, valeur aberrante corrigée | à manier avec précaution |
| **255** | aucune restitution | nuage épais, lune trop forte, données absentes |

````python
# =============================================================================
#  ÉTAPE 11 — LES DRAPEAUX QUALITÉ
# =============================================================================

# Codages de secours, si le fichier ne porte pas d'attribut « description »
CODAGES_DE_SECOURS = {
    "composite": {0: "Bonne qualité", 1: "Qualité médiocre",
                  2: "Valeur comblée (gap-filled)", 255: "Valeur de remplissage"},
    "journalier": {0: "Haute qualité — lumières persistantes",
                   1: "Haute qualité — lumières éphémères",
                   2: "Qualité médiocre — aberration corrigée",
                   255: "Aucune restitution"},
    "terre_eau": {0: "Océan peu profond", 1: "Terre", 2: "Trait de côte",
                  3: "Eau intérieure peu profonde", 4: "Eau temporaire",
                  5: "Eau intérieure profonde", 6: "Continental", 7: "Océan profond"},
}


def analyser_description(texte: str) -> dict:
    """Extrait un dictionnaire {code: libellé} d'un attribut « description ».

    Les descriptions NASA sont de la forme :
        "0 = Good quality 1 = Poor quality 2 = Gap filled 255 = Fill value"
    Le regard en avant (?=...) permet de couper juste avant le code suivant.
    """
    if not texte:
        return {}
    motif = r"(\d+)\s*=\s*([^\n\t0-9][^\n\t]*?)(?=\s+\d+\s*=|$)"
    return {int(code): libelle.strip() for code, libelle in re.findall(motif, str(texte))}


def trouver_couches_drapeaux(chemin) -> list:
    """Liste les couches de drapeaux présentes, dans un ordre de lecture utile."""
    trouvees = []
    with h5py.File(chemin, "r") as f:
        groupe = f[trouver_groupe_donnees(f)]
        for nom, objet in groupe.items():
            if not (isinstance(objet, h5py.Dataset) and objet.ndim == 2):
                continue
            if (nom.endswith("_Quality") or "Quality_Flag" in nom
                    or nom in ("Land_Water_Mask", "QF_Cloud_Mask", "Snow_Flag", "DNB_Platform")):
                trouvees.append(nom)
    priorite = {"Quality": 0, "Land_Water_Mask": 1, "QF_Cloud_Mask": 2}
    return sorted(trouvees, key=lambda n: (0 if "Quality" in n else priorite.get(n, 3), n))


couches_drapeaux = trouver_couches_drapeaux(FICHIER_EXEMPLE)
print(f"Couches de drapeaux détectées : {', '.join(couches_drapeaux)}\n")

# --- Analyse de la couche qualité principale ---------------------------------
nom_qualite = next((n for n in couches_drapeaux if "Quality" in n), None)

if nom_qualite is None:
    print("Aucune couche de qualité dans ce fichier.")
    drapeau_qualite, legende_qualite = None, {}
else:
    with h5py.File(FICHIER_EXEMPLE, "r") as f:
        objet = f[trouver_groupe_donnees(f)][nom_qualite]
        drapeau_qualite = objet[:]
        description = lire_attribut(objet, "description", "") or lire_attribut(objet, "long_name", "")

    legende_qualite = analyser_description(description)
    if not legende_qualite:
        famille = "journalier" if infos["produit"] == "VNP46A2" else "composite"
        legende_qualite = CODAGES_DE_SECOURS[famille]
        print("  (attribut « description » absent — codage de secours appliqué)")

    print(f"COUCHE : {nom_qualite}")
    print(f"description : {str(description)[:110]}\n")

    codes, effectifs = np.unique(drapeau_qualite, return_counts=True)
    repartition = pd.DataFrame({
        "code": codes,
        "signification": [legende_qualite.get(int(k), "code non documenté") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / drapeau_qualite.size, 2),
        "retenu (QA_MAX=%d)" % QA_MAX: ["oui" if int(k) <= QA_MAX else "non" for k in codes],
    })
    display(repartition.set_index("code"))

    conserves = int(np.sum(drapeau_qualite <= QA_MAX))
    print(f"\nAvec QA_MAX = {QA_MAX} : {conserves:,} pixels conservés sur "
          f"{drapeau_qualite.size:,} ({100 * conserves / drapeau_qualite.size:.1f} %)"
          .replace(",", " "))
````

**Deux masques, deux logiques très différentes.**

`Land_Water_Mask` est simple : un code de 0 à 7 par pixel, chaque code désignant un type de surface.
C’est ce qui permet d’exclure l’océan — indispensable, car la mer n’est pas noire : reflets lunaires,
navires et pêcheries y produisent un signal qui gonflerait artificiellement toute somme des lumières côtière.

`QF_Cloud_Mask` (produit journalier A2) est différent : **un seul entier de 16 bits y encode une dizaine
d’informations**, chacune occupant un ou plusieurs bits. Lire ce nombre comme un entier ordinaire n’a
aucun sens ; il faut en **extraire des tranches de bits**. La technique, appelée *masquage binaire*,
tient en une ligne :

```python
valeur_du_champ = (entier >> position_du_premier_bit) & ((1 << nombre_de_bits) - 1)
```

On décale l’entier vers la droite pour amener le champ voulu tout à droite, puis on efface tout ce qui
dépasse. Les principaux champs du `QF_Cloud_Mask` VIIRS :

````python
# =============================================================================
#  ÉTAPE 12 — LES MASQUES : TERRE / EAU, PUIS NUAGES BIT À BIT
# =============================================================================

# --- 1. Le masque terre / eau ------------------------------------------------
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    a_terre_eau = "Land_Water_Mask" in groupe
    if a_terre_eau:
        objet = groupe["Land_Water_Mask"]
        masque_terre_eau = objet[:]
        desc_terre_eau = lire_attribut(objet, "description", "")

if a_terre_eau:
    legende_terre_eau = analyser_description(desc_terre_eau) or CODAGES_DE_SECOURS["terre_eau"]
    codes, effectifs = np.unique(masque_terre_eau, return_counts=True)
    print("MASQUE TERRE / EAU  —  Land_Water_Mask\n")
    display(pd.DataFrame({
        "code": codes,
        "type de surface": [legende_terre_eau.get(int(k), "non documenté") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / masque_terre_eau.size, 2),
    }).set_index("code"))

    # Codes considérés comme de l'eau à exclure (océan et eau profonde)
    CODES_EAU = [0, 7]
    part_eau = 100 * np.mean(np.isin(masque_terre_eau, CODES_EAU))
    print(f"\nAvec MASQUER_EAU = {MASQUER_EAU}, les codes {CODES_EAU} seraient écartés "
          f"({part_eau:.1f} % de la tuile).")
else:
    masque_terre_eau = None
    print("Ce produit ne contient pas de couche Land_Water_Mask.")


# --- 2. Le masque nuage, lu bit à bit ----------------------------------------
CHAMPS_NUAGE = {
    "Jour / nuit":                (0, 1, {0: "Nuit", 1: "Jour"}),
    "Fond terre / eau":           (1, 3, {0: "Terre et désert", 1: "Terre hors désert",
                                          2: "Eau intérieure", 3: "Eau de mer",
                                          5: "Zone côtière"}),
    "Qualité du masque nuage":    (4, 2, {0: "Médiocre", 1: "Faible",
                                          2: "Moyenne", 3: "Élevée"}),
    "Détection de nuage":         (6, 2, {0: "Clair (certain)", 1: "Probablement clair",
                                          2: "Probablement nuageux", 3: "Nuageux (certain)"}),
    "Ombre de nuage":             (8, 1, {0: "Non", 1: "Oui"}),
    "Cirrus":                     (9, 1, {0: "Non", 1: "Oui"}),
}


def extraire_bits(entier, premier_bit, nombre_de_bits):
    """Extrait un champ de bits d'un entier (ou d'un tableau d'entiers)."""
    return (np.asarray(entier) >> premier_bit) & ((1 << nombre_de_bits) - 1)


# On cherche un fichier journalier, seul porteur de QF_Cloud_Mask
journaliers = INVENTAIRE[INVENTAIRE["produit"] == "VNP46A2"]
if journaliers.empty:
    print("\nAucun fichier VNP46A2 dans l'inventaire : "
          "démonstration du décodage binaire sur un entier fabriqué.\n")
    exemple_entier = np.uint16(0b0000_0010_1000_0010)
    tableau_nuage = None
else:
    fichier_journalier = Path(journaliers.iloc[-1]["chemin"])
    with h5py.File(fichier_journalier, "r") as f:
        groupe_j = f[trouver_groupe_donnees(f)]
        tableau_nuage = groupe_j["QF_Cloud_Mask"][:] if "QF_Cloud_Mask" in groupe_j else None
    print(f"\nMASQUE NUAGE  —  QF_Cloud_Mask   ({fichier_journalier.name})\n")
    exemple_entier = (np.uint16(tableau_nuage.flat[0]) if tableau_nuage is not None
                      else np.uint16(0b0000_0010_1000_0010))

# Décodage d'un entier, champ par champ
print(f"Exemple : l'entier {int(exemple_entier)} vaut en binaire "
      f"{int(exemple_entier):016b}\n")
lignes_bits = []
for nom_champ, (premier, largeur, sens) in CHAMPS_NUAGE.items():
    valeur = int(extraire_bits(exemple_entier, premier, largeur))
    position = (f"bit {premier}" if largeur == 1
                else f"bits {premier}–{premier + largeur - 1}")
    lignes_bits.append({"champ": nom_champ, "position": position,
                        "bits": format(valeur, f"0{largeur}b"),
                        "valeur": valeur,
                        "signification": sens.get(valeur, "code non documenté")})
display(pd.DataFrame(lignes_bits).set_index("champ"))

# Statistiques sur toute la tuile
if tableau_nuage is not None:
    detection = extraire_bits(tableau_nuage, 6, 2)
    sens_detection = CHAMPS_NUAGE["Détection de nuage"][2]
    codes, effectifs = np.unique(detection, return_counts=True)
    print("\nRÉPARTITION DE LA DÉTECTION DE NUAGE SUR TOUTE LA TUILE")
    display(pd.DataFrame({
        "code": codes,
        "signification": [sens_detection.get(int(k), "?") for k in codes],
        "pixels": effectifs,
        "part_%": np.round(100 * effectifs / detection.size, 2),
    }).set_index("code"))
    part_nuageuse = 100 * np.mean(detection >= 2)
    print(f"\nNuit analysée : {part_nuageuse:.1f} % de la tuile est probablement ou "
          f"certainement nuageuse.")
    if part_nuageuse > 40:
        print("→ Au-delà de 40 %, une conclusion tirée de cette seule nuit serait imprudente.")
````

**Ce que fait cette étape.** Tout ce que nous avons démonté séparément se remonte ici en une
seule fonction, `lire_ntl()`. Elle applique **dans le bon ordre** les six opérations, et cet ordre n’est
pas négociable :

1. ouvrir le fichier et **trouver** le groupe de données (ne jamais le supposer) ;
2. **choisir** la couche de radiance ;
3. lire les attributs et **décoder** : `valeur × scale_factor + add_offset` ;
4. remplacer le **`_FillValue` par `NaN`** — avant tout calcul ;
5. écarter les pixels dont le **drapeau qualité** dépasse le seuil ;
6. écarter les pixels d’**eau**, puis calculer le géoréférencement.

Inverser les étapes 3 et 4 — décoder avant de neutraliser le `_FillValue` — donne exactement le désastre
chiffré à l’étape 09. La fonction renvoie un dictionnaire contenant la radiance, l’emprise, la
transformation affine et un **journal de traitement** : le nombre de pixels perdus à chaque filtre.

<div style="background:#E8F5EF;border:1px solid #00704A44;border-left:4px solid #00704A;border-radius:8px;
 padding:13px 17px;margin:10px 0;font-family:'Segoe UI',Calibri,system-ui,-apple-system,sans-serif;">
 <div style="color:#00704A;font-weight:700;font-size:13.5px;">&#128273;&nbsp; Pourquoi un journal de traitement ?</div>
 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;">Parce qu’un indicateur sans traçabilité n’est pas publiable. Le champ <code>journal</code> retourné par <code>lire_ntl()</code> vous dit combien de pixels ont été écartés, et par quel filtre. Copié dans vos métadonnées, il répond d’avance à la question que posera le premier relecteur attentif : « sur quoi exactement ce chiffre a-t-il été calculé ? »</div>

````python
# =============================================================================
#  ÉTAPE 13 — LA FONCTION DE LECTURE PROPRE
# =============================================================================
CODES_EAU_A_EXCLURE = [0, 7]      # océan peu profond, océan profond


def _sous_echantillonner(tableau, facteur):
    """Moyenne des blocs facteur × facteur, en ignorant les NaN."""
    if facteur is None or facteur <= 1:
        return tableau
    h, w = tableau.shape
    h2, w2 = (h // facteur) * facteur, (w // facteur) * facteur
    decoupe = tableau[:h2, :w2].reshape(h2 // facteur, facteur, w2 // facteur, facteur)
    return np.nanmean(decoupe, axis=(1, 3))


def lire_ntl(chemin, couche=None, qa_max=None, masquer_eau=None,
             sous_echantillonnage=1, verbeux=False):
    """Lit un fichier VNP46 et renvoie une radiance propre et géoréférencée.

    Paramètres
    ----------
    chemin : str ou Path
        Le fichier .h5 à lire.
    couche : str, optionnel
        Nom de la couche de radiance. Par défaut, choisie automatiquement.
    qa_max : int, optionnel
        Drapeau qualité maximal accepté. Par défaut, la valeur de QA_MAX.
    masquer_eau : bool, optionnel
        Écarter les pixels d'océan. Par défaut, la valeur de MASQUER_EAU.
    sous_echantillonnage : int
        Facteur d'agrégation pour l'affichage (1 = pleine résolution).
    verbeux : bool
        Afficher le journal de traitement au fur et à mesure.

    Retour
    ------
    dict avec les clés : radiance, emprise, transformation, longitudes,
    latitudes, couche, metadonnees, journal.
    """
    chemin = Path(chemin)
    qa_max = QA_MAX if qa_max is None else qa_max
    masquer_eau = MASQUER_EAU if masquer_eau is None else masquer_eau

    infos_nom = decoder_nom_fichier(chemin.name)
    if infos_nom is None:
        raise ValueError(f"Nom de fichier non conforme à la convention VNP46 : {chemin.name}")

    journal = []

    with h5py.File(chemin, "r") as f:
        # --- 1. trouver le groupe, ne jamais le supposer ---------------------
        groupe = f[trouver_groupe_donnees(f)]

        # --- 2. choisir la couche de radiance --------------------------------
        nom_couche = couche or choisir_couche_radiance(groupe)
        jeu = groupe[nom_couche]

        # --- 3. lire les attributs DANS le fichier ---------------------------
        facteur = lire_attribut(jeu, "scale_factor", 1.0)
        offset = lire_attribut(jeu, "add_offset", lire_attribut(jeu, "offset", 0.0))
        vide = lire_attribut(jeu, "_FillValue", None)
        unite = lire_attribut(jeu, "units", "")

        brut = jeu[:].astype("float32")
        total = brut.size
        journal.append(("pixels du fichier", total, 100.0))

        # --- 4. neutraliser le _FillValue AVANT de décoder -------------------
        invalide = ~np.isfinite(brut)
        if vide is not None:
            invalide |= np.isclose(brut, float(vide))
        radiance = brut * facteur + offset
        radiance[invalide] = np.nan
        radiance[radiance < 0] = np.nan          # une radiance négative n'existe pas
        perdus = int(invalide.sum())
        journal.append(("écartés : _FillValue", -perdus, -100.0 * perdus / total))

        # --- 5. filtrer sur le drapeau qualité -------------------------------
        nom_drapeau = next((n for n in (f"{nom_couche}_Quality", "Mandatory_Quality_Flag")
                            if n in groupe), None)
        if nom_drapeau is not None and qa_max is not None:
            drapeau = groupe[nom_drapeau][:]
            a_ecarter = (drapeau > qa_max) & np.isfinite(radiance)
            radiance[a_ecarter] = np.nan
            journal.append((f"écartés : qualité > {qa_max}", -int(a_ecarter.sum()),
                            -100.0 * int(a_ecarter.sum()) / total))

        # --- 6. masquer l'eau -------------------------------------------------
        if masquer_eau and "Land_Water_Mask" in groupe:
            terre_eau = groupe["Land_Water_Mask"][:]
            a_ecarter = np.isin(terre_eau, CODES_EAU_A_EXCLURE) & np.isfinite(radiance)
            radiance[a_ecarter] = np.nan
            journal.append(("écartés : eau", -int(a_ecarter.sum()),
                            -100.0 * int(a_ecarter.sum()) / total))

        metadonnees = {"produit": infos_nom["produit"], "tuile": infos_nom["tuile"],
                       "date": infos_nom["date"].date(), "couche": nom_couche,
                       "scale_factor": facteur, "add_offset": offset,
                       "_FillValue": vide, "unite": unite,
                       "drapeau_qualite": nom_drapeau, "qa_max": qa_max,
                       "eau_masquee": bool(masquer_eau), "fichier": chemin.name}

    # --- 7. géoréférencement ---------------------------------------------------
    o, e, s, n = emprise_tuile(infos_nom["tuile"])
    if sous_echantillonnage and sous_echantillonnage > 1:
        radiance = _sous_echantillonner(radiance, sous_echantillonnage)
    lignes, colonnes = radiance.shape
    res_lon, res_lat = (e - o) / colonnes, (n - s) / lignes

    journal.append(("pixels exploitables", int(np.isfinite(radiance).sum()),
                    100.0 * float(np.isfinite(radiance).mean())))

    if verbeux:
        print(f"{chemin.name}")
        for etiquette, valeur, part in journal:
            print(f"   {etiquette:<28} {valeur:>12,}   {part:>7.2f} %".replace(",", " "))

    return {"radiance": radiance,
            "emprise": (o, e, s, n),
            "transformation": (res_lon, 0.0, o, 0.0, -res_lat, n),
            "longitudes": o + (np.arange(colonnes) + 0.5) * res_lon,
            "latitudes": n - (np.arange(lignes) + 0.5) * res_lat,
            "couche": metadonnees["couche"],
            "metadonnees": metadonnees,
            "journal": pd.DataFrame(journal, columns=["étape", "pixels", "part_%"])}


# --- Mise en oeuvre -----------------------------------------------------------
RESULTAT = lire_ntl(FICHIER_EXEMPLE, sous_echantillonnage=SOUS_ECHANTILLONNAGE)

print("JOURNAL DE TRAITEMENT")
display(RESULTAT["journal"].set_index("étape"))

print("\nMÉTADONNÉES DE SORTIE — à recopier dans votre documentation")
display(pd.Series(RESULTAT["metadonnees"], name="valeur").to_frame())

radiance = RESULTAT["radiance"]
valides = radiance[np.isfinite(radiance)]
print(f"\nRadiance : {valides.size:,} pixels exploitables".replace(",", " "))
print(f"   minimum {valides.min():.3f}   médiane {np.median(valides):.3f}   "
      f"maximum {valides.max():.1f}   (nW·cm⁻²·sr⁻¹)")
print(f"   somme des lumières (SoL) : {valides.sum():,.0f}".replace(",", " "))
````

**Ce qu’il faut absolument comprendre.** La radiance nocturne n’est pas distribuée
normalement. Elle est **extrêmement asymétrique** : l’immense majorité des pixels sont très sombres, et
une poignée de pixels urbains concentre l’essentiel de la lumière totale. Il n’est pas rare que **1 % des
pixels porte plus de la moitié de la somme des lumières**.

Trois conséquences pratiques, qui décident de la qualité de votre analyse :

- **Cartographier en échelle linéaire ne montre rien** : la carte est noire, avec trois points blancs.
  Il faut passer en **échelle logarithmique** — c’est pourquoi toutes les cartes de ce notebook
  utilisent `log1p`.
- **La moyenne est un mauvais résumé.** Elle est dominée par quelques pixels. Pour une comparaison
  entre régions, la **somme** (SoL) et la **médiane** disent deux choses différentes, toutes deux utiles.
- **Un seul pixel aberrant peut déplacer un indicateur national.** Une torchère de gaz, un incendie,
  un navire : cartographiez toujours vos valeurs extrêmes avant de publier.

````python
# =============================================================================
#  ÉTAPE 14 — LA DISTRIBUTION DES RADIANCES
# =============================================================================
valides = radiance[np.isfinite(radiance)]
positifs = valides[valides > 0]

# --- Statistiques descriptives ------------------------------------------------
quantiles = [1, 5, 25, 50, 75, 90, 95, 99, 99.9]
resume = pd.DataFrame({
    "statistique": (["pixels exploitables", "moyenne", "écart-type", "minimum"]
                    + [f"centile {q}" for q in quantiles] + ["maximum", "somme (SoL)"]),
    "valeur": ([f"{valides.size:,}".replace(",", " "),
                f"{valides.mean():.3f}", f"{valides.std():.3f}", f"{valides.min():.3f}"]
               + [f"{np.percentile(valides, q):.3f}" for q in quantiles]
               + [f"{valides.max():.1f}", f"{valides.sum():,.0f}".replace(",", " ")]),
})
display(resume.set_index("statistique"))

# --- La concentration : quelle part de lumière pour quelle part de pixels ? ----
tries = np.sort(valides)[::-1]
cumul = np.cumsum(tries) / tries.sum()
concentration = pd.DataFrame({
    "part des pixels les plus lumineux": ["0,1 %", "1 %", "5 %", "10 %", "25 %", "50 %"],
    "part de la lumière totale": [
        f"{100 * cumul[max(int(p * tries.size / 100) - 1, 0)]:.1f} %"
        for p in (0.1, 1, 5, 10, 25, 50)],
})
print("\nCONCENTRATION DE LA LUMIÈRE")
display(concentration.set_index("part des pixels les plus lumineux"))

seuil_moitie = 100 * (np.searchsorted(cumul, 0.5) + 1) / tries.size
print(f"\nIl suffit des {seuil_moitie:.2f} % de pixels les plus lumineux pour "
      f"atteindre la moitié de la somme des lumières.")
````

 <div style="color:#231F20;font-size:14px;margin-top:4px;">Pourquoi le fichier contient plusieurs versions de la même image</div>

**Ce que révèle cette comparaison.** Un composite propose plusieurs vues de la même scène :
`AllAngle` (toutes les géométries de visée confondues), `NearNadir` (uniquement les passages proches de la
verticale), `OffNadir` (les passages obliques). Elles ne donnent pas exactement le même chiffre.

À cela s’ajoutent les deux couches d’incertitude, qui sont celles que l’on oublie le plus :
`…_Num` dit **sur combien de nuits** le composite a été calculé, et `…_Std` **à quel point ces nuits
divergeaient**. Un pixel à forte radiance et fort écart-type, calculé sur trois nuits, n’a pas la même
valeur probante qu’un pixel stable calculé sur vingt-cinq nuits — et pourtant les deux se ressemblent
sur une carte de radiance.

````python
# =============================================================================
#  ÉTAPE 15 — COMPARER LES COUCHES ENTRE ELLES
# =============================================================================
with h5py.File(FICHIER_EXEMPLE, "r") as f:
    groupe = f[trouver_groupe_donnees(f)]
    disponibles = {nom: obj for nom, obj in groupe.items()
                   if isinstance(obj, h5py.Dataset) and obj.ndim == 2}

    # (1) les variantes de radiance
    variantes = [n for n in disponibles
                 if ("Composite" in n or "NTL" in n)
                 and not n.endswith(("_Num", "_Std", "_Quality"))]

    lignes = []
    tableaux = {}
    for nom in variantes:
        objet = groupe[nom]
        fac = lire_attribut(objet, "scale_factor", 1.0)
        off = lire_attribut(objet, "add_offset", 0.0)
        vid = lire_attribut(objet, "_FillValue", None)
        valeurs = objet[:].astype("float32")
        valeurs = np.where(np.isclose(valeurs, float(vid)) if vid is not None else False,
                           np.nan, valeurs * fac + off)
        tableaux[nom] = valeurs
        ok = valeurs[np.isfinite(valeurs)]
        lignes.append({"couche": nom, "pixels_valides": ok.size,
                       "moyenne": round(float(ok.mean()), 3) if ok.size else np.nan,
                       "médiane": round(float(np.median(ok)), 3) if ok.size else np.nan,
                       "maximum": round(float(ok.max()), 1) if ok.size else np.nan,
                       "somme (SoL)": round(float(ok.sum()), 0) if ok.size else np.nan})

    # (2) les couches d'incertitude
    nom_num = next((n for n in disponibles if n.endswith("_Num")), None)
    nom_std = next((n for n in disponibles if n.endswith("_Std")), None)
    table_num = groupe[nom_num][:].astype("float32") if nom_num else None
    if nom_std is not None:
        objet_std = groupe[nom_std]
        f_std = lire_attribut(objet_std, "scale_factor", 1.0)
        v_std = lire_attribut(objet_std, "_FillValue", None)
        table_std = objet_std[:].astype("float32")
        table_std = np.where(np.isclose(table_std, float(v_std)) if v_std is not None else False,
                             np.nan, table_std * f_std)
    else:
        table_std = None

print("LES VARIANTES DE RADIANCE PRÉSENTES DANS LE FICHIER\n")
comparaison = pd.DataFrame(lignes).sort_values("somme (SoL)", ascending=False)
display(comparaison.set_index("couche"))

# Écart relatif entre la couche de référence et les autres
reference = RESULTAT["couche"]
if reference in tableaux and len(tableaux) > 1:
    base = np.nansum(tableaux[reference])
    print(f"\nÉCART DE SOMME DES LUMIÈRES PAR RAPPORT À « {reference} »")
    for nom, valeurs in tableaux.items():
        if nom == reference:
            continue
        ecart = 100 * (np.nansum(valeurs) - base) / base if base else np.nan
        print(f"   {nom:<44} {ecart:+7.2f} %")

if nom_num:
    fini = table_num[np.isfinite(table_num) & (table_num < 255)]
    print(f"\nNOMBRE D'OBSERVATIONS  ({nom_num})")
    print(f"   médiane {np.median(fini):.0f} nuits   ·   "
          f"minimum {fini.min():.0f}   ·   maximum {fini.max():.0f}")
    faibles = 100 * np.mean(fini <= 5)
    print(f"   {faibles:.1f} % des pixels reposent sur 5 nuits ou moins")
````

plante pas toujours : parfois il tourne, et produit des chiffres faux.

La cellule ci-dessous ouvre un fichier de chaque produit disponible et dresse le comparatif.

````python
# =============================================================================
#  ÉTAPE 16 — COMPARER LES PRODUITS ENTRE EUX
# =============================================================================
lignes = []
apercus = []

for produit in ["VNP46A4", "VNP46A3", "VNP46A2", "VNP46A1"]:
    sous = INVENTAIRE[INVENTAIRE["produit"] == produit]
    if sous.empty:
        continue
    chemin = Path(sous.iloc[-1]["chemin"])
    with h5py.File(chemin, "r") as f:
        chemin_groupe = trouver_groupe_donnees(f)
        groupe = f[chemin_groupe]
        jeux = {n: o for n, o in groupe.items() if isinstance(o, h5py.Dataset)}
        couche = choisir_couche_radiance(groupe)
        jeu = groupe[couche]
        nom_drapeau = next((n for n in (f"{couche}_Quality", "Mandatory_Quality_Flag")
                            if n in groupe), "—")
        lignes.append({
            "produit": produit,
            "famille": FAMILLE[produit],
            "groupe HDF5": chemin_groupe.replace("HDFEOS/GRIDS/", "…/").replace("/Data Fields", ""),
            "couches": len(jeux),
            "couche de radiance": couche,
            "type": str(jeu.dtype),
            "scale_factor": lire_attribut(jeu, "scale_factor", ""),
            "_FillValue": lire_attribut(jeu, "_FillValue", ""),
            "couche qualité": nom_drapeau,
            "forme": "×".join(str(d) for d in jeu.shape),
        })
    try:
        apercus.append((produit, lire_ntl(chemin, sous_echantillonnage=SOUS_ECHANTILLONNAGE)))
    except Exception as erreur:
        print(f"  {produit} : lecture impossible ({erreur})")

COMPARATIF = pd.DataFrame(lignes).set_index("produit")
print("COMPARATIF STRUCTUREL DES PRODUITS PRÉSENTS DANS VOS DONNÉES\n")
display(COMPARATIF.T)

if len(apercus) > 1:
    print("\nCE QUE CHAQUE PRODUIT DONNE SUR LA MÊME ZONE\n")
    resume = []
    for produit, res in apercus:
        v = res["radiance"][np.isfinite(res["radiance"])]
        resume.append({"produit": produit, "date": res["metadonnees"]["date"],
                       "tuile": res["metadonnees"]["tuile"],
                       "pixels exploitables": f"{v.size:,}".replace(",", " "),
                       "couverture_%": round(100 * v.size / res["radiance"].size, 1),
                       "médiane": round(float(np.median(v)), 3) if v.size else np.nan,
                       "SoL": f"{v.sum():,.0f}".replace(",", " ") if v.size else "—"})
    display(pd.DataFrame(resume).set_index("produit"))
````

**Pourquoi exporter.** Le notebook est un lieu d’exploration, pas un lieu de stockage. Le
produit de votre travail doit pouvoir être ouvert par un collègue sous QGIS, joint à un rapport, ou versé
dans un dépôt GitHub. Trois formats couvrent l’essentiel des besoins :

- **GeoTIFF** — le format universel des SIG. Il embarque le géoréférencement : QGIS et ArcGIS le placent
  au bon endroit sans aucun réglage. C’est le format à privilégier pour tout partage.
- **CSV** — un pixel par ligne, avec sa longitude, sa latitude et sa radiance. Lourd, mais compréhensible
  par tout le monde et joignable à des données administratives dans un tableur.
- **NPZ** — l’archive compressée de NumPy : rapide à relire en Python, idéale comme cache intermédiaire
  entre deux étapes d’un traitement.

Si `rasterio` n’est pas installé, la cellule produit tout de même un fichier `.tif` accompagné d’un
**fichier de calage** (`.tfw`) et d’un `.prj` : la plupart des SIG le géoréférenceront correctement.

````python
# =============================================================================
#  ÉTAPE 17 — EXPORTER
# =============================================================================
DOSSIER_SORTIES = Path.cwd() / "sorties_ntl"
DOSSIER_SORTIES.mkdir(exist_ok=True)

meta = RESULTAT["metadonnees"]
base = f"{meta['produit']}_{meta['tuile']}_{meta['date']}"
if MODE_DEMO:
    base = "DEMO_" + base
exportes = []

# --- 1. GeoTIFF ---------------------------------------------------------------
chemin_tif = DOSSIER_SORTIES / f"{base}_radiance.tif"
res_lon, _, ouest_e, _, res_lat_neg, nord_e = RESULTAT["transformation"]

if DISPO.get("rasterio"):
    import rasterio
    from rasterio.transform import Affine

    transformation = Affine(res_lon, 0.0, ouest_e, 0.0, res_lat_neg, nord_e)
    with rasterio.open(
            chemin_tif, "w", driver="GTiff",
            height=radiance.shape[0], width=radiance.shape[1], count=1,
            dtype="float32", crs="EPSG:4326", transform=transformation,
            nodata=np.nan, compress="deflate", tiled=True) as sortie:
        sortie.write(radiance.astype("float32"), 1)
        sortie.update_tags(**{k: str(v) for k, v in meta.items()})
    exportes.append(("GeoTIFF géoréférencé (rasterio)", chemin_tif))
else:
    # Repli sans rasterio : TIFF simple + fichier de calage + projection
    try:
        from PIL import Image
        Image.fromarray(radiance.astype("float32")).save(chemin_tif)
        (DOSSIER_SORTIES / f"{base}_radiance.tfw").write_text(
            f"{res_lon:.12f}\n0.0\n0.0\n{res_lat_neg:.12f}\n"
            f"{ouest_e + res_lon / 2:.12f}\n{nord_e + res_lat_neg / 2:.12f}\n")
        (DOSSIER_SORTIES / f"{base}_radiance.prj").write_text(
            'GEOGCS["WGS 84",DATUM["WGS_1984",SPHEROID["WGS 84",6378137,298.257223563]],'
            'PRIMEM["Greenwich",0],UNIT["degree",0.0174532925199433]]')
        exportes.append(("TIFF + calage .tfw + .prj (sans rasterio)", chemin_tif))
    except Exception as erreur:
        print(f"  export TIFF impossible : {erreur}")

# --- 2. CSV : un pixel par ligne ---------------------------------------------
pas = max(radiance.shape[0] // 220, 1)          # allégé pour rester lisible
sous_radiance = radiance[::pas, ::pas]
LON, LAT = np.meshgrid(RESULTAT["longitudes"][::pas], RESULTAT["latitudes"][::pas])
table_pixels = pd.DataFrame({
    "longitude": np.round(LON.ravel(), 6),
    "latitude": np.round(LAT.ravel(), 6),
    "radiance": np.round(sous_radiance.ravel(), 4),
}).dropna(subset=["radiance"])

chemin_csv = DOSSIER_SORTIES / f"{base}_pixels.csv"
table_pixels.to_csv(chemin_csv, index=False, encoding="utf-8")
exportes.append((f"CSV — {len(table_pixels):,} pixels (1 sur {pas})".replace(",", " "), chemin_csv))

# --- 3. Archive NumPy ---------------------------------------------------------
chemin_npz = DOSSIER_SORTIES / f"{base}_radiance.npz"
np.savez_compressed(chemin_npz, radiance=radiance.astype("float32"),
                    longitudes=RESULTAT["longitudes"], latitudes=RESULTAT["latitudes"],
                    emprise=np.array(RESULTAT["emprise"]))
exportes.append(("Archive NumPy compressée", chemin_npz))

# --- 4. Métadonnées lisibles --------------------------------------------------
chemin_json = DOSSIER_SORTIES / f"{base}_metadonnees.json"
documentation = {
    "source": "NASA Black Marble — suite VNP46, collection 2 (version 002)",
    "avertissement": ("VALEURS SIMULÉES — jeu de démonstration, ne pas utiliser pour produire "
                      "un chiffre" if MODE_DEMO else "Données NASA réelles"),
    "traitement": {k: str(v) for k, v in meta.items()},
    "emprise_WGS84": {"ouest": RESULTAT["emprise"][0], "est": RESULTAT["emprise"][1],
                      "sud": RESULTAT["emprise"][2], "nord": RESULTAT["emprise"][3]},
    "systeme_de_coordonnees": "EPSG:4326 (WGS 84)",
    "transformation_affine": list(RESULTAT["transformation"]),
    "dimensions": {"lignes": int(radiance.shape[0]), "colonnes": int(radiance.shape[1])},
    "sous_echantillonnage": SOUS_ECHANTILLONNAGE,
    "journal_de_traitement": RESULTAT["journal"].to_dict(orient="records"),
    "citation": ("Román, M. O. et al. (2018). NASA's Black Marble nighttime lights product "
                 "suite. Remote Sensing of Environment, 210, 113–143."),
}
chemin_json.write_text(json.dumps(documentation, indent=2, ensure_ascii=False,
                                  default=str), encoding="utf-8")
exportes.append(("Métadonnées JSON", chemin_json))

# --- Restitution --------------------------------------------------------------
print("FICHIERS ÉCRITS")
print("=" * 78)
for etiquette, chemin in exportes:
    print(f"  {etiquette}")
    print(f"     {chemin}   ({chemin.stat().st_size / 1e3:.0f} Ko)")
print("=" * 78)
print(f"\nDossier de sortie : {DOSSIER_SORTIES}")
if not DISPO.get("rasterio"):
    print("\n  (rasterio absent : le GeoTIFF a été produit avec un fichier de calage .tfw.")
    print("   Pour un GeoTIFF pleinement conforme :  pip install rasterio)")

print("\nAPERÇU DU CSV")
display(table_pixels.head(8))
````

 <div style="color:#231F20;font-size:13.5px;margin-top:5px;line-height:1.6;"><b>Halo urbain (<i>blooming</i>)</b> — la lumière déborde au-delà du bâti et gonfle les surfaces « éclairées ». &nbsp;<b>Saturation</b> — les cœurs urbains les plus intenses plafonnent, et une croissance y devient invisible. &nbsp;<b>Torchères de gaz</b> — un site pétrolier peut dominer la radiance d’une région entière sans rapport avec son activité économique. &nbsp;<b>Saisonnalité</b> — nuages et feux de brousse créent des cycles qui ne sont pas économiques. &nbsp;<b>Rupture de capteur</b> — l’arrivée de NOAA-20 aux côtés de Suomi-NPP peut introduire un décrochage dans une série longue. &nbsp;<b>Bruit en zone rurale peu éclairée</b> — près du seuil de détection, le rapport signal/bruit s’effondre : c’est précisément là que vivent les populations que l’indicateur prétend décrire.</div>
