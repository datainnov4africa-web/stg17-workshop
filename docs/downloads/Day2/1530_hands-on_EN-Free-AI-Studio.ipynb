## 1 · Purpose

Generative AI is now available to every statistics office at no licence cost — but **no single model is best at everything**, and free services differ in speed, languages, modalities, limits and data terms. This lab answers three practical questions:

1. **Which service fits which task** in our statistical production?
2. **How good is it on our kind of work** — measured, not assumed?
3. **How do we control the output** before it reaches users?

Each mission is a realistic NSO task, run on a free service chosen for its **signature strength**, and closed by an **automatic verification** recorded in a central register. At the end, the notebook writes an **executive report** you can share with management.

### 🎓 Learning outcomes
By the end, participants can:
- **Select** an appropriate AI service for a task, and justify the choice with evidence (speed, accuracy, cost, data terms).
- **Evaluate** AI outputs with quantitative checks: accuracy against a gold standard, automation rate, word error rate, number preservation, citation validity, test cases.
- **Apply** safeguards: pre-send screening of personal data, local processing for sensitive material, labelling of synthetic media.
- **Produce** communication products (dashboard, explainer video, audio briefing, visual) with every figure traceable to source data.

````python
# @title S4 · Core library: providers, chat(), privacy guard, check register, scoreboard: providers, one chat() for all, scoreboard, house style
import re, json, time, io, math, base64, random, shutil, wave, struct, textwrap, datetime, asyncio, urllib.parse
import concurrent.futures
from dataclasses import dataclass, field
import requests
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw, ImageFont, ImageFilter
from IPython.display import display, Markdown, HTML, Audio, Image as IPImage, Video

# ---------- house style (AfDB-inspired) ----------
AFDB = dict(green="#00A86A", deep="#00704A", forest="#00553A", gold="#F5C242", ochre="#D49A00", teal="#0E7C86",
            terra="#C4621D", brick="#B83B2E", ink="#231F20", slate="#5E6964", mist="#F4F7F5", mint="#E8F5EF", sage="#D5DED9")
import logging
from matplotlib import font_manager
logging.getLogger("matplotlib.font_manager").setLevel(logging.ERROR)
_fonts = {f.name for f in font_manager.fontManager.ttflist}
plt.rcParams.update({"font.family": [f for f in ("Calibri", "Carlito", "Liberation Sans", "DejaVu Sans") if f in _fonts] or ["sans-serif"],
    "axes.spines.top": False, "axes.spines.right": False, "axes.edgecolor": AFDB["sage"], "axes.titleweight": "bold",
    "axes.titlecolor": AFDB["ink"], "axes.labelcolor": AFDB["slate"], "xtick.color": AFDB["slate"], "ytick.color": AFDB["slate"]})

def show(md_text): display(Markdown(md_text))
def banner(kind, text):
    bg, bd, ic = {"ok": ("#E8F5EF", "#00A86A", "✅"), "warn": ("#FBF3DE", "#D49A00", "⚠️"), "bad": ("#FBEFED", "#B83B2E", "⛔"),
                  "info": ("#F4F7F5", "#0E7C86", "💡"), "demo": ("#FFF8E1", "#D49A00", "🟡")}[kind]
    display(HTML(f'<div style="background:{bg};border:1px solid {bd};border-radius:6px;padding:8px 12px;margin:6px 0;color:#231F20;font-family:Calibri,Arial,sans-serif">{ic} {text}</div>'))

def font(size, bold=False):
    names = (["DejaVuSans-Bold.ttf", "LiberationSans-Bold.ttf", "Arial Bold.ttf"] if bold else ["DejaVuSans.ttf", "LiberationSans-Regular.ttf", "Arial.ttf"])
    for n in names:
        try: return ImageFont.truetype(n, size)
        except Exception: pass
    try: return ImageFont.load_default(size=size)
    except TypeError: return ImageFont.load_default()

# ---------- internet check ----------
def internet_ok():
    for url in ("https://www.google.com", "https://huggingface.co"):
        try:
            requests.head(url, timeout=4); return True
        except Exception: pass
    return False
ONLINE = internet_ok()

# ---------- provider registry ----------
PROVIDERS = {
 "groq":       dict(label="Groq", key="GROQ_API_KEY", base="https://api.groq.com/openai/v1",
                    models=["llama-3.3-70b-versatile", "openai/gpt-oss-120b", "llama-3.1-8b-instant"],
                    vision_models=["meta-llama/llama-4-scout-17b-16e-instruct", "meta-llama/llama-4-maverick-17b-128e-instruct"], strength="Speed · Whisper transcription"),
 "cerebras":   dict(label="Cerebras", key="CEREBRAS_API_KEY", base="https://api.cerebras.ai/v1",
                    models=["gpt-oss-120b", "llama-3.3-70b", "llama3.1-8b"], strength="Speed"),
 "gemini":     dict(label="Google Gemini", key="GEMINI_API_KEY", native=True,
                    models=["gemini-flash-latest", "gemini-3-flash-preview", "gemini-2.5-flash", "gemini-flash-lite-latest", "gemini-2.5-flash-lite"],
                    strength="Multimodal · structured output · speech"),
 "mistral":    dict(label="Mistral", key="MISTRAL_API_KEY", base="https://api.mistral.ai/v1",
                    models=["mistral-small-latest", "mistral-medium-latest", "mistral-large-latest"], strength="Multilingual · French"),
 "codestral":  dict(label="Codestral (Mistral)", key="MISTRAL_API_KEY", base="https://api.mistral.ai/v1",
                    models=["codestral-latest", "devstral-small-latest", "mistral-small-latest"], strength="Code"),
 "openrouter": dict(label="OpenRouter (free models)", key="OPENROUTER_API_KEY", base="https://openrouter.ai/api/v1",
                    models=["openrouter/free"], strength="Many open models · reasoning"),
 "github":     dict(label="GitHub Models", key="GITHUB_TOKEN", base="https://models.github.ai/inference",
                    models=["openai/gpt-4.1-mini", "openai/gpt-4o-mini", "meta/Llama-3.3-70B-Instruct"],
                    vision_models=["openai/gpt-4.1-mini", "openai/gpt-4o-mini"], strength="Frontier models for prototyping"),
 "cohere":     dict(label="Cohere", key="COHERE_API_KEY", native=True,
                    models=["command-a-03-2025", "command-r-plus-08-2024", "command-r7b-12-2024"], strength="Grounded answers with citations"),
 "ollama":     dict(label="Ollama (local)", key=None, base="http://localhost:11434/v1", models=[], strength="Private, offline"),
}

def ollama_models():
    try:
        r = requests.get("http://localhost:11434/api/tags", timeout=2)
        return [m["name"] for m in r.json().get("models", [])]
    except Exception:
        return []

def available(p):
    cfg = PROVIDERS[p]
    if p == "ollama": return bool(ollama_models())
    return bool(KEYS.get(cfg["key"])) and ONLINE

# ---------- scoreboard ----------
SCORE = []
def record(mission, provider, model, ok, latency=None, tokens=None, note=""):
    SCORE.append(dict(time=datetime.datetime.now().strftime("%H:%M:%S"), mission=mission, provider=provider, model=model or "-",
                      status="✅ live" if ok is True else ("🟡 fallback" if ok == "fallback" else "❌ failed"),
                      latency_s=None if latency is None else round(latency, 2),
                      tokens_per_s=None if not (tokens and latency) else round(tokens / max(latency, 1e-6)), note=note[:120]))

@dataclass
class Reply:
    ok: bool
    text: str = ""
    provider: str = ""
    model: str = ""
    latency: float = 0.0
    tokens: int = 0
    reasoning: str = ""
    error: str = ""
    raw: dict = field(default_factory=dict)

def _retry(fn, tries=3, wait=8):
    for i in range(tries):
        try:
            return fn()
        except Exception as e:
            s = str(e)
            if any(k in s for k in ("429", "RESOURCE_EXHAUSTED", "rate limit", "Rate limit", "503", "overloaded")) and i < tries - 1:
                print(f"   ⏳ busy/rate-limited — waiting {wait * (i + 1)} s"); time.sleep(wait * (i + 1)); continue
            raise

def _not_found(e): return any(k in str(e) for k in ("404", "not found", "NOT_FOUND", "does not exist", "model_not_found", "decommissioned", "Unknown model", "invalid model", "Invalid model"))

_oai_clients = {}
def _oai(p):
    from openai import OpenAI
    if p not in _oai_clients:
        cfg = PROVIDERS[p]
        _oai_clients[p] = OpenAI(api_key=KEYS.get(cfg["key"]) or "ollama", base_url=cfg["base"], timeout=120)
    return _oai_clients[p]

def _oai_chat(p, prompt, system, json_mode, max_tokens, temperature, model, images):
    cli = _oai(p)
    content = prompt
    if images:
        content = [{"type": "text", "text": prompt}] + [
            {"type": "image_url", "image_url": {"url": "data:image/png;base64," + base64.b64encode(b).decode()}} for b in images]
    msgs = ([{"role": "system", "content": system}] if system else []) + [{"role": "user", "content": content}]
    default = PROVIDERS[p].get("vision_models") if images else None
    models = [model] if model else (ollama_models()[:2] if p == "ollama" else list(default or PROVIDERS[p]["models"]))
    tried, last = set(), None
    while models:
        m = models.pop(0)
        if m in tried: continue
        tried.add(m)
        kw = dict(model=m, messages=msgs, max_tokens=max_tokens, temperature=temperature)
        if json_mode: kw["response_format"] = {"type": "json_object"}
        try:
            try:
                r = _retry(lambda: cli.chat.completions.create(**kw))
            except Exception as e:
                if json_mode and ("response_format" in str(e) or "json" in str(e).lower()):
                    kw.pop("response_format"); r = _retry(lambda: cli.chat.completions.create(**kw))
                else:
                    raise
            msg = r.choices[0].message
            extra = getattr(msg, "model_extra", None) or {}
            reasoning = extra.get("reasoning") or extra.get("reasoning_content") or ""
            toks = getattr(r.usage, "completion_tokens", 0) if r.usage else 0
            return (msg.content or ""), (r.model or m), toks, reasoning
        except Exception as e:
            last = e
            if _not_found(e) and not models and not model:          # discover what exists now
                try:
                    ids = [x.id for x in cli.models.list().data]
                    hint = {"openrouter": ":free", "cerebras": "", "groq": "", "github": "gpt"}.get(p, "")
                    models += [i for i in ids if hint in i and i not in tried and not any(b in i for b in ("whisper", "tts", "guard", "embed", "image"))][:3]
                except Exception: pass
                continue
            if _not_found(e): continue
            raise
    raise last or RuntimeError("no model worked")

_gclient = None
def _gemini():
    global _gclient
    if _gclient is None:
        from google import genai
        _gclient = genai.Client(api_key=KEYS["GEMINI_API_KEY"])
    return _gclient

def _gemini_chat(prompt, system, json_mode, max_tokens, temperature, model, images):
    from google.genai import types as gt
    parts = [gt.Part.from_bytes(data=b, mime_type="image/png") for b in (images or [])] + [prompt]
    cfg = dict(temperature=temperature, max_output_tokens=max(max_tokens, 2048))
    if system: cfg["system_instruction"] = system
    if json_mode: cfg["response_mime_type"] = "application/json"
    models, last = ([model] if model else list(PROVIDERS["gemini"]["models"])), None
    for m in models:
        try:
            r = _retry(lambda: _gemini().models.generate_content(model=m, contents=parts, config=gt.GenerateContentConfig(**cfg)))
            toks = getattr(r.usage_metadata, "candidates_token_count", 0) or 0
            return r.text or "", m, toks, ""
        except Exception as e:
            last = e
            if _not_found(e): continue
            raise
    raise last

def _cohere_chat(prompt, system, max_tokens, temperature, model, documents=None):
    msgs = ([{"role": "system", "content": system}] if system else []) + [{"role": "user", "content": prompt}]
    last = None
    for m in ([model] if model else PROVIDERS["cohere"]["models"]):
        body = dict(model=m, messages=msgs, max_tokens=max_tokens, temperature=temperature)
        if documents: body["documents"] = documents
        r = requests.post("https://api.cohere.com/v2/chat", json=body, timeout=120,
                          headers={"Authorization": f"Bearer {KEYS['COHERE_API_KEY']}", "Content-Type": "application/json"})
        if r.status_code == 404 or (r.status_code == 400 and "model" in r.text.lower()):
            last = RuntimeError(r.text[:200]); continue
        if r.status_code == 429:
            time.sleep(10); r = requests.post("https://api.cohere.com/v2/chat", json=body, timeout=120, headers={"Authorization": f"Bearer {KEYS['COHERE_API_KEY']}"})
        r.raise_for_status()
        j = r.json()
        text = "".join(c.get("text", "") for c in j["message"].get("content", []))
        toks = j.get("usage", {}).get("tokens", {}).get("output_tokens", 0)
        return text, m, toks, j
    raise last

# ---------- privacy guard (pre-send screening) ----------
PII_PATTERNS = {
    "phone number": r"\+?\d{1,3}[\s-]?\d{3}[\s-]?\d{3}[\s-]?\d{3,4}\b",
    "e-mail": r"[\w.+-]+@[\w-]+\.[\w.]+",
    "national ID-like code": r"\b[A-Z]{2}-\d{6,}-\d{3,}\b",
    "titled person name": r"\b(?:Mr|Mrs|Ms|Dr)\.?\s+[A-Z][a-z]+\s+[A-Z][a-z]+",
    "street address": r"\b(?:Plot|House|Street|Avenue|Road)\s*\d+",
}
def privacy_guard(text):
    """Return the list of identifier types found in text (empty list = OK to send)."""
    return [name for name, pat in PII_PATTERNS.items() if re.search(pat, text or "")]

# ---------- check register ----------
CHECKS = []
def check(mission, name, passed, detail="", live=True):
    """Record a verification result and display it."""
    CHECKS.append(dict(mission=mission, check=name, result="PASS" if passed else "FAIL", mode="live" if live else "fallback", detail=re.sub("<[^>]+>", "", str(detail))[:160]))
    banner("ok" if passed else "bad", f"<b>Check · {name}:</b> {'PASS' if passed else 'FAIL'} — {detail}" + ("" if live else " <i>(on fallback data)</i>"))
    return passed

# ---------- run metadata ----------
import platform
RUN_META = dict(started=datetime.datetime.now().isoformat(timespec="seconds"), environment=ENV, python=platform.python_version(),
                online=ONLINE, keys_configured=sum(bool(v) for v in KEYS.values()))

CLOUD = {"groq", "cerebras", "gemini", "mistral", "codestral", "openrouter", "github", "cohere"}

def chat(provider, prompt, *, system=None, json_mode=False, max_tokens=900, temperature=0.3, model=None, images=None, mission="-", quiet=False, **kw):
    """One function for every text/vision provider. Never raises: returns Reply(ok=False, error=...) on failure."""
    label = PROVIDERS[provider]["label"]
    if CONFIG.get("privacy_guard") and provider in CLOUD:
        found = privacy_guard(prompt + " " + (system or "") + " " + json.dumps(kw.get("documents", "")))
        if found:
            record(mission, label, model, False, note="blocked by privacy guard")
            if not quiet: print(f"🛡️ {label}: prompt BLOCKED by privacy guard (found: {', '.join(found)})")
            return Reply(False, provider=label, error="blocked by privacy guard: " + ", ".join(found))
    if not available(provider):
        return Reply(False, provider=label, error="not configured (no key, no internet, or service not running)")
    t0 = time.time()
    try:
        if provider == "gemini":
            text, m, toks, rsn = _gemini_chat(prompt, system, json_mode, max_tokens, temperature, model, images); raw = {}
        elif provider == "cohere":
            text, m, toks, raw = _cohere_chat(prompt, system, max_tokens, temperature, model, kw.get("documents")); rsn = ""
        else:
            text, m, toks, rsn = _oai_chat(provider, prompt, system, json_mode, max_tokens, temperature, model, images); raw = {}
        lat = time.time() - t0
        record(mission, label, m, True, lat, toks)
        if not quiet: print(f"🟢 {label} · {m} · {lat:.1f} s" + (f" · {toks / lat:.0f} tok/s" if toks else ""))
        return Reply(True, text, label, m, lat, toks, rsn, raw=raw)
    except Exception as e:
        record(mission, label, model, False, note=str(e))
        if not quiet: print(f"❌ {label}: {str(e)[:180]}")
        return Reply(False, provider=label, error=str(e)[:300])

def explain_error(err):
    """Turn a raw error into an action the participant can take."""
    e = (err or "").lower()
    if "no key, no internet" in e or "not configured" in e:
        return "no internet" if not ONLINE else "no key loaded → add the secret, then re-run S2 and the next cells"
    if "privacy guard" in e: return err
    if any(k in e for k in ("401", "unauthorized", "invalid api key", "api key not valid", "invalid_api_key", "authentication")):
        return "key rejected → re-copy it fully (no spaces), or create a new one"
    if "403" in e or "permission" in e or "forbidden" in e:
        return "access refused → GitHub: token needs 'Models: read'; Mistral: activate the free Experiment plan"
    if any(k in e for k in ("429", "resource_exhausted", "rate limit", "quota")):
        return "free quota reached → wait 1 minute (per-minute) or until tomorrow (daily)"
    if "404" in e or "not found" in e: return "model unavailable → update the model list in the core cell"
    if "timed out" in e or "timeout" in e or "connection" in e: return "network problem → check internet (Kaggle: Settings → Internet ON)"
    return err[:140]

def chat_first(providers, prompt, *, demo=None, mission="-", **kw):
    """Try providers in order; if none works, return the pre-recorded demo answer and say why."""
    reasons = []
    for p in providers:
        r = chat(p, prompt, mission=mission, **kw)
        if r.ok: return r
        reasons.append(f"<li><b>{PROVIDERS[p]['label']}</b> — {explain_error(r.error)}</li>")
    if demo is not None and CONFIG.get("use_fallbacks", True):
        record(mission, "demo", "-", "fallback", note="pre-recorded answer")
        banner("demo", "No provider available for this mission — showing a <b>pre-recorded example</b>. Why:<ul style='margin:4px 0'>"
               + "".join(reasons) + "</ul>Run the <b>diagnostic cell</b> after the core library for details.")
        return Reply(True, demo, "demo (pre-recorded)", "-")
    return Reply(False, error="no provider available")

def extract_json(text):
    t = re.sub(r"^```(?:json)?|```$", "", (text or "").strip(), flags=re.M).strip()
    try: return json.loads(t)
    except json.JSONDecodeError:
        m = re.search(r"(\{.*\}|\[.*\])", t, flags=re.S)
        if m: return json.loads(m.group(1))
        raise

def numbers_in(text):
    """Extract numbers, normalising French decimal commas (5,8 → 5.8)."""
    t = re.sub(r"(\d),(\d)", r"\1.\2", text or "")
    return sorted(float(x) for x in re.findall(r"\d+(?:\.\d+)?", t))

def save_bytes(name, data):
    p = OUT / name; p.write_bytes(data); return p

# ---------- status table ----------
rows = []
for p, cfg in PROVIDERS.items():
    ok = available(p)
    rows.append(f"| {cfg['label']} | {cfg['strength']} | {'🟢 ready' if ok else ('⚪ not running' if p == 'ollama' else '🟡 no key')} |")
extras = [("Pollinations (images)", "Images without key", "🟢 ready" if ONLINE else "🔴 offline"),
          ("Pollinations (video, music)", "Credit-based media", "🟢 key found" if KEYS.get("POLLINATIONS_API_KEY") else "🟡 no key"),
          ("Hugging Face", "Open image models", "🟢 key found" if KEYS.get("HF_TOKEN") else "🟡 no key"),
          ("Edge neural voices", "Text-to-speech, no key", "🟢 ready" if ONLINE else "🔴 offline")]
rows += [f"| {a} | {b} | {c} |" for a, b, c in extras]
show(f"**Internet:** {'🟢 online' if ONLINE else '🔴 offline — fallbacks will be used'}\n\n| Service | Signature strength | Status |\n|---|---|---|\n" + "\n".join(rows))
````

### 🎯 Goal
Send **the same prompt** to every provider you have configured and **measure** the response time and the generation speed (tokens per second).

### 💡 Concept — why are some providers so fast?
A language model writes text **one token at a time** (a token ≈ ¾ of an English word). Speed depends on the model size *and* on the hardware. **Groq** (LPU chips) and **Cerebras** (wafer-scale chips) are built specifically for this step, which is why they often reach several hundred tokens per second. Speed matters for interactive tools: chat assistants, live coding, real-time transcription.

| Metric | Meaning | Good for |
|---|---|---|
| **Latency** (s) | time until the full answer arrives | user experience |
| **Tokens / second** | generation speed | long outputs, agents |

### 💡 Concept — how offices evaluate automatic coding
Automatic coding is one of the most documented AI use cases in official statistics. It is judged with two numbers that must be read together:

| Metric | Definition | Typical target |
|---|---|---|
| **Automation rate** | share of answers coded automatically (confidence ≥ threshold) | as high as possible |

Raising the threshold increases accuracy but sends more work to humans. The right threshold is a **quality decision**, not a technical one. Here we work at the 1-digit level; production systems code 4 digits and need much larger evaluation sets.

### 🎯 Goal
Let users ask questions such as *"How many regions have internet access above 50%?"*. The model does **not** answer directly: it translates each question into a small **query plan** (JSON), which Python validates and executes on the table. The answer sentence is then written from the **computed** value.

### 💡 Concept — never let the model do the arithmetic
```
question ──► LLM ──► query plan (JSON) ──► validator ──► pandas ──► value ──► templated answer
```
* The plan can only use **known columns and operators** — no arbitrary code is executed.
* Every answer is **reproducible** and **auditable** (the plan is shown).
* The typical error is **methodological**, not arithmetic: e.g. an *unweighted* mean where a population-weighted rate is required. That is exactly what the evaluation catches.

````python
import operator
OPS = {"==": operator.eq, "!=": operator.ne, ">": operator.gt, ">=": operator.ge, "<": operator.lt, "<=": operator.le}
AGGS = {"count", "sum", "mean", "weighted_mean", "max", "min"}
SCHEMA = {c: str(t) for c, t in kivu.drop(columns=["lat", "lon"]).dtypes.items()}

def run_plan(p):
    """Validate and execute a query plan. Raises ValueError on anything not allowed."""
    d = kivu.copy()
    for f in p.get("filters", []) or []:
        if f.get("column") not in SCHEMA or f.get("op") not in OPS: raise ValueError(f"invalid filter {f}")
        val = f["value"] if SCHEMA[f["column"]] == "object" else float(f["value"])
        d = d[OPS[f["op"]](d[f["column"]], val)]
    agg, col, kind = p.get("agg"), p.get("metric"), p.get("answer_type", "number")
    if agg not in AGGS: raise ValueError(f"invalid agg {agg}")
    if agg != "count" and col not in SCHEMA: raise ValueError(f"invalid metric {col}")
    if kind == "region":
        if agg not in ("max", "min"): raise ValueError("region answers need max or min")
        return d.loc[d[col].idxmax() if agg == "max" else d[col].idxmin(), "region"]
    if agg == "count": return int(len(d))
    if agg == "weighted_mean": return round(float((d[col] * d.population_k).sum() / d.population_k.sum()), 1)
    return round(float(getattr(d[col], agg)()), 1)

QP_PROMPT = f"""Translate each question into a query plan over ONE table.
Table columns and types: {json.dumps(SCHEMA)}. One row per region. Rates are percentages.
IMPORTANT: national rates must be population-weighted (agg "weighted_mean", weight = population_k).
Allowed ops: {list(OPS)}. Allowed agg: {sorted(AGGS)}. answer_type: "number" or "region" (region needs agg max or min).
Questions:
{chr(10).join(f"{i}: {q}" for i, q, _ in QUESTIONS)}
Return JSON only: {{"plans": [{{"id": str, "filters": [{{"column": str, "op": str, "value": number or str}}], "metric": str or null, "agg": str, "answer_type": str}}]}}"""

DEMO_PLANS = json.dumps({"plans": [
 {"id": "Q1", "filters": [], "metric": "cpi_infl_2025", "agg": "max", "answer_type": "region"},
 {"id": "Q2", "filters": [], "metric": "cpi_infl_2025", "agg": "mean", "answer_type": "number"},
 {"id": "Q3", "filters": [{"column": "internet_hh_pct", "op": ">", "value": 50}], "metric": None, "agg": "count", "answer_type": "number"},
 {"id": "Q4", "filters": [{"column": "infl_change_pp", "op": "<", "value": -1}], "metric": "dl_speed_mbps", "agg": "mean", "answer_type": "number"},
 {"id": "Q5", "filters": [{"column": "cpi_infl_2025", "op": ">", "value": 6}], "metric": "population_k", "agg": "sum", "answer_type": "number"},
 {"id": "Q6", "filters": [], "metric": "internet_hh_pct", "agg": "min", "answer_type": "region"}]})
r = chat_first(["gemini", "groq", "mistral", "github"], QP_PROMPT, json_mode=True, temperature=0, max_tokens=1500,
               demo=DEMO_PLANS, mission="03 data query")
plans = {p["id"]: p for p in extract_json(r.text)["plans"]}
rows = []
for qid, q, expected in QUESTIONS:
    p = plans.get(qid, {})
    try: got, err = run_plan(p), ""
    except Exception as e: got, err = None, str(e)
    ok = (got == expected) if isinstance(expected, str) else (got is not None and abs(float(got) - float(expected)) < 0.05)
    plan_txt = f"{p.get('agg')}({p.get('metric')})" + "".join(f" · {f['column']} {f['op']} {f['value']}" for f in p.get("filters", []) or [])
    rows.append(dict(id=qid, question=q, plan=plan_txt, answer=got, expected=expected, correct="✅" if ok else "❌", error=err))
qa = pd.DataFrame(rows); display(qa)
rate = (qa.correct == "✅").mean()
check("03", "exact answers on the evaluation questions", rate == 1, f"{r.provider}: {rate:.0%} correct", live=not r.provider.startswith("demo"))
for row in qa[qa.correct == "✅"].head(2).itertuples():
    show(f"💬 *{row.question}* → **{row.answer}** — computed from the published table (plan: `{row.plan}`).")
````

### 💡 Concept — reasoning models
*Reasoning models* (for example the open DeepSeek and GPT-OSS families) spend extra tokens "thinking" before answering. They are stronger on arithmetic chains and logic, but slower. **OpenRouter** gives one key for many free open models; its `openrouter/free` router picks an available one for you. The free line-up changes frequently.

````python
Q = ("A national statistics office plans a household survey to estimate a proportion. Requirements: 95% confidence (z = 1.96), "
     "margin of error 3 percentage points, conservative p = 0.5, design effect 1.8, expected response rate 80%. "
     "Compute the number of households to SELECT. Show the steps briefly, round up only at the end, "
     "and finish with a line: FINAL = <integer>.")
n0 = 1.96**2 * 0.25 / 0.03**2
expected = math.ceil(n0 * 1.8 / 0.8)
DEMO_REASON = (f"1. Simple random sample: n0 = 1.96² × 0.5 × 0.5 / 0.03² = {n0:.1f}\n"
               f"2. Design effect: {n0:.1f} × 1.8 = {n0 * 1.8:.1f}\n"
               f"3. Non-response: {n0 * 1.8:.1f} / 0.8 = {n0 * 1.8 / 0.8:.1f}\nFINAL = {expected}")
r = chat_first(["openrouter", "groq", "gemini", "mistral"], Q, max_tokens=1500, temperature=0, demo=DEMO_REASON, mission="04 reasoning")
show(r.text)
if r.reasoning:
    with_reasoning = textwrap.shorten(r.reasoning, 600)
    show(f"<details><summary>🧠 Show the model's reasoning trace ({len(r.reasoning)} characters)</summary>\n\n{with_reasoning}\n</details>")
m = re.search(r"FINAL\s*=\s*([\d,\s]+)", r.text or "")
final = int(re.sub(r"\D", "", m.group(1))) if m else None
check("04", "final value equals the Python calculation", final == expected, f"{r.provider}: model {final} · Python {expected}", live=not r.provider.startswith("demo"))
````

### 🎯 Goal
Translate the Kivuland press release into **French** and **Kiswahili**, then automatically verify that **every figure survived** the translation.

### 💡 Concept — multilingual models
Mistral AI is a European company whose models are trained with strong emphasis on French and other European languages. For African statistical offices working in French, English, Portuguese, Arabic and Kiswahili, translation quality — and **terminology consistency** — matters. A classic translation error is the **decimal separator**: *5.8%* in English becomes *5,8 %* in French.

````python
def numbers_in(text):
    """Extract numbers, normalising French decimal commas (5,8 → 5.8)."""
    t = re.sub(r"(\d),(\d)", r"\1.\2", text or "")
    return sorted(float(x) for x in re.findall(r"\d+(?:\.\d+)?", t))

GLOSSARY = "inflation → FR: inflation / SW: mfumuko wa bei; household → FR: ménage / SW: kaya; National Statistics Office → FR: Office national de la statistique / SW: Ofisi ya Taifa ya Takwimu"
results = {}
for lang, rule in [("French", "Use French typography: decimal comma and a space before %."), ("Kiswahili", "Keep decimal points as in the source.")]:
    prompt = (f"ROLE: official translator of a national statistics office.\nTASK: Translate into {lang}.\nGLOSSARY: {GLOSSARY}\n"
              f"RULES: {rule} Keep every number. Return only the translation.\n\nTEXT:\n{PRESS_RELEASE}")
    demo = {"French": PRESS_RELEASE.replace("Annual consumer price inflation in Kivuland eased to 5.8% in 2025, down from 6.7% in 2024.",
             "L'inflation annuelle des prix à la consommation au Kivuland a ralenti à 5,8 % en 2025, contre 6,7 % en 2024.")
             .replace("Prices rose fastest in the Highlands region (8.2%) and slowest in the Capital (4.6%).",
             "Les prix ont le plus augmenté dans la région des Hautes-Terres (8,2 %) et le moins dans la Capitale (4,6 %).")
             .replace("Household internet access ranges from 21.7% in the Highlands to 71.4% in the Capital.",
             "L'accès des ménages à Internet va de 21,7 % dans les Hautes-Terres à 71,4 % dans la Capitale.")
             .replace("The full tables are published on the National Statistics Office website.",
             "Les tableaux complets sont publiés sur le site de l'Office national de la statistique."),
             "Kiswahili": "Mfumuko wa bei wa kila mwaka nchini Kivuland ulipungua hadi 5.8% mwaka 2025, kutoka 6.7% mwaka 2024. "
             "Bei zilipanda kwa kasi zaidi katika eneo la Nyanda za Juu (8.2%) na kwa kasi ndogo zaidi Mjini Mkuu (4.6%). "
             "Upatikanaji wa intaneti kwa kaya ni kati ya 21.7% katika Nyanda za Juu na 71.4% Mjini Mkuu. "
             "Majedwali kamili yamechapishwa kwenye tovuti ya Ofisi ya Taifa ya Takwimu."}[lang]
    r = chat_first(["mistral", "gemini", "groq"], prompt, temperature=0.1, demo=demo, mission="05 languages")
    ok = numbers_in(r.text) == numbers_in(PRESS_RELEASE)
    results[lang] = r.text
    show(f"**{lang}** — _{r.provider}_\n\n> {r.text}")
    live = not r.provider.startswith("demo")
    check("05", f"{lang}: all numbers preserved", ok, f"{r.provider} · {numbers_in(r.text)}", live=live)
    if lang == "French":
        terms = ["inflation", "ménage", "Office national de la statistique"]
        hits = [t_ for t_ in terms if t_.lower() in r.text.lower()]
        check("05", "French: glossary terms used", len(hits) >= 2, f"found {hits}", live=live)
(OUT / "M05_translations.md").write_text("\n\n".join(f"## {k}\n{v}" for k, v in results.items()), encoding="utf-8")
````

### 💡 Concept — code models and independent tests
**Codestral** is Mistral's code model; **GitHub Models** gives free prototyping access to frontier models. Generated code is often *plausible but subtly wrong*. The professional habit: define the expected results **before** generating the code.

````python
CODE_PROMPT = ("Write a Python function `laspeyres(p0, p1, q0)` that returns the Laspeyres price index (base = 100) "
               "given lists of base prices p0, current prices p1 and base quantities q0. Raise ValueError if the lists have "
               "different lengths or are empty. Return ONLY one python code block, no explanation.")
DEMO_CODE = """```python
def laspeyres(p0, p1, q0):
    if not (len(p0) == len(p1) == len(q0)) or len(p0) == 0:
        raise ValueError("p0, p1 and q0 must be non-empty and of equal length")
    base = sum(a * q for a, q in zip(p0, q0))
    current = sum(b * q for b, q in zip(p1, q0))
    return 100 * current / base
```"""
r = chat_first(["codestral", "github", "groq", "gemini"], CODE_PROMPT, temperature=0, demo=DEMO_CODE, mission="07 code")
src = re.search(r"```(?:python)?\n(.*?)```", r.text, re.S)
src = src.group(1) if src else r.text
show(f"```python\n{src}\n```")

READ_BEFORE_RUNNING = True     # ← the habit: read the code above first
tests = [  # (arguments, expected)  — written by a human, before seeing the code
    (([10, 20], [11, 22], [1, 1]), 110.0),
    (([2, 4], [3, 4], [5, 0]), 150.0),
    (([1.0], [1.0], [7]), 100.0),
]
if READ_BEFORE_RUNNING:
    ns = {}
    try:
        exec(src, ns)
        f = ns["laspeyres"]
        rows = [dict(test=i + 1, expected=exp, got=round(f(*args), 6), ok="✅" if abs(f(*args) - exp) < 1e-9 else "❌") for i, (args, exp) in enumerate(tests)]
        try: f([], [], []); rows.append(dict(test="empty lists", expected="ValueError", got="no error", ok="❌"))
        except ValueError: rows.append(dict(test="empty lists", expected="ValueError", got="ValueError", ok="✅"))
        tdf = pd.DataFrame(rows); display(tdf)
        (OUT / "M07_laspeyres.py").write_text(src)
        check("07", "independent tests passed", bool((tdf.ok == "✅").all()), f"{(tdf.ok == '✅').sum()}/{len(tdf)} · code by {r.provider}", live=not r.provider.startswith("demo"))
    except Exception as e:
        check("07", "independent tests passed", False, f"generated code failed: {type(e).__name__}: {e}")
````

### 💡 Concept — let the model decide *what*, let code decide *how*
Asking a model to write an entire web page gives unpredictable results. A **specification** is small, checkable and always rendered in the house style. Numbers are computed by pandas, never typed by the model.

````python
DASH_PROMPT = f"""You design dashboards for a national statistics office.
Columns available (numeric unless stated): region (text), {', '.join(c for c in kivu.columns if c not in ('region',))}.
Design a dashboard about inflation and connectivity for ministry officials.
Return JSON only:
{{"title": str, "subtitle": str,
  "kpis": [{{"label": str, "column": str, "agg": "mean"|"weighted_mean"|"max"|"min"}}] (exactly 3),
  "charts": [{{"type": "bar"|"scatter"|"map", "x": column, "y": column, "title": finding-style str}}] (exactly 4)}}
For weighted_mean the weight is population_k. For "map" use x="lon", y="lat" and colour by the given "color" column (add a "color" key)."""
DEMO_DASH = json.dumps({"title": "Kivuland — Prices & Connectivity 2025", "subtitle": "Synthetic data · regional overview for ministry officials",
  "kpis": [{"label": "National inflation 2025 (%)", "column": "cpi_infl_2025", "agg": "weighted_mean"},
           {"label": "Highest regional inflation (%)", "column": "cpi_infl_2025", "agg": "max"},
           {"label": "Avg. household internet access (%)", "column": "internet_hh_pct", "agg": "weighted_mean"}],
  "charts": [{"type": "bar", "x": "region", "y": "cpi_infl_2025", "title": "Highlands face the fastest price growth"},
             {"type": "bar", "x": "region", "y": "infl_change_pp", "title": "Inflation eased in every region"},
             {"type": "scatter", "x": "internet_hh_pct", "y": "dl_speed_mbps", "title": "Better-connected regions also have faster speeds"},
             {"type": "map", "x": "lon", "y": "lat", "color": "internet_hh_pct", "title": "Connectivity is concentrated in the Capital and Coast"}]})
r = chat_first(["gemini", "groq", "mistral"], DASH_PROMPT, json_mode=True, temperature=0.4, demo=DEMO_DASH, mission="09 dashboard")
spec = extract_json(r.text)

def validate_dash(s):
    errs = []
    cols = set(kivu.columns)
    for k in s.get("kpis", []):
        if k.get("column") not in cols: errs.append(f"KPI column {k.get('column')} unknown")
        if k.get("agg") not in {"mean", "weighted_mean", "max", "min"}: errs.append(f"agg {k.get('agg')} unknown")
    for c in s.get("charts", []):
        if c.get("type") not in {"bar", "scatter", "map"}: errs.append(f"chart type {c.get('type')} unknown")
        for key in ("x", "y", "color"):
            if c.get(key) and c[key] not in cols: errs.append(f"{key}={c[key]} unknown")
    if len(s.get("kpis", [])) != 3 or len(s.get("charts", [])) != 4: errs.append("need exactly 3 KPIs and 4 charts")
    return errs

errs = validate_dash(spec)
check("09", "dashboard specification valid", not errs, "; ".join(errs[:3]) or f"3 KPIs and 4 charts from {r.provider}", live=not r.provider.startswith("demo"))
if errs:
    banner("warn", "Spec from the model rejected (" + "; ".join(errs[:4]) + ") — using the reference spec instead.")
    spec = json.loads(DEMO_DASH)
display(pd.json_normalize(spec["charts"]))
````

The next cell defines the media engines. Each one tries **several free services in order** and ends with an **offline fallback**, so the pipeline never breaks:

| Function | Tries, in order | Offline fallback |
|---|---|---|
| `make_image()` | Pollinations FLUX (key or anonymous) → Hugging Face FLUX.1-schnell | generated placeholder artwork |
| `speak()` | Edge neural voices → Gemini TTS → Google gTTS | synthetic "voice-like" placeholder |
| `transcribe()` | Groq Whisper → Pollinations transcription | — |
| `compose_video()` | local assembly with FFmpeg (bundled via `imageio-ffmpeg`) | — |

````python
# @title M · Media engines (image · voice · transcription · video assembly)
import imageio.v2 as imageio
import imageio_ffmpeg
FFMPEG = shutil.which("ffmpeg") or imageio_ffmpeg.get_ffmpeg_exe()
POLL_KEY, HF_TOKEN = KEYS.get("POLLINATIONS_API_KEY"), KEYS.get("HF_TOKEN")
W_, H_ = CONFIG["video_size"]
_last_anon = [0.0]

def ai_label(img, text="AI-generated · training use"):
    """Stamp a visible label — good practice for any synthetic media."""
    img = img.convert("RGB"); d = ImageDraw.Draw(img); f = font(max(12, img.width // 70))
    w = d.textlength(text, font=f); pad = 6
    d.rectangle([img.width - w - 3 * pad, img.height - f.size - 3 * pad, img.width - pad, img.height - pad], fill=(35, 31, 32))
    d.text((img.width - w - 2 * pad, img.height - f.size - 2 * pad), text, fill=(245, 194, 66), font=f)
    return img

def placeholder_image(prompt, w=1024, h=576, seed=0):
    rng = random.Random(seed); img = Image.new("RGB", (w, h)); d = ImageDraw.Draw(img)
    c1, c2 = np.array([0, 85, 58]), np.array([0, 168, 106])
    for y in range(h):
        d.line([(0, y), (w, y)], fill=tuple(int(v) for v in c1 + (c2 - c1) * y / h))
    for _ in range(14):
        r = rng.randint(20, 160); x, y = rng.randint(0, w), rng.randint(0, h)
        d.ellipse([x - r, y - r, x + r, y + r], outline=(255, 255, 255), width=2)
    d.rectangle([0, h - 110, w, h], fill=(0, 0, 0))
    d.text((24, h - 100), "OFFLINE PLACEHOLDER (no image service reachable)", fill=(245, 194, 66), font=font(20, True))
    d.multiline_text((24, h - 70), textwrap.fill(prompt, 95)[:190], fill="white", font=font(16))
    return img

def make_image(prompt, w=1024, h=576, seed=7, engines=("pollinations", "huggingface"), mission="10 image"):
    q = urllib.parse.quote(prompt[:900])
    for eng in engines:
        t0 = time.time()
        try:
            if eng == "pollinations" and ONLINE:
                if POLL_KEY:
                    url = f"https://gen.pollinations.ai/image/{q}?model=flux&width={w}&height={h}&seed={seed}&nologo=true"
                    r = requests.get(url, headers={"Authorization": f"Bearer {POLL_KEY}"}, timeout=180)
                else:
                    wait = 16 - (time.time() - _last_anon[0])
                    if wait > 0: time.sleep(wait)                 # anonymous tier ≈ 1 request / 15 s
                    url = f"https://image.pollinations.ai/prompt/{q}?model=flux&width={w}&height={h}&seed={seed}&nologo=true"
                    r = requests.get(url, timeout=180); _last_anon[0] = time.time()
                if r.status_code == 200 and r.headers.get("content-type", "").startswith("image"):
                    record(mission, "Pollinations", "flux", True, time.time() - t0)
                    return Image.open(io.BytesIO(r.content)).convert("RGB"), "Pollinations · FLUX"
                raise RuntimeError(f"HTTP {r.status_code}: {r.text[:120]}")
            if eng == "huggingface" and HF_TOKEN and ONLINE:
                from huggingface_hub import InferenceClient
                img = InferenceClient(provider="auto", api_key=HF_TOKEN).text_to_image(prompt, model="black-forest-labs/FLUX.1-schnell", width=w, height=h)
                record(mission, "Hugging Face", "FLUX.1-schnell", True, time.time() - t0)
                return img.convert("RGB"), "Hugging Face · FLUX.1-schnell"
        except Exception as e:
            record(mission, eng, "-", False, note=str(e)); print(f"   ⚠️ {eng}: {str(e)[:140]}")
    record(mission, "placeholder", "-", "fallback")
    return placeholder_image(prompt, w, h, seed), "offline placeholder"

# ---------- audio ----------
def to_wav(src, dst, sr=24000):
    subprocess.run([FFMPEG, "-y", "-loglevel", "error", "-i", str(src), "-ar", str(sr), "-ac", "1", str(dst)], check=True)
    return Path(dst)

def read_wav(path):
    with wave.open(str(path)) as w:
        sr, n, sw = w.getframerate(), w.getnframes(), w.getsampwidth()
        data = np.frombuffer(w.readframes(n), dtype=np.int16 if sw == 2 else np.uint8).astype(np.float32)
    return data / 32768.0, sr

def write_wav(path, x, sr=24000):
    x = np.clip(x, -1, 1)
    with wave.open(str(path), "wb") as w:
        w.setnchannels(1); w.setsampwidth(2); w.setframerate(sr); w.writeframes((x * 32767).astype(np.int16).tobytes())
    return Path(path)

def _run_async(coro):
    with concurrent.futures.ThreadPoolExecutor(1) as ex:   # works inside Jupyter's running event loop
        return ex.submit(asyncio.run, coro).result()

def placeholder_voice(text, path, sr=24000):
    """Offline stand-in: syllable-like tones so that timing and lip-sync still work. Not speech."""
    rng = np.random.default_rng(len(text)); out = []
    for word in text.split():
        for _ in range(max(1, len(word) // 3)):
            dur = rng.uniform(0.11, 0.18); t = np.arange(int(sr * dur)) / sr
            f0 = rng.uniform(150, 210)
            out.append(0.25 * np.sin(2 * np.pi * f0 * t) * np.sin(np.pi * t / dur) ** 2)
        out.append(np.zeros(int(sr * 0.07)))
    return write_wav(path, np.concatenate(out), sr)

def speak(text, path_stem, voice=None, engines=("edge", "gemini", "gtts"), gemini_voice="Kore", lang="en", mission="11 voice"):
    """Return (wav_path, engine_label). Always returns something."""
    voice = voice or CONFIG["voice_en"]
    wav = OUT / f"{path_stem}.wav"
    for eng in engines:
        t0 = time.time()
        try:
            if not ONLINE: break
            if eng == "edge":
                import edge_tts
                mp3 = OUT / f"{path_stem}.mp3"
                _run_async(edge_tts.Communicate(text, voice).save(str(mp3)))
                to_wav(mp3, wav); record(mission, "Edge neural TTS", voice, True, time.time() - t0)
                return wav, f"Edge neural voice · {voice}"
            if eng == "gemini" and KEYS.get("GEMINI_API_KEY"):
                from google.genai import types as gt
                last = None
                for m in ("gemini-3.1-flash-tts-preview", "gemini-2.5-flash-preview-tts"):
                    try:
                        r = _gemini().models.generate_content(model=m, contents=f"Read clearly, like a calm news presenter: {text}",
                            config=gt.GenerateContentConfig(response_modalities=["AUDIO"], speech_config=gt.SpeechConfig(
                                voice_config=gt.VoiceConfig(prebuilt_voice_config=gt.PrebuiltVoiceConfig(voice_name=gemini_voice)))))
                        pcm = r.candidates[0].content.parts[0].inline_data.data
                        with wave.open(str(wav), "wb") as w:
                            w.setnchannels(1); w.setsampwidth(2); w.setframerate(24000); w.writeframes(pcm)
                        record(mission, "Gemini TTS", m, True, time.time() - t0)
                        return wav, f"Gemini TTS · {m} · voice {gemini_voice}"
                    except Exception as e:
                        last = e
                raise last
            if eng == "gtts":
                from gtts import gTTS
                mp3 = OUT / f"{path_stem}_gtts.mp3"
                gTTS(text, lang=lang).save(str(mp3)); to_wav(mp3, wav)
                record(mission, "gTTS", lang, True, time.time() - t0)
                return wav, "Google gTTS"
        except Exception as e:
            record(mission, eng, "-", False, note=str(e)); print(f"   ⚠️ {eng}: {str(e)[:140]}")
    for junk in OUT.glob(f"{path_stem}*.mp3"):
        if junk.stat().st_size == 0: junk.unlink()
    record(mission, "placeholder voice", "-", "fallback")
    return placeholder_voice(text, wav), "offline placeholder (tones, not speech)"

def transcribe(path, mission="11 voice"):
    t0 = time.time()
    if available("groq"):
        for m in ("whisper-large-v3-turbo", "whisper-large-v3"):
            try:
                with open(path, "rb") as f:
                    txt = _oai("groq").audio.transcriptions.create(model=m, file=f).text
                record(mission, "Groq Whisper", m, True, time.time() - t0)
                return txt, f"Groq · {m}"
            except Exception as e:
                last = e
        record(mission, "Groq Whisper", "-", False, note=str(last))
    return None, "no transcription service available"

def wer(ref, hyp):
    norm = lambda s: re.sub(r"[^\w\s.]", " ", s.lower()).replace(". ", " ").split()
    r, h = norm(ref), norm(hyp)
    d = np.arange(len(h) + 1)
    for i in range(1, len(r) + 1):
        prev, d[0] = d[0], i
        for j in range(1, len(h) + 1):
            cur = d[j]
            d[j] = min(d[j] + 1, d[j - 1] + 1, prev + (r[i - 1] != h[j - 1]))
            prev = cur
    return d[len(h)] / max(len(r), 1)

def show_audio(path, caption):
    show(f"🔈 **{caption}**"); display(Audio(str(path)))

# ---------- video assembly ----------
def envelope(x, sr, fps):
    hop = int(sr / fps); n = len(x) // hop
    e = np.array([np.sqrt(np.mean(x[i * hop:(i + 1) * hop] ** 2)) for i in range(n)])
    return e / (e.max() + 1e-9)

def mux(video_path, wav_path, out_path):
    subprocess.run([FFMPEG, "-y", "-loglevel", "error", "-i", str(video_path), "-i", str(wav_path),
                    "-c:v", "copy", "-c:a", "aac", "-b:a", "128k", "-shortest", str(out_path)], check=True)
    return Path(out_path)

def concat_wavs(paths, out, gap=0.45, sr=24000):
    parts, spans, t = [], [], 0.0
    for p in paths:
        x, _ = read_wav(to_wav(p, OUT / f"_tmp_{Path(p).stem}.wav", sr))
        spans.append((t, t + len(x) / sr)); parts += [x, np.zeros(int(gap * sr))]; t += len(x) / sr + gap
    return write_wav(out, np.concatenate(parts), sr), spans

def show_video(path, width=720):
    data = Path(path).read_bytes()
    if len(data) < 25_000_000:
        display(HTML(f'<video width="{width}" controls src="data:video/mp4;base64,{base64.b64encode(data).decode()}"></video>'))
    else:
        print(f"Video saved to {path} (too large to preview inline)")
print(f"🎛️ Media toolkit ready · FFmpeg: {Path(FFMPEG).name}")
````

### 🎯 Goal
Turn a short communication brief into a **detailed image prompt** (text model), then generate the illustration with **two free image services** and compare.

### 💡 Concept — diffusion models
Image generators such as **FLUX** start from noise and remove it step by step, guided by your text. They respond best to concrete visual descriptions: subject, setting, style, light, composition, colours. A text model is an excellent **prompt writer** for them.

> ⚠️ Avoid real people, logos, flags and copyrighted characters. Always **label** synthetic images (the helper stamps a label automatically).

````python
BRIEF = "Cover illustration for a statistical bulletin on household internet access in rural East Africa. Optimistic, respectful, no text."
PW_PROMPT = (f"Write ONE image-generation prompt (max 70 words) for this brief: {BRIEF}\n"
             "Describe subject, setting, style (flat editorial illustration), lighting, composition, and a palette of deep green, "
             "emerald and warm gold. Fictional people only. No text, logos or flags. Return only the prompt.")
DEMO_PW = ("Flat editorial illustration of a fictional rural East African family sitting outside a small home at golden hour, "
           "a grandmother and a teenager looking together at a smartphone, subtle glowing network lines connecting nearby homes "
           "and a distant cell tower, rolling green hills, palette of deep green, emerald and warm gold, soft light, wide composition, no text")
img_prompt = chat_first(["gemini", "groq", "mistral"], PW_PROMPT, temperature=0.7, demo=DEMO_PW, mission="10 image").text.strip()
show(f"**Image prompt written by the model:**\n\n> {img_prompt}")

gallery = []
for eng in ("pollinations", "huggingface"):
    im, label = make_image(img_prompt, 1024, 576, seed=11, engines=(eng,))
    im = ai_label(im, f"AI-generated · {label}")
    path = OUT / f"M10_image_{eng}.png"; im.save(path); gallery.append((path, label))
fig, axes = plt.subplots(1, 2, figsize=(14, 4.4))
for ax, (p, lab) in zip(axes, gallery):
    ax.imshow(Image.open(p)); ax.set_title(lab, loc="left"); ax.axis("off")
plt.tight_layout(); plt.show()
check("10", "synthetic-media label applied", True, "each image carries a visible 'AI-generated' stamp", live=any("placeholder" not in lab for _, lab in gallery))
````

### 💡 Concept — TTS and STT
* **Text-to-speech (TTS)** turns text into audio. Modern neural voices are natural enough for accessibility services, radio spots and phone lines.

````python
JINGLE_BRIEF = "10-second uplifting jingle for a statistics release: marimba and soft pads, warm, optimistic, modern African feel, no vocals"
music_path = None
if POLL_KEY and ONLINE and CONFIG["try_pollinations_music"]:
    t0 = time.time()
    try:
        mlist = []
        for u in ("https://gen.pollinations.ai/audio/models", "https://gen.pollinations.ai/v1/models"):
            try:
                j = requests.get(u, timeout=20).json()
                items = j if isinstance(j, list) else j.get("data", [])
                mlist += [(m.get("id") or m.get("name")) if isinstance(m, dict) else m for m in items]
            except Exception: pass
        music_models = [m for m in mlist if m and "music" in str(m).lower()] or [None]
        for mm in music_models[:3]:
            url = f"https://gen.pollinations.ai/audio/{urllib.parse.quote(JINGLE_BRIEF)}" + (f"?model={mm}" if mm else "")
            r = requests.get(url, headers={"Authorization": f"Bearer {POLL_KEY}"}, timeout=300)
            if r.status_code == 200 and r.headers.get("content-type", "").startswith("audio"):
                music_path = save_bytes("M14_jingle_pollinations.mp3", r.content)
                record("14 music", "Pollinations", mm, True, time.time() - t0)
                show_audio(music_path, f"Generated by Pollinations · {mm}"); break
            print(f"   ⚠️ {mm}: HTTP {r.status_code} {r.text[:120]}")
    except Exception as e:
        record("14 music", "Pollinations", "-", False, note=str(e)); print("   ⚠️", str(e)[:150])
if not music_path:
    banner("info", "Pollinations music not available (no key, no credit or no internet) — using the LLM-composer route below.")
````

````python
COMPOSE_PROMPT = ("Compose a 10-second, 4-bar jingle in C major pentatonic, tempo 112 BPM, uplifting. "
                  'Return JSON only: {"bpm": int, "melody": [[midi_note, beats], ...], "bass": [[midi_note, beats], ...]}. '
                  "Melody notes between 60 and 84, bass between 36 and 55, each track totalling 16 beats. End on C.")
DEMO_SONG = json.dumps({"bpm": 112, "melody": [[72,1],[74,.5],[76,.5],[79,1],[76,1],[74,1],[76,.5],[79,.5],[81,2],[79,1],[76,1],[74,.5],[76,.5],[72,1],[74,1],[72,3]],
                        "bass": [[48,2],[48,2],[45,2],[45,2],[41,2],[43,2],[48,4]]})
r = chat_first(["gemini", "groq", "mistral", "openrouter"], COMPOSE_PROMPT, json_mode=True, temperature=0.8, demo=DEMO_SONG, mission="14 music")
song = extract_json(r.text)
ok = all(60 <= n <= 84 for n, _ in song["melody"]) and all(36 <= n <= 55 for n, _ in song["bass"])
check("14", "composition respects notation rules", ok, f"{r.provider}: {len(song['melody'])} melody notes, ranges respected", live=not r.provider.startswith("demo"))

def synth(track, bpm, sr=24000, kind="marimba"):
    out = []
    for note, beats in track:
        dur = 60 / bpm * float(beats); t = np.arange(int(sr * dur)) / sr
        f = 440 * 2 ** ((int(note) - 69) / 12)
        if kind == "marimba":
            w = (np.sin(2 * np.pi * f * t) + 0.35 * np.sin(2 * np.pi * 4 * f * t) * np.exp(-t * 30)) * np.exp(-t * 5)
        else:
            w = 0.6 * np.sin(2 * np.pi * f * t) * np.minimum(1, t * 30) * np.exp(-t * 1.2)
        out.append(w)
    return np.concatenate(out)

bpm = int(song.get("bpm", 112))
mel, bass = synth(song["melody"], bpm), synth(song["bass"], bpm, kind="bass")
n = max(len(mel), len(bass)); mix = np.zeros(n); mix[:len(mel)] += 0.55 * mel; mix[:len(bass)] += 0.35 * bass
mix = mix / (np.abs(mix).max() + 1e-9) * 0.9
jingle = write_wav(OUT / "M14_jingle_composed.wav", mix)
show_audio(jingle, f"Composed by {r.provider} as notes · performed by Python")

fig, ax = plt.subplots(figsize=(10, 2.6)); t = 0
for note, beats in song["melody"]:
    ax.add_patch(plt.Rectangle((t, note - .4), beats * .95, .8, color=AFDB["green"])); t += beats
t = 0
for note, beats in song["bass"]:
    ax.add_patch(plt.Rectangle((t, note - .4), beats * .95, .8, color=AFDB["ochre"])); t += beats
ax.set_xlim(0, 16); ax.set_ylim(34, 86); ax.set_xlabel("beats"); ax.set_ylabel("MIDI note")
ax.set_title("Piano roll — melody (green) and bass (gold) written by the model", loc="left"); plt.show()
````

### 🎯 Goal
1. Show that the **privacy guard** stops a record with identifiers from reaching a cloud service.
2. Anonymise the (fictional) record with a **small open model running locally**, and compare with a **rule-based** method.

### 💡 Concept — local models
**Ollama** downloads open models (Llama, Qwen, Gemma, Mistral…) and serves them on `localhost`. Nothing is sent over the internet after the download, which makes local models the only option in this notebook that could, under your office's rules, be considered for **confidential** material. Small models (0.5–3 billion parameters) run on an ordinary laptop but are less accurate — always check their output.

| Where | How |
|---|---|
| **Local** | install from [ollama.com](https://ollama.com), then `ollama pull qwen2.5:1.5b` |
| **Colab** | set `CONFIG["install_ollama_in_colab"] = True` in the control panel (≈ 2 min, CPU is enough for small models) |
| **Kaggle** | usually not possible (no background services) — use the rule-based result |

````python
LOCAL_MODEL = CONFIG["local_model"]
if CONFIG["install_ollama_in_colab"] and ENV == "colab" and not ollama_models():
    subprocess.run("curl -fsSL https://ollama.com/install.sh | sh", shell=True, check=False)
    subprocess.Popen(["ollama", "serve"], stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL); time.sleep(6)
    subprocess.run(["ollama", "pull", LOCAL_MODEL], check=False)

RECORD = ("Interviewer note, 14 March 2026: Mrs Grace Uwimana (ID KV-19840312-4471), phone +250 788 123 456, "
          "lives at Plot 27, Nyarutarama Road, Lakes region. Household of 6; head aged 43; monthly spending 185,000.")
PII = ["Grace Uwimana", "KV-19840312-4471", "+250 788 123 456", "Plot 27, Nyarutarama Road"]

def rule_based(text):
    text = re.sub(r"\+?\d[\d\s]{8,}\d", "[PHONE]", text)
    text = re.sub(r"\b[A-Z]{2}-\d{8}-\d{4}\b", "[ID]", text)
    text = re.sub(r"\b(Mr|Mrs|Ms|Dr)\.?\s+[A-Z][a-z]+(\s+[A-Z][a-z]+)?", r"\1 [NAME]", text)
    text = re.sub(r"Plot \d+,[^,]+", "[ADDRESS]", text)
    return text

# 1 · the guard in action: try to send the raw record to a cloud service
attempt = chat("gemini", "Summarise this interview note: " + RECORD, mission="15 local")
check("15", "privacy guard blocks identifiers before a cloud call", "privacy guard" in attempt.error, attempt.error or "NOT BLOCKED")

rb = rule_based(RECORD)
llm_out = None
if ollama_models():
    prompt = ("Anonymise this record. Replace person names with [NAME], ID numbers with [ID], phone numbers with [PHONE], "
              "street addresses with [ADDRESS]. Keep region, household size, age and spending unchanged. Return only the text.\n\n" + RECORD)
    r = chat("ollama", prompt, temperature=0, mission="15 local")
    llm_out = r.text if r.ok else None
else:
    banner("info", "Ollama is not running here — <b>this is normal in Colab/Kaggle</b>, not an error. Only the rule-based method is shown. To try the local model: in Colab set the install option to True in this mission (≈ 2 min), or run it on your laptop with Ollama installed.")

def leak_check(text):
    return [p for p in PII if p.lower() in (text or "").lower()]
rows = [dict(method="rule-based (regex)", output=rb, leaks=", ".join(leak_check(rb)) or "none ✅",
             kept_facts="✅" if all(k in rb for k in ("Lakes", "6", "43", "185,000")) else "⚠️")]
if llm_out:
    rows.append(dict(method=f"local LLM ({r.model})", output=llm_out, leaks=", ".join(leak_check(llm_out)) or "none ✅",
                     kept_facts="✅" if all(k in llm_out for k in ("Lakes", "6", "43", "185,000")) else "⚠️"))
pd.set_option("display.max_colwidth", 200); display(pd.DataFrame(rows))
check("15", "no identifier left after rule-based anonymisation", not leak_check(rb), f"guard re-scan: {privacy_guard(rb) or 'clean'}")
if llm_out:
    check("15", "no identifier left after local-LLM anonymisation", not leak_check(llm_out), f"{r.model}")
````

> 🧠 **Lesson.** For well-structured identifiers, **rules are faster, cheaper and more reliable** than an LLM. Models help with messy free text (names without titles, addresses written in many ways). Mature anonymisation combines both — and is always validated before release.

| `429` / *RESOURCE_EXHAUSTED* / *rate limit* | free quota reached | wait a minute (helpers retry); reduce `benchmark_runs`; try tomorrow for daily caps |
| `404` / *model not found* | model retired or renamed | helpers try alternatives and discover available models; update the list in cell S4 |
| *blocked by privacy guard* | identifiers detected in the prompt | intended behaviour — remove personal data or process locally (mission 15) |
| Kaggle: connection errors | Internet switched off | *Settings → Internet → On* (phone-verified account) |
| No sound / placeholder voice | TTS service unreachable | check internet; the pipeline still runs with a labelled placeholder |
| Video cell slow | many frames on a small CPU | lower `CONFIG["video_fps"]` or `video_size` |
| Ollama not detected | service not running | local: `ollama serve`; Colab: enable `install_ollama_in_colab` |

## 🧠 Check yourself

<details><summary><b>1.</b> In mission 02, service A has 92 % overall accuracy and service B 88 %. Which should you choose?</summary>
It depends on the <b>accuracy of automatic codes at your threshold</b> and the <b>automation rate</b>. B may be better calibrated (low confidence when wrong) and therefore safer to automate. Compare the trade-off curves, on a larger evaluation set.
</details>

<details><summary><b>2.</b> Why did mission 03 ask for a query plan instead of the answer?</summary>
So that the computation is done by pandas, the method is explicit (e.g. weighted vs unweighted mean), nothing arbitrary is executed, and each answer can be tested and audited.
</details>

<details><summary><b>3.</b> A colleague wants to summarise interview notes containing names with a free cloud model. What do you advise?</summary>
No. The privacy guard would block it, rightly. Anonymise first (rules, local model), validate, and follow the office's data-protection policy.
</details>

<details><summary><b>4.</b> What makes the explainer video in mission 12 acceptable for an official release?</summary>
The narration uses only verified facts (checked automatically), the media are labelled, and the product still goes through editorial approval.
</details>

<details><summary><b>5.</b> Which two numbers would you report to management about a transcription service?</summary>
The word error rate on representative recordings, and the error rate on numbers and place names specifically.
</details>

## 📖 Glossary
| Term | Meaning |
|---|---|
| **Automation rate** | Share of cases handled automatically without human review |
| **Confidence threshold** | Minimum model confidence required to accept an output automatically |
| **Gold standard** | Reference labels produced by experts, used to measure accuracy |
| **Grounding / RAG** | Answering only from supplied documents, with citations |
| **Latency** | Time until the answer arrives |
| **Multimodal** | Accepting several input types (text, image, audio…) |
| **Query plan** | A structured description of a computation, executed by code |
| **Token** | A piece of a word; models read and write tokens |
| **TTS / STT** | Text-to-speech / speech-to-text |
| **WER** | Word error rate — share of words wrong in a transcript |
