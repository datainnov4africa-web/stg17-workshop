## 1 · Understand what we measure

When you send a text to a language model, this is what really happens:

```
 YOUR TEXT ──► split into TOKENS ──► network ──► queue ──► ① reading the prompt ──► ② writing, token by token ──► RESPONSE
               (≈ pieces of words)                          ("prefill", fast)          ("decode", the longest part)
               │                                                                      │
               └──── billed at the INPUT PRICE                                         └──── billed at the OUTPUT PRICE
 ◄──────────── TTFT: time to first token ─────────────────────────────────────────►◄── duration = output tokens ÷ THROUGHPUT ──►
 ◄──────────────────────────────────────────────── TOTAL LATENCY ─────────────────────────────────────────────────────────────►
```

| Quantity | Definition | Why it matters for an NSO |
|---|---|---|
| **Token** | Unit of text read by the model (≈ ¾ of a word in English) | Everything is billed in tokens. **Watch out:** Arabic, Kiswahili and less-represented languages often use **more tokens for the same content**, so they cost more. This is the "language tax". |
| **TTFT** (*time to first token*) | Delay before the first word appears | What a user *feels*: website assistant, tool for journalists, chatbot |
| **Throughput** (tokens/s) | Writing speed once started | Decisive for long texts and batch processing |
| **Total latency** | ≈ TTFT + output tokens ÷ throughput | The time for one document, which sets the time for a whole campaign |
| **Cost per 1,000 documents** | (input tokens × input price + output tokens × output price) × 1,000 | **The comparison unit for procurement**, independent of the provider |
| **Rate limits** | Per-minute or per-day caps (RPM, TPM, RPD) | The real ceiling on scaling up, often forgotten |
| **Quality** | Here: accuracy of the figures and a reviewer's score | A release with a wrong figure is worse than no release |
| **Sovereignty** | Where the data goes, who can read it, can we take the model in-house? | Statistics law, confidentiality, embargo, dependence on one vendor |

#### Worked example, to do in your head
One press release = **350 tokens** in (prompt + text) and **400 tokens** out. Price: $0.15 / $0.60 per million tokens.

$$\text{cost} = 350 \times \frac{0.15}{10^6} + 400 \times \frac{0.60}{10^6} \approx 0.00029 \text{ dollars per release}$$

that is, **$0.29 for 1,000 press releases**.

The unit cost is often **negligible**. What makes the difference is **volume**, **speed** and **guarantees**.

#### Why is Groq fast?
Most providers run models on **GPUs**, general-purpose graphics processors designed mainly for training. Groq designed its own chip, the **LPU** (*Language Processing Unit*), dedicated to **inference**: memory sits on the chip itself, and execution is scheduled in advance by the compiler. This greatly reduces the main bottleneck of token-by-token generation, which is memory access.

**What it means for sovereignty.** Groq serves **open-weight** models (gpt-oss, Qwen...). The model you test on Groq today can therefore, in principle, be **downloaded and deployed on your own servers** tomorrow (see the *Infrastructure* session on Day 1). You are testing a **model**, not just a **provider**.

### The prompt: every rule has a reason

## 3 · The stopwatch: how we measure one call

The function below is the **heart of the benchmark**. It does not just wait for the response: it receives it by **streaming**, piece by piece, which lets us time two things separately:
- the **first token** (TTFT), that is, responsiveness;
- the **end of the response**, that is, total latency.

It then reads the **official token count** returned by the provider, the one used for billing, and computes the cost. **Read the comments**, then run it.

````python
def measure(pid, text, language, on_token=None, max_attempts=4):
    """Translates `text` into `language` with provider `pid` and returns every measurement."""
    p = PROVIDERS[pid]
    messages = build_messages(text, language)
    timestamp = dt.datetime.now().isoformat(timespec="seconds")

    if MODES[pid] == "demo":                                       # no key: simulated response
        r = simulate_call(pid, messages, language, on_token)
    else:
        for attempt in range(max_attempts):
            try:
                t0 = time.perf_counter()                            # ⏱ start
                ttft, pieces, usage = None, [], None
                stream = CLIENTS[pid].chat.completions.create(
                    model=p["model"], messages=messages,
                    temperature=0,                                  # most stable answers possible
                    max_tokens=2048, stream=True,
                    stream_options={"include_usage": True},         # ask for the official count at the end of the stream
                    extra_body=p["params"] or None)
                for chunk in stream:
                    if chunk.choices and chunk.choices[0].delta.content:
                        if ttft is None:
                            ttft = time.perf_counter() - t0          # ⏱ first visible token
                        pieces.append(chunk.choices[0].delta.content)
                        if on_token: on_token(chunk.choices[0].delta.content)
                    extra = getattr(chunk, "model_extra", None) or {}
                    u = chunk.usage or (extra.get("x_groq") or {}).get("usage")
                    if u: usage = u if isinstance(u, dict) else u.model_dump()
                total = time.perf_counter() - t0                    # ⏱ finish
                out_text = "".join(pieces).strip()
                if usage:
                    det = usage.get("completion_tokens_details") or {}
                    r = dict(tokens_in=usage.get("prompt_tokens", 0), tokens_out=usage.get("completion_tokens", 0),
                             reasoning_tokens=(det or {}).get("reasoning_tokens") or 0, usage_estimated=False)
                else:                                               # no count: rough estimate (≈ 4 characters/token)
                    r = dict(tokens_in=sum(len(m["content"]) for m in messages) // 4,
                             tokens_out=len(out_text) // 4, reasoning_tokens=0, usage_estimated=True)
                r.update(text=out_text, ttft=ttft if ttft is not None else total, total=total)
                break
            except openai.RateLimitError as e:                       # 429: rate limit reached
                wait = float(e.response.headers.get("retry-after", 0) or 0) or 5 * 2 ** attempt
                print(f"   ⏸ {p['name']}: rate limit reached, retrying in {wait:.0f} s")
                time.sleep(min(wait, 60))
            except Exception as e:
                return dict(provider=pid, language=language, error=f"{type(e).__name__}: {str(e)[:150]}",
                            mode=MODES[pid], timestamp=timestamp)
        else:
            return dict(provider=pid, language=language, error="persistent rate limit", mode=MODES[pid], timestamp=timestamp)

    # --- derived indicators ---
    r["generation"] = max(r["total"] - r["ttft"], 1e-6)
    r["throughput"] = r["tokens_out"] / r["generation"]                      # tokens/s while writing
    r["cost_usd"] = (r["tokens_in"] * p["price_in"] + r["tokens_out"] * p["price_out"]) / 1e6
    r["cost_1000_docs"] = r["cost_usd"] * 1000
    r.update(provider=pid, model=p["model"], language=language, mode=MODES[pid], timestamp=timestamp, error=None)
    return r

print("✅ measure() function ready")
````

## 4 · 🏁 The side-by-side race

Both providers receive **exactly the same request at the same moment**. Watch **who starts** writing (TTFT) and **who finishes** (total latency). Re-run the cell with a different `RACE_LANGUAGE` to see whether the gap holds in Arabic or Kiswahili.

````python
RACE_LANGUAGE = "French"      # ✏️ also try "Arabic" or "Kiswahili"

def race(text, language, providers=None):
    providers = providers or COMPARE
    state = {pid: dict(text="", start=None, ttft=None, end=None, res=None) for pid in providers}
    rtl = "rtl" if language == "Arabic" else "ltr"

    def runner(pid):
        def on_token(t):
            if state[pid]["ttft"] is None: state[pid]["ttft"] = time.perf_counter() - state[pid]["start"]
            state[pid]["text"] += t
        state[pid]["start"] = time.perf_counter()
        state[pid]["res"] = measure(pid, text, language, on_token=on_token)
        state[pid]["end"] = time.perf_counter() - state[pid]["start"]

    def render():
        cols = []
        for pid in providers:
            s = state[pid]; elapsed = (time.perf_counter() - s["start"]) if s["start"] else 0
            if s["end"] is not None: status, clock = "✅ finished", f"{num(s['end'])} s"
            elif s["ttft"] is not None: status, clock = "✍️ writing...", f"{num(elapsed)} s"
            else: status, clock = "⏳ thinking...", f"{num(elapsed)} s"
            ttft = f"1st token: {num(s['ttft'])} s" if s["ttft"] is not None else "1st token: ..."
            cols.append(
                f'<div style="flex:1;min-width:260px;border:1px solid {C["sage"]};border-top:6px solid {COLOUR[pid]};border-radius:8px;padding:12px 14px;background:white">'
                f'<div style="display:flex;justify-content:space-between;align-items:baseline">'
                f'<b style="font-size:16px">{NAME[pid]}</b><span style="font-size:26px;font-weight:bold;color:{COLOUR[pid]}">{clock}</span></div>'
                f'<div style="font-size:12px;color:{C["slate"]}">{PROVIDERS[pid]["model"]} · {status} · {ttft}{" · 🎭 demo" if MODES[pid]=="demo" else ""}</div>'
                f'<div dir="{rtl}" style="margin-top:8px;min-height:150px;line-height:1.5;font-size:13.5px">{html.escape(s["text"])}</div></div>')
        return HTML(f'<div style="font-family:{FONT};display:flex;gap:12px;flex-wrap:wrap">{"".join(cols)}</div>')

    threads = [threading.Thread(target=runner, args=(pid,)) for pid in providers]
    for t in threads: t.start()
    area = display(render(), display_id=True)
    while any(t.is_alive() for t in threads):
        time.sleep(0.12); area.update(render())
    area.update(render())

    ok = {pid: state[pid] for pid in providers if state[pid]["res"] and not state[pid]["res"].get("error")}
    if len(ok) >= 2:
        ranking = sorted(ok, key=lambda pid: ok[pid]["end"])
        w, l = ranking[0], ranking[-1]
        box(f"🏆 {NAME[w]} finishes first",
            f"{NAME[w]}: <b>{num(ok[w]['end'])} s</b> · {NAME[l]}: <b>{num(ok[l]['end'])} s</b>, i.e. "
            f"<b>{num(ok[l]['end']/ok[w]['end'], 1)} times faster</b> on this document. "
            "A single race is only an anecdote: step 5 turns it into a measurement.", "ok")
    return state

_ = race(TEXT, RACE_LANGUAGE)
````

## 5 · The systematic benchmark

We move from anecdote to measurement, with three method rules:
1. **A warm-up**: one uncounted first call per provider, because the first connection is always slower.
2. **Repetitions**: each provider × language combination is measured several times, and we keep the **median**, which is insensitive to outliers.
3. **Alternation**: we alternate providers on every call so both face the **same network conditions**.

With 4 languages, 2 providers and 2 repetitions, that makes **16 calls**: about 1 minute.

````python
N_REPETITIONS = 2        # ✏️ 3 or more if you have the time (and the quota)

def benchmark(text, languages, providers=None, repetitions=2):
    providers = providers or COMPARE
    plan = [(rep, lang, pid) for rep in range(repetitions) for lang in languages for pid in providers]
    area = display(HTML(""), display_id=True)
    for pid in providers:                                      # warm-up, not counted
        measure(pid, text, languages[0])
    results = []
    for i, (rep, lang, pid) in enumerate(plan, 1):
        r = measure(pid, text, lang); r["repetition"] = rep + 1
        results.append(r)
        pct = i / len(plan) * 100
        area.update(HTML(f'<div style="font-family:{FONT}"><div style="background:{C["sage"]};border-radius:6px;height:14px;width:100%">'
                         f'<div style="background:{C["green"]};height:14px;border-radius:6px;width:{pct:.0f}%"></div></div>'
                         f'<span style="font-size:12px;color:{C["slate"]}">{i}/{len(plan)} · {NAME[pid]} → {lang} · '
                         f'{"❌ " + r["error"] if r.get("error") else "%.2f s" % r["total"]}</span></div>'))
    return pd.DataFrame(results)

df_runs = benchmark(TEXT, TARGET_LANGUAGES, repetitions=N_REPETITIONS)
errors = df_runs[df_runs["error"].notna()]
df_ok = df_runs[df_runs["error"].isna()].copy()

# Safeguard: a provider with NO successful measurement is replaced by demo measurements,
# so the rest of the notebook stays readable (these rows are never sent to the shared sheet).
for pid in COMPARE:
    if (df_ok["provider"] == pid).sum() == 0:
        MODES[pid] = "demo"
        box(f"{NAME[pid]}: no successful measurement, switching to demo mode",
            "Read the errors below, fix the configuration (key, model), then re-run from step 0.", "alert")
        df_runs = pd.concat([df_runs, benchmark(TEXT, TARGET_LANGUAGES, [pid], N_REPETITIONS)], ignore_index=True)
df_ok = df_runs[df_runs["error"].isna()].copy()
if len(errors):
    box(f"{len(errors)} call(s) failed", "<br>".join(f"{NAME.get(a, a)} / {b}: {c}" for a, b, c in errors[["provider", "language", "error"]].values), "warning")

summary = (df_ok.groupby(["provider", "language"])
           .agg(n=("total", "size"), ttft=("ttft", "median"), latency=("total", "median"),
                latency_p90=("total", lambda s: s.quantile(0.9)), throughput=("throughput", "median"),
                tok_in=("tokens_in", "mean"), tok_out=("tokens_out", "mean"),
                cost_1000=("cost_1000_docs", "mean"))
           .reset_index())
summary["_order"] = summary["provider"].map({p: i for i, p in enumerate(COMPARE)}) * 100 + summary["language"].map({l: i for i, l in enumerate(TARGET_LANGUAGES)})
summary = summary.sort_values("_order").drop(columns="_order").reset_index(drop=True)
view = summary.assign(provider=summary["provider"].map(NAME)).rename(columns={
    "provider": "Provider", "language": "Language", "n": "Runs", "ttft": "Median TTFT (s)",
    "latency": "Median latency (s)", "latency_p90": "P90 latency (s)", "throughput": "Throughput (tok/s)",
    "tok_in": "Input tokens", "tok_out": "Output tokens", "cost_1000": "Cost / 1,000 docs ($)"})
table(view, formats={"Median TTFT (s)": "{:.2f}", "Median latency (s)": "{:.2f}", "P90 latency (s)": "{:.2f}",
                     "Throughput (tok/s)": "{:.0f}", "Input tokens": "{:.0f}", "Output tokens": "{:.0f}",
                     "Cost / 1,000 docs ($)": "{:.3f}"}, title="Benchmark results (medians)")
````

## 6 · Quality: fast, but correct?

A fast, cheap engine that alters a figure is **unusable** for publication. We assess quality at two levels:

| Level | Method | Strength | Limit |
|---|---|---|---|
| **① Automatic** | Does every number in the source text appear in the translation? | Objective, instant, works on thousands of documents | Judges neither style nor terminology |
| **② Human** | A bilingual reviewer from the team scores from 1 to 5 | Judges meaning, register and terminology | Slow and subjective: does not scale |

In production, you combine the two: **the automatic check filters everything, and a human reviews a sample plus every flagged case.** This is the *verification before publication* principle from Day 1.

````python
ARABIC_DIGITS = str.maketrans("٠١٢٣٤٥٦٧٨٩٫٬", "0123456789.,")

def extract_numbers(t):
    """Every number in a text, normalised (thousands separators removed, decimal comma → point, Arabic-Indic digits → 0-9)."""
    t = t.translate(ARABIC_DIGITS)
    t = re.sub(r"(?<=\d)[,   ](?=\d{3}(?!\d))", "", t)      # 12,480 / 12 480 → 12480
    return Counter(v.replace(",", ".") for v in re.findall(r"\d+(?:[.,]\d+)?", t))

def check_numbers(source, translation):
    s, t = extract_numbers(source), extract_numbers(translation)
    missing, added = s - t, t - s
    rate = 100 * (1 - sum(missing.values()) / max(sum(s.values()), 1))
    return rate, sorted(missing.elements()), sorted(added.elements())

ref_text = TEXT if not ("demo" in MODES.values()) else TEXTS["cpi"]
print("Numbers in the source text:", sorted(extract_numbers(ref_text).elements()))

# Latest translation for each provider × language
latest = df_ok.sort_values("timestamp").groupby(["provider", "language"]).tail(1)
rows = []
for _, r in latest.iterrows():
    rate, miss, add = check_numbers(ref_text, r["text"])
    rows.append({"provider": r["provider"], "language": r["language"], "numbers_ok": rate,
                 "missing": ", ".join(miss) or "–", "added": ", ".join(add) or "–"})
df_numbers = pd.DataFrame(rows)
view = df_numbers.assign(provider=df_numbers["provider"].map(NAME)).rename(columns={
    "provider": "Provider", "language": "Language", "numbers_ok": "Numbers preserved (%)",
    "missing": "Missing from translation", "added": "Appeared in translation"})
table(view, formats={"Numbers preserved (%)": "{:.0f}"}, title="① Automatic check of the figures")

flags = df_numbers[df_numbers["numbers_ok"] < 100]
if len(flags):
    box("Discrepancy detected: this document must not be published without correction",
        "<br>".join(f"<b>{NAME[a]} → {b}</b>: expected {m}, found {j}" for a, b, m, j in flags[["provider", "language", "missing", "added"]].values)
        + "<br><i>Speed does not guarantee accuracy. Without this check, the error would have gone unnoticed.</i>", "alert")
else:
    box("All figures preserved", "No discrepancy detected. Still check meaning and terminology in step ②.", "ok")
````

## 7 · Scaling up: what speed makes *feasible*

Two very different usage regimes:

| Regime | Example in an NSO | What matters |
|---|---|---|
| **Interactive** | Website assistant, help for journalists, search across publications | **TTFT**: beyond 2 to 3 s, users give up |
| **Batch** | Translating the whole metadata catalogue, the archives, 54 country profiles | **Overall throughput** = latency × **concurrency** (parallel calls), capped by the **provider's limits** |

Overall throughput, in documents per minute, is the **smallest** of three ceilings:

$$\text{docs/min} = \min\Big(\underbrace{\tfrac{\text{concurrency} \times 60}{\text{latency}}}_{\text{engine speed}},\ \underbrace{\text{RPM}}_{\text{requests/min allowed}},\ \underbrace{\tfrac{\text{TPM}}{\text{tokens per doc}}}_{\text{tokens/min allowed}}\Big)$$

On top of this comes a **daily cap** (RPD). On a free tier, it is often **the limit, not the speed, that sets the duration**. That is why limits must be negotiated in a contract, not just the price.

````python
# ✏️ EDIT ME: your production scenarios
SCENARIOS = {
    "Monthly release (4 languages)": 4,
    "1,000 documents": 1_000,
    "Metadata catalogue: 2,000 sheets × 4 languages": 8_000,
    "Publication archive: 50,000 paragraphs × 4 languages": 200_000,
}
CONCURRENCY = 8     # simultaneous calls your pipeline would send

# Free-tier rate limits: ORDERS OF MAGNITUDE to be checked in each provider's console.
# Paid tiers raise these ceilings considerably (to be negotiated in the contract).
FREE_TIER_LIMITS = {
    "groq":       dict(rpm=30, tpm=8_000, rpd=1_000),
    "gemini":     dict(rpm=15, tpm=250_000, rpd=1_000),
    "openrouter": dict(rpm=20, tpm=None, rpd=50),
    "local":      dict(rpm=None, tpm=None, rpd=None),
}

def project(pid, n_docs, regime):
    s = summary[summary.provider == pid]
    latency, tokens = s["latency"].median(), (s["tok_in"] + s["tok_out"]).mean()
    cost_doc = s["cost_1000"].mean() / 1000
    ceilings = [CONCURRENCY * 60 / latency]
    if regime == "free":
        L = FREE_TIER_LIMITS.get(pid, {})
        if L.get("rpm"): ceilings.append(L["rpm"])
        if L.get("tpm"): ceilings.append(L["tpm"] / tokens)
    docs_min = min(ceilings)
    seconds = n_docs / docs_min * 60
    if regime == "free" and FREE_TIER_LIMITS.get(pid, {}).get("rpd") and n_docs > FREE_TIER_LIMITS[pid]["rpd"]:
        seconds = max(seconds, (math.ceil(n_docs / FREE_TIER_LIMITS[pid]["rpd"]) - 1) * 86400)
    return seconds, n_docs * cost_doc, docs_min

rows = []
for sc_name, n in SCENARIOS.items():
    row = {"Scenario": sc_name, "Documents": n}
    for pid in COMPARE:
        s_unl, cost, _ = project(pid, n, "unlimited")
        s_free, _, _ = project(pid, n, "free")
        row[f"{NAME[pid]} · time (limits lifted)"] = readable_duration(s_unl)
        row[f"{NAME[pid]} · time (free tier)"] = readable_duration(s_free)
        row[f"{NAME[pid]} · cost ($)"] = cost
    rows.append(row)
df_scale = pd.DataFrame(rows)
table(df_scale, formats=dict({c: "{:,.2f}" for c in df_scale.columns if "cost" in c}, Documents="{:,}"),
      title=f"Production projection · concurrency = {CONCURRENCY}")
````

## 8 · Sovereignty and confidentiality

Performance is not enough. For an NSO, the decisive question is often: **which data am I *allowed* to send to this provider?** The answer depends on the **class of data**, not on the engine:

| Data class | Example | Question to ask |
|---|---|---|
| 🟢 **Public** | Release already published, public metadata | No particular constraint |
| 🟠 **Under embargo** | Release **before** its publication date | Does the provider keep the data? Who can access it? Risk of a leak before the markets |
| 🔴 **Confidential** | Microdata, individual responses, registers | Does the statistics law allow a transfer outside the institution, or outside the country? |

Four questions for every provider: **Where** is the data processed? **How long** is it kept? Is it **used** to train models? **Can I leave** with the same model (open weights) if the contract ends?

````python
# ✏️ EDIT ME: answer "yes", "no" or "?" after reading the terms of use / the data processing agreement (DPA)
# Pre-filled: only what follows from the configuration. Your team checks the rest.
SOVEREIGNTY_GRID = {
    pid: {
        "Open-weight model (reversibility, on-premises deployment possible)": "yes" if PROVIDERS[pid]["open_weights"] else "no",
        "No retention of the data sent (or can be switched off)": "?",
        "Data not used to train models": "?",
        "Processing location known and contractual": "?",
        "Contract / DPA compatible with the national statistics law": "?",
        "Payment possible by the institution (invoice, purchase order)": "?",
    } for pid in COMPARE
}
if "local" in COMPARE:
    SOVEREIGNTY_GRID["local"] = {k: "yes" for k in SOVEREIGNTY_GRID["local"]}

# ✏️ EDIT ME: your team's decision for each data class
ALLOWED_USE = {pid: {"public": "yes", "embargo": "?", "confidential": "no"} for pid in COMPARE}

def sov_score(pid):
    ans = list(SOVEREIGNTY_GRID[pid].values())
    return 100 * sum(a == "yes" for a in ans) / len(ans)

sym = {"yes": "✅", "no": "❌", "?": "❔"}
rows = [{"Criterion": k, **{NAME[p]: sym.get(SOVEREIGNTY_GRID[p][k], "❔") for p in COMPARE}} for k in SOVEREIGNTY_GRID[COMPARE[0]]]
rows.append({"Criterion": "SOVEREIGNTY SCORE", **{NAME[p]: f"{sov_score(p):.0f}%" for p in COMPARE}})
for cls in ["public", "embargo", "confidential"]:
    rows.append({"Criterion": f"→ {cls.capitalize()} data allowed?", **{NAME[p]: sym.get(ALLOWED_USE[p][cls], "❔") for p in COMPARE}})
table(pd.DataFrame(rows), title="Sovereignty grid")
if any(v == "?" for p in COMPARE for v in SOVEREIGNTY_GRID[p].values()):
    box("Some \"?\" remain", "That is normal at this stage, and it is a result in itself: these are exactly the questions to put to the provider, "
        "and to your legal adviser, before any purchase.", "warning")
````

There is no "best engine" in absolute terms, only a best engine **for a given use**. Each criterion is brought to a scale of 100, where 100 is the best of the providers compared, then weighted. Change the **weights** to fit your office's context, and watch the ranking change across three typical profiles.

````python
# ✏️ EDIT ME: how much each criterion matters to YOUR office (any total, normalised automatically)
WEIGHTS = {"speed": 30, "cost": 20, "quality": 30, "sovereignty": 20}

PROFILES = {
    "Your weighting":               WEIGHTS,
    "Public interactive assistant": {"speed": 50, "cost": 15, "quality": 25, "sovereignty": 10},
    "Mass archive translation":     {"speed": 20, "cost": 40, "quality": 30, "sovereignty": 10},
    "Documents under embargo":      {"speed": 10, "cost": 10, "quality": 30, "sovereignty": 50},
}

base = df_ok.groupby("provider").agg(latency=("total", "median"), cost=("cost_1000_docs", "mean")).reindex(COMPARE)
q = df_numbers.groupby("provider").agg(numbers=("numbers_ok", "mean"), score=("human_score", "mean")).reindex(COMPARE)
q["quality"] = [0.5 * c + 0.5 * ((n - 1) / 4 * 100) if pd.notna(n) else c for c, n in q[["numbers", "score"]].values]
scores = pd.DataFrame({
    "speed": base["latency"].min() / base["latency"] * 100,
    "cost": (base["cost"].min() / base["cost"] * 100) if base["cost"].min() > 0 else 100,
    "quality": q["quality"] / q["quality"].max() * 100,
    "sovereignty": [sov_score(p) for p in COMPARE],
}, index=COMPARE)

def overall(profile):
    w = pd.Series(profile); w = w / w.sum()
    return (scores[w.index] * w).sum(axis=1)

res = pd.DataFrame({name: overall(w) for name, w in PROFILES.items()})
view = scores.round(0).join(res.round(0))
view.insert(0, "Provider", [NAME[p] for p in view.index])
view = view.reset_index(drop=True)
table(view, formats={c: "{:.0f}" for c in view.columns if c != "Provider"},
      title="Normalised scores (100 = best) and overall score by usage profile")

fig, ax = plt.subplots(figsize=(10, 0.5 * len(PROFILES) * len(COMPARE) + 0.8))
hh = 0.8 / len(COMPARE)
for i, pid in enumerate(COMPARE):
    y = [k + i * hh for k in range(len(PROFILES))]
    b = ax.barh(y, res.loc[pid], height=hh - 0.06, color=COLOUR[pid], label=NAME[pid]); bar_labels(ax, b, "{:.0f}")
ax.set_yticks([k + hh * (len(COMPARE) - 1) / 2 for k in range(len(PROFILES))]); ax.set_yticklabels(list(PROFILES))
ax.invert_yaxis(); ax.set_xlim(0, 118); ax.set_xticks([]); ax.spines["bottom"].set_visible(False)
ax.set_title("Overall score (out of 100) by usage profile", loc="left"); ax.legend(loc="lower right")
plt.tight_layout(); plt.show()

winner = res["Your weighting"].idxmax()
box(f"With your weighting: {NAME[winner]} ({res.loc[winner, 'Your weighting']:.0f}/100)",
    "Does the ranking change with the usage profile? If so, your office may need <b>two engines</b>: "
    "a fast engine for public data and a sovereign engine for sensitive data.", "info")
````

1. **Speed changes what is feasible, not just comfort.** Publishing every language version at the same time, translating archives overnight, offering an assistant that answers in under a second: these are *service decisions*, made possible by the infrastructure.
2. **Unit cost is low; real cost depends on volume, language and hidden tokens.** The language tax and reasoning tokens must appear in every quote.
3. **Rate limits matter as much as price.** On a free tier, they cap everything. In a contract, they are negotiable.
4. **Fast does not mean correct.** An automatic check of the figures is mandatory before any publication, backed by human review of a sample.
5. **Sovereignty is decided by data class.** A fast engine for public data and a sovereign engine (local, or open-weight on premises) for sensitive data can coexist. Open-weight models keep that door open.

### Limits of this benchmark, to state honestly
- **Small sample**: a few documents and a few repetitions. Orders of magnitude are reliable; decimals are not.
- **Local conditions**: latencies depend on the room's network, the time of day and the provider's load.
- **Dated prices and limits**: checked on the date shown, they change often.
- **Partial quality**: the number check does not judge style, and the human score depends on the reviewer.
- **Different models**: comparing Groq and Gemini compares an infrastructure *and* a model at once. To isolate the infrastructure, compare the same model on two hosts (Groq and OpenRouter).

## Annex C · Glossary

| Term | Definition |
|---|---|
| **Concurrency** | Number of calls a pipeline sends at the same time |
| **DPA** | *Data Processing Agreement*: contractual agreement on the processing of personal and confidential data |
| **Inference** | Using an already-trained model to produce an answer (as opposed to training) |
| **Language tax** | Extra tokens needed for languages that are less represented in the tokeniser's training data |
| **LPU** | *Language Processing Unit*: chip designed by Groq, dedicated to language-model inference |
| **Median / P90** | Middle value / value below which 90% of measurements fall (the "common bad case") |
| **OpenAI-compatible API** | Exchange format that has become a de facto standard: the same code works with many providers |
| **Open weights** (*open-weight*) | Model whose parameters can be downloaded and deployed on your own servers |
| **Reasoning tokens** | A model's internal "thinking" tokens, invisible in the answer but billed |
| **RPM / TPM / RPD** | Requests per minute / tokens per minute / requests per day allowed by the provider |
| **Streaming** | Receiving the answer piece by piece, as it is generated |
| **Token** | Unit of text processed and billed by the model (a piece of a word) |
| **TTFT** | *Time to first token*: delay before the first token of the answer appears |
