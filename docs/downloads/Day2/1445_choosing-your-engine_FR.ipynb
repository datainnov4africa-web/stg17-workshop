## 1 · Comprendre ce que l'on mesure

Quand vous envoyez un texte à un modèle de langage, voici ce qui se passe réellement :

```
 VOTRE TEXTE ──► découpé en TOKENS ──► réseau ──► file d'attente ──► ① lecture du prompt ──► ② écriture, token par token ──► RÉPONSE
                 (≈ morceaux de mots)                                  (« prefill », rapide)     (« decode », le plus long)
                 │                                                                              │
                 └──── facturé au PRIX D'ENTRÉE                                                  └──── facturé au PRIX DE SORTIE
 ◄──────────── TTFT : temps jusqu'au premier token ────────────────────────────────────────►◄── durée = tokens de sortie ÷ DÉBIT ──►
 ◄──────────────────────────────────────────────── LATENCE TOTALE ──────────────────────────────────────────────────────────────────►
```

| Grandeur | Définition | Pourquoi elle compte pour un INS |
|---|---|---|
| **Token** | Unité de texte lue par le modèle (≈ ¾ de mot en anglais) | Tout se facture en tokens. **Attention :** l'arabe, le kiswahili ou les langues moins représentées consomment souvent **plus de tokens pour le même contenu**, donc coûtent plus cher. C'est la « taxe linguistique ». |
| **TTFT** (*time to first token*) | Délai avant que le premier mot n'apparaisse | Ce que *ressent* un utilisateur : assistant du site web, outil d'aide aux journalistes, agent conversationnel |
| **Débit** (tokens/s) | Vitesse d'écriture une fois lancée | Décisif pour les textes longs et les traitements par lots |
| **Latence totale** | ≈ TTFT + tokens de sortie ÷ débit | Le temps d'un document, qui fixe le temps d'une campagne |
| **Coût par 1 000 documents** | (tokens d'entrée × prix d'entrée + tokens de sortie × prix de sortie) × 1 000 | **L'unité de comparaison pour les achats**, indépendante du fournisseur |
| **Limites de débit** | Plafonds par minute ou par jour (RPM, TPM, RPD) | Le vrai plafond du passage à l'échelle, souvent oublié |
| **Qualité** | Ici : fidélité des chiffres et note d'un relecteur | Un communiqué avec un chiffre faux est pire que pas de communiqué |
| **Souveraineté** | Où vont les données, qui peut les lire, peut-on reprendre le modèle chez soi ? | Loi statistique, confidentialité, embargo, dépendance à un fournisseur |

#### Exemple de calcul, à faire de tête
Un communiqué = **350 tokens** en entrée (prompt + texte) et **400 tokens** en sortie. Tarif : 0,15 $ / 0,60 $ par million de tokens.

$$\text{coût} = 350 \times \frac{0{,}15}{10^6} + 400 \times \frac{0{,}60}{10^6} \approx 0{,}000\,29 \text{ dollar par communiqué}$$

soit **0,29 $ pour 1 000 communiqués**.

Le coût unitaire est souvent **dérisoire**. Ce qui fait la différence, ce sont le **volume**, la **vitesse** et les **garanties**.

#### Pourquoi Groq est-il rapide ?
La plupart des fournisseurs font tourner les modèles sur des **GPU**, des processeurs graphiques polyvalents conçus d'abord pour l'entraînement. Groq a conçu sa propre puce, le **LPU** (*Language Processing Unit*), dédiée à **l'inférence** : la mémoire est placée sur la puce elle-même, et l'exécution est planifiée à l'avance par le compilateur. Le goulot d'étranglement principal de la génération token par token, l'accès à la mémoire, s'en trouve fortement réduit.

**Conséquence pour la souveraineté.** Groq sert des modèles **à poids ouverts** (gpt-oss, Qwen…). Le modèle que vous testez aujourd'hui sur Groq peut donc, en principe, être **téléchargé et déployé sur vos propres serveurs** demain (voir la session *Infrastructure* du Jour 1). Vous testez un **modèle**, pas seulement un **fournisseur**.

### Le prompt : chaque règle a une raison

Vous reconnaîtrez les principes vus ce matin : **rôle**, **contraintes explicites**, **format de sortie strict**. Le prompt est **identique pour tous les fournisseurs**. C'est la condition d'une comparaison honnête.

|---|---|
| Rôle : traducteur officiel d'un INS | Un registre familier ou journalistique |

## 3 · Le chronomètre : comment on mesure un appel

La fonction ci-dessous est le **cœur du benchmark**. Elle ne se contente pas d'attendre la réponse, elle la reçoit en **streaming**, morceau par morceau, ce qui permet de chronométrer séparément :
- le **premier token** (TTFT), c'est-à-dire la réactivité ;
- la **fin de la réponse**, c'est-à-dire la latence totale.

Elle récupère ensuite le **compteur officiel de tokens** renvoyé par le fournisseur, celui qui sert à la facturation, et calcule le coût. **Lisez les commentaires**, puis exécutez.

````python
def mesurer(fid, texte, langue, on_token=None, essais_max=4):
    """Traduit `texte` vers `langue` avec le fournisseur `fid` et renvoie toutes les mesures."""
    f = FOURNISSEURS[fid]
    messages = construire_messages(texte, langue)
    horodatage = dt.datetime.now().isoformat(timespec="seconds")

    if MODES[fid] == "démo":                                       # pas de clé : réponse simulée
        r = simuler_appel(fid, messages, langue, on_token)
    else:
        for essai in range(essais_max):
            try:
                t0 = time.perf_counter()                            # ⏱ top départ
                ttft, morceaux, usage = None, [], None
                flux = CLIENTS[fid].chat.completions.create(
                    model=f["modele"], messages=messages,
                    temperature=0,                                  # réponses les plus stables possibles
                    max_tokens=2048, stream=True,
                    stream_options={"include_usage": True},         # demande le compteur officiel en fin de flux
                    extra_body=f["params"] or None)
                for chunk in flux:
                    if chunk.choices and chunk.choices[0].delta.content:
                        if ttft is None:
                            ttft = time.perf_counter() - t0          # ⏱ premier token visible
                        morceaux.append(chunk.choices[0].delta.content)
                        if on_token: on_token(chunk.choices[0].delta.content)
                    extra = getattr(chunk, "model_extra", None) or {}
                    u = chunk.usage or (extra.get("x_groq") or {}).get("usage")
                    if u: usage = u if isinstance(u, dict) else u.model_dump()
                total = time.perf_counter() - t0                    # ⏱ arrivée
                texte_sortie = "".join(morceaux).strip()
                if usage:
                    det = usage.get("completion_tokens_details") or {}
                    r = dict(tokens_entree=usage.get("prompt_tokens", 0), tokens_sortie=usage.get("completion_tokens", 0),
                             tokens_raisonnement=(det or {}).get("reasoning_tokens") or 0, usage_estime=False)
                else:                                               # pas de compteur : estimation grossière (≈ 4 caractères/token)
                    r = dict(tokens_entree=sum(len(m["content"]) for m in messages) // 4,
                             tokens_sortie=len(texte_sortie) // 4, tokens_raisonnement=0, usage_estime=True)
                r.update(texte=texte_sortie, ttft=ttft if ttft is not None else total, total=total)
                break
            except openai.RateLimitError as e:                       # 429 : limite de débit atteinte
                attente = float(e.response.headers.get("retry-after", 0) or 0) or 5 * 2 ** essai
                print(f"   ⏸ {f['nom']} : limite de débit atteinte, nouvel essai dans {attente:.0f} s")
                time.sleep(min(attente, 60))
            except Exception as e:
                return dict(fournisseur=fid, langue=langue, erreur=f"{type(e).__name__}: {str(e)[:150]}",
                            mode=MODES[fid], horodatage=horodatage)
        else:
            return dict(fournisseur=fid, langue=langue, erreur="limite de débit persistante", mode=MODES[fid], horodatage=horodatage)

    # --- indicateurs dérivés ---
    r["generation"] = max(r["total"] - r["ttft"], 1e-6)
    r["debit"] = r["tokens_sortie"] / r["generation"]                           # tokens/s pendant l'écriture
    r["cout_usd"] = (r["tokens_entree"] * f["prix_entree"] + r["tokens_sortie"] * f["prix_sortie"]) / 1e6
    r["cout_1000_docs"] = r["cout_usd"] * 1000
    r.update(fournisseur=fid, modele=f["modele"], langue=langue, mode=MODES[fid], horodatage=horodatage, erreur=None)
    return r

print("✅ Fonction mesurer() prête")
````

## 4 · 🏁 La course côte à côte

Les deux fournisseurs reçoivent **exactement la même demande au même instant**. Regardez **qui commence** à écrire (TTFT) et **qui termine** (latence totale). Relancez la cellule en changeant `LANGUE_COURSE` pour voir si l'écart se maintient en arabe ou en kiswahili.

````python
LANGUE_COURSE = "anglais"      # ✏️ essayez aussi "arabe" ou "kiswahili"

def course(texte, langue, fournisseurs=None):
    fournisseurs = fournisseurs or COMPARER
    etat = {fid: dict(texte="", debut=None, ttft=None, fin=None, res=None) for fid in fournisseurs}
    rtl = "rtl" if langue == "arabe" else "ltr"

    def coureur(fid):
        def on_token(t):
            if etat[fid]["ttft"] is None: etat[fid]["ttft"] = time.perf_counter() - etat[fid]["debut"]
            etat[fid]["texte"] += t
        etat[fid]["debut"] = time.perf_counter()
        etat[fid]["res"] = mesurer(fid, texte, langue, on_token=on_token)
        etat[fid]["fin"] = time.perf_counter() - etat[fid]["debut"]

    def rendu():
        cols = []
        for fid in fournisseurs:
            e = etat[fid]; ecoule = (time.perf_counter() - e["debut"]) if e["debut"] else 0
            if e["fin"] is not None: statut, chrono = "✅ terminé", f"{fr(e['fin'])} s"
            elif e["ttft"] is not None: statut, chrono = "✍️ écrit…", f"{fr(ecoule)} s"
            else: statut, chrono = "⏳ réfléchit…", f"{fr(ecoule)} s"
            ttft = f"1er token : {fr(e['ttft'])} s" if e["ttft"] is not None else "1er token : …"
            cols.append(
                f'<div style="flex:1;min-width:260px;border:1px solid {C["sage"]};border-top:6px solid {COULEUR[fid]};border-radius:8px;padding:12px 14px;background:white">'
                f'<div style="display:flex;justify-content:space-between;align-items:baseline">'
                f'<b style="font-size:16px">{NOM[fid]}</b><span style="font-size:26px;font-weight:bold;color:{COULEUR[fid]}">{chrono}</span></div>'
                f'<div style="font-size:12px;color:{C["slate"]}">{FOURNISSEURS[fid]["modele"]} · {statut} · {ttft}{" · 🎭 démo" if MODES[fid]=="démo" else ""}</div>'
                f'<div dir="{rtl}" style="margin-top:8px;min-height:150px;line-height:1.5;font-size:13.5px">{html.escape(e["texte"])}</div></div>')
        return HTML(f'<div style="font-family:{FONT};display:flex;gap:12px;flex-wrap:wrap">{"".join(cols)}</div>')

    fils = [threading.Thread(target=coureur, args=(fid,)) for fid in fournisseurs]
    for t in fils: t.start()
    zone = display(rendu(), display_id=True)
    while any(t.is_alive() for t in fils):
        time.sleep(0.12); zone.update(rendu())
    zone.update(rendu())

    ok = {fid: etat[fid] for fid in fournisseurs if etat[fid]["res"] and not etat[fid]["res"].get("erreur")}
    if len(ok) >= 2:
        classement = sorted(ok, key=lambda fid: ok[fid]["fin"])
        g, p = classement[0], classement[-1]
        encadre(f"🏆 {NOM[g]} termine en premier",
                f"{NOM[g]} : <b>{fr(ok[g]['fin'])} s</b> · {NOM[p]} : <b>{fr(ok[p]['fin'])} s</b>, soit "
                f"<b>{fr(ok[p]['fin']/ok[g]['fin'], 1)} fois plus rapide</b> sur ce document. "
                "Une course isolée n'est qu'une anecdote : l'étape 5 la transforme en mesure.", "ok")
    return etat

_ = course(TEXTE, LANGUE_COURSE)
````

## 5 · Le benchmark systématique

On passe de l'anecdote à la mesure, avec trois règles de méthode :
1. **Un échauffement** : un premier appel non compté par fournisseur, car la première connexion est toujours plus lente.
2. **Des répétitions** : chaque combinaison fournisseur × langue est mesurée plusieurs fois, et on retient la **médiane**, insensible aux valeurs aberrantes.
3. **L'alternance** : on alterne les fournisseurs à chaque appel pour que les deux subissent les **mêmes conditions de réseau**.

Avec 4 langues, 2 fournisseurs et 2 répétitions, cela fait **16 appels** : environ 1 minute.

````python
N_REPETITIONS = 2        # ✏️ 3 ou plus si vous avez le temps (et le quota)

def benchmark(texte, langues, fournisseurs=None, repetitions=2):
    fournisseurs = fournisseurs or COMPARER
    plan = [(rep, lang, fid) for rep in range(repetitions) for lang in langues for fid in fournisseurs]
    zone = display(HTML(""), display_id=True)
    for fid in fournisseurs:                                   # échauffement, non compté
        mesurer(fid, texte, langues[0])
    resultats = []
    for i, (rep, lang, fid) in enumerate(plan, 1):
        r = mesurer(fid, texte, lang); r["repetition"] = rep + 1
        resultats.append(r)
        pct = i / len(plan) * 100
        zone.update(HTML(f'<div style="font-family:{FONT}"><div style="background:{C["sage"]};border-radius:6px;height:14px;width:100%">'
                         f'<div style="background:{C["green"]};height:14px;border-radius:6px;width:{pct:.0f}%"></div></div>'
                         f'<span style="font-size:12px;color:{C["slate"]}">{i}/{len(plan)} · {NOM[fid]} → {lang} · '
                         f'{"❌ " + r["erreur"] if r.get("erreur") else "%.2f s" % r["total"]}</span></div>'))
    return pd.DataFrame(resultats)

df_mesures = benchmark(TEXTE, LANGUES_CIBLES, repetitions=N_REPETITIONS)
erreurs = df_mesures[df_mesures["erreur"].notna()]
df_ok = df_mesures[df_mesures["erreur"].isna()].copy()

# Garde-fou : un fournisseur sans AUCUNE mesure réussie est remplacé par des mesures démo,
# pour que la suite du notebook reste lisible (ces lignes ne seront jamais envoyées dans la feuille).
for fid in COMPARER:
    if (df_ok["fournisseur"] == fid).sum() == 0:
        MODES[fid] = "démo"
        encadre(f"{NOM[fid]} : aucune mesure réussie, bascule en mode démo",
                "Lisez les erreurs ci-dessous, corrigez la configuration (clé, modèle), puis réexécutez depuis l'étape 0.", "alerte")
        df_mesures = pd.concat([df_mesures, benchmark(TEXTE, LANGUES_CIBLES, [fid], N_REPETITIONS)], ignore_index=True)
df_ok = df_mesures[df_mesures["erreur"].isna()].copy()
if len(erreurs):
    encadre(f"{len(erreurs)} appel(s) en erreur", "<br>".join(f"{NOM.get(a,a)} / {b} : {c}" for a, b, c in erreurs[["fournisseur","langue","erreur"]].values), "attention")

synthese = (df_ok.groupby(["fournisseur", "langue"])
            .agg(n=("total", "size"), ttft=("ttft", "median"), latence=("total", "median"),
                 latence_p90=("total", lambda s: s.quantile(0.9)), debit=("debit", "median"),
                 tok_in=("tokens_entree", "mean"), tok_out=("tokens_sortie", "mean"),
                 cout_1000=("cout_1000_docs", "mean"))
            .reset_index())
synthese["_ordre"] = synthese["fournisseur"].map({f: i for i, f in enumerate(COMPARER)}) * 100 + synthese["langue"].map({l: i for i, l in enumerate(LANGUES_CIBLES)})
synthese = synthese.sort_values("_ordre").drop(columns="_ordre").reset_index(drop=True)
vue = synthese.assign(fournisseur=synthese["fournisseur"].map(NOM)).rename(columns={
    "fournisseur": "Fournisseur", "langue": "Langue", "n": "Mesures", "ttft": "TTFT médian (s)",
    "latence": "Latence médiane (s)", "latence_p90": "Latence P90 (s)", "debit": "Débit (tok/s)",
    "tok_in": "Tokens entrée", "tok_out": "Tokens sortie", "cout_1000": "Coût / 1 000 docs ($)"})
tableau(vue, formats={"TTFT médian (s)": "{:.2f}", "Latence médiane (s)": "{:.2f}", "Latence P90 (s)": "{:.2f}",
                      "Débit (tok/s)": "{:.0f}", "Tokens entrée": "{:.0f}", "Tokens sortie": "{:.0f}",
                      "Coût / 1 000 docs ($)": "{:.3f}"}, titre="Résultats du benchmark (médianes)")
````

## 6 · La qualité : vite, mais juste ?

Un moteur rapide et bon marché qui altère un chiffre est **inutilisable** pour la publication. On évalue la qualité à deux niveaux :

| Niveau | Méthode | Force | Limite |
|---|---|---|---|
| **① Automatique** | Chaque nombre du texte source se retrouve-t-il dans la traduction ? | Objectif, instantané, applicable à des milliers de documents | Ne juge ni le style ni la terminologie |
| **② Humain** | Un relecteur bilingue de l'équipe note de 1 à 5 | Juge le sens, le registre et la terminologie | Lent, subjectif : ne passe pas à l'échelle |

En production, on combine les deux : **le contrôle automatique filtre tout, l'humain relit un échantillon et tous les cas signalés.** C'est le principe de la *vérification avant publication* vu au Jour 1.

````python
CHIFFRES_ARABES = str.maketrans("٠١٢٣٤٥٦٧٨٩٫٬", "0123456789.,")

def extraire_nombres(t):
    """Tous les nombres d'un texte, normalisés (séparateurs de milliers retirés, virgule décimale → point, chiffres arabo-indiens → 0-9)."""
    t = t.translate(CHIFFRES_ARABES)
    t = re.sub(r"(?<=\d)[,\u202f\u00a0 ](?=\d{3}(?!\d))", "", t)      # 12 480 / 12,480 → 12480
    return Counter(v.replace(",", ".") for v in re.findall(r"\d+(?:[.,]\d+)?", t))

def verifier_chiffres(source, traduction):
    s, t = extraire_nombres(source), extraire_nombres(traduction)
    manquants, ajoutes = s - t, t - s
    taux = 100 * (1 - sum(manquants.values()) / max(sum(s.values()), 1))
    return taux, sorted(manquants.elements()), sorted(ajoutes.elements())

texte_ref = TEXTE if not ("démo" in MODES.values()) else TEXTES["ipc"]
print("Nombres du texte source :", sorted(extraire_nombres(texte_ref).elements()))

# Dernière traduction obtenue pour chaque fournisseur × langue
derniere = df_ok.sort_values("horodatage").groupby(["fournisseur", "langue"]).tail(1)
lignes = []
for _, r in derniere.iterrows():
    taux, manq, ajo = verifier_chiffres(texte_ref, r["texte"])
    lignes.append({"fournisseur": r["fournisseur"], "langue": r["langue"], "chiffres_ok": taux,
                   "manquants": ", ".join(manq) or "–", "ajoutés": ", ".join(ajo) or "–"})
df_chiffres = pd.DataFrame(lignes)
vue = df_chiffres.assign(fournisseur=df_chiffres["fournisseur"].map(NOM)).rename(columns={
    "fournisseur": "Fournisseur", "langue": "Langue", "chiffres_ok": "Chiffres conservés (%)",
    "manquants": "Absents de la traduction", "ajoutés": "Apparus dans la traduction"})
tableau(vue, formats={"Chiffres conservés (%)": "{:.0f}"}, titre="① Contrôle automatique des chiffres")

alertes = df_chiffres[df_chiffres["chiffres_ok"] < 100]
if len(alertes):
    encadre("Écart détecté : ce document ne doit pas être publié sans correction",
            "<br>".join(f"<b>{NOM[a]} → {b}</b> : attendu {m}, trouvé {j}" for a, b, m, j in alertes[["fournisseur","langue","manquants","ajoutés"]].values)
            + "<br><i>La vitesse ne garantit pas la fidélité. Sans ce contrôle, l'erreur serait passée inaperçue.</i>", "alerte")
else:
    encadre("Tous les chiffres sont conservés", "Aucun écart détecté. Vérifiez tout de même le sens et la terminologie à l'étape ②.", "ok")
````

### ② Relecture humaine
Le relecteur de l'équipe qui maîtrise la langue lit les traductions ci-dessous, côte à côte, **sans regarder quel fournisseur est le plus rapide**, puis note chaque traduction :

| Note | Signification |
|:---:|---|
| **5** | Publiable telle quelle |
| **4** | Retouches mineures (une tournure, un mot) |
| **3** | Plusieurs corrections nécessaires, sens préservé |

## 7 · Passage à l'échelle : ce que la vitesse rend *faisable*

Deux régimes d'usage très différents :

| Régime | Exemple dans un INS | Ce qui compte |
|---|---|---|
| **Interactif** | Assistant sur le site web, aide aux journalistes, recherche dans les publications | Le **TTFT** : au-delà de 2 à 3 s, l'utilisateur décroche |
| **Par lots** | Traduire tout le catalogue de métadonnées, les archives, les 54 profils pays | Le **débit global** = latence × **concurrence** (appels en parallèle), plafonné par les **limites du fournisseur** |

Le débit global, en documents par minute, est le **plus petit** de trois plafonds :

$$\text{docs/min} = \min\Big(\underbrace{\tfrac{\text{concurrence} \times 60}{\text{latence}}}_{\text{vitesse du moteur}},\ \underbrace{\text{RPM}}_{\text{requêtes/min autorisées}},\ \underbrace{\tfrac{\text{TPM}}{\text{tokens par doc}}}_{\text{tokens/min autorisés}}\Big)$$

À cela s'ajoute un **plafond journalier** (RPD). Sur une offre gratuite, c'est souvent **la limite, et non la vitesse, qui fixe la durée**. D'où l'importance de négocier les limites dans un contrat, et pas seulement le prix.

````python
# ✏️ À MODIFIER : vos scénarios de production
SCENARIOS = {
    "Communiqué mensuel (4 langues)": 4,
    "1 000 documents": 1_000,
    "Catalogue de métadonnées : 2 000 fiches × 4 langues": 8_000,
    "Archives de publications : 50 000 paragraphes × 4 langues": 200_000,
}
CONCURRENCE = 8     # appels simultanés que votre pipeline enverrait

# Limites de débit de l'offre gratuite : ORDRES DE GRANDEUR à vérifier dans la console de chaque fournisseur.
# Les offres payantes relèvent fortement ces plafonds (à négocier dans le contrat).
LIMITES_GRATUIT = {
    "groq":       dict(rpm=30, tpm=8_000, rpd=1_000),
    "gemini":     dict(rpm=15, tpm=250_000, rpd=1_000),
    "openrouter": dict(rpm=20, tpm=None, rpd=50),
    "local":      dict(rpm=None, tpm=None, rpd=None),
}

def projeter(fid, n_docs, regime):
    s = synthese[synthese.fournisseur == fid]
    latence, tokens = s["latence"].median(), (s["tok_in"] + s["tok_out"]).mean()
    cout_doc = s["cout_1000"].mean() / 1000
    plafonds = [CONCURRENCE * 60 / latence]
    if regime == "gratuit":
        L = LIMITES_GRATUIT.get(fid, {})
        if L.get("rpm"): plafonds.append(L["rpm"])
        if L.get("tpm"): plafonds.append(L["tpm"] / tokens)
    docs_min = min(plafonds)
    secondes = n_docs / docs_min * 60
    if regime == "gratuit" and LIMITES_GRATUIT.get(fid, {}).get("rpd") and n_docs > LIMITES_GRATUIT[fid]["rpd"]:
        secondes = max(secondes, (math.ceil(n_docs / LIMITES_GRATUIT[fid]["rpd"]) - 1) * 86400)
    return secondes, n_docs * cout_doc, docs_min

lignes = []
for nom_sc, n in SCENARIOS.items():
    ligne = {"Scénario": nom_sc, "Documents": n}
    for fid in COMPARER:
        s_lib, cout, _ = projeter(fid, n, "sans limite")
        s_grat, _, _ = projeter(fid, n, "gratuit")
        ligne[f"{NOM[fid]} · durée (limites levées)"] = duree_lisible(s_lib)
        ligne[f"{NOM[fid]} · durée (offre gratuite)"] = duree_lisible(s_grat)
        ligne[f"{NOM[fid]} · coût ($)"] = cout
    lignes.append(ligne)
df_echelle = pd.DataFrame(lignes)
tableau(df_echelle, formats=dict({c: "{:,.2f}" for c in df_echelle.columns if "coût" in c}, Documents="{:,}"),
        titre=f"Projection en production · concurrence = {CONCURRENCE}")
````

## 8 · Souveraineté et confidentialité

La performance ne suffit pas. Pour un INS, la question décisive est souvent : **quelles données ai-je le *droit* d'envoyer à ce fournisseur ?** La réponse dépend de la **classe de données**, pas du moteur :

| Classe de données | Exemple | Question à se poser |
|---|---|---|
| 🟢 **Publique** | Communiqué déjà publié, métadonnées publiques | Aucune contrainte particulière |
| 🟠 **Sous embargo** | Communiqué **avant** sa date de publication | Le fournisseur conserve-t-il les données ? Qui y a accès ? Risque de fuite avant les marchés |
| 🔴 **Confidentielle** | Microdonnées, réponses individuelles, registres | La loi statistique autorise-t-elle un transfert hors de l'institution, ou hors du pays ? |

Quatre questions pour chaque fournisseur : **Où** les données sont-elles traitées ? **Combien de temps** sont-elles conservées ? **Servent-elles** à entraîner des modèles ? **Puis-je partir** avec le même modèle (poids ouverts) si le contrat s'arrête ?

````python
# ✏️ À MODIFIER : répondez "oui", "non" ou "?" après lecture des conditions d'utilisation / de l'accord de traitement des données (DPA)
# Pré-rempli : uniquement ce qui découle de la configuration. Le reste est à vérifier par votre équipe.
GRILLE_SOUVERAINETE = {
    fid: {
        "Modèle à poids ouverts (réversibilité, déploiement sur site possible)": "oui" if FOURNISSEURS[fid]["poids_ouverts"] else "non",
        "Aucune conservation des données envoyées (ou désactivable)": "?",
        "Données non utilisées pour entraîner des modèles": "?",
        "Lieu de traitement connu et contractuel": "?",
        "Contrat / DPA compatible avec la loi statistique nationale": "?",
        "Paiement possible par l'institution (facture, bon de commande)": "?",
    } for fid in COMPARER
}
if "local" in COMPARER:
    GRILLE_SOUVERAINETE["local"] = {k: "oui" for k in GRILLE_SOUVERAINETE["local"]}

# ✏️ À MODIFIER : la décision de votre équipe pour chaque classe de données
USAGE_AUTORISE = {fid: {"publique": "oui", "embargo": "?", "confidentielle": "non"} for fid in COMPARER}

def score_souv(fid):
    rep = list(GRILLE_SOUVERAINETE[fid].values())
    return 100 * sum(r == "oui" for r in rep) / len(rep)

sym = {"oui": "✅", "non": "❌", "?": "❔"}
lignes = [{"Critère": k, **{NOM[f]: sym.get(GRILLE_SOUVERAINETE[f][k], "❔") for f in COMPARER}} for k in GRILLE_SOUVERAINETE[COMPARER[0]]]
lignes.append({"Critère": "SCORE DE SOUVERAINETÉ", **{NOM[f]: f"{score_souv(f):.0f} %" for f in COMPARER}})
for classe in ["publique", "embargo", "confidentielle"]:
    lignes.append({"Critère": f"→ Données {classe} autorisées ?", **{NOM[f]: sym.get(USAGE_AUTORISE[f][classe], "❔") for f in COMPARER}})
tableau(pd.DataFrame(lignes), titre="Grille de souveraineté")
if any(v == "?" for f in COMPARER for v in GRILLE_SOUVERAINETE[f].values()):
    encadre("Des « ? » subsistent", "C'est normal à ce stade et c'est un résultat en soi : ce sont exactement les questions à poser au fournisseur, "
            "et à votre juriste, avant tout achat.", "attention")
````

## 9 · Décision : la fiche de synthèse

Il n'existe pas de « meilleur moteur » dans l'absolu, seulement un meilleur moteur **pour un usage donné**. Chaque critère est ramené sur 100, où 100 correspond au meilleur des fournisseurs comparés, puis pondéré. Modifiez les **poids** selon le contexte de votre office, et regardez le classement changer selon trois profils types.

````python
# ✏️ À MODIFIER : l'importance de chaque critère pour VOTRE office (total libre, normalisé automatiquement)
POIDS = {"vitesse": 30, "coût": 20, "qualité": 30, "souveraineté": 20}

PROFILS = {
    "Votre pondération":            POIDS,
    "Assistant interactif public":  {"vitesse": 50, "coût": 15, "qualité": 25, "souveraineté": 10},
    "Traduction massive d'archives": {"vitesse": 20, "coût": 40, "qualité": 30, "souveraineté": 10},
    "Documents sous embargo":        {"vitesse": 10, "coût": 10, "qualité": 30, "souveraineté": 50},
}

base = df_ok.groupby("fournisseur").agg(latence=("total", "median"), cout=("cout_1000_docs", "mean")).reindex(COMPARER)
q = df_chiffres.groupby("fournisseur").agg(chiffres=("chiffres_ok", "mean"), note=("note_humaine", "mean")).reindex(COMPARER)
q["qualite"] = [0.5 * c + 0.5 * ((n - 1) / 4 * 100) if pd.notna(n) else c for c, n in q[["chiffres", "note"]].values]
scores = pd.DataFrame({
    "vitesse": base["latence"].min() / base["latence"] * 100,
    "coût": (base["cout"].min() / base["cout"] * 100) if base["cout"].min() > 0 else 100,
    "qualité": q["qualite"] / q["qualite"].max() * 100,
    "souveraineté": [score_souv(f) for f in COMPARER],
}, index=COMPARER)

def total(profil):
    p = pd.Series(profil); p = p / p.sum()
    return (scores[p.index] * p).sum(axis=1)

res = pd.DataFrame({nom: total(p) for nom, p in PROFILS.items()})
vue = scores.round(0).join(res.round(0))
vue.insert(0, "Fournisseur", [NOM[f] for f in vue.index])
vue = vue.reset_index(drop=True)
tableau(vue, formats={c: "{:.0f}" for c in vue.columns if c != "Fournisseur"},
        titre="Scores normalisés (100 = meilleur) et score global selon le profil d'usage")

fig, ax = plt.subplots(figsize=(10, 0.5 * len(PROFILS) * len(COMPARER) + 0.8))
hh = 0.8 / len(COMPARER)
for i, fid in enumerate(COMPARER):
    y = [k + i * hh for k in range(len(PROFILS))]
    b = ax.barh(y, res.loc[fid], height=hh - 0.06, color=COULEUR[fid], label=NOM[fid]); etiquettes_barres(ax, b, "{:.0f}")
ax.set_yticks([k + hh * (len(COMPARER) - 1) / 2 for k in range(len(PROFILS))]); ax.set_yticklabels(list(PROFILS))
ax.invert_yaxis(); ax.set_xlim(0, 118); ax.set_xticks([]); ax.spines["bottom"].set_visible(False)
ax.set_title("Score global (sur 100) selon le profil d'usage", loc="left"); ax.legend(loc="lower right")
plt.tight_layout(); plt.show()

gagnant = res["Votre pondération"].idxmax()
encadre(f"Avec votre pondération : {NOM[gagnant]} ({res.loc[gagnant, 'Votre pondération']:.0f}/100)",
        "Le classement change-t-il selon le profil d'usage ? Si oui, votre office aura peut-être besoin de <b>deux moteurs</b> : "
        "un moteur rapide pour les données publiques et un moteur souverain pour les données sensibles.", "info")
````

1. **La vitesse change ce qui est faisable, pas seulement le confort.** Publier toutes les versions linguistiques à la même heure, traduire des archives en une nuit, offrir un assistant qui répond en moins d'une seconde : ce sont des *décisions de service*, rendues possibles par l'infrastructure.
2. **Le coût unitaire est faible ; le coût réel dépend du volume, de la langue et des tokens cachés.** La taxe linguistique et les tokens de raisonnement doivent figurer dans tout devis.
3. **Les limites de débit sont aussi importantes que le prix.** En offre gratuite, elles plafonnent tout. Dans un contrat, elles se négocient.
4. **Rapide ne veut pas dire juste.** Un contrôle automatique des chiffres est obligatoire avant toute publication, complété par une relecture humaine sur échantillon.
5. **La souveraineté se décide par classe de données.** Un moteur rapide pour les données publiques et un moteur souverain (local, ou à poids ouverts sur site) pour les données sensibles : les deux peuvent coexister. Les modèles à poids ouverts gardent cette porte ouverte.

### Limites de ce benchmark, à mentionner honnêtement
- **Petit échantillon** : quelques documents et quelques répétitions. Les ordres de grandeur sont fiables, les décimales ne le sont pas.
- **Conditions locales** : les latences dépendent du réseau de la salle, de l'heure et de la charge du fournisseur.
- **Prix et limites datés** : vérifiés le jour indiqué, ils évoluent souvent.
- **Qualité partielle** : le contrôle des chiffres ne juge pas le style, et la note humaine dépend du relecteur.
- **Modèles différents** : comparer Groq et Gemini compare à la fois une infrastructure *et* un modèle. Pour isoler l'infrastructure, comparez le même modèle sur deux hébergeurs (Groq et OpenRouter).

const COLONNES = ["horodatage","equipe","pays","mode","fournisseur","modele","poids_ouverts","texte",
  "langue_source","langue_cible","n_mesures","ttft_median_s","latence_mediane_s","latence_p90_s",

function doPost(e) {
  const classeur = SpreadsheetApp.getActiveSpreadsheet();
  const feuille = classeur.getSheetByName("Résultats") || classeur.insertSheet("Résultats");
  if (feuille.getLastRow() === 0) feuille.appendRow(COLONNES);

## Annexe C · Glossaire

| Terme | Définition |
|---|---|
| **API compatible OpenAI** | Format d'échange devenu standard de fait : le même code fonctionne avec de nombreux fournisseurs |
| **Concurrence** | Nombre d'appels envoyés simultanément par un pipeline |
| **DPA** | *Data Processing Agreement* : accord contractuel sur le traitement des données personnelles et confidentielles |
| **Inférence** | Utilisation d'un modèle déjà entraîné pour produire une réponse (par opposition à l'entraînement) |
| **LPU** | *Language Processing Unit* : puce conçue par Groq, dédiée à l'inférence des modèles de langage |
| **Médiane / P90** | Valeur centrale / valeur sous laquelle se trouvent 90 % des mesures (le « mauvais cas courant ») |
| **Poids ouverts** (*open-weight*) | Modèle dont les paramètres sont téléchargeables et déployables sur ses propres serveurs |
| **RPM / TPM / RPD** | Requêtes par minute / tokens par minute / requêtes par jour autorisés par le fournisseur |
| **Streaming** | Réception de la réponse morceau par morceau, au fil de sa génération |
| **Taxe linguistique** | Surcoût en tokens des langues moins représentées dans les données d'entraînement du tokeniseur |
| **Token** | Unité de texte traitée et facturée par le modèle (morceau de mot) |
| **Tokens de raisonnement** | Tokens de « réflexion » interne d'un modèle, invisibles dans la réponse mais facturés |
| **TTFT** | *Time to first token* : délai avant l'apparition du premier token de la réponse |
