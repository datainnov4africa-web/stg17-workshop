#### 0.1 &nbsp;Environnement et d&eacute;pendances

````python
# OUTIL — détecte Colab / Kaggle / local, installe ce qui manque, fixe le dossier de travail.
import importlib, os, subprocess, sys, textwrap, warnings
from pathlib import Path

warnings.filterwarnings("ignore", category=UserWarning)


def detect_environment() -> str:
    if "google.colab" in sys.modules or Path("/content").exists():
        return "colab"
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE") or Path("/kaggle/working").exists():
        return "kaggle"
    return "local"


ENV = detect_environment()
WORKDIR = {"colab": Path("/content/lab02"),
           "kaggle": Path("/kaggle/working/lab02")}.get(ENV, Path.cwd() / "lab02")
WORKDIR.mkdir(parents=True, exist_ok=True)
os.chdir(WORKDIR)

REQUIRED = {"requests": "requests", "pdfplumber": "pdfplumber", "pandas": "pandas"}
OPTIONAL = {"fitz": "pymupdf",        # page images, only needed for scanned PDFs
            "reportlab": "reportlab"}  # builds the bundled demonstration document


def ensure(packages: dict, hard: bool) -> list:
    missing = [pip for mod, pip in packages.items() if not importlib.util.find_spec(mod)]
    if not missing:
        return []
    print(f"installation : {', '.join(missing)} …")
    cmd = [sys.executable, "-m", "pip", "install", "-q", *missing]
    if ENV == "kaggle":
        cmd.insert(4, "--no-warn-script-location")
    try:
        subprocess.run(cmd, check=True, capture_output=True, timeout=600)
    except Exception as exc:
        msg = f"installation impossible de {missing} : {exc}"
        if hard:
            raise RuntimeError(msg + "\n(sur Kaggle, activez Internet dans le panneau de réglages)")
        print("  optionnel, ignoré —", msg)
    return missing


ensure(REQUIRED, hard=True)
ensure(OPTIONAL, hard=False)

import pandas as pd  # noqa: E402
pd.set_option("display.max_rows", 120, "display.width", 200,
              "display.max_colwidth", 60, "display.float_format", lambda v: f"{v:,.2f}")

print(f"environnement  : {ENV}")
print(f"python         : {sys.version.split()[0]}")
print(f"dossier travail: {WORKDIR}")
````

````python
# OUTIL — un seul endroit où chercher un secret, quel que soit l'environnement.
def get_secret(name: str, prompt: bool = True, blank_means: str = "mode hors-ligne"):
    """variable d'environnement → secret Colab → secret Kaggle → fichier .env → saisie."""
    if os.environ.get(name):
        return os.environ[name].strip()

    if ENV == "colab":
        try:
            from google.colab import userdata
            v = userdata.get(name)
            if v:
                return v.strip()
        except Exception:
            pass

    if ENV == "kaggle":
        try:
            from kaggle_secrets import UserSecretsClient
            v = UserSecretsClient().get_secret(name)
            if v:
                return v.strip()
        except Exception:
            pass

    for candidate in (WORKDIR / ".env", WORKDIR.parent / ".env", Path.cwd() / ".env"):
        if candidate.exists():
            for line in candidate.read_text().splitlines():
                k, _, v = line.partition("=")
                if k.strip() == name and v.strip():
                    return v.strip().strip("\"'")

    if prompt:
        try:                                   # échoue en exécution automatisée — c'est normal
            import getpass
            v = getpass.getpass(f"{name} (vide = {blank_means}) : ").strip()
            if v:
                os.environ[name] = v
                return v
        except Exception:
            pass
    return None


print(textwrap.dedent(f"""\
    Où mettre votre clé :
      Colab   ▸ icône clé dans la barre latérale ▸ ajouter un secret, accès notebook ACTIVÉ
      Kaggle  ▸ Add-ons ▸ Secrets ▸ rattacher à ce notebook
      Local   ▸ un fichier .env à côté du notebook :  GROQ_API_KEY=gsk_…
      Partout ▸ ou exécutez la cellule suivante et collez la clé (elle n'est pas conservée)

    Fournisseur choisi : {PROVIDERS[PROVIDER]['label']}
    Obtenir une clé    : {PROVIDERS[PROVIDER]['console'] or '—'}"""))
````

#### 0.3 &nbsp;Le client : une fonction, tous les fournisseurs

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#E8F5EF;border:1px solid #00A86A;

<div style="font-size:14px;color:#231F20;line-height:1.55">Trois choses m&eacute;ritent l&rsquo;attention dans la cellule suivante, parce que ce sont elles qui distinguent un notebook qui survit &agrave; un atelier d&rsquo;un notebook qui n&rsquo;y survit pas. Il <b>demande au fournisseur quels mod&egrave;les existent</b> plut&ocirc;t que de coder en dur un nom p&eacute;rim&eacute; le mois dernier. Il <b>retente apr&egrave;s un d&eacute;passement de quota</b>, ce que vous rencontrerez &agrave; trente personnes sur un seul wifi. Et il <b>enregistre la latence et le nombre de tokens de chaque appel</b>, de sorte qu&rsquo;&agrave; la partie 2 vous disposerez de chiffres r&eacute;els pour la discussion d&rsquo;achat de cet apr&egrave;s-midi.</div></div>


````python
import json, random, re, time
import requests

CALLS = []          # telemetry: one row per API call
OFFLINE = False     # set automatically below if no engine is reachable


class LLMError(RuntimeError):
    pass


def _headers(key):
    h = {"Authorization": f"Bearer {key}", "Content-Type": "application/json"}
    if PROVIDER == "openrouter":
        h["HTTP-Referer"] = "https://stg17.workshop"
        h["X-Title"] = "STG17 Lab 02"
    return h


def list_models(provider=None, key=None, timeout=30):
    """Demande au fournisseur ce qu'il sert aujourd'hui. Renvoie une liste d'identifiants."""
    p = PROVIDERS[provider or PROVIDER]
    if not p["base"]:
        return []
    r = requests.get(f"{p['base']}/models", headers=_headers(key or API_KEY), timeout=timeout)
    r.raise_for_status()
    payload = r.json()
    items = payload.get("data", payload if isinstance(payload, list) else [])
    return sorted({(m.get("id") or m.get("name")) for m in items if isinstance(m, dict)} - {None})


def pick_model(available, prefer):
    for pattern in prefer:
        for m in available:
            if re.search(pattern, m, re.I):
                return m
    return available[0] if available else None


def llm(messages, *, json_mode=False, temperature=0.0, max_tokens=4096,
        model=None, retries=4, label=""):
    """Une complétion. Renvoie le texte du modèle (ou un dict si json_mode)."""
    if OFFLINE:
        raise LLMError("mode hors-ligne : aucun modèle configuré")

    body = {"model": model or MODEL, "messages": messages,
            "temperature": temperature, "max_tokens": max_tokens}
    if json_mode:
        body["response_format"] = {"type": "json_object"}

    url = f"{PROVIDERS[PROVIDER]['base']}/chat/completions"
    last = None
    for attempt in range(retries):
        t0 = time.time()
        try:
            r = requests.post(url, headers=_headers(API_KEY), json=body, timeout=180)
        except requests.RequestException as exc:
            last = exc
            time.sleep(1.5 * 2 ** attempt)
            continue
        dt = time.time() - t0

        if r.status_code == 200:
            data = r.json()
            usage = data.get("usage") or {}
            CALLS.append(dict(label=label, provider=PROVIDER, model=body["model"],
                              seconds=round(dt, 2),
                              prompt_tokens=usage.get("prompt_tokens"),
                              completion_tokens=usage.get("completion_tokens")))
            text = data["choices"][0]["message"].get("content") or ""
            return parse_json(text) if json_mode else text

        # la sortie structurée n'est pas universelle — se dégrader, ne pas échouer
        if r.status_code in (400, 404, 422) and "response_format" in body:
            body.pop("response_format")
            continue
        if r.status_code in (408, 409, 429, 500, 502, 503, 504):
            wait = float(r.headers.get("retry-after", 0)) or (1.5 * 2 ** attempt + random.random())
            print(f"  {r.status_code} — waiting {wait:.1f}s (attempt {attempt + 1}/{retries})")
            time.sleep(wait)
            last = r.text[:300]
            continue
        raise LLMError(f"{r.status_code}: {r.text[:400]}")
    raise LLMError(f"abandon après {retries} tentatives : {last}")


def parse_json(text):
    """Les modèles emballent le JSON dans de la prose ou des balises plus souvent qu'on ne l'admet."""
    text = (text or "").strip()
    text = re.sub(r"^```(?:json)?|```$", "", text, flags=re.M).strip()
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass
    start = min((i for i in (text.find("{"), text.find("[")) if i >= 0), default=-1)
    if start >= 0:
        for end in range(len(text), start, -1):
            try:
                return json.loads(text[start:end])
            except json.JSONDecodeError:
                continue
    raise LLMError(f"aucun JSON dans la réponse : {text[:200]}")
````

#### 1.2 &nbsp;Lire les pages

````python
import unicodedata

NBSP = "\u00a0\u202f\u2009"


def norm(s: str) -> str:
    """Normalise une page pour que les comparaisons ultérieures survivent aux bizarreries du PDF."""
    s = unicodedata.normalize("NFKC", s or "")
    for ch in NBSP:
        s = s.replace(ch, " ")
    s = s.replace("\u2019", "'").replace("\u2013", "-").replace("\u2014", "-")
    return re.sub(r"[ \t]+", " ", s)


def read_pdf(path, max_pages=MAX_PAGES):
    import pdfplumber
    pages = []
    with pdfplumber.open(path) as pdf:
        total = len(pdf.pages)
        for i, page in enumerate(pdf.pages[:max_pages], 1):
            pages.append({"page": i, "text": norm(page.extract_text() or ""),
                          "chars": len(page.chars), "images": len(page.images)})
    return pages, total


PAGES, TOTAL_PAGES = read_pdf(PDF_PATH)
scanned = [p["page"] for p in PAGES if p["chars"] < 40 and p["images"]]

print(f"{TOTAL_PAGES} pages dans le document, {len(PAGES)} lues "
      f"({'toutes' if len(PAGES) == TOTAL_PAGES else f'plafonné à MAX_PAGES={MAX_PAGES}'})")
print(f"caractères récupérés : {sum(len(p['text']) for p in PAGES):,}")
if scanned:
    print(f"\npages sans couche texte (images scannées) : {scanned}")
    print("  → il leur faut un modèle vision ou de l'OCR ; voir la note ci-dessous.")
else:
    print("toutes les pages ont une couche texte — la voie économique est ouverte")
````

#### 1.3 &nbsp;Tri : quelles pages m&eacute;ritent un appel au mod&egrave;le&nbsp;?

````python
CUES = ("tableau", "table", "cuadro", "%", "source", "enqu", "survey")


def table_score(text: str) -> float:
    """
    Économique, explicable, sans modèle. Trois signaux : la densité de chiffres de la
    page, le nombre de lignes qui ressemblent à des lignes de tableau, et la fréquence
    du vocabulaire des tableaux.
    """
    if not text:
        return 0.0
    digits = sum(c.isdigit() for c in text) / max(len(text), 1)
    rows = sum(1 for l in text.splitlines()
               if len(re.findall(r"\d[\d  .,]*", l)) >= 3 and re.search(r"[A-Za-zÀ-ÿ]{3}", l))
    cues = sum(text.lower().count(c) for c in CUES)
    return round(digits * 100 + rows * 2 + cues, 1)


inventory = pd.DataFrame([{
    "page": p["page"],
    "score": table_score(p["text"]),
    "chars": len(p["text"]),
    "first line": (p["text"].splitlines() or [""])[0][:58],
} for p in PAGES]).sort_values("score", ascending=False)

display(inventory.head(12).style.hide(axis="index")
        .bar(subset=["score"], color="#9ED9C0")
        .set_caption("Pages les plus susceptibles de contenir des tableaux"))

# ============================== YOUR TURN ==================================
# Restez bref. Trois bonnes pages valent mieux que trente médiocres.
TARGET_PAGES = [int(p) for p in inventory.head(3)["page"]]
# TARGET_PAGES = [4, 5, 11]      # ← ou désignez-les vous-même après avoir regardé
# ===========================================================================
print("\npages retenues :", sorted(TARGET_PAGES))
````

Le score r&eacute;compense donc une grille dens&eacute;ment remplie, r&eacute;compense un tableau dont la ligne de total s&rsquo;additionne, et r&eacute;trograde toute lecture pr&eacute;sentant la signature d&rsquo;un s&eacute;parateur de milliers bris&eacute; &mdash; un petit entier coll&eacute; &agrave; gauche d&rsquo;un nombre &agrave; trois chiffres, ligne apr&egrave;s ligne.</div></div>



````python
# OUTIL — trois lecteurs et un arbitre. C'est ce qui permet au notebook de fonctionner
# sur un document que personne n'a réglé pour lui.
import pdfplumber

SP = r"[ \u00a0\u202f\u2009]"


def _num_pattern(max_groups=None):
    """Les nombres tels que les publications statistiques les impriment réellement."""
    grouped = (rf"[+\-\u2212]?\d{{1,3}}(?:{SP}\d{{3}})"
               + (rf"{{1,{max_groups}}}" if max_groups else "+") + r"(?:[.,]\d+)?%?")
    plain = rf"[+\-\u2212]?\d+(?:[.,]\d+)?%?(?:{SP}?\(\d+\))?"
    missing = r"…|\.\.\.|[–—-]"
    return rf"(?:{plain}|{missing})" if max_groups == 0 else rf"(?:{grouped}|{plain}|{missing})"


def parse_number(raw):
    """
    '12,8' → 12.8      '18 742' → 18742      '+14,2' → 14.2      '15,7%' → 15.7
    '17,0 (1)' → 17.0  '…' / '-' / '' → None (manquant, jamais zéro)
    """
    if raw is None:
        return None
    s = norm(str(raw)).strip()
    s = re.sub(r"\(\d+\)", "", s).replace("%", "").strip()
    if s in {"", "-", "–", "—", "…", "...", "..", "n.d.", "nd", "N/A", "na"}:
        return None
    s = s.replace(" ", "").replace("\u2212", "-")
    if "," in s and "." in s:                       # 1,234.5  vs  1.234,5
        s = s.replace(",", "") if s.rfind(".") > s.rfind(",") else s.replace(".", "").replace(",", ".")
    elif "," in s:
        s = s.replace(",", ".") if re.search(r",\d{1,2}$", s) else s.replace(",", "")
    try:
        return float(s)
    except ValueError:
        return None


def split_numbers(chunk, max_groups=None):
    return [parse_number(t) for t in re.findall(_num_pattern(max_groups), chunk)]


NUM = _num_pattern()
# Un code de nomenclature devant le libellé (« 01.1.7 Légumes ») est capturé à part
# plutôt que de laisser ses chiffres faire échouer la reconnaissance.
ROW = re.compile(rf"^(?:(?P<code>\d[\d.]{{0,9}}){SP}+)?(?P<label>[^\d\n]{{2,64}}?){SP}+"
                 rf"(?P<nums>{NUM}(?:{SP}+{NUM})+){SP}*$")
CAPTION = re.compile(r"^\s*(tableau|table|cuadro|jadual|annex|annexe|appendix)\s*"
                     r"(?P<no>[\dIVX]+)?\s*[—:\-–.]\s*(?P<title>.+)$", re.I)
YEAR = re.compile(r"\b(?:1[89]|20)\d{2}\b")
TOTAL_WORDS = re.compile(r"^(total|ensemble|whole|national|pays|country|tous|all|general|"
                         r"g[ée]n[ée]ral|overall|toutes)\b", re.I)


def _is_header(values):
    vs = [v for v in values if v is not None]
    return bool(vs) and all(float(v).is_integer() and 1800 <= v <= 2100 for v in vs)


def _dedupe(columns):
    """
    Deux colonnes nommées « 2025 » sont pires que deux colonnes « col3 » et « col4 » :
    tout l'aval regroupe par période, donc les doublons fusionnent silencieusement les
    trimestres en une seule série et toutes les réconciliations échouent ensuite.
    Ne jamais les laisser passer.
    """
    seen, out = {}, []
    for c in columns:
        c = str(c)
        seen[c] = seen.get(c, 0) + 1
        out.append(c if seen[c] == 1 else f"{c} ({seen[c]})")
    return out


# « T1 2025 T2 2025 » — garder le qualificatif devant chaque année, pas seulement l'année
YEAR_WITH_PREFIX = re.compile(r"((?:[^\s\d]{1,10}\s+){0,2})((?:1[89]|20)\d{2})")


def _columns_from_header(line):
    hits = YEAR_WITH_PREFIX.findall(line)
    labels = [(pre + yr).strip() for pre, yr in hits]
    if len(set(labels)) < len(labels):
        labels = [yr for _, yr in hits]
    return labels


def _repair_widths(rows, width):
    """
    « 18 742 100,0 114 » est ambigu : un nombre à huit chiffres, ou un total et deux
    colonnes ? Une ligne plus étroite que le reste de son tableau est relue en relâchant
    progressivement la règle « espace = séparateur de milliers » jusqu'à ce que les
    largeurs concordent.
    """
    for r in rows:
        if len(r["values"]) >= width:
            continue
        for mg in (2, 1, 0):
            alt = split_numbers(r["nums"], mg)
            if len(alt) == width:
                r["values"], r["repaired"] = alt, True
                break


# ---------------------------------------------------------------- reader 1 & 2: geometry
def _clean(c):
    return norm(c or "").replace("\n", " ").strip()


def split_code(label):
    """
    « 01.1.7 Légumes » → ('01.1.7', 'Légumes').
    « 65 ans et plus » → ('', '65 ans et plus') — ce 65 est la donnée, pas un code.
    Seul compte un code pointé, ou complété par un zéro de tête.
    """
    m = re.match(r"^(\d[\d.]{0,9})\s+(.{2,})$", label or "")
    if not m:
        return "", label
    head = m.group(1)
    looks_like_code = "." in head or (len(head) >= 2 and head.startswith("0"))
    return (head, m.group(2)) if looks_like_code else ("", label)


def _all_years(vals):
    n = [parse_number(v) for v in vals]
    n = [x for x in n if x is not None]
    return bool(n) and all(float(x).is_integer() and 1800 <= x <= 2100 for x in n)


def _is_label_row(row):
    """Surtout des mots plutôt que des chiffres — ou une ligne d'années nues, qui reste un en-tête."""
    vals = [_clean(c) for c in row[1:] if _clean(c)]
    if not vals or _all_years(vals):
        return True
    return sum(parse_number(v) is not None for v in vals) <= len(vals) / 2


def _merge_headers(header_rows, width):
    """Propage un libellé fusionné vers la droite, puis joint : « Hommes » + « 2020 » → « Hommes 2020 »."""
    filled = []
    for row in header_rows:
        out, last = [], ""
        for i in range(width):
            c = _clean(row[i]) if i < len(row) else ""
            last = c or last
            out.append(last)
        filled.append(out)
    cols = []
    for i in range(1, width):
        parts, seen = [], set()
        for row in filled:
            if row[i] and row[i] not in seen:
                parts.append(row[i])
                seen.add(row[i])
        cols.append(" ".join(parts) or f"col{i}")
    return _dedupe(cols)


def _caption_above(page, bbox):
    try:
        above = page.crop((0, max(bbox[1] - 60, 0), page.width, max(bbox[1] - 1, 1)))
        lines = [norm(l).strip() for l in (above.extract_text() or "").splitlines()]
    except Exception:
        return ""
    lines = [l for l in lines if len(l) > 6]
    return lines[-1] if lines else ""


LAYOUT_STRATEGIES = (
    ("lines", {"vertical_strategy": "lines", "horizontal_strategy": "lines"}),
    ("text", {"vertical_strategy": "text", "horizontal_strategy": "text",
              "text_x_tolerance": 2, "intersection_tolerance": 8}),
)


def layout_tables(pdf_path, only=None, min_rows=2, min_numeric=0.55):
    out = []
    with pdfplumber.open(pdf_path) as pdf:
        for pno, page in enumerate(pdf.pages, 1):
            if only and pno not in only:
                continue
            best, best_score = [], 0.0
            for name, settings in LAYOUT_STRATEGIES:
                try:
                    found = page.find_tables(settings) or []
                except Exception:
                    continue
                blocks, score = [], 0.0
                for tb in found:
                    rows = [[_clean(c) for c in r] for r in (tb.extract() or [])]
                    rows = [r for r in rows if any(r)]
                    if len(rows) < min_rows + 1 or len(rows[0]) < 2:
                        continue
                    cells = [c for r in rows[1:] for c in r[1:] if _clean(c)]
                    share = (sum(parse_number(c) is not None for c in cells) / len(cells)
                             if cells else 0)
                    if share < min_numeric:
                        continue
                    blocks.append((name, tb, rows))
                    score += share * len(rows)
                if score > best_score:
                    best, best_score = blocks, score

            for name, tb, rows in best:
                width = max(len(r) for r in rows)
                rows = [r + [""] * (width - len(r)) for r in rows]
                header, i = [], 0
                while i < len(rows) and _is_label_row(rows[i]) and i < 3:
                    header.append(rows[i]); i += 1
                columns = _merge_headers(header, width) if header \
                    else [f"col{j}" for j in range(1, width)]
                body = []
                for r in rows[i:]:
                    label, values = _clean(r[0]), [parse_number(c) for c in r[1:]]
                    if not label or all(v is None for v in values):
                        continue
                    code, label = split_code(label)
                    body.append({"label": label, "code": code, "values": values,
                                 "nums": "", "line": " ".join(x for x in r if x),
                                 "repaired": False})
                if len(body) >= min_rows:
                    cap = _caption_above(page, tb.bbox)
                    out.append({"caption": cap or f"table on page {pno}", "title": cap,
                                "page": pno, "columns": columns, "rows": body,
                                "origin": f"layout:{name}"})
    return out


# ---------------------------------------------------------------- reader 2b: captions
def caption_tables(page_text, page_no=None):
    """Le lecteur classique : une légende annonce le tableau, les lignes suivent jusqu'à la note de source."""
    stop = re.compile(r"^\s*(source|note|champ|lecture|\(\d\)|…\s*:|\*)", re.I)
    tables, cur = [], None

    def close():
        nonlocal cur
        if cur and cur["rows"]:
            widths = [len(r["values"]) for r in cur["rows"]]
            _repair_widths(cur["rows"], max(set(widths), key=widths.count))
            width = max(len(r["values"]) for r in cur["rows"])
            cols = cur["columns"][:width]
            cur["columns"] = _dedupe(cols + [f"col{i+1}" for i in range(len(cols), width)])
            tables.append(cur)
        cur = None

    for line in page_text.splitlines():
        s_ = norm(line).strip()
        cap = CAPTION.match(s_)
        if cap:
            close()
            years = YEAR.findall(s_)
            cur = {"caption": s_, "title": cap.group("title").strip(), "page": page_no,
                   "rows": [], "columns": years if len(years) >= 2 else [],
                   "origin": "caption"}
            continue
        if cur is None:
            continue
        if stop.match(s_):
            close(); continue
        m = ROW.match(s_)
        if not m:
            continue
        vals = split_numbers(m.group("nums"))
        if len(vals) < 2:
            continue
        if _is_header(vals) and not cur["rows"]:
            cur["columns"] = [str(int(v)) for v in vals]
            continue
        cur["rows"].append({"label": m.group("label").strip(" .:—–-"),
                            "code": (m.group("code") or "").strip(), "values": vals,
                            "nums": m.group("nums"), "line": s_, "repaired": False})
    close()
    return tables


# ---------------------------------------------------------------- reader 3: lines of text
def block_tables(page_text, page_no=None, min_rows=2):
    lines = [norm(l).rstrip() for l in page_text.splitlines()]
    matches = []
    for ln in lines:
        m = ROW.match(ln.strip())
        vals = split_numbers(m.group("nums")) if m else []
        matches.append((len(matches), m, vals) if m and len(vals) >= 2 else None)

    tables, run = [], []

    def flush():
        if len(run) < min_rows:
            run.clear(); return
        widths = [len(v) for _, _, v in run]
        width = max(set(widths), key=widths.count)
        rows = []
        for i, m, v in run:
            code, label = (m.group("code") or "").strip(), m.group("label").strip(" .:—–-")
            if code and not ("." in code or (len(code) >= 2 and code.startswith("0"))):
                code, label = "", f"{code} {label}"      # '65 years and over'
            rows.append({"label": label, "code": code, "values": v, "nums": m.group("nums"),
                         "line": lines[i].strip(), "repaired": False})
        _repair_widths(rows, width)
        width = max(len(r["values"]) for r in rows)
        first, columns = run[0][0], []
        for back in range(first - 1, max(first - 4, -1), -1):
            cand = lines[back].strip()
            if not cand:
                continue
            hm = ROW.match(cand)
            if hm:
                hv = split_numbers(hm.group("nums"))
                if _is_header(hv) and len(hv) >= width - 1:
                    columns = [str(int(v)) for v in hv if v is not None]; break
            years = _columns_from_header(cand)
            if len(years) >= width - 1:
                columns = years; break
        caption = ""
        for back in range(first - 1, max(first - 6, -1), -1):
            cand = lines[back].strip()
            if len(cand) > 8 and not ROW.match(cand):
                caption = cand; break
        columns = _dedupe(columns[:width]
                          + [f"col{i+1}" for i in range(len(columns), width)])
        tables.append({"caption": caption or f"table on page {page_no}", "title": caption,
                       "page": page_no, "columns": columns, "rows": rows,
                       "origin": "text-block"})
        run.clear()

    prev = None
    for item in matches:
        if item is None:
            flush(); prev = None; continue
        if prev and abs(len(item[2]) - prev) > 1:
            flush()
        run.append(item); prev = len(item[2])
    flush()
    return tables
````

````python
# OUTIL — l'arbitre : quelle lecture de chaque page conserve-t-on ?
def _reconciles(t):
    """La ligne de total égale-t-elle la somme de ses parties ? Une lecture éclatée échouera."""
    totals = [r for r in t["rows"] if TOTAL_WORDS.match(r["label"])]
    parts = [r for r in t["rows"] if not TOTAL_WORDS.match(r["label"])]
    if not totals or len(parts) < 2:
        return False
    for i in range(len(totals[0]["values"])):
        tv = totals[0]["values"][i]
        pv = [r["values"][i] for r in parts
              if i < len(r["values"]) and r["values"][i] is not None]
        if tv is None or len(pv) < 2:
            continue
        if abs(sum(pv) - tv) <= max(1.0, abs(tv) * 0.005) or abs(sum(pv) - 100) <= 1.0:
            return True
    return False


def _thousands_split(tables):
    """Un entier inférieur à 1000 juste à gauche d'un nombre à trois chiffres, ligne après ligne."""
    hits = rows = 0
    for t in tables:
        for r in t["rows"]:
            rows += 1
            for a, b in zip(r["values"], r["values"][1:]):
                if (a is not None and b is not None and float(a).is_integer()
                        and 0 < a < 1000 and 100 <= abs(b) < 1000):
                    hits += 1
                    break
    return bool(rows) and hits / rows >= 0.5


def reading_score(tables):
    """
    Pas « qui a trouvé le plus de nombres » — un lecteur qui coupe un chiffre en trois
    en trouve davantage et se trompe. On récompense les cellules remplies, on pénalise
    une grille creuse, on récompense les colonnes nommées, une vraie légende, et un
    tableau dont le total s'additionne.
    """
    if not tables:
        return 0.0
    cells = slots = 0
    for t in tables:
        for r in t["rows"]:
            cells += sum(v is not None for v in r["values"])
            slots += max(len(t["columns"]), len(r["values"]))
    density = cells / slots if slots else 0
    named = sum(1 for t in tables for c in t["columns"] if not str(c).startswith("col"))
    titled = sum(1 for t in tables if CAPTION.match(t.get("caption") or ""))
    bonus = 1.6 if any(_reconciles(t) for t in tables) else 1.0
    return (cells * density + named * 0.5 + titled * 2) * bonus


def _labels(tables):
    return {r["label"].lower().strip() for t in tables for r in t["rows"] if r["label"]}


def read_document(pdf_path, pages, only=None):
    """Exécute chaque lecteur sur chaque page, garde la meilleure lecture, et dit laquelle a gagné."""
    by_page = {}
    for t in layout_tables(pdf_path, only=only):
        by_page.setdefault(t["page"], []).append(t)

    tables, report = [], []
    for p in pages:
        if only and p["page"] not in only:
            continue
        candidates = {
            "layout": by_page.get(p["page"], []),
            "caption": [dict(t, origin="caption") for t in caption_tables(p["text"], p["page"])],
            "text-block": block_tables(p["text"], p["page"]),
        }
        ranked = sorted(candidates.items(), key=lambda kv: reading_score(kv[1]), reverse=True)
        name, chosen = ranked[0]
        if chosen and _thousands_split(chosen):
            mine = _labels(chosen)

            def same(a, b):
                a, b = a[:14], b[:14]
                return len(a) >= 6 and (a.startswith(b[:len(a)]) or b.startswith(a[:len(b)]))

            for alt_name, alt in ranked[1:]:
                if not alt or _thousands_split(alt):
                    continue
                theirs = _labels(alt)
                overlap = sum(any(same(x, y) for x in mine) for y in theirs) / max(len(theirs), 1)
                narrower = (max((len(t["columns"]) for t in alt), default=0)
                            < max((len(t["columns"]) for t in chosen), default=0))
                if overlap >= 0.6 and narrower:
                    name, chosen = alt_name, alt
                    break
        report.append({"page": p["page"], **{k: len(v) for k, v in candidates.items()},
                       "kept": name if chosen else "none", "tables": len(chosen),
                       "cells": int(sum(sum(v is not None for v in r["values"])
                                        for t in chosen for r in t["rows"]))})
        tables += chosen
    return tables, report


def baseline_facts(tables):
    out = []
    for t in tables:
        for r in t["rows"]:
            for i, v in enumerate(r["values"]):
                if v is None:
                    continue
                out.append(dict(table=t["title"] or t["caption"], caption=t["caption"],
                                page=t["page"], label=r["label"], code=r.get("code", ""),
                                position=i, value=v, line=r["line"],
                                origin=t.get("origin", ""),
                                column=t["columns"][i] if i < len(t["columns"]) else f"col{i+1}"))
    return out
````

````python
EXTRACTION_SYSTEM = """Vous transcrivez des tableaux issus de publications statistiques officielles.
Vous êtes un transcripteur, pas un analyste. Vous recopiez ce qui est imprimé. Vous
n'estimez jamais, n'interpolez jamais, ne complétez jamais une série, ne convertissez
jamais une unité.

Règles qui comptent plus que la fluidité :
- Une virgule dans un nombre est un séparateur DÉCIMAL : 12,8 vaut 12.8.
- Un espace dans un nombre est un séparateur de MILLIERS : 18 742 vaut 18742.
- « … », « - », « n.d. » et une cellule vide signifient tous MANQUANT. Renvoyez null.
  Ne renvoyez jamais 0.
- Un appel de note collé à une valeur (« 17,0 (1) ») ne fait pas partie de la valeur :
  renvoyez 17.0.
- Si vous ne pouvez pas lire une cellule avec certitude, renvoyez null. Une omission est
  un petit problème ; un chiffre inventé est un problème grave.
- source_line doit être RECOPIÉE CARACTÈRE PAR CARACTÈRE depuis le texte de la page qui
  vous est fourni. Ne la nettoyez pas, ne la reformatez pas, ne la traduisez pas. Elle
  sera vérifiée automatiquement contre la page, et une ligne introuvable entraînera le
  rejet de tout l'enregistrement.

Renvoyez un seul objet JSON : {"facts": [ ... ]} et rien d'autre. Sans prose, sans balises."""

EXTRACTION_USER = """PAGE {page} de la publication « {title} ».

Transcrivez chaque cellule de données de chaque tableau de cette page dans le schéma
ci-dessous. Un objet par cellule : un indicateur, une zone, une période, une valeur.

{{"facts": [{{
  "indicator": "clé courte en snake_case, réutilisée pour toutes les cellules du même tableau",
  "indicator_label": "le libellé tel qu'imprimé, dans la langue du document",
  "area": "le libellé de ligne, recopié exactement tel qu'imprimé",
  "period": "le libellé de colonne : 2025, T4 2025, 2021-2025 …",
  "value": 12.8,
  "unit": "%",
  "page": {page},
  "source_line": "la ligne du texte de la page dont cette cellule a été lue, mot pour mot",
  "is_total": false
}}]}}

TEXTE DE LA PAGE — c'est le seul matériau que vous puissiez utiliser :
<<<
{text}
>>>"""
````

````python
def extract_page(page, title, max_chars=12000):
    text = page["text"][:max_chars]
    out = llm([{"role": "system", "content": EXTRACTION_SYSTEM},
               {"role": "user", "content": EXTRACTION_USER.format(
                   page=page["page"], title=title, text=text)}],
              json_mode=True, temperature=0.0, max_tokens=6000,
              label=f"extract p{page['page']}")
    facts = out.get("facts", out if isinstance(out, list) else [])
    for f in facts:
        f["page"] = page["page"]
        f["value"] = parse_number(f.get("value")) if isinstance(f.get("value"), str) \
            else f.get("value")
    return facts


_head = [l for l in PAGES[0]["text"].splitlines() if l.strip()][:3]
DOC_TITLE = (_head[1] if len(_head) > 1 else (_head[0] if _head else PDF_PATH.stem))[:90]
EXTRACTED = []

if OFFLINE:
    print("hors-ligne — enregistrements dérivés de la lecture par règles, sans modèle.\n"
          "La différence se verra plus loin : les règles savent quels nombres sont sur quelle\n"
          "ligne, mais pas que la colonne 3 est une densité, ni que « Total » et « Ensemble\n"
          "du pays » désignent la même ligne. Nommer les choses est précisément le rôle du\n"
          "modèle.\n")
    for b in BASELINE:
        EXTRACTED.append(dict(indicator=re.sub(r"\W+", "_", b["table"].lower())[:28] or "series",
                              indicator_label=b["table"], area=b["label"], period=str(b["column"]),
                              value=b["value"], unit="%" if "%" in b["caption"] else "",
                              page=b["page"], source_line=b["line"],
                              is_total=bool(re.match(r"(total|ensemble|whole)", b["label"], re.I))))
else:
    for pno in sorted(TARGET_PAGES):
        page = next(p for p in PAGES if p["page"] == pno)
        try:
            got = extract_page(page, DOC_TITLE)
            EXTRACTED += got
            print(f"page {pno} : {len(got)} cellules  ({CALLS[-1]['seconds']}s)")
        except Exception as exc:
            print(f"page {pno} : ÉCHEC — {exc}")

df = pd.DataFrame(EXTRACTED)
print(f"\n{len(df)} cellules issues du modèle, {len(BASELINE)} issues des règles")
display(df.head(10))
````

#### 3.1 &nbsp;Portes 1 et 2 &mdash; la preuve est-elle r&eacute;elle&nbsp;?

````python
def number_variants(v):
    """
    Toutes les façons plausibles d'imprimer `v`, et aucune qui arrondirait à un autre
    chiffre — ainsi 22,3 ne produit jamais « 22 » et ne peut pas correspondre au « 22,4 »
    voisin.
    """
    v = float(v)
    out = set()
    for d in (0, 1, 2, 3):
        if round(v, d) != round(v, 6):
            continue
        ip, _, fp = f"{abs(v):.{d}f}".partition(".")
        ints = {ip}
        if len(ip) > 3:
            ints |= {f"{int(ip):,}".replace(",", " "), f"{int(ip):,}",
                     f"{int(ip):,}".replace(",", ".")}
        for i in ints:
            out.add(f"{i},{fp}" if fp else i)
            out.add(f"{i}.{fp}" if fp else i)
    return {o for o in out if o}


def value_on_page(v, text) -> bool:
    if v is None or not text:
        return False
    hay = norm(text)
    return any(re.search(r"(?<![\d.,])" + re.escape(c) + r"(?![\d]|[.,]\d)", hay)
               for c in number_variants(v))


PAGE_TEXT = {p["page"]: p["text"] for p in PAGES}


def gate_provenance(f):
    ok = value_on_page(f.get("value"), PAGE_TEXT.get(f.get("page"), ""))
    return ok, "" if ok else f"{f.get('value')} n'est pas imprimé sur la page {f.get('page')}"


def gate_evidence(f):
    line = norm(f.get("source_line") or "").strip()
    if not line:
        return False, "aucune ligne source citée"
    ok = line in norm(PAGE_TEXT.get(f.get("page"), ""))
    return ok, "" if ok else "la ligne source citée n'apparaît pas sur cette page"
````

#### 3.2 &nbsp;Porte 3 &mdash; unit&eacute;s et plages plausibles

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#FDF6E4;border:1px solid #D49A00;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#8A6400;margin-bottom:5px">YOUR TURN</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Renseignez <code>MY_UNIT_RULES</code> pour votre propre document. C&rsquo;est une connaissance m&eacute;tier qu&rsquo;aucun mod&egrave;le ne poss&egrave;de : vous seul savez que votre s&eacute;rie de ch&ocirc;mage est un pourcentage entre 0 et 100, que votre colonne de population est en milliers, et qu&rsquo;une valeur de 4&nbsp;912 dans une colonne intitul&eacute;e &laquo;&nbsp;%&nbsp;&raquo; n&rsquo;a aucun sens. &Eacute;crite une fois, elle prot&egrave;ge toutes les ex&eacute;cutions futures.</div></div>


````python
PERCENT = {"%", "percent", "pourcentage", "pour cent", "p.100"}


def seed_unit_rules(facts):
    """
    Construit une règle de départ par indicateur à partir de ce qui a réellement été
    extrait, pour que la porte 3 fasse quelque chose sur un document pour lequel personne
    n'a écrit de règles. Un pourcentage reçoit la borne 0-100 ; le reste reçoit une borne
    large autour de sa propre dispersion, ce qui attrape une erreur d'ordre de grandeur
    sans prétendre à une connaissance du domaine.
    """
    rules = {}
    for f in facts:
        ind = f.get("indicator")
        if ind is None:
            continue
        r = rules.setdefault(ind, {"units": set(), "values": []})
        if f.get("unit"):
            r["units"].add(str(f["unit"]).strip().lower())
        if f.get("value") is not None:
            r["values"].append(f["value"])
    out = {}
    for ind, r in rules.items():
        units = r["units"] or None
        vals = r["values"] or [0]
        if units and units <= PERCENT:
            out[ind] = dict(units=units, lo=-100, hi=100, source="auto : pourcentage")
        else:
            lo, hi = min(vals), max(vals)
            span = max(abs(hi), abs(lo), 1)
            out[ind] = dict(units=units, lo=lo - span * 10, hi=hi + span * 10,
                            source="auto : dispersion observée ×10")
    return out


AUTO_RULES = seed_unit_rules(EXTRACTED)

# ============================== YOUR TURN ==================================
# Les règles que vous écrivez vous-même battent tout ce qui est dérivé des données,
# parce que vous seul savez que la colonne population est en milliers et ne peut pas
# dépasser 2 millions. Ce que vous omettez retombe sur AUTO_RULES, plus faible mais
# jamais inerte.
MY_UNIT_RULES = {
    # "unemployment": dict(units={"%", "pourcentage"}, lo=0, hi=100),
    # "population":   dict(units={"milliers", "thousands"}, lo=0, hi=2_000_000),
}
# ===========================================================================
UNIT_RULES = {**AUTO_RULES, **MY_UNIT_RULES}
print(f"{len(UNIT_RULES)} indicateurs sous règle de plage "
      f"({len(MY_UNIT_RULES)} écrites à la main, {len(UNIT_RULES) - len(MY_UNIT_RULES)} dérivées)")
for k, v in list(UNIT_RULES.items())[:6]:
    print(f"   {str(k)[:34]:<34} {str(v.get('units') or 'toute unité')[:26]:<26} "
          f"[{v['lo']:,.0f} … {v['hi']:,.0f}]")


def gate_unit_range(f):
    rule, problems = UNIT_RULES.get(f.get("indicator")), []
    unit = (f.get("unit") or "").strip().lower()
    v = f.get("value")
    if rule:
        if rule["units"] is not None and unit not in {u.lower() for u in rule["units"]}:
            problems.append(f"unité « {f.get('unit')} » non autorisée pour {f.get('indicator')}")
        if v is not None and not (rule["lo"] <= v <= rule["hi"]):
            problems.append(f"{v} est hors de la plage plausible pour {f.get('indicator')}")
    elif unit in PERCENT and v is not None and not (-100 <= v <= 100):
        # aucune règle écrite pour cet indicateur, mais l'unité seule suffit
        problems.append(f"{v} ne peut pas être un pourcentage")
    return not problems, "; ".join(problems)


def gate_is_active(indicator):
    return indicator in MY_UNIT_RULES
````

#### 3.3 &nbsp;Porte 4 &mdash; les parties s&rsquo;additionnent-elles&nbsp;?

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#E8F5EF;border:1px solid #00A86A;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#00704A;margin-bottom:5px">Why this step matters</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Une ligne de total est un cadeau : la publication vous a d&eacute;j&agrave; dit &agrave; combien les parties doivent sommer. Quand ce n&rsquo;est pas le cas, vous savez que quelque chose cloche &mdash; mais pas <i>quoi</i>. La cellule ci-dessous va donc un cran plus loin et demande, pour chaque ligne tour &agrave; tour : si cette seule cellule contenait la valeur qui comblerait l&rsquo;&eacute;cart, cette valeur serait-elle r&eacute;ellement imprim&eacute;e sur la page&nbsp;? En g&eacute;n&eacute;ral une ou deux lignes r&eacute;pondent oui, et ce sont les cellules &agrave; relire. Du triage automatique, sans aucun mod&egrave;le.</div></div>


````python
TOTAL_WORDS = re.compile(r"^(total|ensemble|whole|national|pays|country|tous|all|general|"
                         r"g[ée]n[ée]ral|overall|toutes)\b", re.I)


def is_total(f):
    return bool(f.get("is_total")) or bool(TOTAL_WORDS.match(str(f.get("area", ""))))


def gate_arithmetic(facts, tolerance=0.005):
    """Réconcilie chaque groupe (indicateur, période) doté d'une ligne de total."""
    checks = []
    groups = {}
    for f in facts:
        if f.get("value") is None:
            continue
        groups.setdefault((f.get("indicator"), f.get("period")), []).append(f)

    for (ind, period), rows in groups.items():
        totals = [r for r in rows if is_total(r)]
        parts = [r for r in rows if not is_total(r)]
        if not totals or len(parts) < 2:
            continue
        published = totals[0]["value"]
        computed = sum(r["value"] for r in parts)
        biggest = max(r["value"] for r in parts)
        additive = abs(computed - published) <= max(1.0, abs(published) * tolerance)
        shares = abs(computed - 100) <= 1.0
        if additive or shares:
            checks.append(dict(indicator=ind, period=period, kind="sum" if additive else "shares",
                               computed=round(computed, 2), published=published,
                               gap=round(computed - published, 2), passed=True,
                               candidates=[], note=""))
            continue

        # UN TAUX N'EST PAS ADDITIF. Les taux de chômage régionaux ne somment pas au taux
        # national et ne le feront jamais : le chiffre national est une moyenne pondérée,
        # pas un total. Les tester produirait un mur de fausses alertes ; la porte
        # s'abstient donc et le dit, ce qui n'est pas la même chose que réussir.
        unit = (parts[0].get("unit") or "").strip().lower()
        if unit in PERCENT:
            checks.append(dict(indicator=ind, period=period, kind="rate — not additive",
                               computed=round(computed, 2), published=published, gap=None,
                               passed=None, candidates=[],
                               note="un taux ne se réconcilie pas par addition ; cette porte s'abstient. "
                                    "Pondérez par la population pour le contrôler correctement."))
            continue
        if published < biggest:
            # Un total plus petit que sa plus grande partie n'est pas un total. C'est une
            # moyenne, une densité, un indice ou un taux. S'abstenir plutôt que de crier.
            checks.append(dict(indicator=ind, period=period, kind="not a sum",
                               computed=round(computed, 2), published=published, gap=None,
                               passed=None, candidates=[],
                               note=f"le chiffre publié ({published:,.1f}) est plus petit que la plus "
                                    f"grande partie ({biggest:,.1f}) : c'est une moyenne ou une "
                                    f"densité, pas un total ; cette porte s'abstient".replace(",", " ")))
            continue

        gap = published - computed
        candidates = [dict(area=r["area"], read=r["value"], should_be=round(r["value"] + gap, 2))
                      for r in parts
                      if value_on_page(r["value"] + gap, PAGE_TEXT.get(r.get("page"), ""))]
        checks.append(dict(indicator=ind, period=period, kind="sum",
                           computed=round(computed, 2), published=published,
                           gap=round(gap, 2), passed=False, candidates=candidates[:3], note=""))
        note = (f"les parties somment à {computed:,.1f} mais le document publie {published:,.1f}"
                .replace(",", " "))
        for r in parts + totals:
            r.setdefault("_arith", []).append(note)
        for c in candidates[:3]:
            for r in parts:
                if r["area"] == c["area"]:
                    r["_arith"].append(f"cellule à relire : {c['should_be']:,.1f} est imprimé page "
                                       f"{r.get('page')} et comblerait l'écart"
                                       .replace(",", " "))
    return checks
````

#### 3.4 &nbsp;Porte 5 &mdash; le second avis gratuit

````python
def row_cells(label, page=None, line=None):
    """
    Le même libellé de ligne apparaît souvent dans plusieurs tableaux d'une même page.
    Le modèle nous a déjà dit quelle LIGNE il a lue, et le lecteur par règles a enregistré
    la ligne de chaque rangée trouvée — on apparie donc d'abord sur la ligne. C'est à cela
    que sert la preuve citée.
    """
    if line:
        exact = [b for b in BASELINE if norm(b["line"]) == norm(line)]
        if exact:
            return exact
    lab = norm(str(label)).lower().strip()
    hits = [b for b in BASELINE
            if norm(b["label"]).lower().strip() == lab and (page is None or b["page"] == page)]
    lines = {b["line"] for b in hits}
    return hits if len(lines) <= 1 else []      # ligne ambiguë → pas d'avis, pas un échec


def gate_cross_read(f):
    """
    match     même valeur, même colonne, même ligne  → accord fort
    mismatch  cette colonne de cette ligne contient un autre nombre
    row-only  la valeur est sur la ligne mais dans une autre colonne  (dérive de colonne)
    absent    la valeur n'est pas du tout sur cette ligne            (décalage de ligne)
    no-row    le lecteur par règles n'a jamais trouvé cette ligne    (pas d'avis, pas un échec)
    """
    cells = row_cells(f.get("area"), f.get("page"), f.get("source_line"))
    v = f.get("value")
    if not cells or v is None:
        return True, "", "no-row"
    same_col = [c for c in cells if str(c["column"]) == str(f.get("period"))]
    if same_col:
        got = same_col[0]["value"]
        if abs(got - v) < 1e-9:
            return True, "", "match"
        return False, f"la lecture par règles de cette cellule donne {got}, le modèle a dit {v}", "mismatch"
    if any(abs(c["value"] - v) < 1e-9 for c in cells):
        return True, "", "row-only"
    return False, (f"{v} n'est nulle part sur la ligne « {f.get('area')} » ; cette ligne contient "
                   f"{[c['value'] for c in cells][:6]}"), "absent"
````

#### 3.5 &nbsp;Ex&eacute;cuter toutes les portes et appliquer la r&egrave;gle de publication

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#FBEDEB;border:1px solid #B83B2E;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#8C2A20;margin-bottom:5px">La r&egrave;gle de publication</div>
<div style="font-size:14px;color:#231F20;line-height:1.55"><b>&Eacute;chec &agrave; la porte 1 ou 2 &rarr; le chiffre est &eacute;cart&eacute;.</b> Aucune preuve dans la page signifie aucune raison de croire qu&rsquo;il existe. Il n&rsquo;atteint jamais le tableau de bord.<br><b>&Eacute;chec &agrave; la porte 3, 4 ou 5 &rarr; le chiffre est publi&eacute; mais marqu&eacute;</b> &laquo;&nbsp;&agrave; revoir&nbsp;&raquo;, avec la raison visible pour le lecteur, jusqu&rsquo;&agrave; ce qu&rsquo;un statisticien tranche.<br><br>Cette asym&eacute;trie est d&eacute;lib&eacute;r&eacute;e. Un chiffre invent&eacute; est une fausse affirmation. Un chiffre qui ne se r&eacute;concilie pas peut n&rsquo;&ecirc;tre qu&rsquo;un arrondi, une note de bas de page, ou une propri&eacute;t&eacute; r&eacute;elle des donn&eacute;es &mdash; mais le lecteur a le droit de le savoir avant de le citer.</div></div>


````python
ARITHMETIC = gate_arithmetic(EXTRACTED)

rows = []
for f in EXTRACTED:
    p_ok, p_msg = gate_provenance(f)
    e_ok, e_msg = gate_evidence(f)
    u_ok, u_msg = gate_unit_range(f)
    x_ok, x_msg, verdict = gate_cross_read(f)
    a_msgs = f.get("_arith", [])
    a_ok = not a_msgs

    if not (p_ok and e_ok):
        status = "dropped"
    elif p_ok and e_ok and u_ok and a_ok and x_ok:
        status = "verified"
    else:
        status = "review"

    f["status"] = status
    f["reasons"] = [m for m in (p_msg, e_msg, u_msg, x_msg, *a_msgs) if m]
    f["cross_read"] = verdict
    rows.append(dict(indicator=f.get("indicator"), area=f.get("area"), period=f.get("period"),
                     value=f.get("value"), unit=f.get("unit"), page=f.get("page"),
                     provenance=p_ok, evidence=e_ok, unit_range=u_ok,
                     arithmetic=a_ok, cross_read=verdict, status=status,
                     reason=" | ".join(f["reasons"])))

REPORT = pd.DataFrame(rows)
VERIFIED = [f for f in EXTRACTED if f["status"] != "dropped"]
REPORT.to_csv(WORKDIR / "verification_report.csv", index=False)

counts = REPORT.status.value_counts().reindex(["verified", "review", "dropped"]).fillna(0).astype(int)

# Combien de portes ont réellement eu un avis ? Une porte qui s'abstient n'est pas une
# porte qui passe, et un rapport qui masque la différence est pire que pas de rapport.
COVERAGE = {
    "provenance": 1.0,
    "preuve citée": 1.0,
    "unité et plage": sum(f.get("indicator") in UNIT_RULES for f in EXTRACTED) / max(len(EXTRACTED), 1),
    "arithmétique": sum(1 for f in EXTRACTED if f.get("_arith") is not None
                      or any(c["passed"] is not None and c["indicator"] == f.get("indicator")
                             for c in ARITHMETIC)) / max(len(EXTRACTED), 1),
    "relecture croisée": sum(f.get("cross_read") not in (None, "no-row") for f in EXTRACTED)
                  / max(len(EXTRACTED), 1),
}

print(f"{len(REPORT)} cellules extraites")
for k, v in counts.items():
    share = v / max(len(REPORT), 1) * 100
    print(f"  {k:<9} {v:>4}  {'█' * int(share / 2.5)} {share:.0f}%")
print(f"\n{len(VERIFIED)} cellules atteindront le tableau de bord "
      f"(dont {counts.get('review', 0)} signalées à revoir)")
print("rapport complet écrit dans verification_report.csv")

print("\ncouverture des portes — à quelle fréquence chacune a eu un avis :")
for gate, cov in COVERAGE.items():
    bar = "█" * int(cov * 20)
    flag = "" if cov >= 0.5 else "   ← faible sur ce document"
    print(f"   {gate:<14} {cov * 100:5.0f}%  {bar:<20}{flag}")
if not MY_UNIT_RULES:
    print("\n   La porte 3 ne tourne que sur des plages auto-dérivées. Elles attrapent une")
    print("   erreur d'ordre de grandeur, pas une unité fausse — vous seul savez que votre")
    print("   colonne de population est en milliers. Écrire deux ou trois règles dans")
    print("   MY_UNIT_RULES est le meilleur emploi de trois minutes dans ce notebook.")
if COVERAGE["relecture croisée"] < 0.5:
    print("\n   Les lecteurs par règles n'ont pas trouvé les lignes rapportées par le modèle :")
    print("   le second avis gratuit est donc largement indisponible ici. Le décalage de ligne")
    print("   — la seule erreur que la provenance ne peut pas attraper — est donc peu gardé.")
    print("   Dites-le dans LIMITATIONS.md plutôt que de laisser croire le contraire.")

if len(ARITHMETIC):
    arith = pd.DataFrame(ARITHMETIC)
    arith["verdict"] = arith.passed.map({True: "réconcilie", False: "NE RÉCONCILIE PAS",
                                         None: "s'abstient"}).fillna("s'abstient")
    display(arith[["indicator", "period", "kind", "computed", "published", "gap",
                   "verdict", "note"]].style.hide(axis="index")
            .set_caption("Réconciliation avec les totaux publiés"))
````

````python
# Tout ce que les portes n'ont pas aimé, avec la raison. C'est la liste de travail du relecteur.
problems = REPORT[REPORT.status != "verified"]
if len(problems):
    display(problems[["status", "indicator", "area", "period", "value", "page", "reason"]]
            .style.hide(axis="index")
            .set_caption(f"{len(problems)} cellules exigeant un humain — lisez chaque ligne"))
else:
    print("toutes les portes passent sur toutes les cellules.")
    print("C'est possible sur un tableau propre — mais vérifiez que vous ne testez pas le vide :")
    print("exécutez la cellule suivante pour confirmer que les portes se déclenchent.")
````

#### 3.6 &nbsp;Prouver que les portes fonctionnent

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#E8F5EF;border:1px solid #00A86A;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#00704A;margin-bottom:5px">Why this step matters</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Une suite de tests qui ne tombe jamais en &eacute;chec n&rsquo;est pas une preuve de qualit&eacute; ; c&rsquo;est la preuve d&rsquo;une suite de tests qui ne teste rien. Avant de faire confiance &agrave; un rapport tout vert, corrompez les donn&eacute;es expr&egrave;s et confirmez que chaque porte attrape son propre type d&rsquo;erreur. Faites ceci chaque fois que vous changez un prompt ou un mod&egrave;le.</div></div>


````python
import copy

def probe(mutate, description):
    sample = copy.deepcopy(EXTRACTED)
    target = mutate(sample)
    if target is None:
        return print(f"  – {description} : aucune cellule adaptée dans ce document, ignoré")
    p_ok, _ = gate_provenance(target)
    e_ok, _ = gate_evidence(target)
    u_ok, _ = gate_unit_range(target)
    x_ok, _, _ = gate_cross_read(target)
    caught = [n for n, ok in (("provenance", p_ok), ("preuve", e_ok),
                              ("unité/plage", u_ok), ("relecture croisée", x_ok)) if not ok]
    mark = "✓" if caught else "✗ NON ATTRAPÉ"
    print(f"  {mark} {description:<44} → {', '.join(caught) or "aucune porte déclenchée"}")


def first(sample, pred=lambda f: f.get("value") is not None):
    return next((f for f in sample if pred(f)), None)


print("injection d'erreurs du type de celles que les modèles commettent réellement :\n")
probe(lambda s: (f := first(s)) and f.update(value=(f["value"] or 0) + 0.07) or f,
      "une valeur décalée vers une qui n'est pas imprimée")
probe(lambda s: (f := first(s)) and f.update(source_line="Une ligne que le modèle a inventée")
      or f, "une ligne source fabriquée")
probe(lambda s: (f := first(s, lambda x: gate_is_active(x.get("indicator")))) and
      f.update(unit="tonnes") or f, "un nombre plausible avec la mauvaise unité")
if not any(gate_is_active(f.get("indicator")) for f in EXTRACTED):
    print("\n  note : aucune entrée UNIT_RULES ne correspond aux clés d'indicateur produites,")
    print("        la porte 3 n'applique donc que son contrôle générique de pourcentage.")
    print("        Clés à utiliser :", sorted({f.get("indicator") for f in EXTRACTED})[:6])
probe(lambda s: (f := first(s)) and f.update(
    value=next((b["value"] for b in BASELINE
                if abs(b["value"] - (f["value"] or 0)) > 1e-9), (f["value"] or 0) + 1)) or f,
      "un vrai nombre pris sur la mauvaise ligne")
````

 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#0A5C64;margin-bottom:5px">Point de contr&ocirc;le &mdash; partie 3</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Vous voulez un <code>✓</code> sur chaque ligne, et au moins une porte nomm&eacute;e pour chacune. Si la derni&egrave;re sonde annonce &laquo;&nbsp;rien ne s&rsquo;est d&eacute;clench&eacute;&nbsp;&raquo;, c&rsquo;est que votre lecture par r&egrave;gles n&rsquo;a pas trouv&eacute; cette ligne &mdash; la porte 5 n&rsquo;a alors aucun avis, ce qui est honn&ecirc;te mais signifie que le d&eacute;calage de ligne n&rsquo;est pas gard&eacute; sur ce document. Dites-le dans l&rsquo;&eacute;nonc&eacute; des limites en partie 6.<br><br>Vous devez aussi avoir <code>verification_report.csv</code> sur le disque. Ce fichier est un livrable, pas un artefact de d&eacute;bogage : c&rsquo;est lui qui permet &agrave; quelqu&rsquo;un d&rsquo;autre de rejouer votre jugement.</div></div>

````python
# ============================== YOUR TURN ==================================
# Amorcez ceci depuis la liste terminologique de votre institut. Elle bat tout modèle.
GLOSSARY = {
    "taux de chômage": "unemployment rate",
    "population active": "labour force",
    "population résidente": "resident population",
    "ménage": "household",
    "ménages ordinaires": "ordinary households",
    "ensemble du pays": "whole country",
    "milliers": "thousands",
    "milliers d'habitants": "thousands of inhabitants",
    "part": "share",
    "densité": "density",
    "variation": "change",
    "points": "points",
    "provisoire": "provisional",
    "donnée non disponible": "data not available",
    "enquête nationale sur l'emploi": "national labour force survey",
    "littoral nord": "north coast", "littoral sud": "south coast",
    "plateau central": "central plateau", "vallée de l'est": "eastern valley",
    "hauts plateaux": "highlands", "extrême ouest": "far west",
}
SOURCE_LANG = "fr"      # la langue dans laquelle votre publication est rédigée
# ===========================================================================
GLOSSARY = {k.lower(): v for k, v in GLOSSARY.items()}
````

````python
TRANSLATE_SYSTEM = """Vous traduisez la terminologie de la statistique officielle entre le français et
l'anglais. Vous traduisez des libellés pour un institut national de la statistique, pas
de la communication commerciale.

- Employez le terme normalisé de la statistique officielle, pas une traduction littérale.
- Restez bref : ce sont des libellés de graphiques et d'en-têtes, pas des phrases.
- Préservez les conventions de capitalisation ainsi que les chiffres et unités du libellé.
- Si un terme est un nom propre (région, enquête, institution), translittérez plutôt que
  de traduire, et gardez l'orthographe d'origine lorsqu'elle est déjà dans la langue cible.
- Renvoyez uniquement du JSON : {"terms": {"terme source": "traduction", ...}}. Sans prose."""


def translate_terms(terms, src="fr", dst="en", check=True):
    """Glossaire d'abord ; modèle pour le reste seulement ; rétrotraduction de ce que le modèle a écrit."""
    out, unknown = {}, []
    for t in terms:
        hit = GLOSSARY.get(str(t).lower().strip())
        out[t] = hit if (hit and src == "fr" and dst == "en") else None
        if out[t] is None:
            unknown.append(t)

    if unknown and not OFFLINE:
        got = llm([{"role": "system", "content": TRANSLATE_SYSTEM},
                   {"role": "user", "content":
                    f"Translate these {src} terms into {dst}:\n"
                    + json.dumps(unknown, ensure_ascii=False)}],
                  json_mode=True, temperature=0.0, label="translate")
        got = got.get("terms", got)
        for t in unknown:
            out[t] = got.get(t) or t

        if check and got:
            back = llm([{"role": "system", "content": TRANSLATE_SYSTEM},
                        {"role": "user", "content":
                         f"Translate these {dst} terms back into {src}:\n"
                         + json.dumps(list(got.values()), ensure_ascii=False)}],
                       json_mode=True, temperature=0.0, label="back-translate")
            back = back.get("terms", back)
            drift = [(t, out[t], back.get(out[t]))
                     for t in unknown
                     if back.get(out[t], "").lower().strip() != str(t).lower().strip()]
            if drift:
                print("la rétrotraduction ne revient pas à l'original — à vérifier à la main :")
                for a, b, c in drift[:12]:
                    print(f"   {a!r} → {b!r} → {c!r}")
    else:
        for t in unknown:
            out[t] = out[t] or t
    return out


AREAS = sorted({f["area"] for f in VERIFIED if f.get("area")})
INDICATOR_LABELS = {f["indicator"]: f.get("indicator_label") or f["indicator"] for f in VERIFIED}
UNITS = sorted({(f.get("unit") or "").strip() for f in VERIFIED} - {""})

to_translate = AREAS + list(INDICATOR_LABELS.values()) + UNITS
EN = translate_terms(to_translate)

LABELS = {
    "areas": {a: {"fr": a, "en": EN.get(a, a)} for a in AREAS},
    "indicators": {k: {"fr": v, "en": EN.get(v, v)} for k, v in INDICATOR_LABELS.items()},
    "units": {u: {"fr": u, "en": EN.get(u, u)} for u in UNITS},
}
display(pd.DataFrame([{"source": k, "en": v["en"]}
                      for group in LABELS.values() for k, v in group.items()])
        .style.hide(axis="index").set_caption("Libellés qui apparaîtront sur le tableau de bord"))
````

````python
DESIGN_SYSTEM = """Vous êtes le rédacteur du tableau de bord public d'un institut national de la statistique.
Vous écrivez en anglais et en français, au niveau d'une publication officielle : sobre,
précis, sans communication, sans adjectif qu'un statisticien ne signerait pas.

On vous donne la STRUCTURE d'un jeu de données vérifié — clés d'indicateurs, noms de zones,
périodes, unités et étendue de chaque série. On ne vous donne PAS les chiffres, et vous ne
devez jamais en énoncer un.

Your job:
  - name the page and write a one-sentence standfirst, in both languages;
  - for each indicator, choose a chart and write its tab label, title, caption and a one-line
    reading note, in both languages;
  - choose at most four headline figures by POINTING AT CELLS (indicator, area, period).
    Write only the label; the value is filled in from the verified data;
  - write the limitations paragraph in both languages, using only what the notes below say.

Choix du graphique : « line » seulement si les libellés de période forment une vraie
séquence temporelle (années, trimestres, mois) — une courbe affirme que l'axe des x est
ordonné, ne l'employez donc jamais pour des catégories comme « Hommes 2020 / Femmes 2020 ».
« grouped » pour quelques catégories, « bar » pour une colonne unique.

Return JSON only, in this shape, and nothing else:
{"kicker":{"en":"","fr":""},"title":{"en":"","fr":""},"subtitle":{"en":"","fr":""},
 "ovTitle":{"en":"","fr":""},"ovLede":{"en":"","fr":""},
 "views":[{"id":"","chart":"line","tab":{"en":"","fr":""},"title":{"en":"","fr":""},
           "sub":{"en":"","fr":""},"lede":{"en":"","fr":""},"src":{"en":"","fr":""}}],
 "kpis":[{"indicator":"","area":"","period":"","compare_period":null,
          "labels":{"en":"","fr":""},"good":"down","highlight":false}],
 "limBody":{"en":"","fr":""}}"""


def dataset_shape(facts):
    """Ce que le modèle a le droit de voir : la structure, jamais les valeurs."""
    shape = {}
    for f in facts:
        s = shape.setdefault(f["indicator"], {"label": f.get("indicator_label"),
                                              "unit": f.get("unit"), "periods": set(),
                                              "areas": set(), "n": 0})
        s["periods"].add(str(f.get("period")))
        s["areas"].add(str(f.get("area")))
        s["n"] += 1
    return {k: {"label": v["label"], "unit": v["unit"],
                "periods": sorted(v["periods"]), "areas": sorted(v["areas"]),
                "cells": v["n"]} for k, v in shape.items()}


SHAPE = dataset_shape(VERIFIED)
REVIEW_NOTES = sorted({r for f in VERIFIED for r in f.get("reasons", [])})[:8]
print(json.dumps(SHAPE, ensure_ascii=False, indent=1)[:1200])
````

#### 5.3 &nbsp;Remplir les chiffres (depuis les donn&eacute;es v&eacute;rifi&eacute;es, pas depuis le mod&egrave;le)

````python
import unicodedata
from datetime import date


def slug(s):
    s = unicodedata.normalize("NFKD", str(s)).encode("ascii", "ignore").decode()
    return re.sub(r"[^a-z0-9]+", "-", s.lower()).strip("-") or "x"


def lookup(indicator, area, period):
    for f in VERIFIED:
        if (f.get("indicator") == indicator and str(f.get("area")) == str(area)
                and str(f.get("period")) == str(period)):
            return f
    return None


def resolve_kpis(spec):
    """Le modèle a désigné des cellules ; nous lisons les valeurs dans le jeu vérifié."""
    out = []
    for k in spec[:4]:
        cell = lookup(k.get("indicator"), k.get("area"), k.get("period"))
        if not cell or cell.get("value") is None:
            continue
        delta = None
        if k.get("compare_period"):
            prev = lookup(k["indicator"], k["area"], k["compare_period"])
            if prev and prev.get("value") is not None:
                delta = round(cell["value"] - prev["value"], 2)
        unit = cell.get("unit") or ""
        out.append({"labels": k.get("labels") or {"en": k.get("indicator"), "fr": k.get("indicator")},
                    "value": cell["value"],
                    "unit": {"en": LABELS["units"].get(unit, {}).get("en", unit), "fr": unit},
                    "delta": delta,
                    "deltaLabel": {"en": f"vs {k.get('compare_period')}",
                                   "fr": f"vs {k.get('compare_period')}"},
                    "good": k.get("good", "down"), "highlight": bool(k.get("highlight"))})
    return out


KPIS = resolve_kpis(DESIGN.get("kpis", []))
if not KPIS:                                     # aucune suggestion exploitable → on prend les cellules les plus récentes
    for ind, s in list(SHAPE.items())[:3]:
        period = s["periods"][-1]
        cell = next((f for f in VERIFIED if f["indicator"] == ind
                     and str(f["period"]) == period and is_total(f)), None) \
            or next((f for f in VERIFIED if f["indicator"] == ind and str(f["period"]) == period), None)
        if cell:
            KPIS.append({"labels": {"en": f'{LABELS["indicators"][ind]["en"]}, {period}',
                                    "fr": f'{LABELS["indicators"][ind]["fr"]}, {period}'},
                         "value": cell["value"], "unit": {"en": cell.get("unit", ""),
                                                          "fr": cell.get("unit", "")},
                         "delta": None, "good": "down", "highlight": not KPIS})

for k in KPIS:
    print(f"  {k['labels']['en']:<44} {k['value']}  {k['unit']['en']}")
````

#### 5.4 &nbsp;Assembler, valider, pr&eacute;visualiser

````python
def inject(template, payload):
    blob = json.dumps(payload, ensure_ascii=False, indent=1)
    html, n = re.subn(r"/\*__DATA__\*/.*?/\*__ENDDATA__\*/",
                      lambda _: "/*__DATA__*/" + blob + "/*__ENDDATA__*/",
                      template, count=1, flags=re.S)
    if not n:
        raise RuntimeError("marqueurs d'injection absents du gabarit")
    return html


def validate_dashboard(html, payload):
    """Contrôles structurels économiques. Ne remplacent pas l'ouverture de la page."""
    problems, notes = [], []
    if "/*__DATA__*/" not in html:
        problems.append("les données n'ont pas été injectées")
    m = re.search(r"/\*__DATA__\*/(.*?)/\*__ENDDATA__\*/", html, re.S)
    try:
        back = json.loads(m.group(1))
    except Exception as exc:
        problems.append(f"les données injectées ne sont pas du JSON valide : {exc}")
        back = {}
    if back.get("facts") and len(back["facts"]) != len(payload["facts"]):
        problems.append("le nombre de cellules a changé pendant l'injection")

    external = [u for u in re.findall(r'<script[^>]+src="([^"]+)"', html)
                if not u.startswith("https://cdn.jsdelivr.net/")]
    if external:
        problems.append(f"script externe inattendu : {external}")
    if re.search(r"fetch\(|XMLHttpRequest", html):
        notes.append("la page fait un appel réseau — vérifiez lequel avant publication")
    for need, why in (("data-lang=\"fr\"", "bascule française"), ("data-lang=\"en\"", "bascule anglaise"),
                      ("prefers-color-scheme", "compatibilité mode sombre"), ("<title>", "titre de page")):
        if need not in html:
            notes.append(f"{why} manquant")
    if len(html) > 5_000_000:
        problems.append("page au-delà de 5 Mo — trop lourde pour un site public")

    # les chiffres de la page doivent être ceux qui ont passé la vérification
    published = {(f["indicator"], f["area"], f["period"], f["value"]) for f in back.get("facts", [])}
    expected = {(f["indicator"], f["area"], f["period"], f["value"]) for f in payload["facts"]}
    if published != expected:
        problems.append(f"{len(expected ^ published)} cellules diffèrent entre le jeu vérifié "
                        "et la page")
    return problems, notes


HTML = inject(DASHBOARD_TEMPLATE, PAYLOAD)
problems, notes = validate_dashboard(HTML, PAYLOAD)

OUT = WORKDIR / "index.html"
if problems:
    print("VALIDATION ÉCHOUÉE — rien n'a été écrit :")
    for p in problems:
        print("  ✗", p)
else:
    OUT.write_text(HTML, encoding="utf-8")
    print(f"✓ {OUT}  ({len(HTML) / 1024:,.0f} ko, autonome)")
for n in notes:
    print("  !", n)
````

<div style="font-size:14px;color:#231F20;line-height:1.55">Vous pr&eacute;f&eacute;rez un jeton <b>fine-grained</b>&nbsp;? Cela fonctionne aussi, mais il faut trois permissions au lieu d&rsquo;une : <i>Administration</i> lecture et &eacute;criture (pour cr&eacute;er le d&eacute;p&ocirc;t), <i>Contents</i> lecture et &eacute;criture (pour t&eacute;l&eacute;verser) et <i>Pages</i> lecture et &eacute;criture (pour activer l&rsquo;h&eacute;bergement).
````python
import base64
import time

GH_API = "https://api.github.com"


def gh(method, path, token=None, **kw):
    """Une enveloppe fine pour que chaque appel porte les bons en-têtes et une vraie erreur."""
    headers = {"Accept": "application/vnd.github+json",
               "X-GitHub-Api-Version": "2022-11-28"}
    if token:
        headers["Authorization"] = f"Bearer {token}"
    return requests.request(method, GH_API + path, headers=headers, timeout=60, **kw)


def github_reachable():
    """
    Les réseaux d'institution et de conférence bloquent ou brident api.github.com plus
    souvent qu'on ne le croit. Autant le savoir avant de demander un jeton à quiconque.
    """
    try:
        r = gh("GET", "/rate_limit")
    except requests.RequestException as exc:
        return False, f"api.github.com totalement injoignable ({exc.__class__.__name__})"
    if r.status_code == 200:
        core = r.json().get("resources", {}).get("core", {})
        return True, f"joignable — {core.get('remaining', '?')}/{core.get('limit', '?')} " \
                     f"requêtes anonymes restantes"
    return False, f"api.github.com répond {r.status_code} — un proxy ou un pare-feu bloque"


def _repo_exists(owner, repo, token):
    """200 veut dire oui, 404 veut dire non. Tout le reste n'est pas une réponse et ne se devine pas."""
    r = gh("GET", f"/repos/{owner}/{repo}", token)
    if r.status_code == 200:
        return True
    if r.status_code == 404:
        return False
    raise RuntimeError(f"GitHub répond {r.status_code} à propos de {owner}/{repo} : "
                       f"{r.text[:200]}\nCe n'est ni un oui ni un non, donc rien n'a été modifié. "
                       f"Vérifiez les scopes du jeton, la limite de requêtes et le réseau.")


def github_identity(token):
    r = gh("GET", "/user", token)
    if r.status_code == 401:
        raise RuntimeError("GitHub a rejeté le jeton. Expiré, ou mal recopié ?")
    r.raise_for_status()
    who = r.json()
    scopes = r.headers.get("x-oauth-scopes", "(fine-grained token)")
    return who["login"], scopes


def upload_folder(folder, owner, repo, token, branch="main", skip=(".git",)):
    """
    Téléverse chaque fichier via l'API Contents. Un commit par fichier — plus bruyant
    qu'un vrai push git, mais cela n'exige aucun binaire git, fonctionne à l'identique
    dans Colab, Kaggle et un navigateur, et se lit facilement quand ça se passe mal.
    """
    sent = []
    for path in sorted(p for p in folder.rglob("*") if p.is_file()):
        rel = path.relative_to(folder).as_posix()
        if any(rel.startswith(s) for s in skip):
            continue
        body = {"message": f"lab02: publish {rel}",
                "content": base64.b64encode(path.read_bytes()).decode(),
                "branch": branch}
        cur = gh("GET", f"/repos/{owner}/{repo}/contents/{rel}", token,
                 params={"ref": branch})
        if cur.status_code == 200:
            body["sha"] = cur.json()["sha"]          # mettre à jour plutôt qu'échouer
        elif cur.status_code not in (404,):
            raise RuntimeError(f"vérification impossible de {rel} avant téléversement : "
                               f"{cur.status_code} {cur.text[:200]}")
        put = gh("PUT", f"/repos/{owner}/{repo}/contents/{rel}", token, json=body)
        if put.status_code not in (200, 201):
            raise RuntimeError(f"téléversement de {rel} échoué : {put.status_code} {put.text[:300]}")
        sent.append(rel)
        print(f"   téléversé {rel}")
    return sent


def enable_pages(owner, repo, token, branch="main", wait=150):
    r = gh("POST", f"/repos/{owner}/{repo}/pages", token,
           json={"source": {"branch": branch, "path": "/"}})
    if r.status_code not in (201, 409):               # 409 = déjà activé
        hint = ""
        if r.status_code in (403, 404):
            hint = ("\n   L'API Pages exige le scope « repo » complet sur un jeton classique — "
                    "« public_repo » vous mène jusqu'ici et pas plus loin. Sur un jeton "
                    "fine-grained, il faut Pages : lecture et écriture. Tout est déjà "
                    "téléversé, donc activer Pages à la main dans Settings marche aussi.")
        raise RuntimeError(f"activation de Pages impossible : {r.status_code} {r.text[:300]}{hint}")
    print("   Pages activé, attente du premier build …")
    deadline = time.time() + wait
    url = None
    while time.time() < deadline:
        s = gh("GET", f"/repos/{owner}/{repo}/pages", token)
        if s.status_code == 200:
            data = s.json()
            url = data.get("html_url") or url
            if data.get("status") == "built":
                return url, True
        time.sleep(5)
    return url, False


def publish_to_github(folder, repo_name, token, org=None, description="", branch="main"):
    login, scopes = github_identity(token)
    owner = org or login
    print(f"authentifié comme {login}   scopes : {scopes}")

    if _repo_exists(owner, repo_name, token):
        print("   le dépôt existe déjà — mise à jour sur place")
    else:
        payload = {"name": repo_name, "description": description[:350],
                   "private": False, "auto_init": False, "has_issues": True}
        r = gh("POST", f"/orgs/{org}/repos" if org else "/user/repos", token, json=payload)
        if r.status_code == 422 and "already exists" in r.text:
            print("   le dépôt existait déjà — mise à jour sur place")
        elif r.status_code != 201:
            hint = ""
            if r.status_code == 403:
                hint = ("\n   Un 403 signifie en général que le jeton n'a pas le scope « repo », "
                        "ou que l'organisation ne l'a pas autorisé.")
            raise RuntimeError(f"création impossible de {owner}/{repo_name} : "
                               f"{r.status_code} {r.text[:300]}{hint}")
        else:
            print(f"   créé https://github.com/{owner}/{repo_name}")

    upload_folder(folder, owner, repo_name, token, branch)
    url, built = enable_pages(owner, repo_name, token, branch)
    return {"owner": owner, "repo": repo_name, "url": url, "built": built,
            "repo_url": f"https://github.com/{owner}/{repo_name}"}
````
