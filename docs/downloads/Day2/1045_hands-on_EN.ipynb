#### 0.1 &nbsp;Environment and dependencies

````python
# UTILITY — detects Colab / Kaggle / local, installs what is missing, sets the working directory.
import importlib, os, subprocess, sys, textwrap, warnings
from pathlib import Path

warnings.filterwarnings("ignore", category=UserWarning)


def detect_environment() -> str:
    if "google.colab" in sys.modules or Path("/content").exists():
        return "colab"
    if os.environ.get("KAGGLE_KERNEL_RUN_TYPE") or Path("/kaggle/working").exists():
        return "kaggle"
    return "local"


ENV = detect_environment()
WORKDIR = {"colab": Path("/content/lab02"),
           "kaggle": Path("/kaggle/working/lab02")}.get(ENV, Path.cwd() / "lab02")
WORKDIR.mkdir(parents=True, exist_ok=True)
os.chdir(WORKDIR)

REQUIRED = {"requests": "requests", "pdfplumber": "pdfplumber", "pandas": "pandas"}
OPTIONAL = {"fitz": "pymupdf",        # page images, only needed for scanned PDFs
            "reportlab": "reportlab"}  # builds the bundled demonstration document


def ensure(packages: dict, hard: bool) -> list:
    missing = [pip for mod, pip in packages.items() if not importlib.util.find_spec(mod)]
    if not missing:
        return []
    print(f"installing: {', '.join(missing)} …")
    cmd = [sys.executable, "-m", "pip", "install", "-q", *missing]
    if ENV == "kaggle":
        cmd.insert(4, "--no-warn-script-location")
    try:
        subprocess.run(cmd, check=True, capture_output=True, timeout=600)
    except Exception as exc:
        msg = f"could not install {missing}: {exc}"
        if hard:
            raise RuntimeError(msg + "\n(on Kaggle, switch Internet on in the settings panel)")
        print("  optional, skipped —", msg)
    return missing


ensure(REQUIRED, hard=True)
ensure(OPTIONAL, hard=False)

import pandas as pd  # noqa: E402
pd.set_option("display.max_rows", 120, "display.width", 200,
              "display.max_colwidth", 60, "display.float_format", lambda v: f"{v:,.2f}")

print(f"environment : {ENV}")
print(f"python      : {sys.version.split()[0]}")
print(f"working dir : {WORKDIR}")
````

#### 0.3 &nbsp;The client: one function, any provider

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#E8F5EF;border:1px solid #00A86A;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#00704A;margin-bottom:5px">Why this step matters</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Three things are worth noticing in the next cell, because they are what separates a notebook that survives a workshop from one that does not. It <b>asks the provider which models exist</b> rather than hard-coding a name that expired last month. It <b>retries on rate limits</b>, which you will hit with thirty people on one wifi. And it <b>records latency and token counts for every call</b>, so that by Part 2 you have real numbers to take into this afternoon's procurement discussion.</div></div>


````python
import json, random, re, time
import requests

CALLS = []          # telemetry: one row per API call
OFFLINE = False     # set automatically below if no engine is reachable


class LLMError(RuntimeError):
    pass


def _headers(key):
    h = {"Authorization": f"Bearer {key}", "Content-Type": "application/json"}
    if PROVIDER == "openrouter":
        h["HTTP-Referer"] = "https://stg17.workshop"
        h["X-Title"] = "STG17 Lab 02"
    return h


def list_models(provider=None, key=None, timeout=30):
    """Ask the provider what it serves today. Returns a list of model ids."""
    p = PROVIDERS[provider or PROVIDER]
    if not p["base"]:
        return []
    r = requests.get(f"{p['base']}/models", headers=_headers(key or API_KEY), timeout=timeout)
    r.raise_for_status()
    payload = r.json()
    items = payload.get("data", payload if isinstance(payload, list) else [])
    return sorted({(m.get("id") or m.get("name")) for m in items if isinstance(m, dict)} - {None})


def pick_model(available, prefer):
    for pattern in prefer:
        for m in available:
            if re.search(pattern, m, re.I):
                return m
    return available[0] if available else None


def llm(messages, *, json_mode=False, temperature=0.0, max_tokens=4096,
        model=None, retries=4, label=""):
    """One chat completion. Returns the assistant's text (or a dict if json_mode)."""
    if OFFLINE:
        raise LLMError("offline mode: no model configured")

    body = {"model": model or MODEL, "messages": messages,
            "temperature": temperature, "max_tokens": max_tokens}
    if json_mode:
        body["response_format"] = {"type": "json_object"}

    url = f"{PROVIDERS[PROVIDER]['base']}/chat/completions"
    last = None
    for attempt in range(retries):
        t0 = time.time()
        try:
            r = requests.post(url, headers=_headers(API_KEY), json=body, timeout=180)
        except requests.RequestException as exc:
            last = exc
            time.sleep(1.5 * 2 ** attempt)
            continue
        dt = time.time() - t0

        if r.status_code == 200:
            data = r.json()
            usage = data.get("usage") or {}
            CALLS.append(dict(label=label, provider=PROVIDER, model=body["model"],
                              seconds=round(dt, 2),
                              prompt_tokens=usage.get("prompt_tokens"),
                              completion_tokens=usage.get("completion_tokens")))
            text = data["choices"][0]["message"].get("content") or ""
            return parse_json(text) if json_mode else text

        # structured outputs are not universally supported — degrade, do not fail
        if r.status_code in (400, 404, 422) and "response_format" in body:
            body.pop("response_format")
            continue
        if r.status_code in (408, 409, 429, 500, 502, 503, 504):
            wait = float(r.headers.get("retry-after", 0)) or (1.5 * 2 ** attempt + random.random())
            print(f"  {r.status_code} — waiting {wait:.1f}s (attempt {attempt + 1}/{retries})")
            time.sleep(wait)
            last = r.text[:300]
            continue
        raise LLMError(f"{r.status_code}: {r.text[:400]}")
    raise LLMError(f"gave up after {retries} attempts: {last}")


def parse_json(text):
    """Models wrap JSON in prose or fences more often than anyone admits."""
    text = (text or "").strip()
    text = re.sub(r"^```(?:json)?|```$", "", text, flags=re.M).strip()
    try:
        return json.loads(text)
    except json.JSONDecodeError:
        pass
    start = min((i for i in (text.find("{"), text.find("[")) if i >= 0), default=-1)
    if start >= 0:
        for end in range(len(text), start, -1):
            try:
                return json.loads(text[start:end])
            except json.JSONDecodeError:
                continue
    raise LLMError(f"no JSON in response: {text[:200]}")
````

#### 1.2 &nbsp;Read the pages

````python
import unicodedata

NBSP = "\u00a0\u202f\u2009"


def norm(s: str) -> str:
    """Normalise a page so later comparisons survive PDF quirks."""
    s = unicodedata.normalize("NFKC", s or "")
    for ch in NBSP:
        s = s.replace(ch, " ")
    s = s.replace("\u2019", "'").replace("\u2013", "-").replace("\u2014", "-")
    return re.sub(r"[ \t]+", " ", s)


def read_pdf(path, max_pages=MAX_PAGES):
    import pdfplumber
    pages = []
    with pdfplumber.open(path) as pdf:
        total = len(pdf.pages)
        for i, page in enumerate(pdf.pages[:max_pages], 1):
            pages.append({"page": i, "text": norm(page.extract_text() or ""),
                          "chars": len(page.chars), "images": len(page.images)})
    return pages, total


PAGES, TOTAL_PAGES = read_pdf(PDF_PATH)
scanned = [p["page"] for p in PAGES if p["chars"] < 40 and p["images"]]

print(f"{TOTAL_PAGES} pages in the document, {len(PAGES)} read "
      f"({'all' if len(PAGES) == TOTAL_PAGES else f'capped at MAX_PAGES={MAX_PAGES}'})")
print(f"characters recovered: {sum(len(p['text']) for p in PAGES):,}")
if scanned:
    print(f"\npages with no text layer (scanned images): {scanned}")
    print("  → these need a vision model or OCR; see the note below.")
else:
    print("every page has a text layer — the cheap path is open")
````

#### 1.3 &nbsp;Triage: which pages are worth a model call?

````python
CUES = ("tableau", "table", "cuadro", "%", "source", "enqu", "survey")


def table_score(text: str) -> float:
    """
    Cheap, explainable, no model. Three signals: how dense in digits the page is,
    how many lines look like table rows, and how often table vocabulary appears.
    """
    if not text:
        return 0.0
    digits = sum(c.isdigit() for c in text) / max(len(text), 1)
    rows = sum(1 for l in text.splitlines()
               if len(re.findall(r"\d[\d  .,]*", l)) >= 3 and re.search(r"[A-Za-zÀ-ÿ]{3}", l))
    cues = sum(text.lower().count(c) for c in CUES)
    return round(digits * 100 + rows * 2 + cues, 1)


inventory = pd.DataFrame([{
    "page": p["page"],
    "score": table_score(p["text"]),
    "chars": len(p["text"]),
    "first line": (p["text"].splitlines() or [""])[0][:58],
} for p in PAGES]).sort_values("score", ascending=False)

display(inventory.head(12).style.hide(axis="index")
        .bar(subset=["score"], color="#9ED9C0")
        .set_caption("Pages most likely to contain tables"))

# ============================== YOUR TURN ==================================
# Keep it small. Three good pages beat thirty mediocre ones.
TARGET_PAGES = [int(p) for p in inventory.head(3)["page"]]
# TARGET_PAGES = [4, 5, 11]      # ← or name them yourself once you have looked
# ===========================================================================
print("\nworking on pages:", sorted(TARGET_PAGES))
````

#### 2.1 &nbsp;Reading tables without being told where they are

<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#8A6400;margin-bottom:5px">Why the obvious scoring rule is wrong</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Choosing between the three cannot simply be &ldquo;who found the most numbers&rdquo;: a reader that breaks one figure into three <i>finds more numbers</i> and is worse. So the score rewards a densely filled grid, rewards a table whose own total row adds up, and demotes any reading that shows the signature of a split thousands separator &mdash; a small whole number sitting immediately left of a three-digit one, row after row.</div></div>



````python
# UTILITY — three readers and a referee. This is the part that makes the notebook
# work on a document nobody tuned it for.
import pdfplumber

SP = r"[ \u00a0\u202f\u2009]"


def _num_pattern(max_groups=None):
    """Numbers as statistical publications actually print them."""
    grouped = (rf"[+\-\u2212]?\d{{1,3}}(?:{SP}\d{{3}})"
               + (rf"{{1,{max_groups}}}" if max_groups else "+") + r"(?:[.,]\d+)?%?")
    plain = rf"[+\-\u2212]?\d+(?:[.,]\d+)?%?(?:{SP}?\(\d+\))?"
    missing = r"…|\.\.\.|[–—-]"
    return rf"(?:{plain}|{missing})" if max_groups == 0 else rf"(?:{grouped}|{plain}|{missing})"


def parse_number(raw):
    """
    '12,8' → 12.8      '18 742' → 18742      '+14,2' → 14.2      '15.7%' → 15.7
    '17,0 (1)' → 17.0  '…' / '-' / '' → None (missing, never zero)
    """
    if raw is None:
        return None
    s = norm(str(raw)).strip()
    s = re.sub(r"\(\d+\)", "", s).replace("%", "").strip()
    if s in {"", "-", "–", "—", "…", "...", "..", "n.d.", "nd", "N/A", "na"}:
        return None
    s = s.replace(" ", "").replace("\u2212", "-")
    if "," in s and "." in s:                       # 1,234.5  vs  1.234,5
        s = s.replace(",", "") if s.rfind(".") > s.rfind(",") else s.replace(".", "").replace(",", ".")
    elif "," in s:
        s = s.replace(",", ".") if re.search(r",\d{1,2}$", s) else s.replace(",", "")
    try:
        return float(s)
    except ValueError:
        return None


def split_numbers(chunk, max_groups=None):
    return [parse_number(t) for t in re.findall(_num_pattern(max_groups), chunk)]


NUM = _num_pattern()
# A classification code in front of the label ("01.1.7 Vegetables") is captured
# separately rather than letting its digits defeat the match.
ROW = re.compile(rf"^(?:(?P<code>\d[\d.]{{0,9}}){SP}+)?(?P<label>[^\d\n]{{2,64}}?){SP}+"
                 rf"(?P<nums>{NUM}(?:{SP}+{NUM})+){SP}*$")
CAPTION = re.compile(r"^\s*(tableau|table|cuadro|jadual|annex|annexe|appendix)\s*"
                     r"(?P<no>[\dIVX]+)?\s*[—:\-–.]\s*(?P<title>.+)$", re.I)
YEAR = re.compile(r"\b(?:1[89]|20)\d{2}\b")
TOTAL_WORDS = re.compile(r"^(total|ensemble|whole|national|pays|country|tous|all|general|"
                         r"g[ée]n[ée]ral|overall|toutes)\b", re.I)


def _is_header(values):
    vs = [v for v in values if v is not None]
    return bool(vs) and all(float(v).is_integer() and 1800 <= v <= 2100 for v in vs)


def _dedupe(columns):
    """
    Two columns called '2025' are worse than two columns called 'col3' and 'col4':
    everything downstream groups by period, so duplicates silently merge quarters
    into one series and every reconciliation then fails. Never let them through.
    """
    seen, out = {}, []
    for c in columns:
        c = str(c)
        seen[c] = seen.get(c, 0) + 1
        out.append(c if seen[c] == 1 else f"{c} ({seen[c]})")
    return out


# 'T1 2025 T2 2025' — keep the qualifier in front of each year, not just the year
YEAR_WITH_PREFIX = re.compile(r"((?:[^\s\d]{1,10}\s+){0,2})((?:1[89]|20)\d{2})")


def _columns_from_header(line):
    hits = YEAR_WITH_PREFIX.findall(line)
    labels = [(pre + yr).strip() for pre, yr in hits]
    if len(set(labels)) < len(labels):
        labels = [yr for _, yr in hits]
    return labels


def _repair_widths(rows, width):
    """
    '18 742 100,0 114' is ambiguous: one eight-digit number, or a total and two
    columns? A row narrower than the rest of its table is re-read with the
    space-as-thousands rule relaxed until the widths agree.
    """
    for r in rows:
        if len(r["values"]) >= width:
            continue
        for mg in (2, 1, 0):
            alt = split_numbers(r["nums"], mg)
            if len(alt) == width:
                r["values"], r["repaired"] = alt, True
                break


# ---------------------------------------------------------------- reader 1 & 2: geometry
def _clean(c):
    return norm(c or "").replace("\n", " ").strip()


def split_code(label):
    """
    '01.1.7 Vegetables' → ('01.1.7', 'Vegetables').
    '65 years and over' → ('', '65 years and over') — that 65 is the data, not a code.
    Only a dotted code, or one padded with a leading zero, counts.
    """
    m = re.match(r"^(\d[\d.]{0,9})\s+(.{2,})$", label or "")
    if not m:
        return "", label
    head = m.group(1)
    looks_like_code = "." in head or (len(head) >= 2 and head.startswith("0"))
    return (head, m.group(2)) if looks_like_code else ("", label)


def _all_years(vals):
    n = [parse_number(v) for v in vals]
    n = [x for x in n if x is not None]
    return bool(n) and all(float(x).is_integer() and 1800 <= x <= 2100 for x in n)


def _is_label_row(row):
    """Mostly words rather than figures — or a row of bare years, still a header."""
    vals = [_clean(c) for c in row[1:] if _clean(c)]
    if not vals or _all_years(vals):
        return True
    return sum(parse_number(v) is not None for v in vals) <= len(vals) / 2


def _merge_headers(header_rows, width):
    """Carry a spanning label rightwards, then join: 'Male' + '2020' → 'Male 2020'."""
    filled = []
    for row in header_rows:
        out, last = [], ""
        for i in range(width):
            c = _clean(row[i]) if i < len(row) else ""
            last = c or last
            out.append(last)
        filled.append(out)
    cols = []
    for i in range(1, width):
        parts, seen = [], set()
        for row in filled:
            if row[i] and row[i] not in seen:
                parts.append(row[i])
                seen.add(row[i])
        cols.append(" ".join(parts) or f"col{i}")
    return _dedupe(cols)


def _caption_above(page, bbox):
    try:
        above = page.crop((0, max(bbox[1] - 60, 0), page.width, max(bbox[1] - 1, 1)))
        lines = [norm(l).strip() for l in (above.extract_text() or "").splitlines()]
    except Exception:
        return ""
    lines = [l for l in lines if len(l) > 6]
    return lines[-1] if lines else ""


LAYOUT_STRATEGIES = (
    ("lines", {"vertical_strategy": "lines", "horizontal_strategy": "lines"}),
    ("text", {"vertical_strategy": "text", "horizontal_strategy": "text",
              "text_x_tolerance": 2, "intersection_tolerance": 8}),
)


def layout_tables(pdf_path, only=None, min_rows=2, min_numeric=0.55):
    out = []
    with pdfplumber.open(pdf_path) as pdf:
        for pno, page in enumerate(pdf.pages, 1):
            if only and pno not in only:
                continue
            best, best_score = [], 0.0
            for name, settings in LAYOUT_STRATEGIES:
                try:
                    found = page.find_tables(settings) or []
                except Exception:
                    continue
                blocks, score = [], 0.0
                for tb in found:
                    rows = [[_clean(c) for c in r] for r in (tb.extract() or [])]
                    rows = [r for r in rows if any(r)]
                    if len(rows) < min_rows + 1 or len(rows[0]) < 2:
                        continue
                    cells = [c for r in rows[1:] for c in r[1:] if _clean(c)]
                    share = (sum(parse_number(c) is not None for c in cells) / len(cells)
                             if cells else 0)
                    if share < min_numeric:
                        continue
                    blocks.append((name, tb, rows))
                    score += share * len(rows)
                if score > best_score:
                    best, best_score = blocks, score

            for name, tb, rows in best:
                width = max(len(r) for r in rows)
                rows = [r + [""] * (width - len(r)) for r in rows]
                header, i = [], 0
                while i < len(rows) and _is_label_row(rows[i]) and i < 3:
                    header.append(rows[i]); i += 1
                columns = _merge_headers(header, width) if header \
                    else [f"col{j}" for j in range(1, width)]
                body = []
                for r in rows[i:]:
                    label, values = _clean(r[0]), [parse_number(c) for c in r[1:]]
                    if not label or all(v is None for v in values):
                        continue
                    code, label = split_code(label)
                    body.append({"label": label, "code": code, "values": values,
                                 "nums": "", "line": " ".join(x for x in r if x),
                                 "repaired": False})
                if len(body) >= min_rows:
                    cap = _caption_above(page, tb.bbox)
                    out.append({"caption": cap or f"table on page {pno}", "title": cap,
                                "page": pno, "columns": columns, "rows": body,
                                "origin": f"layout:{name}"})
    return out


# ---------------------------------------------------------------- reader 2b: captions
def caption_tables(page_text, page_no=None):
    """The classic reader: a caption announces the table, rows follow until a source note."""
    stop = re.compile(r"^\s*(source|note|champ|lecture|\(\d\)|…\s*:|\*)", re.I)
    tables, cur = [], None

    def close():
        nonlocal cur
        if cur and cur["rows"]:
            widths = [len(r["values"]) for r in cur["rows"]]
            _repair_widths(cur["rows"], max(set(widths), key=widths.count))
            width = max(len(r["values"]) for r in cur["rows"])
            cols = cur["columns"][:width]
            cur["columns"] = _dedupe(cols + [f"col{i+1}" for i in range(len(cols), width)])
            tables.append(cur)
        cur = None

    for line in page_text.splitlines():
        s_ = norm(line).strip()
        cap = CAPTION.match(s_)
        if cap:
            close()
            years = YEAR.findall(s_)
            cur = {"caption": s_, "title": cap.group("title").strip(), "page": page_no,
                   "rows": [], "columns": years if len(years) >= 2 else [],
                   "origin": "caption"}
            continue
        if cur is None:
            continue
        if stop.match(s_):
            close(); continue
        m = ROW.match(s_)
        if not m:
            continue
        vals = split_numbers(m.group("nums"))
        if len(vals) < 2:
            continue
        if _is_header(vals) and not cur["rows"]:
            cur["columns"] = [str(int(v)) for v in vals]
            continue
        cur["rows"].append({"label": m.group("label").strip(" .:—–-"),
                            "code": (m.group("code") or "").strip(), "values": vals,
                            "nums": m.group("nums"), "line": s_, "repaired": False})
    close()
    return tables


# ---------------------------------------------------------------- reader 3: lines of text
def block_tables(page_text, page_no=None, min_rows=2):
    lines = [norm(l).rstrip() for l in page_text.splitlines()]
    matches = []
    for ln in lines:
        m = ROW.match(ln.strip())
        vals = split_numbers(m.group("nums")) if m else []
        matches.append((len(matches), m, vals) if m and len(vals) >= 2 else None)

    tables, run = [], []

    def flush():
        if len(run) < min_rows:
            run.clear(); return
        widths = [len(v) for _, _, v in run]
        width = max(set(widths), key=widths.count)
        rows = []
        for i, m, v in run:
            code, label = (m.group("code") or "").strip(), m.group("label").strip(" .:—–-")
            if code and not ("." in code or (len(code) >= 2 and code.startswith("0"))):
                code, label = "", f"{code} {label}"      # '65 years and over'
            rows.append({"label": label, "code": code, "values": v, "nums": m.group("nums"),
                         "line": lines[i].strip(), "repaired": False})
        _repair_widths(rows, width)
        width = max(len(r["values"]) for r in rows)
        first, columns = run[0][0], []
        for back in range(first - 1, max(first - 4, -1), -1):
            cand = lines[back].strip()
            if not cand:
                continue
            hm = ROW.match(cand)
            if hm:
                hv = split_numbers(hm.group("nums"))
                if _is_header(hv) and len(hv) >= width - 1:
                    columns = [str(int(v)) for v in hv if v is not None]; break
            years = _columns_from_header(cand)
            if len(years) >= width - 1:
                columns = years; break
        caption = ""
        for back in range(first - 1, max(first - 6, -1), -1):
            cand = lines[back].strip()
            if len(cand) > 8 and not ROW.match(cand):
                caption = cand; break
        columns = _dedupe(columns[:width]
                          + [f"col{i+1}" for i in range(len(columns), width)])
        tables.append({"caption": caption or f"table on page {page_no}", "title": caption,
                       "page": page_no, "columns": columns, "rows": rows,
                       "origin": "text-block"})
        run.clear()

    prev = None
    for item in matches:
        if item is None:
            flush(); prev = None; continue
        if prev and abs(len(item[2]) - prev) > 1:
            flush()
        run.append(item); prev = len(item[2])
    flush()
    return tables
````

````python
# UTILITY — the referee: which reading of each page do we keep?
def _reconciles(t):
    """Does a total row equal the sum of its parts? A shattered reading will not."""
    totals = [r for r in t["rows"] if TOTAL_WORDS.match(r["label"])]
    parts = [r for r in t["rows"] if not TOTAL_WORDS.match(r["label"])]
    if not totals or len(parts) < 2:
        return False
    for i in range(len(totals[0]["values"])):
        tv = totals[0]["values"][i]
        pv = [r["values"][i] for r in parts
              if i < len(r["values"]) and r["values"][i] is not None]
        if tv is None or len(pv) < 2:
            continue
        if abs(sum(pv) - tv) <= max(1.0, abs(tv) * 0.005) or abs(sum(pv) - 100) <= 1.0:
            return True
    return False


def _thousands_split(tables):
    """A whole number under 1000 immediately left of a three-digit one, row after row."""
    hits = rows = 0
    for t in tables:
        for r in t["rows"]:
            rows += 1
            for a, b in zip(r["values"], r["values"][1:]):
                if (a is not None and b is not None and float(a).is_integer()
                        and 0 < a < 1000 and 100 <= abs(b) < 1000):
                    hits += 1
                    break
    return bool(rows) and hits / rows >= 0.5


def reading_score(tables):
    """
    Not "who found the most numbers" — a reader that splits one figure into three
    finds more numbers and is worse. Reward filled cells, penalise a sparse grid,
    reward named columns and a real caption, reward a table that adds up.
    """
    if not tables:
        return 0.0
    cells = slots = 0
    for t in tables:
        for r in t["rows"]:
            cells += sum(v is not None for v in r["values"])
            slots += max(len(t["columns"]), len(r["values"]))
    density = cells / slots if slots else 0
    named = sum(1 for t in tables for c in t["columns"] if not str(c).startswith("col"))
    titled = sum(1 for t in tables if CAPTION.match(t.get("caption") or ""))
    bonus = 1.6 if any(_reconciles(t) for t in tables) else 1.0
    return (cells * density + named * 0.5 + titled * 2) * bonus


def _labels(tables):
    return {r["label"].lower().strip() for t in tables for r in t["rows"] if r["label"]}


def read_document(pdf_path, pages, only=None):
    """Run every reader on every page, keep the best reading, and say which won."""
    by_page = {}
    for t in layout_tables(pdf_path, only=only):
        by_page.setdefault(t["page"], []).append(t)

    tables, report = [], []
    for p in pages:
        if only and p["page"] not in only:
            continue
        candidates = {
            "layout": by_page.get(p["page"], []),
            "caption": [dict(t, origin="caption") for t in caption_tables(p["text"], p["page"])],
            "text-block": block_tables(p["text"], p["page"]),
        }
        ranked = sorted(candidates.items(), key=lambda kv: reading_score(kv[1]), reverse=True)
        name, chosen = ranked[0]
        if chosen and _thousands_split(chosen):
            mine = _labels(chosen)

            def same(a, b):
                a, b = a[:14], b[:14]
                return len(a) >= 6 and (a.startswith(b[:len(a)]) or b.startswith(a[:len(b)]))

            for alt_name, alt in ranked[1:]:
                if not alt or _thousands_split(alt):
                    continue
                theirs = _labels(alt)
                overlap = sum(any(same(x, y) for x in mine) for y in theirs) / max(len(theirs), 1)
                narrower = (max((len(t["columns"]) for t in alt), default=0)
                            < max((len(t["columns"]) for t in chosen), default=0))
                if overlap >= 0.6 and narrower:
                    name, chosen = alt_name, alt
                    break
        report.append({"page": p["page"], **{k: len(v) for k, v in candidates.items()},
                       "kept": name if chosen else "none", "tables": len(chosen),
                       "cells": int(sum(sum(v is not None for v in r["values"])
                                        for t in chosen for r in t["rows"]))})
        tables += chosen
    return tables, report


def baseline_facts(tables):
    out = []
    for t in tables:
        for r in t["rows"]:
            for i, v in enumerate(r["values"]):
                if v is None:
                    continue
                out.append(dict(table=t["title"] or t["caption"], caption=t["caption"],
                                page=t["page"], label=r["label"], code=r.get("code", ""),
                                position=i, value=v, line=r["line"],
                                origin=t.get("origin", ""),
                                column=t["columns"][i] if i < len(t["columns"]) else f"col{i+1}"))
    return out
````

````python
EXTRACTION_SYSTEM = """You transcribe tables from official statistical publications.
You are a transcriber, not an analyst. You copy what is printed. You never estimate,
never interpolate, never complete a series, never convert a unit.

Rules that matter more than fluency:
- A comma inside a number is a DECIMAL separator in French-language documents: 12,8 is 12.8.
- A space inside a number is a THOUSANDS separator: 18 742 is 18742.
- "…", "-", "n.d." and an empty cell all mean MISSING. Return null. Never return 0.
- A footnote marker glued to a value ("17,0 (1)") is not part of the value: return 17.0.
- If you cannot read a cell with certainty, return null for it. An omission is a small
  problem; an invented figure is a serious one.
- source_line must be COPIED CHARACTER FOR CHARACTER from the page text you were given.
  Do not tidy it, do not reflow it, do not translate it. It will be checked automatically
  against the page and a line that is not found will cause the whole row to be discarded.

Return a single JSON object: {"facts": [ ... ]} and nothing else. No prose, no code fences."""

EXTRACTION_USER = """PAGE {page} of the publication "{title}".

Transcribe every data cell of every table on this page into the schema below.
One object per cell: one indicator, one area, one period, one value.

{{"facts": [{{
  "indicator": "short_snake_case_key, reused across all cells of the same table",
  "indicator_label": "the label as printed, in the language of the document",
  "area": "the row label, copied exactly as printed",
  "period": "the column label: 2025, T4 2025, 2021-2025 …",
  "value": 12.8,
  "unit": "%",
  "page": {page},
  "source_line": "the line of the page text this cell was read from, verbatim",
  "is_total": false
}}]}}

PAGE TEXT — this is the only material you may use:
<<<
{text}
>>>"""
````

````python
def extract_page(page, title, max_chars=12000):
    text = page["text"][:max_chars]
    out = llm([{"role": "system", "content": EXTRACTION_SYSTEM},
               {"role": "user", "content": EXTRACTION_USER.format(
                   page=page["page"], title=title, text=text)}],
              json_mode=True, temperature=0.0, max_tokens=6000,
              label=f"extract p{page['page']}")
    facts = out.get("facts", out if isinstance(out, list) else [])
    for f in facts:
        f["page"] = page["page"]
        f["value"] = parse_number(f.get("value")) if isinstance(f.get("value"), str) \
            else f.get("value")
    return facts


_head = [l for l in PAGES[0]["text"].splitlines() if l.strip()][:3]
DOC_TITLE = (_head[1] if len(_head) > 1 else (_head[0] if _head else PDF_PATH.stem))[:90]
EXTRACTED = []

if OFFLINE:
    print("offline — deriving records from the rules-only read instead of a model.\n"
          "Expect the difference to be visible later: the rules know which numbers are in\n"
          "which row, but not that column 3 is a density, or that 'Total' and 'Ensemble du\n"
          "pays' are the same row. Naming things is precisely what the model is for.\n")
    for b in BASELINE:
        EXTRACTED.append(dict(indicator=re.sub(r"\W+", "_", b["table"].lower())[:28] or "series",
                              indicator_label=b["table"], area=b["label"], period=str(b["column"]),
                              value=b["value"], unit="%" if "%" in b["caption"] else "",
                              page=b["page"], source_line=b["line"],
                              is_total=bool(re.match(r"(total|ensemble|whole)", b["label"], re.I))))
else:
    for pno in sorted(TARGET_PAGES):
        page = next(p for p in PAGES if p["page"] == pno)
        try:
            got = extract_page(page, DOC_TITLE)
            EXTRACTED += got
            print(f"page {pno}: {len(got)} cells  ({CALLS[-1]['seconds']}s)")
        except Exception as exc:
            print(f"page {pno}: FAILED — {exc}")

df = pd.DataFrame(EXTRACTED)
print(f"\n{len(df)} cells from the model, {len(BASELINE)} from the rules")
display(df.head(10))
````

#### 3.1 &nbsp;Gate 1 and 2 &mdash; is the evidence real?

````python
def number_variants(v):
    """
    Every way `v` could plausibly be printed, and no way that would round to a
    different figure — so 22.3 never produces '22' and cannot match a neighbouring '22,4'.
    """
    v = float(v)
    out = set()
    for d in (0, 1, 2, 3):
        if round(v, d) != round(v, 6):
            continue
        ip, _, fp = f"{abs(v):.{d}f}".partition(".")
        ints = {ip}
        if len(ip) > 3:
            ints |= {f"{int(ip):,}".replace(",", " "), f"{int(ip):,}",
                     f"{int(ip):,}".replace(",", ".")}
        for i in ints:
            out.add(f"{i},{fp}" if fp else i)
            out.add(f"{i}.{fp}" if fp else i)
    return {o for o in out if o}


def value_on_page(v, text) -> bool:
    if v is None or not text:
        return False
    hay = norm(text)
    return any(re.search(r"(?<![\d.,])" + re.escape(c) + r"(?![\d]|[.,]\d)", hay)
               for c in number_variants(v))


PAGE_TEXT = {p["page"]: p["text"] for p in PAGES}


def gate_provenance(f):
    ok = value_on_page(f.get("value"), PAGE_TEXT.get(f.get("page"), ""))
    return ok, "" if ok else f"{f.get('value')} is not printed on page {f.get('page')}"


def gate_evidence(f):
    line = norm(f.get("source_line") or "").strip()
    if not line:
        return False, "no source line quoted"
    ok = line in norm(PAGE_TEXT.get(f.get("page"), ""))
    return ok, "" if ok else "the quoted source line does not appear on that page"
````

#### 3.2 &nbsp;Gate 3 &mdash; units and plausible ranges

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#FDF6E4;border:1px solid #D49A00;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#8A6400;margin-bottom:5px">YOUR TURN</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">Edit <code>UNIT_RULES</code> for your own document. This is domain knowledge and no model has it: only you know that your unemployment series is a percentage between 0 and 100, that your population column is in thousands, and that a value of 4 912 in a column labelled &lsquo;%&rsquo; is nonsense. Written down once, it protects every future run.</div></div>


````python
PERCENT = {"%", "percent", "pourcentage", "pour cent", "p.100"}


def seed_unit_rules(facts):
    """
    Build a starting rule per indicator from what was actually extracted, so that
    gate 3 does something on a document nobody has written rules for. A percentage
    gets the 0-100 box; anything else gets a wide box around its own spread, which
    catches an order-of-magnitude slip without pretending to domain knowledge.
    """
    rules = {}
    for f in facts:
        ind = f.get("indicator")
        if ind is None:
            continue
        r = rules.setdefault(ind, {"units": set(), "values": []})
        if f.get("unit"):
            r["units"].add(str(f["unit"]).strip().lower())
        if f.get("value") is not None:
            r["values"].append(f["value"])
    out = {}
    for ind, r in rules.items():
        units = r["units"] or None
        vals = r["values"] or [0]
        if units and units <= PERCENT:
            out[ind] = dict(units=units, lo=-100, hi=100, source="auto: percentage")
        else:
            lo, hi = min(vals), max(vals)
            span = max(abs(hi), abs(lo), 1)
            out[ind] = dict(units=units, lo=lo - span * 10, hi=hi + span * 10,
                            source="auto: observed spread ×10")
    return out


AUTO_RULES = seed_unit_rules(EXTRACTED)

# ============================== YOUR TURN ==================================
# Rules you write yourself beat anything derived from the data, because only you
# know that the population column is in thousands and cannot exceed 2 million.
# Whatever you leave out falls back on AUTO_RULES, which is weaker but never inert.
MY_UNIT_RULES = {
    # "unemployment": dict(units={"%", "pourcentage"}, lo=0, hi=100),
    # "population":   dict(units={"milliers", "thousands"}, lo=0, hi=2_000_000),
}
# ===========================================================================
UNIT_RULES = {**AUTO_RULES, **MY_UNIT_RULES}
print(f"{len(UNIT_RULES)} indicators under a range rule "
      f"({len(MY_UNIT_RULES)} written by hand, {len(UNIT_RULES) - len(MY_UNIT_RULES)} derived)")
for k, v in list(UNIT_RULES.items())[:6]:
    print(f"   {str(k)[:34]:<34} {str(v.get('units') or 'any unit')[:26]:<26} "
          f"[{v['lo']:,.0f} … {v['hi']:,.0f}]")


def gate_unit_range(f):
    rule, problems = UNIT_RULES.get(f.get("indicator")), []
    unit = (f.get("unit") or "").strip().lower()
    v = f.get("value")
    if rule:
        if rule["units"] is not None and unit not in {u.lower() for u in rule["units"]}:
            problems.append(f"unit '{f.get('unit')}' not allowed for {f.get('indicator')}")
        if v is not None and not (rule["lo"] <= v <= rule["hi"]):
            problems.append(f"{v} is outside the plausible range for {f.get('indicator')}")
    elif unit in PERCENT and v is not None and not (-100 <= v <= 100):
        # no rule written for this indicator, but the unit alone says enough
        problems.append(f"{v} cannot be a percentage")
    return not problems, "; ".join(problems)


def gate_is_active(indicator):
    return indicator in MY_UNIT_RULES
````

#### 3.3 &nbsp;Gate 4 &mdash; do the parts add up?

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#E8F5EF;border:1px solid #00A86A;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#00704A;margin-bottom:5px">Why this step matters</div>
<div style="font-size:14px;color:#231F20;line-height:1.55">A total row is a gift: the publication has already told you what the parts must sum to. When they do not, you know something is wrong &mdash; but not <i>what</i>. So the cell below goes one step further and asks, for each row in turn: if this one cell held the value that would close the gap, is that value actually printed on the page? Usually one or two rows answer yes, and those are the cells to re-read. Automated triage, no model involved.</div></div>


````python
TOTAL_WORDS = re.compile(r"^(total|ensemble|whole|national|pays|country|tous|all|general|"
                         r"g[ée]n[ée]ral|overall|toutes)\b", re.I)


def is_total(f):
    return bool(f.get("is_total")) or bool(TOTAL_WORDS.match(str(f.get("area", ""))))


def gate_arithmetic(facts, tolerance=0.005):
    """Reconcile every (indicator, period) group that has a total row. Returns checks."""
    checks = []
    groups = {}
    for f in facts:
        if f.get("value") is None:
            continue
        groups.setdefault((f.get("indicator"), f.get("period")), []).append(f)

    for (ind, period), rows in groups.items():
        totals = [r for r in rows if is_total(r)]
        parts = [r for r in rows if not is_total(r)]
        if not totals or len(parts) < 2:
            continue
        published = totals[0]["value"]
        computed = sum(r["value"] for r in parts)
        biggest = max(r["value"] for r in parts)
        additive = abs(computed - published) <= max(1.0, abs(published) * tolerance)
        shares = abs(computed - 100) <= 1.0
        if additive or shares:
            checks.append(dict(indicator=ind, period=period, kind="sum" if additive else "shares",
                               computed=round(computed, 2), published=published,
                               gap=round(computed - published, 2), passed=True,
                               candidates=[], note=""))
            continue

        # A RATE IS NOT ADDITIVE. Regional unemployment rates do not sum to the national
        # rate and never will: the national figure is a weighted average, not a total.
        # Testing them would produce a wall of false alarms, so the gate abstains and
        # says so, which is different from passing.
        unit = (parts[0].get("unit") or "").strip().lower()
        if unit in PERCENT:
            checks.append(dict(indicator=ind, period=period, kind="rate — not additive",
                               computed=round(computed, 2), published=published, gap=None,
                               passed=None, candidates=[],
                               note="a rate cannot be reconciled by addition; this gate abstains. "
                                    "Weight by population to check it properly."))
            continue
        if published < biggest:
            # A total smaller than its own largest part is not a total. It is a mean, a
            # density, an index or a rate. Abstain rather than raise a false alarm.
            checks.append(dict(indicator=ind, period=period, kind="not a sum",
                               computed=round(computed, 2), published=published, gap=None,
                               passed=None, candidates=[],
                               note=f"the published figure ({published:,.1f}) is smaller than the "
                                    f"largest part ({biggest:,.1f}), so it is an average or a "
                                    f"density, not a total; this gate abstains".replace(",", " ")))
            continue

        gap = published - computed
        candidates = [dict(area=r["area"], read=r["value"], should_be=round(r["value"] + gap, 2))
                      for r in parts
                      if value_on_page(r["value"] + gap, PAGE_TEXT.get(r.get("page"), ""))]
        checks.append(dict(indicator=ind, period=period, kind="sum",
                           computed=round(computed, 2), published=published,
                           gap=round(gap, 2), passed=False, candidates=candidates[:3], note=""))
        note = (f"parts sum to {computed:,.1f} but the document publishes {published:,.1f}"
                .replace(",", " "))
        for r in parts + totals:
            r.setdefault("_arith", []).append(note)
        for c in candidates[:3]:
            for r in parts:
                if r["area"] == c["area"]:
                    r["_arith"].append(f"re-read this cell: {c['should_be']:,.1f} is printed on "
                                       f"page {r.get('page')} and would close the gap"
                                       .replace(",", " "))
    return checks
````

#### 3.4 &nbsp;Gate 5 &mdash; the free second opinion

````python
def row_cells(label, page=None, line=None):
    """
    The same row label often appears in several tables on one page. The model already
    told us which LINE it read, and the rules reader recorded the line of every row it
    found — so match on the line first. That is what the quoted evidence is for.
    """
    if line:
        exact = [b for b in BASELINE if norm(b["line"]) == norm(line)]
        if exact:
            return exact
    lab = norm(str(label)).lower().strip()
    hits = [b for b in BASELINE
            if norm(b["label"]).lower().strip() == lab and (page is None or b["page"] == page)]
    lines = {b["line"] for b in hits}
    return hits if len(lines) <= 1 else []      # ambiguous row → no opinion, not a failure


def gate_cross_read(f):
    """
    match     same value, same column, same row  → strong agreement
    mismatch  that column of that row holds another number
    row-only  the value is on the row but in a different column   (column drift)
    absent    the value is not on that row at all                 (row misalignment)
    no-row    the rules reader never found that row               (no opinion, not a failure)
    """
    cells = row_cells(f.get("area"), f.get("page"), f.get("source_line"))
    v = f.get("value")
    if not cells or v is None:
        return True, "", "no-row"
    same_col = [c for c in cells if str(c["column"]) == str(f.get("period"))]
    if same_col:
        got = same_col[0]["value"]
        if abs(got - v) < 1e-9:
            return True, "", "match"
        return False, f"the rules reading of that cell is {got}, the model said {v}", "mismatch"
    if any(abs(c["value"] - v) < 1e-9 for c in cells):
        return True, "", "row-only"
    return False, (f"{v} is nowhere on the '{f.get('area')}' row; that row reads "
                   f"{[c['value'] for c in cells][:6]}"), "absent"
````

#### 3.5 &nbsp;Run every gate and apply the publication rule

<div style="font-family:Calibri, Carlito, 'Segoe UI', system-ui, sans-serif;background:#FBEDEB;border:1px solid #B83B2E;
 border-left-width:4px;border-radius:8px;padding:13px 17px;margin:10px 0">
<div style="font-size:12.5px;font-weight:700;letter-spacing:.4px;color:#8C2A20;margin-bottom:5px">The publication rule</div>
<div style="font-size:14px;color:#231F20;line-height:1.55"><b>Fails gate 1 or 2 &rarr; the figure is dropped.</b> No page evidence means no reason to believe it exists. It never reaches the dashboard.<br><b>Fails gate 3, 4 or 5 &rarr; the figure is published but flagged</b> &lsquo;to review&rsquo;, with the reason visible to the reader, until a statistician clears it.<br><br>That asymmetry is deliberate. An invented figure is a false statement. A figure that does not reconcile may simply be rounding, or a footnote, or a real feature of the data &mdash; but the reader is entitled to know before they quote it.</div></div>


````python
ARITHMETIC = gate_arithmetic(EXTRACTED)

rows = []
for f in EXTRACTED:
    p_ok, p_msg = gate_provenance(f)
    e_ok, e_msg = gate_evidence(f)
    u_ok, u_msg = gate_unit_range(f)
    x_ok, x_msg, verdict = gate_cross_read(f)
    a_msgs = f.get("_arith", [])
    a_ok = not a_msgs

    if not (p_ok and e_ok):
        status = "dropped"
    elif p_ok and e_ok and u_ok and a_ok and x_ok:
        status = "verified"
    else:
        status = "review"

    f["status"] = status
    f["reasons"] = [m for m in (p_msg, e_msg, u_msg, x_msg, *a_msgs) if m]
    f["cross_read"] = verdict
    rows.append(dict(indicator=f.get("indicator"), area=f.get("area"), period=f.get("period"),
                     value=f.get("value"), unit=f.get("unit"), page=f.get("page"),
                     provenance=p_ok, evidence=e_ok, unit_range=u_ok,
                     arithmetic=a_ok, cross_read=verdict, status=status,
                     reason=" | ".join(f["reasons"])))

REPORT = pd.DataFrame(rows)
VERIFIED = [f for f in EXTRACTED if f["status"] != "dropped"]
REPORT.to_csv(WORKDIR / "verification_report.csv", index=False)

counts = REPORT.status.value_counts().reindex(["verified", "review", "dropped"]).fillna(0).astype(int)

# How many gates actually had an opinion? A gate that abstains is not a gate that
# passed, and a verification report that hides the difference is worse than none.
COVERAGE = {
    "provenance": 1.0,
    "evidence": 1.0,
    "unit & range": sum(f.get("indicator") in UNIT_RULES for f in EXTRACTED) / max(len(EXTRACTED), 1),
    "arithmetic": sum(1 for f in EXTRACTED if f.get("_arith") is not None
                      or any(c["passed"] is not None and c["indicator"] == f.get("indicator")
                             for c in ARITHMETIC)) / max(len(EXTRACTED), 1),
    "cross-read": sum(f.get("cross_read") not in (None, "no-row") for f in EXTRACTED)
                  / max(len(EXTRACTED), 1),
}

print(f"{len(REPORT)} cells extracted")
for k, v in counts.items():
    share = v / max(len(REPORT), 1) * 100
    print(f"  {k:<9} {v:>4}  {'█' * int(share / 2.5)} {share:.0f}%")
print(f"\n{len(VERIFIED)} cells will reach the dashboard "
      f"({counts.get('review', 0)} of them flagged for review)")
print("full report written to verification_report.csv")

print("\ngate coverage — how often each gate had an opinion at all:")
for gate, cov in COVERAGE.items():
    bar = "█" * int(cov * 20)
    flag = "" if cov >= 0.5 else "   ← weak on this document"
    print(f"   {gate:<14} {cov * 100:5.0f}%  {bar:<20}{flag}")
if not MY_UNIT_RULES:
    print("\n   Gate 3 is running on auto-derived ranges only. They catch an order-of-")
    print("   magnitude slip, not a wrong unit — only you know that your population")
    print("   column is in thousands. Writing two or three rules into MY_UNIT_RULES is")
    print("   the highest-value three minutes in this notebook.")
if COVERAGE["cross-read"] < 0.5:
    print("\n   The rules-only readers could not find the rows the model reported, so the")
    print("   free second opinion is mostly unavailable here. Row misalignment — the one")
    print("   error class provenance cannot catch — is therefore largely unguarded.")
    print("   Say so in LIMITATIONS.md rather than letting the green ticks imply otherwise.")

if len(ARITHMETIC):
    arith = pd.DataFrame(ARITHMETIC)
    arith["verdict"] = arith.passed.map({True: "reconciles", False: "DOES NOT RECONCILE",
                                         None: "abstains"}).fillna("abstains")
    display(arith[["indicator", "period", "kind", "computed", "published", "gap",
                   "verdict", "note"]].style.hide(axis="index")
            .set_caption("Reconciliation against published totals"))
````

````python
# Everything the gates did not like, with the reason. This is the reviewer's worklist.
problems = REPORT[REPORT.status != "verified"]
if len(problems):
    display(problems[["status", "indicator", "area", "period", "value", "page", "reason"]]
            .style.hide(axis="index")
            .set_caption(f"{len(problems)} cells needing a human — read every line"))
else:
    print("every gate passed on every cell.")
    print("That is possible on a clean table — but check you are not testing nothing:")
    print("run the next cell to confirm the gates actually fire.")
````

<div style="font-size:14px;color:#231F20;line-height:1.55">A test suite that never fails is not evidence of quality; it is evidence of a test suite that is not testing. Before trusting a green verification report, corrupt the data on purpose and confirm each gate catches its own kind of error. Do this every time you change a prompt or a model.</div></div>


````python
import copy

def probe(mutate, description):
    sample = copy.deepcopy(EXTRACTED)
    target = mutate(sample)
    if target is None:
        return print(f"  – {description}: no suitable cell in this document, skipped")
    p_ok, _ = gate_provenance(target)
    e_ok, _ = gate_evidence(target)
    u_ok, _ = gate_unit_range(target)
    x_ok, _, _ = gate_cross_read(target)
    caught = [n for n, ok in (("provenance", p_ok), ("evidence", e_ok),
                              ("unit/range", u_ok), ("cross-read", x_ok)) if not ok]
    mark = "✓" if caught else "✗ NOT CAUGHT"
    print(f"  {mark} {description:<44} → {', '.join(caught) or 'nothing fired'}")


def first(sample, pred=lambda f: f.get("value") is not None):
    return next((f for f in sample if pred(f)), None)


print("injecting errors of the kind models actually make:\n")
probe(lambda s: (f := first(s)) and f.update(value=(f["value"] or 0) + 0.07) or f,
      "a value nudged to one that is not printed")
probe(lambda s: (f := first(s)) and f.update(source_line="Une ligne que le modèle a inventée")
      or f, "a fabricated source line")
probe(lambda s: (f := first(s, lambda x: gate_is_active(x.get("indicator")))) and
      f.update(unit="tonnes") or f, "a plausible number with the wrong unit")
if not any(gate_is_active(f.get("indicator")) for f in EXTRACTED):
    print("\n  note: no UNIT_RULES entry matches the indicator keys this document produced,")
    print("        so gate 3 only applies its generic percentage check. Add entries above —")
    print("        the keys to use are:", sorted({f.get("indicator") for f in EXTRACTED})[:6])
probe(lambda s: (f := first(s)) and f.update(
    value=next((b["value"] for b in BASELINE
                if abs(b["value"] - (f["value"] or 0)) > 1e-9), (f["value"] or 0) + 1)) or f,
      "a real number taken from the wrong row")
````

````python
# ============================== YOUR TURN ==================================
# Seed this from your office's own terminology list. It beats any model.
GLOSSARY = {
    "taux de chômage": "unemployment rate",
    "population active": "labour force",
    "population résidente": "resident population",
    "ménage": "household",
    "ménages ordinaires": "ordinary households",
    "ensemble du pays": "whole country",
    "milliers": "thousands",
    "milliers d'habitants": "thousands of inhabitants",
    "part": "share",
    "densité": "density",
    "variation": "change",
    "points": "points",
    "provisoire": "provisional",
    "donnée non disponible": "data not available",
    "enquête nationale sur l'emploi": "national labour force survey",
    "littoral nord": "north coast", "littoral sud": "south coast",
    "plateau central": "central plateau", "vallée de l'est": "eastern valley",
    "hauts plateaux": "highlands", "extrême ouest": "far west",
}
SOURCE_LANG = "fr"      # the language your publication is written in
# ===========================================================================
GLOSSARY = {k.lower(): v for k, v in GLOSSARY.items()}
````

````python
TRANSLATE_SYSTEM = """You translate the terminology of official statistics between French and
English. You are translating labels for a national statistical institute, not marketing copy.

- Use the standard term of official statistics, not a literal rendering.
- Keep it short: these are chart labels and table headers, not sentences.
- Preserve capitalisation conventions and any figures or units inside the label.
- If a term is a proper name (a region, a survey, an institution), transliterate rather than
  translate it, and keep the original spelling where it is already in the target language.
- Return JSON only: {"terms": {"source term": "translation", ...}}. No prose, no fences."""


def translate_terms(terms, src="fr", dst="en", check=True):
    """Glossary first; model only for the remainder; back-translation on what the model wrote."""
    out, unknown = {}, []
    for t in terms:
        hit = GLOSSARY.get(str(t).lower().strip())
        out[t] = hit if (hit and src == "fr" and dst == "en") else None
        if out[t] is None:
            unknown.append(t)

    if unknown and not OFFLINE:
        got = llm([{"role": "system", "content": TRANSLATE_SYSTEM},
                   {"role": "user", "content":
                    f"Translate these {src} terms into {dst}:\n"
                    + json.dumps(unknown, ensure_ascii=False)}],
                  json_mode=True, temperature=0.0, label="translate")
        got = got.get("terms", got)
        for t in unknown:
            out[t] = got.get(t) or t

        if check and got:
            back = llm([{"role": "system", "content": TRANSLATE_SYSTEM},
                        {"role": "user", "content":
                         f"Translate these {dst} terms back into {src}:\n"
                         + json.dumps(list(got.values()), ensure_ascii=False)}],
                       json_mode=True, temperature=0.0, label="back-translate")
            back = back.get("terms", back)
            drift = [(t, out[t], back.get(out[t]))
                     for t in unknown
                     if back.get(out[t], "").lower().strip() != str(t).lower().strip()]
            if drift:
                print("back-translation did not return to the original — check these by hand:")
                for a, b, c in drift[:12]:
                    print(f"   {a!r} → {b!r} → {c!r}")
    else:
        for t in unknown:
            out[t] = out[t] or t
    return out


AREAS = sorted({f["area"] for f in VERIFIED if f.get("area")})
INDICATOR_LABELS = {f["indicator"]: f.get("indicator_label") or f["indicator"] for f in VERIFIED}
UNITS = sorted({(f.get("unit") or "").strip() for f in VERIFIED} - {""})

to_translate = AREAS + list(INDICATOR_LABELS.values()) + UNITS
EN = translate_terms(to_translate)

LABELS = {
    "areas": {a: {"fr": a, "en": EN.get(a, a)} for a in AREAS},
    "indicators": {k: {"fr": v, "en": EN.get(v, v)} for k, v in INDICATOR_LABELS.items()},
    "units": {u: {"fr": u, "en": EN.get(u, u)} for u in UNITS},
}
display(pd.DataFrame([{"source": k, "en": v["en"]}
                      for group in LABELS.values() for k, v in group.items()])
        .style.hide(axis="index").set_caption("Labels that will appear on the dashboard"))
````

````python
DESIGN_SYSTEM = """You are the editor of a national statistical institute's public dashboard.
You write in English and French, to the standard of an official publication: plain, precise,
no marketing, no adjectives that a statistician would not sign.

You are given the STRUCTURE of a verified dataset — indicator keys, area names, periods, units
and the range of each series. You are NOT given the figures, and you must never state one.

Your job:
  - name the page and write a one-sentence standfirst, in both languages;
  - for each indicator, choose a chart and write its tab label, title, caption and a one-line
    reading note, in both languages;
  - choose at most four headline figures by POINTING AT CELLS (indicator, area, period).
    Write only the label; the value is filled in from the verified data;
  - write the limitations paragraph in both languages, using only what the notes below say.

Chart choice: "line" only when the period labels are a real time sequence (years, quarters,
months) — a line chart claims the x axis is ordered, so never use one for categories such as
"Male 2020 / Female 2020". "grouped" for a handful of categories, "bar" for a single column.

Return JSON only, in this shape, and nothing else:
{"kicker":{"en":"","fr":""},"title":{"en":"","fr":""},"subtitle":{"en":"","fr":""},
 "ovTitle":{"en":"","fr":""},"ovLede":{"en":"","fr":""},
 "views":[{"id":"","chart":"line","tab":{"en":"","fr":""},"title":{"en":"","fr":""},
           "sub":{"en":"","fr":""},"lede":{"en":"","fr":""},"src":{"en":"","fr":""}}],
 "kpis":[{"indicator":"","area":"","period":"","compare_period":null,
          "labels":{"en":"","fr":""},"good":"down","highlight":false}],
 "limBody":{"en":"","fr":""}}"""


def dataset_shape(facts):
    """What the model is allowed to see: structure, never values."""
    shape = {}
    for f in facts:
        s = shape.setdefault(f["indicator"], {"label": f.get("indicator_label"),
                                              "unit": f.get("unit"), "periods": set(),
                                              "areas": set(), "n": 0})
        s["periods"].add(str(f.get("period")))
        s["areas"].add(str(f.get("area")))
        s["n"] += 1
    return {k: {"label": v["label"], "unit": v["unit"],
                "periods": sorted(v["periods"]), "areas": sorted(v["areas"]),
                "cells": v["n"]} for k, v in shape.items()}


SHAPE = dataset_shape(VERIFIED)
REVIEW_NOTES = sorted({r for f in VERIFIED for r in f.get("reasons", [])})[:8]
print(json.dumps(SHAPE, ensure_ascii=False, indent=1)[:1200])
````

````python
def fallback_design(shape):
    """Used offline, and as the shape the model is asked to fill."""
    def temporal(periods):
        """A line chart claims the x axis is ordered. Only draw one if it really is."""
        pat = re.compile(r"^(?:(?:19|20)\d{2}|[TQ][1-4][ -]?(?:19|20)?\d{0,4}|"
                         r"\w{3,9}[ -]\d{2,4})$", re.I)
        return sum(bool(pat.match(str(p).strip())) for p in periods) >= len(periods) - 1

    views = []
    for k, s in shape.items():
        n = len(s["periods"])
        chart = ("line" if (n > 2 and temporal(s["periods"]))
                 else "grouped" if n > 1 else "bar")
        views.append({"id": k, "chart": chart,
                      "tab": {"en": s["label"] or k, "fr": s["label"] or k},
                      "title": {"en": s["label"] or k, "fr": s["label"] or k},
                      "sub": {"en": s["unit"] or "", "fr": s["unit"] or ""},
                      "lede": {"en": "", "fr": ""},
                      "src": {"en": f"Source: {DOC_TITLE}", "fr": f"Source : {DOC_TITLE}"}})
    return {"kicker": {"en": DOC_TITLE, "fr": DOC_TITLE},
            "title": {"en": "Statistics dashboard", "fr": "Tableau de bord statistique"},
            "subtitle": {"en": "", "fr": ""},
            "ovTitle": {"en": "Overview", "fr": "Vue d'ensemble"},
            "ovLede": {"en": "", "fr": ""}, "views": views, "kpis": [],
            "limBody": {"en": " ".join(REVIEW_NOTES), "fr": " ".join(REVIEW_NOTES)}}


if OFFLINE:
    DESIGN = fallback_design(SHAPE)
    print("offline — using the neutral fallback design.")
else:
    try:
        DESIGN = llm([{"role": "system", "content": DESIGN_SYSTEM},
                      {"role": "user", "content":
                       f"Publication: {DOC_TITLE}\n"
                       f"Source language: {SOURCE_LANG}\n\n"
                       f"Dataset structure:\n{json.dumps(SHAPE, ensure_ascii=False, indent=1)}\n\n"
                       f"Verification notes that must be reflected in limBody:\n"
                       f"{json.dumps(REVIEW_NOTES, ensure_ascii=False, indent=1)}"}],
                     json_mode=True, temperature=0.3, max_tokens=4000, label="design")
        base = fallback_design(SHAPE)
        for k, v in base.items():
            DESIGN.setdefault(k, v)
        known = {v["id"] for v in base["views"]}
        DESIGN["views"] = [v for v in DESIGN.get("views", []) if v.get("id") in known] \
            or base["views"]
        print("the model proposed:")
        for v in DESIGN["views"]:
            print(f"   {v['id']:<18} {v.get('chart','?'):<8} “{v.get('title',{}).get('en','')}”")
        print(f"   {len(DESIGN.get('kpis', []))} headline figures suggested")
    except Exception as exc:
        print("design call failed, using the fallback:", exc)
        DESIGN = fallback_design(SHAPE)
````

#### 5.3 &nbsp;Fill in the numbers (from the verified data, not the model)

````python
import unicodedata
from datetime import date


def slug(s):
    s = unicodedata.normalize("NFKD", str(s)).encode("ascii", "ignore").decode()
    return re.sub(r"[^a-z0-9]+", "-", s.lower()).strip("-") or "x"


def lookup(indicator, area, period):
    for f in VERIFIED:
        if (f.get("indicator") == indicator and str(f.get("area")) == str(area)
                and str(f.get("period")) == str(period)):
            return f
    return None


def resolve_kpis(spec):
    """The model pointed at cells; we read the values out of the verified dataset."""
    out = []
    for k in spec[:4]:
        cell = lookup(k.get("indicator"), k.get("area"), k.get("period"))
        if not cell or cell.get("value") is None:
            continue
        delta = None
        if k.get("compare_period"):
            prev = lookup(k["indicator"], k["area"], k["compare_period"])
            if prev and prev.get("value") is not None:
                delta = round(cell["value"] - prev["value"], 2)
        unit = cell.get("unit") or ""
        out.append({"labels": k.get("labels") or {"en": k.get("indicator"), "fr": k.get("indicator")},
                    "value": cell["value"],
                    "unit": {"en": LABELS["units"].get(unit, {}).get("en", unit), "fr": unit},
                    "delta": delta,
                    "deltaLabel": {"en": f"vs {k.get('compare_period')}",
                                   "fr": f"vs {k.get('compare_period')}"},
                    "good": k.get("good", "down"), "highlight": bool(k.get("highlight"))})
    return out


KPIS = resolve_kpis(DESIGN.get("kpis", []))
if not KPIS:                                     # no usable suggestion → pick the latest cells
    for ind, s in list(SHAPE.items())[:3]:
        period = s["periods"][-1]
        cell = next((f for f in VERIFIED if f["indicator"] == ind
                     and str(f["period"]) == period and is_total(f)), None) \
            or next((f for f in VERIFIED if f["indicator"] == ind and str(f["period"]) == period), None)
        if cell:
            KPIS.append({"labels": {"en": f'{LABELS["indicators"][ind]["en"]}, {period}',
                                    "fr": f'{LABELS["indicators"][ind]["fr"]}, {period}'},
                         "value": cell["value"], "unit": {"en": cell.get("unit", ""),
                                                          "fr": cell.get("unit", "")},
                         "delta": None, "good": "down", "highlight": not KPIS})

for k in KPIS:
    print(f"  {k['labels']['en']:<44} {k['value']}  {k['unit']['en']}")
````

#### 5.4 &nbsp;Assemble, validate, preview

````python
def inject(template, payload):
    blob = json.dumps(payload, ensure_ascii=False, indent=1)
    html, n = re.subn(r"/\*__DATA__\*/.*?/\*__ENDDATA__\*/",
                      lambda _: "/*__DATA__*/" + blob + "/*__ENDDATA__*/",
                      template, count=1, flags=re.S)
    if not n:
        raise RuntimeError("injection markers missing from the template")
    return html


def validate_dashboard(html, payload):
    """Cheap structural checks. Not a substitute for opening the page and using it."""
    problems, notes = [], []
    if "/*__DATA__*/" not in html:
        problems.append("data was not injected")
    m = re.search(r"/\*__DATA__\*/(.*?)/\*__ENDDATA__\*/", html, re.S)
    try:
        back = json.loads(m.group(1))
    except Exception as exc:
        problems.append(f"injected data is not valid JSON: {exc}")
        back = {}
    if back.get("facts") and len(back["facts"]) != len(payload["facts"]):
        problems.append("cell count changed during injection")

    external = [u for u in re.findall(r'<script[^>]+src="([^"]+)"', html)
                if not u.startswith("https://cdn.jsdelivr.net/")]
    if external:
        problems.append(f"unexpected external script: {external}")
    if re.search(r"fetch\(|XMLHttpRequest", html):
        notes.append("the page makes a network call — check what it is before publishing")
    for need, why in (("data-lang=\"fr\"", "French toggle"), ("data-lang=\"en\"", "English toggle"),
                      ("prefers-color-scheme", "dark-mode safety"), ("<title>", "page title")):
        if need not in html:
            notes.append(f"missing {why}")
    if len(html) > 5_000_000:
        problems.append("page is over 5 MB — too heavy for a public site")

    # the figures on the page must be the figures that passed verification
    published = {(f["indicator"], f["area"], f["period"], f["value"]) for f in back.get("facts", [])}
    expected = {(f["indicator"], f["area"], f["period"], f["value"]) for f in payload["facts"]}
    if published != expected:
        problems.append(f"{len(expected ^ published)} cells differ between the verified set "
                        "and the page")
    return problems, notes


HTML = inject(DASHBOARD_TEMPLATE, PAYLOAD)
problems, notes = validate_dashboard(HTML, PAYLOAD)

OUT = WORKDIR / "index.html"
if problems:
    print("VALIDATION FAILED — not written:")
    for p in problems:
        print("  ✗", p)
else:
    OUT.write_text(HTML, encoding="utf-8")
    print(f"✓ {OUT}  ({len(HTML) / 1024:,.0f} kB, self-contained)")
for n in notes:
    print("  !", n)
````

<div style="font-size:14px;color:#231F20;line-height:1.55">Prefer a <b>fine-grained</b> token? It works too, but needs three permissions rather than one: <i>Administration</i> read and write (to create the repository), <i>Contents</i> read and write (to upload the files) and <i>Pages</i> read and write (to switch hosting on).
````python
import base64
import time

GH_API = "https://api.github.com"


def gh(method, path, token=None, **kw):
    """One thin wrapper so every call carries the right headers and a real error."""
    headers = {"Accept": "application/vnd.github+json",
               "X-GitHub-Api-Version": "2022-11-28"}
    if token:
        headers["Authorization"] = f"Bearer {token}"
    return requests.request(method, GH_API + path, headers=headers, timeout=60, **kw)


def github_reachable():
    """
    Institutional and conference networks block or throttle api.github.com more often
    than anyone expects. Find out before asking anyone for a token.
    """
    try:
        r = gh("GET", "/rate_limit")
    except requests.RequestException as exc:
        return False, f"cannot reach api.github.com at all ({exc.__class__.__name__})"
    if r.status_code == 200:
        core = r.json().get("resources", {}).get("core", {})
        return True, f"reachable — {core.get('remaining', '?')}/{core.get('limit', '?')} " \
                     f"anonymous requests left"
    return False, f"api.github.com answered {r.status_code} — a proxy or firewall is in the way"


def _repo_exists(owner, repo, token):
    """200 means yes, 404 means no. Anything else is not an answer and must not be guessed."""
    r = gh("GET", f"/repos/{owner}/{repo}", token)
    if r.status_code == 200:
        return True
    if r.status_code == 404:
        return False
    raise RuntimeError(f"GitHub answered {r.status_code} when asked about {owner}/{repo}: "
                       f"{r.text[:200]}\nThat is not a yes or a no, so nothing was changed. "
                       f"Check the token scopes, the rate limit, and the network.")


def github_identity(token):
    r = gh("GET", "/user", token)
    if r.status_code == 401:
        raise RuntimeError("GitHub rejected the token. Is it expired, or mistyped?")
    r.raise_for_status()
    who = r.json()
    scopes = r.headers.get("x-oauth-scopes", "(fine-grained token)")
    return who["login"], scopes


def upload_folder(folder, owner, repo, token, branch="main", skip=(".git",)):
    """
    Upload every file through the Contents API. One commit per file — noisier than a
    real git push, but it needs no git binary, works identically in Colab, Kaggle and
    a browser, and is easy to read when it goes wrong.
    """
    sent = []
    for path in sorted(p for p in folder.rglob("*") if p.is_file()):
        rel = path.relative_to(folder).as_posix()
        if any(rel.startswith(s) for s in skip):
            continue
        body = {"message": f"lab02: publish {rel}",
                "content": base64.b64encode(path.read_bytes()).decode(),
                "branch": branch}
        cur = gh("GET", f"/repos/{owner}/{repo}/contents/{rel}", token,
                 params={"ref": branch})
        if cur.status_code == 200:
            body["sha"] = cur.json()["sha"]          # update rather than fail
        elif cur.status_code not in (404,):
            raise RuntimeError(f"could not check {rel} before uploading: "
                               f"{cur.status_code} {cur.text[:200]}")
        put = gh("PUT", f"/repos/{owner}/{repo}/contents/{rel}", token, json=body)
        if put.status_code not in (200, 201):
            raise RuntimeError(f"upload of {rel} failed: {put.status_code} {put.text[:300]}")
        sent.append(rel)
        print(f"   uploaded {rel}")
    return sent


def enable_pages(owner, repo, token, branch="main", wait=150):
    r = gh("POST", f"/repos/{owner}/{repo}/pages", token,
           json={"source": {"branch": branch, "path": "/"}})
    if r.status_code not in (201, 409):               # 409 = already switched on
        hint = ""
        if r.status_code in (403, 404):
            hint = ("\n   The Pages API needs the full 'repo' scope on a classic token — "
                    "'public_repo' gets you this far and no further. On a fine-grained token, "
                    "it needs Pages: read and write. Everything is already uploaded, so "
                    "switching Pages on by hand in Settings works too.")
        raise RuntimeError(f"could not enable Pages: {r.status_code} {r.text[:300]}{hint}")
    print("   Pages enabled, waiting for the first build …")
    deadline = time.time() + wait
    url = None
    while time.time() < deadline:
        s = gh("GET", f"/repos/{owner}/{repo}/pages", token)
        if s.status_code == 200:
            data = s.json()
            url = data.get("html_url") or url
            if data.get("status") == "built":
                return url, True
        time.sleep(5)
    return url, False


def publish_to_github(folder, repo_name, token, org=None, description="", branch="main"):
    login, scopes = github_identity(token)
    owner = org or login
    print(f"authenticated as {login}   scopes: {scopes}")

    if _repo_exists(owner, repo_name, token):
        print("   repository already exists — updating it in place")
    else:
        payload = {"name": repo_name, "description": description[:350],
                   "private": False, "auto_init": False, "has_issues": True}
        r = gh("POST", f"/orgs/{org}/repos" if org else "/user/repos", token, json=payload)
        if r.status_code == 422 and "already exists" in r.text:
            print("   repository already existed — updating it in place")
        elif r.status_code != 201:
            hint = ""
            if r.status_code == 403:
                hint = ("\n   403 usually means the token lacks the 'repo' scope, or the "
                        "organisation has not authorised it.")
            raise RuntimeError(f"could not create {owner}/{repo_name}: "
                               f"{r.status_code} {r.text[:300]}{hint}")
        else:
            print(f"   created https://github.com/{owner}/{repo_name}")

    upload_folder(folder, owner, repo_name, token, branch)
    url, built = enable_pages(owner, repo_name, token, branch)
    return {"owner": owner, "repo": repo_name, "url": url, "built": built,
            "repo_url": f"https://github.com/{owner}/{repo_name}"}
````
