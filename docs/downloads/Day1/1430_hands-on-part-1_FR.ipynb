```python
# @title ⚙️ Mise en forme { display-mode: "form" }
# ⚙️ Cellule technique : mise en forme des tableaux, cartes et graphiques.
import html as _html
import re, time, json, math, unicodedata
from dataclasses import dataclass
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import HTML, display

C = dict(vert="#00A86A", vert_fonce="#00704A", foret="#00553A", or_="#F5C242", ocre="#D49A00",
         bleu="#0E7C86", terre="#C4621D", brique="#B83B2E", encre="#231F20", ardoise="#5E6964",
         brume="#F4F7F5", menthe="#E8F5EF", sauge="#D5DED9", gris="#A9B5B0")

_CSS = f"""<style>
.rg {{font-family:'Segoe UI',Calibri,system-ui,sans-serif;color:{C['encre']};}}
.rg-titre {{font-weight:700;font-size:15px;margin:6px 0 6px 0;color:{C['encre']};}}
.rg-note {{font-size:12.5px;color:{C['ardoise']};font-style:italic;margin:6px 0 2px 0;}}
table.rg-t {{border-collapse:collapse;font-size:13px;width:100%;background:#fff;}}
table.rg-t th {{background:{C['foret']};color:#fff;padding:7px 10px;text-align:left;font-weight:600;}}
table.rg-t td {{padding:6px 10px;border-bottom:1px solid {C['sauge']};color:{C['encre']};background:#fff;vertical-align:top;}}
table.rg-t tr:nth-child(even) td {{background:{C['brume']};}}
.rg-carte {{background:#fff;border:1px solid {C['sauge']};border-radius:12px;padding:14px 18px;margin:10px 0;color:{C['encre']};}}
.rg-pill {{display:inline-block;border-radius:12px;padding:2px 10px;font-size:11.5px;font-weight:700;margin-right:6px;}}
.rg-src {{font-size:12.5px;color:{C['ardoise']};border-top:1px dashed {C['sauge']};padding-top:6px;margin-top:6px;}}
</style>"""

def pastille(texte, couleur):
    return (f'<span class="rg-pill" style="background:{couleur}22;color:{couleur};'
            f'border:1px solid {couleur};">{_html.escape(str(texte))}</span>')

def tableau(df, titre=None, note=None):
    corps = df.to_html(index=False, escape=False, classes="rg-t", border=0)
    t = f'<div class="rg-titre">{titre}</div>' if titre else ""
    n = f'<div class="rg-note">{note}</div>' if note else ""
    display(HTML(f'{_CSS}<div class="rg">{t}<div style="overflow-x:auto">{corps}</div>{n}</div>'))

def encadre(texte, genre="info"):
    styles = {"info": (C["bleu"], "💡"), "succes": (C["vert"], "✅"),
              "alerte": (C["ocre"], "⚠️"), "erreur": (C["brique"], "⛔")}
    couleur, icone = styles[genre]
    display(HTML(f'{_CSS}<div class="rg" style="background:{couleur}14;border:1px solid {couleur};'
                 f'border-radius:10px;padding:10px 16px;margin:8px 0;">{icone}&nbsp; {texte}</div>'))

def carte_reponse(r):
    couleur = C["ardoise"] if r["approche"].startswith("A") else C["vert"]
    if r["refus"]:
        etat = pastille("REFUS", C["ocre"])
    else:
        etat = pastille("RÉPONSE", C["vert_fonce"])
    conf = "" if r.get("confiance") is None else pastille(f"confiance {r['confiance']:.2f}", C["bleu"])
    mode = pastille(r["mode"], C["gris"])
    motif = f'<div class="rg-note">Motif : {_html.escape(r["motif"])}</div>' if r.get("motif") else ""
    sources = "".join(
        f'<div>[{i}] <b>{_html.escape(s["citation"])}</b> · score {s["score"]:.3f}<br>'
        f'<span style="color:{C["ardoise"]}">« {_html.escape(s["extrait"])}… »</span></div>'
        for i, s in enumerate(r["sources"], 1))
    display(HTML(
        f'{_CSS}<div class="rg rg-carte" style="border-left:6px solid {couleur};">'
        f'<div style="font-size:11.5px;letter-spacing:1.5px;color:{couleur};font-weight:700;">'
        f'{_html.escape(r["approche"]).upper()}</div>'
        f'<div style="font-size:15.5px;font-weight:700;margin:4px 0 8px 0;">❓ {_html.escape(r["question"])}</div>'
        f'<div>{etat}{conf}{mode}</div>'
        f'<div style="font-size:14.5px;margin:10px 0;line-height:1.5;">{_html.escape(r["reponse"])}</div>'
        f'{motif}<div class="rg-src">{sources or "Aucune source transmise."}</div></div>'))

plt.rcParams.update({
    "figure.dpi": 110, "figure.facecolor": "white", "axes.facecolor": "white",
    "axes.spines.top": False, "axes.spines.right": False, "axes.edgecolor": C["sauge"],
    "axes.labelcolor": C["ardoise"], "xtick.color": C["ardoise"], "ytick.color": C["ardoise"],
    "axes.titleweight": "bold", "axes.titlesize": 12.5, "axes.titlecolor": C["encre"],
    "font.size": 10, "legend.frameon": False,
})
encadre("Mise en forme prête.", "succes")
```

Le modèle n'est **pas indispensable** : sans lui, le notebook fonctionne en **mode extractif** (il cite la phrase la plus pertinente des documents). Mais pour observer le vrai comportement d'un RAG — et surtout sa capacité à **refuser** — il faut un modèle. Voici les options, de la plus adaptée à un usage intensif à la moins adaptée :

| Option | Coût | Volume d'essais | Mise en place | Pour qui ? |
|---|---|---|---|---|
| 🥇 **Ollama** (modèle ouvert sur la machine) | Gratuit | **Illimité** | 3 à 5 min (installation + téléchargement ~2 Go) | Ateliers, entraînement intensif, données sensibles |
| 🥈 **Groq** (API) | Gratuit, sans carte bancaire | **1 000 requêtes/jour par modèle**, 30/min, 8 000 tokens/min | 1 min (clé sur `console.groq.com/keys`) | Démarrage rapide, réponses très rapides |
| Anthropic, OpenAI (API) | Payant | Selon le budget | 1 min | Comparaison avec des modèles commerciaux |

**Ordre de grandeur.** Une exécution complète du notebook demande environ **30 appels** au modèle. Avec Groq, cela représente une vingtaine d'exécutions par jour et par modèle. Avec Ollama, aucune limite.

> ⚠️ **Quotas.** Les limites de Groq indiquées ci-dessus sont celles de sa documentation officielle en septembre 2026. Elles évoluent : consultez `console.groq.com/settings/limits` pour les valeurs actives sur votre compte. En cas de dépassement, le notebook **attend automatiquement** le délai demandé par le service, puis réessaie.

```python
# @title 🤖 Connexion au modèle { display-mode: "form" }
import urllib.error

def _secret(nom):
    valeur = os.environ.get(nom)
    if valeur:
        return valeur
    if ENV["nom"] == "colab":
        try:
            from google.colab import userdata
            return userdata.get(nom)
        except Exception:
            return None
    if ENV["nom"] == "kaggle":
        try:
            from kaggle_secrets import UserSecretsClient
            return UserSecretsClient().get_secret(nom)
        except Exception:
            return None
    return None

def _post(url, charge, entetes, delai=120):
    requete = urllib.request.Request(url, data=json.dumps(charge).encode("utf-8"),
                                     headers={"Content-Type": "application/json", **entetes})
    with urllib.request.urlopen(requete, timeout=delai) as reponse:
        return json.loads(reponse.read().decode("utf-8"))

class ClientLLM:
    """Interface unique vers Ollama, Groq, Anthropic ou OpenAI.
    Température 0, respect des quotas (pause entre appels, nouvelle tentative après une erreur 429)."""

    PAUSE_MIN = {"groq": 2.2, "anthropic": 0.5, "openai": 0.5, "ollama": 0.0}

    def __init__(self, choix="auto"):
        self.presence = {"ollama": _ollama_actif(),
                         "groq": bool(_secret("GROQ_API_KEY")),
                         "anthropic": bool(_secret("ANTHROPIC_API_KEY")),
                         "openai": bool(_secret("OPENAI_API_KEY"))}
        ordre = ["ollama", "groq", "anthropic", "openai"] if choix == "auto" else [choix]
        self.nom = next((f for f in ordre if self.presence.get(f)), "aucun")
        self.modele = {"ollama": MODELE_OLLAMA, "groq": MODELE_GROQ, "anthropic": MODELE_ANTHROPIC,
                       "openai": MODELE_OPENAI}.get(self.nom, "—")
        self.appels, self.attentes, self._dernier = 0, 0.0, 0.0

    @property
    def disponible(self):
        return self.nom != "aucun"

    def _appel(self, systeme, message, max_tokens):
        if self.nom == "ollama":
            d = _post("http://localhost:11434/api/chat",
                      {"model": self.modele, "stream": False, "options": {"temperature": 0},
                       "messages": [{"role": "system", "content": systeme},
                                    {"role": "user", "content": message}]}, {}, delai=600)
            return d["message"]["content"]
        if self.nom == "groq":
            charge = {"model": self.modele, "temperature": 0, "max_completion_tokens": max_tokens + 600,
                      "messages": [{"role": "system", "content": systeme},
                                   {"role": "user", "content": message}]}
            if "gpt-oss" in self.modele:
                charge["reasoning_effort"] = "low"          # raisonnement court : moins de tokens consommés
            d = _post("https://api.groq.com/openai/v1/chat/completions", charge,
                      {"Authorization": f"Bearer {_secret('GROQ_API_KEY')}"})
            return d["choices"][0]["message"].get("content") or ""
        if self.nom == "anthropic":
            d = _post("https://api.anthropic.com/v1/messages",
                      {"model": self.modele, "max_tokens": max_tokens, "temperature": 0,
                       "system": systeme, "messages": [{"role": "user", "content": message}]},
                      {"x-api-key": _secret("ANTHROPIC_API_KEY"), "anthropic-version": "2023-06-01"})
            return "".join(b.get("text", "") for b in d["content"])
        if self.nom == "openai":
            d = _post("https://api.openai.com/v1/chat/completions",
                      {"model": self.modele, "temperature": 0, "max_completion_tokens": max_tokens,
                       "messages": [{"role": "system", "content": systeme},
                                    {"role": "user", "content": message}]},
                      {"Authorization": f"Bearer {_secret('OPENAI_API_KEY')}"})
            return d["choices"][0]["message"]["content"]
        raise RuntimeError("Aucun fournisseur de modèle disponible.")

    def generer(self, systeme, message, max_tokens=400, essais=5):
        for tentative in range(essais):
            attente = self.PAUSE_MIN.get(self.nom, 0) - (time.time() - self._dernier)
            if attente > 0:
                time.sleep(attente)
            try:
                self._dernier = time.time()
                texte = self._appel(systeme, message, max_tokens)
                self.appels += 1
                return re.sub(r"<think>.*?</think>", "", texte, flags=re.S).strip()
            except urllib.error.HTTPError as erreur:
                if erreur.code == 429 and tentative < essais - 1:
                    delai = float(erreur.headers.get("retry-after") or 2 ** (tentative + 2))
                    delai = min(delai, 90)
                    print(f"⏳ Quota momentanément atteint ({self.nom}) : nouvelle tentative dans {delai:.0f} s…")
                    self.attentes += delai
                    time.sleep(delai)
                    continue
                detail = erreur.read().decode("utf-8", "ignore")[:200]
                raise RuntimeError(f"HTTP {erreur.code} : {detail}") from erreur

LLM = ClientLLM("aucun" if FOURNISSEUR == "aucun" else FOURNISSEUR)

tableau(pd.DataFrame([
    {"Fournisseur": "Ollama (local)", "Coût": "gratuit · illimité", "Détecté": "✅" if LLM.presence["ollama"] else "—",
     "Comment l'activer": "INSTALLER_OLLAMA = True, ou Ollama installé en local"},
    {"Fournisseur": "Groq", "Coût": "gratuit · 1 000 req./jour", "Détecté": "✅" if LLM.presence["groq"] else "—",
     "Comment l'activer": "secret ou variable GROQ_API_KEY"},
    {"Fournisseur": "Anthropic", "Coût": "payant", "Détecté": "✅" if LLM.presence["anthropic"] else "—",
     "Comment l'activer": "secret ou variable ANTHROPIC_API_KEY"},
    {"Fournisseur": "OpenAI", "Coût": "payant", "Détecté": "✅" if LLM.presence["openai"] else "—",
     "Comment l'activer": "secret ou variable OPENAI_API_KEY"},
]), titre="Fournisseurs de modèle détectés")

if LLM.disponible:
    try:
        debut = time.time()
        test = LLM.generer("Tu réponds en un seul mot.", "Réponds uniquement : OK", max_tokens=20)
        encadre(f"Fournisseur retenu : <b>{LLM.nom}</b> · modèle <code>{LLM.modele}</code> · "
                f"test de connexion réussi en {time.time() - debut:.1f} s (réponse : « {_html.escape(test[:30])} »).",
                "succes")
    except Exception as erreur:
        encadre(f"Le fournisseur <b>{LLM.nom}</b> a été détecté mais le test a échoué : "
                f"<code>{_html.escape(str(erreur)[:220])}</code><br>Le notebook passe en <b>mode extractif</b>. "
                "Vérifiez la clé, le nom du modèle et l'accès à Internet.", "erreur")
        LLM.nom, LLM.modele = "aucun", "—"
else:
    encadre("Aucun fournisseur : le notebook fonctionne en <b>mode extractif</b>. Toutes les étapes s'exécutent. "
            "Pour tester un vrai modèle gratuitement : clé Groq (1 min) ou Ollama (essais illimités).", "alerte")
```

```python
# @title ⚙️ T2 · Diagnostic et extraction structurée { display-mode: "form" }
from dataclasses import dataclass, field

def diagnostiquer(doc):
    lignes = []
    for page in doc:
        caracteres = len(page.get_text().strip())
        images = len(page.get_images(full=True))
        etat = "texte" if caracteres >= 50 else ("scannée" if images else "vide")
        lignes.append({"Page": page.number + 1, "Caractères": caracteres, "Images": images, "État": etat})
    return pd.DataFrame(lignes)

@dataclass
class Element:
    type: str           # "titre", "paragraphe", "tableau" ou "ocr"
    texte: str
    page: int
    zone: tuple         # rectangle (x0, y0, x1, y1) sur la page
    niveau: int = 0     # niveau de titre (1 = le plus important)
    section: str = ""

def _nfkc(t):
    return unicodedata.normalize("NFKC", t)

def lignes_de_page(page):
    lignes = []
    for bloc in page.get_text("dict")["blocks"]:
        if bloc.get("type") != 0:
            continue
        for ligne in bloc["lines"]:
            spans = [s for s in ligne["spans"] if s["text"].strip()]
            if not spans:
                continue
            lignes.append({"texte": _nfkc("".join(s["text"] for s in ligne["spans"])).strip(),
                           "taille": round(max(s["size"] for s in spans), 1),
                           "zone": pymupdf.Rect(ligne["bbox"]), "bloc": bloc["number"]})
    return lignes

def _signature(texte):
    return re.sub(r"\d+", "#", texte.lower()).strip()

def _tableaux(page):
    if not EXTRAIRE_TABLEAUX:
        return []
    with contextlib.redirect_stdout(io.StringIO()):
        trouves = page.find_tables().tables
    resultat = []
    for t in trouves:
        lignes = [[_nfkc((c or "").replace("\n", " ")).strip() for c in r] for r in t.extract()]
        entetes = [_nfkc((c or "").replace("\n", " ")).strip() for c in t.header.names]
        if lignes and lignes[0] == entetes:
            lignes = lignes[1:]
        if lignes and len(entetes) >= 2:
            resultat.append({"zone": pymupdf.Rect(t.bbox), "entetes": entetes, "lignes": lignes})
    return resultat

def lineariser(tab, legende):
    phrases = []
    for ligne in tab["lignes"]:
        if not any(ligne):
            continue
        valeurs = " ; ".join(f"{tab['entetes'][j]} : {ligne[j]}" for j in range(1, len(ligne))
                             if j < len(tab["entetes"]) and ligne[j])
        phrases.append(f"{ligne[0]} — {valeurs}.")
    return (f"{legende}\n" if legende else "") + "\n".join(phrases)

def ocr_page(page):
    installer("pytesseract", "pytesseract")
    if not shutil.which("tesseract") and shutil.which("apt-get") and ENV["nom"] in ("colab", "kaggle"):
        print("Installation de Tesseract (une minute)…")
        subprocess.run("apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1 || "
                       "(apt-get update -qq > /dev/null 2>&1 && apt-get install -y -qq tesseract-ocr tesseract-ocr-fra > /dev/null 2>&1)",
                       shell=True)
    import pytesseract
    from PIL import Image
    langues = set(pytesseract.get_languages(config=""))
    langue = "+".join(l for l in LANGUE_OCR.split("+") if l in langues) or "eng"
    if langue != LANGUE_OCR:
        print(f"⚠️ Langue OCR « {LANGUE_OCR} » indisponible : « {langue} » utilisée.")
    pix = page.get_pixmap(dpi=300)
    image = Image.frombytes("RGB", (pix.width, pix.height), pix.samples)
    texte = _nfkc(pytesseract.image_to_string(image, lang=langue))
    return [re.sub(r"\s*\n\s*", " ", p).strip() for p in re.split(r"\n\s*\n", texte) if len(p.strip()) > 3]

def extraire_structure(doc):
    pages = {p.number + 1: lignes_de_page(p) for p in doc}
    avec_texte = [n for n, l in pages.items() if l]
    # 1. en-têtes et pieds de page répétés
    repetes = set()
    if SUPPRIMER_ENTETES and len(avec_texte) >= 2:
        compte = Counter()
        for n in avec_texte:
            hauteur = doc[n - 1].rect.height
            vus = {_signature(l["texte"]) for l in pages[n]
                   if l["zone"].y1 < hauteur * 0.1 or l["zone"].y0 > hauteur * 0.9}
            compte.update(vus)
        repetes = {s for s, c in compte.items() if c >= max(2, 0.5 * len(avec_texte))}
    # 2. taille du corps de texte et tailles de titres
    poids = Counter()
    for l in (l for n in avec_texte for l in pages[n]):
        poids[l["taille"]] += len(l["texte"])
    corps = poids.most_common(1)[0][0] if poids else 10
    tailles_titres = sorted({l["taille"] for n in avec_texte for l in pages[n]
                             if l["taille"] >= corps * 1.2 and len(l["texte"]) <= 120}, reverse=True)
    niveau_de = {t: i + 1 for i, t in enumerate(tailles_titres)}
    elements, retirees = [], 0
    for n in sorted(pages):
        page = doc[n - 1]
        if not pages[n]:
            if UTILISER_OCR and page.get_images():
                for paragraphe in ocr_page(page):
                    elements.append(Element("ocr", paragraphe, n, tuple(page.rect)))
            continue
        tableaux = _tableaux(page)
        morceaux = []                         # (ordonnée, élément)
        courant = None
        for l in pages[n]:
            if _signature(l["texte"]) in repetes:
                retirees += 1
                continue
            if any(l["zone"].intersects(t["zone"]) for t in tableaux):
                continue
            if l["taille"] in niveau_de:
                if courant: morceaux.append(courant); courant = None
                if morceaux and morceaux[-1][1].type == "titre" and morceaux[-1][2] == l["bloc"]:
                    titre = morceaux[-1][1]
                    titre.texte += " " + l["texte"]
                    titre.zone = tuple(pymupdf.Rect(titre.zone) | l["zone"])
                else:
                    morceaux.append((l["zone"].y0, Element("titre", l["texte"], n, tuple(l["zone"]),
                                                          niveau_de[l["taille"]]), l["bloc"]))
                continue
            if courant and courant[2] == l["bloc"]:
                el = courant[1]
                if el.texte.endswith("-") and l["texte"][:1].islower():
                    el.texte = el.texte[:-1] + l["texte"]
                else:
                    el.texte += " " + l["texte"]
                el.zone = tuple(pymupdf.Rect(el.zone) | l["zone"])
            else:
                if courant: morceaux.append(courant)
                courant = (l["zone"].y0, Element("paragraphe", l["texte"], n, tuple(l["zone"])), l["bloc"])
        if courant: morceaux.append(courant)
        for t in tableaux:
            morceaux.append((t["zone"].y0, Element("tableau", "", n, tuple(t["zone"])), None, t))
        morceaux.sort(key=lambda m: m[0])
        for i, m in enumerate(morceaux):
            el = m[1]
            if el.type == "tableau":
                legende = ""
                precedent = elements[-1] if elements else None
                if precedent and precedent.type == "paragraphe" and re.match(r"^(Tableau|Table|Graphique|Figure)\b", precedent.texte):
                    legende = precedent.texte
                    elements.pop()
                el.texte = lineariser(m[3], legende)
                el.tableau = m[3]
            elements.append(el)
    # 3. rattacher chaque élément à sa section
    section = ""
    for i, el in enumerate(elements):
        if el.type == "titre":
            est_titre_du_document = i == 0 and el.niveau == 1 and len(tailles_titres) > 1
            section = "" if est_titre_du_document else el.texte
        el.section = section if el.type != "ocr" else f"Page {el.page} (texte reconnu par OCR)"
    return elements, {"en-têtes retirés": retirees, "taille du corps": corps, "tailles de titres": tailles_titres}


print("✅ T2 prêt.")
```

```python
# @title ⚙️ T3 · Découpage en passages { display-mode: "form" }
@dataclass
class Passage:
    id: int
    doc_id: int
    doc: str             # titre court du document
    section: str
    texte: str
    texte_index: str
    zones: list          # [(page, rectangle), …]
    type: str = "texte"

    @property
    def pages(self):
        return sorted({p for p, _ in self.zones})

    @property
    def citation(self):
        pages = self.pages
        p = f"p. {pages[0]}" if len(pages) == 1 else f"p. {pages[0]}–{pages[-1]}"
        return f"{self.doc} · {self.section or 'début du document'} · {p}"

def _phrases(texte):
    return [p.strip() for p in re.split(r"(?<=[.!?;])\s+(?=[A-ZÀ-Ý0-9«•–-])", texte) if p.strip()]

def decouper(document, depart=0, taille_max=700, chevauchement=1):
    passages = []
    def ajouter(section, unites, genre="texte"):
        texte = ("\n" if genre == "tableau" else " ").join(u[0] for u in unites)
        zones = list(dict.fromkeys((u[1], u[2]) for u in unites))
        entete = f"{document.titre} — {section}" if section else document.titre
        passages.append(Passage(depart + len(passages), document.id, document.court, section, texte,
                                f"{entete}. {texte}", zones, genre))
    tampon, section = [], None
    def vider():
        nonlocal tampon
        i = 0
        while i < len(tampon):
            lot, longueur, j = [], 0, i
            while j < len(tampon) and (not lot or longueur + len(tampon[j][0]) <= taille_max):
                lot.append(tampon[j]); longueur += len(tampon[j][0]) + 1; j += 1
            ajouter(section, lot)
            if j >= len(tampon):
                break
            i = max(j - chevauchement, i + 1)
        tampon = []
    for el in document.elements:
        if el.type == "titre":
            continue
        if el.section != section:
            vider()
            section = el.section
        if el.type == "tableau":
            vider()
            lignes = el.texte.split("\n")
            lot = []
            for ligne in lignes:
                if lot and sum(len(x) for x in lot) + len(ligne) > taille_max:
                    ajouter(section, [(x, el.page, el.zone) for x in ([lignes[0]] if lot[0] != lignes[0] else []) + lot], "tableau")
                    lot = []
                lot.append(ligne)
            if lot:
                ajouter(section, [(x, el.page, el.zone) for x in ([lignes[0]] if lot[0] != lignes[0] else []) + lot], "tableau")
            continue
        tampon += [(p, el.page, el.zone) for p in _phrases(el.texte)]
    vider()
    return passages
print("✅ T3 prêt.")
```

```python
# @title ⚙️ T4 · Recherche { display-mode: "form" }
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
from sklearn.preprocessing import Normalizer

MOTS_VIDES = set("""le la les l un une des de du d et en a au aux est sont ete ce cet cette ces
qui que quoi quel quelle quels quelles comment combien quand ou il elle on se sa son ses leur leurs
ne pas plus par pour sur dans avec y t n s qu etait sera peut peuvent fait faire lors entre apres
avant tout toute tous toutes ainsi dont tres document documents rapport dit indique selon information
informations donne donnent parle concernant sujet ils elles eux lui quoi chez aussi meme""".split())

SUFFIXES = sorted(["ements", "ement", "ations", "ation", "itions", "ition", "ances", "ance", "ences",
                   "ence", "euses", "euse", "eurs", "eur", "ites", "ite", "iques", "ique", "ives",
                   "ive", "ifs", "if", "elles", "elle", "iels", "iel", "aux", "ants", "ant", "ents",
                   "ent", "ees", "ee", "es", "er", "age", "s", "e", "x"], key=len, reverse=True)

def sans_accents(texte):
    return "".join(c for c in unicodedata.normalize("NFD", texte) if unicodedata.category(c) != "Mn")

def raciner(mot):
    if any(ch.isdigit() for ch in mot):
        return mot
    for suffixe in SUFFIXES:
        if mot.endswith(suffixe) and len(mot) - len(suffixe) >= 4:
            return mot[: -len(suffixe)]
    return mot

def analyser(texte):
    t = sans_accents(texte.lower())
    t = re.sub(r"(?<=\d)[\s\u00a0\u202f](?=\d{3}\b)", "", t)
    return [raciner(m) for m in re.findall(r"[a-z0-9]+(?:,\d+)?", t) if m not in MOTS_VIDES and len(m) > 1]

GLOSSAIRE = {
    "habitants": "population residente personnes", "gens": "population personnes",
    "vivent": "population residente", "combien de personnes": "population",
    "ville": "urbain", "villes": "urbain", "campagne": "rural",
    "sans emploi": "chomage chomeur", "chomeurs": "chomage", "travail": "emploi",
    "jeunes": "15 24 ans", "proportion": "taux part", "pourcentage": "taux part",
    "microdonnees": "fichiers donnees individuelles anonymises",
    "chercheur": "chercheurs", "obtenir": "acceder accessibles acces",
    "sort": "publie publies diffusion paraitra", "sortent": "publie publies diffusion paraitra",
    "sortira": "paraitra publie", "publication": "publies paraitra", "quand": "date publies paraitra",
    "pas repondu": "non reponse repondération", "non-reponse": "repondération",
    "premier trimestre": "t1", "deuxieme trimestre": "t2", "second trimestre": "t2",
    "troisieme trimestre": "t3", "quatrieme trimestre": "t4",
    "evolue": "hausse baisse", "evolution": "hausse baisse", "varie": "hausse baisse",
    "informel": "informelle protection sociale contrat", "salaire": "revenu",
    "echantillon": "logements tirage zones", "definition": "glossaire", "signifie": "glossaire",
    "prochain recensement": "recensement programme", "taille des menages": "taille moyenne menage",
}

def reformuler(question):
    q = " " + re.sub(r"[^a-z0-9-]+", " ", sans_accents(question.lower())) + " "
    ajouts = [(cle, termes) for cle, termes in GLOSSAIRE.items() if f" {cle} " in q]
    return question + " " + " ".join(t for _, t in ajouts), ajouts

def concepts(question):
    _, ajouts = reformuler(question)
    synonymes = {}
    for cle, termes in ajouts:
        for mot in analyser(cle) or [raciner(sans_accents(cle.split()[-1]))]:
            synonymes.setdefault(mot, set()).update(analyser(termes))
    return [(t, {t} | synonymes.get(t, set())) for t in dict.fromkeys(analyser(question))]

class IndexBM25:
    def __init__(self, passages, k1=1.5, b=0.75):
        self.passages, self.k1, self.b = passages, k1, b
        self.docs = [analyser(p.texte_index) for p in passages]
        self.tf = [Counter(d) for d in self.docs]
        self.longueur_moy = max(1e-9, np.mean([len(d) for d in self.docs]))
        df = Counter(t for d in self.docs for t in set(d))
        n = len(self.docs)
        self.idf = {t: math.log(1 + (n - f + 0.5) / (f + 0.5)) for t, f in df.items()}
        self.idf_max = max(self.idf.values()) if self.idf else 1.0

    def chercher(self, question, k=30):
        termes = analyser(reformuler(question)[0])
        scores = np.zeros(len(self.docs))
        for i, (tf, d) in enumerate(zip(self.tf, self.docs)):
            norme = self.k1 * (1 - self.b + self.b * len(d) / self.longueur_moy)
            for t in termes:
                f = tf.get(t, 0)
                if f:
                    scores[i] += self.idf[t] * f * (self.k1 + 1) / (f + norme)
        ordre = [i for i in np.argsort(-scores) if scores[i] > 0][:k]
        return [(self.passages[i], float(scores[i])) for i in ordre]

_MODELE_EMB = {}
class IndexSemantique:
    def __init__(self, passages):
        self.passages, self.mode, self.erreur = passages, None, None
        textes = [p.texte_index for p in passages]
        if UTILISER_EMBEDDINGS:
            try:
                if INSTALLER_EMBEDDINGS:
                    installer("sentence_transformers", "sentence-transformers")
                from sentence_transformers import SentenceTransformer
                if MODELE_EMBEDDINGS not in _MODELE_EMB:
                    _MODELE_EMB[MODELE_EMBEDDINGS] = SentenceTransformer(MODELE_EMBEDDINGS)
                self.modele = _MODELE_EMB[MODELE_EMBEDDINGS]
                self.matrice = self.modele.encode(textes, normalize_embeddings=True)
                self.mode, self.nom = "embeddings", f"Embeddings · {MODELE_EMBEDDINGS.split('/')[-1]}"
            except Exception as erreur:
                self.erreur = type(erreur).__name__
        if self.mode is None:
            self.vectoriseur = TfidfVectorizer(analyzer="char_wb", ngram_range=(3, 5),
                                               strip_accents="unicode", sublinear_tf=True)
            m = self.vectoriseur.fit_transform(textes)
            self.svd = TruncatedSVD(n_components=max(1, min(96, m.shape[0] - 1, m.shape[1] - 1)), random_state=0)
            self.normaliseur = Normalizer()
            self.matrice = self.normaliseur.fit_transform(self.svd.fit_transform(m))
            self.mode, self.nom = "lsa", "Sémantique approchée · LSA hors ligne"

    def _encoder(self, texte):
        if self.mode == "embeddings":
            return self.modele.encode([texte], normalize_embeddings=True)
        return self.normaliseur.transform(self.svd.transform(self.vectoriseur.transform([texte])))

    def chercher(self, question, k=30):
        scores = (self.matrice @ self._encoder(reformuler(question)[0]).T).ravel()
        ordre = np.argsort(-scores)[:k]
        return [(self.passages[i], float(scores[i])) for i in ordre]

def fusion_rrf(listes, k=60, top=12):
    scores, objets = {}, {}
    for liste in listes:
        for rang, (p, _) in enumerate(liste, start=1):
            scores[p.id] = scores.get(p.id, 0.0) + 1.0 / (k + rang)
            objets[p.id] = p
    ordre = sorted(scores, key=scores.get, reverse=True)[:top]
    return [(objets[i], scores[i]) for i in ordre]

class Reclasseur:
    def __init__(self, bm25):
        self.bm25 = bm25

    def poids(self, terme):
        return self.bm25.idf.get(terme, self.bm25.idf_max * 1.5)

    def couverture(self, question, passage):
        present = set(analyser(passage.texte_index))
        liste = concepts(question)
        total = sum(self.poids(t) for t, _ in liste)
        return sum(self.poids(t) for t, g in liste if g & present) / total if total else 0.0

    def reclasser(self, question, candidats):
        scores = [self.couverture(question, p) for p, _ in candidats]
        ordre = sorted(range(len(candidats)), key=lambda i: (-round(scores[i], 6), i))
        return [(candidats[i][0], scores[i]) for i in ordre]

def diversifier(classes, k, plusieurs_documents):
    """Au plus k-1 passages d'un même document quand plusieurs documents sont interrogés."""
    plafond = max(1, k - 1) if plusieurs_documents else k
    choisis, compte = [], Counter()
    for p, s in classes:
        if compte[p.doc_id] < plafond:
            choisis.append((p, s)); compte[p.doc_id] += 1
        if len(choisis) == k:
            return choisis
    for p, s in classes:
        if (p, s) not in choisis and len(choisis) < k:
            choisis.append((p, s))
    return choisis
print("✅ T4 prêt.")
```

```python
# @title ⚙️ T5 · Mémoire de conversation { display-mode: "form" }
DEBUTS_DE_SUIVI = ("et ", "et,", "pour ", "qu en est il", "qu'en est-il", "meme question", "même question",
                   "idem", "aussi ", "meme chose", "même chose", "et si ")
CATEGORIES = {
    "région": r"\b(littoral|centre|nord|sud)\b",
    "sexe": r"\b(femmes?|hommes?)\b",
    "trimestre": r"\b(t[1-4]|(?:premier|deuxieme|second|troisieme|quatrieme) trimestre)\b",
    "milieu": r"\b(urbain|rural|villes?|campagnes?)\b",
    "âge": r"\b(jeunes|adultes|seniors|15 a 24 ans)\b",
    "année": r"\b((?:19|20)\d{2})\b",
    "indicateur": r"\b(taux (?:de chomage|d'activite|d'emploi)|sous-emploi|emploi informel)\b",
}

def _aligne(texte):
    """Minuscules sans accents, en conservant la longueur (positions identiques)."""
    return "".join((sans_accents(ch) or ch)[:1] for ch in texte.lower())

def est_suivi(question, historique):
    if not historique:
        return False
    q = _aligne(question).strip()
    if q.startswith(tuple(_aligne(d) for d in DEBUTS_DE_SUIVI)):
        return True
    interrogative = re.match(r"^(quel|quelle|quels|quelles|combien|comment|qui|quand|ou|pourquoi|est-ce|que |qu'|y a-t-il)", q)
    return len(analyser(question)) <= 2 and not interrogative

PROMPT_SUIVI = """Tu réécris des questions. À partir de l'historique, réécris la DERNIÈRE question de
l'utilisateur pour qu'elle soit compréhensible seule, sans l'historique. Ne réponds pas à la question.
Renvoie uniquement la question réécrite, en français, sur une seule ligne."""

def reformuler_suivi(question, historique, utiliser_modele=True):
    """Renvoie (question autonome, reformulée ?, méthode)."""
    if not est_suivi(question, historique):
        return question, False, None
    precedente = historique[-1]["question_autonome"]
    if utiliser_modele and LLM.disponible:
        try:
            echanges = "\n".join(f"Utilisateur : {h['question_autonome']}\nAssistant : {h['reponse'][:200]}"
                                 for h in historique[-3:])
            texte = LLM.generer(PROMPT_SUIVI, f"HISTORIQUE\n{echanges}\n\nDERNIÈRE QUESTION\n{question}",
                                max_tokens=120).strip().strip('"«» ').splitlines()[0]
            if 5 < len(texte) < 300:
                return texte, True, "modèle"
        except Exception:
            pass
    resultat, q_al = precedente, _aligne(question)
    remplacements = 0
    for motif in CATEGORIES.values():
        nouveau = re.search(motif, q_al)
        if not nouveau:
            continue
        ancien = re.search(motif, _aligne(resultat))
        valeur = question[nouveau.start():nouveau.end()]
        if ancien:
            resultat = resultat[:ancien.start()] + valeur + resultat[ancien.end():]
            remplacements += 1
    if not remplacements:
        reste = re.sub(r"^\s*(et|pour|qu'en est-il|même question|aussi)\b[\s,]*", "", question, flags=re.I)
        resultat = precedente.rstrip(" ?") + " — " + reste.rstrip(" ?") + " ?"
    return resultat, True, "règles"
print("✅ T5 prêt.")
```

```python
# @title ⚙️ T6 · La bibliothèque : chargement, indexation et réponses { display-mode: "form" }
REFUS = "Information non disponible dans les documents."

PROMPT_STRICT = f"""Tu es l'assistant documentaire d'un institut national de la statistique.
Tu réponds à partir d'extraits de un ou plusieurs documents PDF.
Règles, par ordre de priorité :
1. Réponds UNIQUEMENT à partir des sources numérotées fournies. Si elles ne contiennent pas la réponse,
   écris exactement : « {REFUS} » et rien d'autre.
2. N'utilise jamais tes connaissances générales, même si tu crois connaître la réponse.
3. Recopie les chiffres exactement, avec leur unité, leur période et le document dont ils proviennent.
4. Si la question compare des périodes ou des documents, donne chaque valeur avec sa source.
5. Termine chaque phrase par le numéro de sa source entre crochets, par exemple [1].
6. Les sources de type tableau sont écrites « libellé — colonne : valeur ; … ».
7. Réponds en français, en quatre phrases au maximum."""

COMPARAISON = re.compile(r"\b(evolu\w*|entre|compar\w*|par rapport|difference|hausse|baisse|progress\w*|versus|vs)\b")

@dataclass
class DocumentPDF:
    id: int
    fichier: str
    titre: str
    octets: bytes
    pages: int
    diagnostic: object
    elements: list
    infos: dict

    @property
    def court(self):
        t = re.sub(r"\s*\((fictif|fictive)\)\s*$", "", self.titre)
        return t if len(t) <= 42 else t[:40].rstrip() + "…"

    @property
    def pages_scannees(self):
        return self.diagnostic.loc[self.diagnostic["État"] == "scannée", "Page"].tolist()

def est_refus(texte):
    t = sans_accents(texte.lower())
    return any(m in t for m in ["information non disponible", "ne contient pas", "ne contiennent pas",
                                "pas d'information", "aucune information", "je ne sais pas",
                                "ne permet pas de repondre", "ne precise pas", "ne mentionne pas"])

def _decouper_phrases(texte):
    return [p.strip() for p in re.split(r"(?<=[.!?])\s+|\n", texte) if len(p.strip()) > 15]

def _meilleure_phrase(liste, passage, poids):
    phrases = _decouper_phrases(passage.texte)
    meilleure, score_max = None, 0.0
    for n, phrase in enumerate(phrases):
        termes = set(analyser(phrase))
        if passage.type == "tableau" and n > 0:
            termes |= set(analyser(phrases[0]))
        score = sum(poids(t) for t, g in liste if g & termes)
        if score > score_max:
            if n > 0 and re.match(r"^(Il|Elle|Ils|Elles|Ce|Cette|Ces|Sa|Son)\b", phrase):
                phrase = phrases[n - 1] + " " + phrase
            if passage.type == "tableau" and n > 0:
                phrase = f"{phrases[0]} : {phrase}"
            meilleure, score_max = phrase, score
    return meilleure, score_max

def _normaliser_nombres(texte):
    t = texte.replace("\u00a0", " ").replace("\u202f", " ")
    return re.sub(r"(?<=\d) (?=\d{3}\b)", "", t)

def verifier_chiffres(texte, passages):
    """Chaque nombre de la réponse doit figurer dans au moins une des sources transmises."""
    sans_refs = re.sub(r"\[\d+\]", " ", _normaliser_nombres(texte))
    nombres = list(dict.fromkeys(n.replace(".", ",") for n in re.findall(r"\d+(?:[.,]\d+)?", sans_refs)))
    source = _normaliser_nombres(" ".join(p.texte for p, _ in passages))
    absents = [n for n in nombres if not re.search(rf"(?<![\d,]){re.escape(n)}(?!\d|,\d)", source)]
    return {"nombres": len(nombres), "absents": absents}

def termes_absents(question, passage):
    """Mots de la question dont aucun équivalent ne figure dans le passage retenu."""
    present = set(analyser(passage.texte_index))
    manquants = {t for t, g in concepts(question) if not g & present}
    mots = []
    for mot in re.findall(r"[\w'-]+", question):
        if any(t in manquants for t in analyser(mot)) and mot not in mots:
            mots.append(mot)
    return mots

AVIS = []
def enregistrer_avis(echange, utile, commentaire="", canal="notebook"):
    ligne = {"horodatage": time.strftime("%Y-%m-%d %H:%M:%S"), "canal": canal, "utile": bool(utile),
             "question": echange.get("question", ""), "question_autonome": echange.get("question_autonome", ""),
             "reponse": echange.get("reponse", ""), "sources": " | ".join(echange.get("citations", [])),
             "commentaire": commentaire}
    AVIS.append(ligne)
    chemin = DOSSIER_SORTIE / "avis_utilisateurs.csv"
    pd.DataFrame([ligne]).to_csv(chemin, mode="a", header=not chemin.exists(), index=False, encoding="utf-8-sig")
    return ligne

class Bibliotheque:
    def __init__(self):
        self.documents, self.passages = [], []
        self.seuil, self.calibration, self.pret = 0.5, None, False
        self.dossier = ENV["racine"] / "pdf"
        self.dossier.mkdir(exist_ok=True)

    # --- chargement -------------------------------------------------------------------------
    def _resoudre(self, sources):
        chemins = []
        for s in ([sources] if isinstance(sources, (str, Path)) else sources):
            s = str(s)
            if s == "exemple":
                chemins += creer_documents_exemple(self.dossier / "exemples")
            elif s.startswith(("http://", "https://")):
                nom = re.sub(r"[^\w.-]", "_", s.split("?")[0].rstrip("/").split("/")[-1]) or "document"
                cible = self.dossier / (nom if nom.lower().endswith(".pdf") else nom + ".pdf")
                requete = urllib.request.Request(s, headers={"User-Agent": "Mozilla/5.0"})
                with urllib.request.urlopen(requete, timeout=60) as reponse:
                    cible.write_bytes(reponse.read())
                chemins.append(cible)
            elif Path(s).is_dir():
                chemins += sorted(Path(s).rglob("*.pdf"))
            elif Path(s).exists():
                chemins.append(Path(s))
            else:
                print(f"⚠️ Source introuvable, ignorée : {s}")
        return chemins

    def charger(self, sources):
        nouveaux, erreurs = [], []
        deja = {d.fichier for d in self.documents}
        for chemin in self._resoudre(sources):
            try:
                octets = Path(chemin).read_bytes()
                if not octets.startswith(b"%PDF"):
                    raise ValueError("ce n'est pas un PDF")
                doc = pymupdf.open(stream=octets, filetype="pdf")
                if doc.needs_pass:
                    raise ValueError("PDF protégé par mot de passe")
                if Path(chemin).name in deja:
                    continue
                elements, infos = extraire_structure(doc)
                titre = ((doc.metadata or {}).get("title") or "").strip()
                if not titre:
                    titre = elements[0].texte if elements and elements[0].type == "titre" else Path(chemin).stem.replace("_", " ")
                d = DocumentPDF(len(self.documents), Path(chemin).name, titre, octets, doc.page_count,
                                diagnostiquer(doc), elements, infos)
                self.documents.append(d); nouveaux.append(d); deja.add(d.fichier)
            except Exception as erreur:
                erreurs.append((Path(chemin).name, str(erreur)[:120]))
        self.pret = False
        return nouveaux, erreurs

    # --- indexation --------------------------------------------------------------------------
    def indexer(self):
        self.passages = []
        for d in self.documents:
            self.passages += decouper(d, len(self.passages), TAILLE_MAX, CHEVAUCHEMENT_PHRASES)
        if not self.passages:
            raise ValueError("Aucun passage : vérifiez que les PDF contiennent du texte (ou activez l'OCR).")
        self.bm25 = IndexBM25(self.passages)
        self.semantique = IndexSemantique(self.passages)
        self.reclasseur = Reclasseur(self.bm25)
        self.calibrer()
        self.pret = True

    def calibrer(self):
        hors_sujet = ["Quelle équipe a gagné la dernière coupe du monde de football ?",
                      "Quelle est la recette traditionnelle du couscous ?",
                      "Quel temps fera-t-il demain dans la capitale ?",
                      "Quel est le prix du maïs au kilogramme sur les marchés ?",
                      "Combien de vaccins ont été administrés contre la rougeole ?",
                      "Quelle est la production nationale de cacao ?",
                      "Qui est l'auteur du roman le plus vendu cette année ?",
                      "Combien d'hôpitaux compte la région du Nord ?"]
        titres = [re.sub(r"^\s*[\dIVX]+[.)]?\s*", "", e.texte) for d in self.documents
                  for e in d.elements if e.type == "titre" and e.niveau > 1]
        avec = [f"Que dit le document sur : {t} ?" for t in dict.fromkeys(titres)][:15]
        lignes = [{"Question": q, "Réponse existe": True, "Confiance": round(self.confiance(q), 3)} for q in avec]
        lignes += [{"Question": q, "Réponse existe": False, "Confiance": round(self.confiance(q), 3)} for q in hors_sujet]
        self.calibration = pd.DataFrame(lignes)
        plafond = self.calibration.loc[~self.calibration["Réponse existe"], "Confiance"].max()
        self.seuil_calibre = round(min(plafond + 0.05, 0.95), 2)
        self.seuil = SEUIL_MANUEL if SEUIL_MANUEL is not None else self.seuil_calibre

    # --- recherche et réponse ----------------------------------------------------------------
    def confiance(self, question):
        r = self.rechercher(question, 1)
        return r[0][1] if r else 0.0

    def rechercher(self, question, k=None, docs=None):
        k = k or K_RESULTATS
        autorises = set(range(len(self.documents))) if docs is None else set(docs)
        filtre = lambda liste: [(p, s) for p, s in liste if p.doc_id in autorises]
        candidats = fusion_rrf([filtre(self.bm25.chercher(question, 40)),
                                filtre(self.semantique.chercher(question, 40))], top=14)
        classes = self.reclasseur.reclasser(question, candidats)
        return diversifier(classes, k, len(autorises) > 1)

    def repondre(self, question, historique=None, docs=None, k=None, seuil=None, garde_fous=True, memoire=True):
        if not self.pret:
            self.indexer()
        debut = time.time()
        seuil = self.seuil if seuil is None else seuil
        autonome, reformulee, methode = (reformuler_suivi(question, historique or [], MEMOIRE_PAR_MODELE)
                                         if memoire else (question, False, None))
        resultats = self.rechercher(autonome, k, docs)
        conf = resultats[0][1] if resultats else 0.0
        alternative = None
        if docs is not None and set(docs) != set(range(len(self.documents))):
            meilleur = self.rechercher(autonome, 1, None)
            if meilleur and meilleur[0][0].doc_id not in set(docs) and meilleur[0][1] >= max(conf + 0.15, seuil):
                alternative = meilleur[0][0].doc
        r = {"question": question, "question_autonome": autonome, "reformulee": reformulee,
             "methode_reformulation": methode, "confiance": conf, "passages": resultats, "motif": None,
             "garde_fous": garde_fous, "heure": time.strftime("%H:%M:%S"), "alternative": alternative,
             "termes_absents": termes_absents(autonome, resultats[0][0]) if resultats else [],
             "verification": None, "citations": [p.citation for p, _ in resultats]}
        if garde_fous:
            annees = set(re.findall(r"\b(?:19|20)\d{2}\b", autonome))
            contenu = " ".join(p.texte_index for p, _ in resultats)
            manquantes = sorted(a for a in annees if a not in contenu)
            if conf < seuil:
                r["motif"] = f"confiance {conf:.2f} inférieure au seuil {seuil:.2f}"
            elif manquantes:
                r["motif"] = f"année(s) {', '.join(manquantes)} absente(s) des passages"
            if r["motif"]:
                return {**r, "reponse": REFUS, "refus": True, "mode": "garde-fou",
                        "duree_s": round(time.time() - debut, 2)}
        texte, mode = None, "extractif"
        if LLM.disponible:
            try:
                sources = "\n\n".join(f"[{i}] ({p.citation}{', tableau' if p.type == 'tableau' else ''})\n{p.texte}"
                                      for i, (p, _) in enumerate(resultats, 1))
                texte = LLM.generer(PROMPT_STRICT, f"SOURCES\n\n{sources}\n\nQUESTION\n{autonome}")
                mode = f"{LLM.nom} · {LLM.modele}"
            except Exception as erreur:
                r["motif"] = f"modèle indisponible ({str(erreur)[:80]}) : mode extractif"
        if texte is None:
            texte = self._extraire(autonome, resultats)
        refus = est_refus(texte)
        if not refus:
            r["verification"] = verifier_chiffres(texte, resultats)
        return {**r, "reponse": texte, "refus": refus, "mode": mode,
                "duree_s": round(time.time() - debut, 2)}

    def _extraire(self, question, resultats):
        liste = concepts(question)
        candidats = []
        for j, (p, _) in enumerate(resultats):
            phrase, score = _meilleure_phrase(liste, p, self.reclasseur.poids)
            if phrase:
                candidats.append((score * (1 - 0.02 * j), j, p, phrase))
        if not candidats:
            return REFUS
        candidats.sort(key=lambda c: -c[0])
        score0, j0, p0, phrase0 = candidats[0]
        texte = f"{phrase0} [{j0 + 1}]"
        if COMPARAISON.search(sans_accents(question.lower())):
            autres = [c for c in candidats[1:] if c[2].doc_id != p0.doc_id and c[0] >= 0.5 * score0]
            if autres:
                _, j1, p1, phrase1 = autres[0]
                texte += f" — {p1.doc} : {phrase1} [{j1 + 1}]"
        return texte

    # --- affichage ------------------------------------------------------------------------------
    def vignette(self, passage, **kw):
        doc = self.documents[passage.doc_id]
        page0 = passage.pages[0]
        return vignette_html(doc.octets, page0, [z for pg, z in passage.zones if pg == page0], **kw)

    def carte_html(self, r, vignettes=True):
        etat = pastille("REFUS", C["ocre"]) if r["refus"] else pastille("RÉPONSE", C["vert_fonce"])
        badges = etat + pastille(f"confiance {r['confiance']:.2f}", C["bleu"]) + pastille(r["mode"], C["gris"])
        if not r["garde_fous"]:
            badges += pastille("garde-fous désactivés", C["brique"])
        v = r.get("verification")
        if v and v["nombres"]:
            badges += (pastille(f"chiffres vérifiés {v['nombres']}/{v['nombres']}", C["vert_fonce"]) if not v["absents"]
                       else pastille(f"⚠ chiffre(s) absent(s) des sources : {', '.join(v['absents'])}", C["brique"]))
        alertes = ""
        if not r["refus"] and r.get("termes_absents"):
            alertes += (f'<div style="font-size:12.5px;color:{C["ocre"]};margin-top:6px;">⚠️ Non trouvé dans la source '
                        f'principale : <b>{_html.escape(", ".join(r["termes_absents"]))}</b> — vérifiez que la réponse '
                        f'porte bien sur votre question.</div>')
        if r.get("alternative"):
            alertes += (f'<div style="font-size:12.5px;color:{C["bleu"]};margin-top:6px;">💡 Une réponse plus pertinente '
                        f'figure probablement dans « <b>{_html.escape(r["alternative"])}</b> », exclu par le filtre.</div>')
        comprise = ""
        if r["reformulee"]:
            comprise = (f'<div style="font-size:12.5px;color:{C["bleu"]};margin-top:6px;">🧠 Compris comme : '
                        f'<i>{_html.escape(r["question_autonome"])}</i> '
                        f'<span style="color:{C["ardoise"]};">({r["methode_reformulation"]})</span></div>')
        motif = f'<div class="rg-note">Motif : {_html.escape(r["motif"])}</div>' if r.get("motif") else ""
        reponse = re.sub(r"\[(\d+)\]", lambda m: f'<sup style="background:{C["menthe"]};color:{C["vert_fonce"]};'
                         f'border-radius:4px;padding:0 4px;font-weight:700;">{m.group(1)}</sup>', _html.escape(r["reponse"]))
        sources = []
        for i, (p, s) in enumerate(r["passages"], 1):
            vign = self.vignette(p, dpi=55, largeur=105) if vignettes and i <= 2 and not r["refus"] else ""
            sources.append(
                f'<div style="display:flex;gap:10px;align-items:flex-start;margin:6px 0;">{vign}<div style="flex:1;">'
                f'<b>[{i}]</b> {pastille(p.doc, C["vert_fonce"])}<b>{_html.escape(p.section or "début")}</b> · '
                f'p. {", ".join(map(str, p.pages))} · {p.type} · confiance {s:.2f}<br>'
                f'<span style="color:{C["ardoise"]};">« {_html.escape(" ".join(p.texte.split())[:220])}… »</span></div></div>')
        return (f'<div class="rg rg-carte" style="border-left:6px solid {C["ocre"] if r["refus"] else C["vert"]};margin:0;">'
                f'<div>{badges}<span style="float:right;color:{C["ardoise"]};font-size:11px;">{r["heure"]} · {r["duree_s"]} s</span></div>'
                f'{comprise}{alertes}<div style="font-size:14.5px;margin:10px 0;line-height:1.55;">{reponse}</div>{motif}'
                f'<details {"open" if not r["refus"] else ""}><summary style="cursor:pointer;color:{C["vert_fonce"]};'
                f'font-weight:600;font-size:13px;">Sources ({len(r["passages"])})</summary>'
                f'<div class="rg-src">{"".join(sources) or "Aucune source."}</div></details></div>')

    def afficher(self, r):
        display(HTML(f'{_CSS}<div class="rg" style="margin:10px 0;"><div style="font-weight:700;font-size:15px;'
                     f'margin-bottom:6px;">❓ {_html.escape(r["question"])}</div>{self.carte_html(r)}</div>'))

    def resume(self):
        return pd.DataFrame([{"N°": d.id, "Document": d.titre, "Fichier": d.fichier, "Pages": d.pages,
                              "Pages scannées": ", ".join(map(str, d.pages_scannees)) or "—",
                              "Tableaux": sum(e.type == "tableau" for e in d.elements),
                              "Passages": sum(p.doc_id == d.id for p in self.passages) if self.passages else "—"}
                             for d in self.documents])

BIB = Bibliotheque()
_HISTORIQUE_DEMO = []

def demander(question, docs=None, memoire=True):
    """Pose une question à la bibliothèque, avec mémoire de la conversation précédente."""
    r = BIB.repondre(question, historique=_HISTORIQUE_DEMO if memoire else None, docs=docs)
    _HISTORIQUE_DEMO.append(r)
    BIB.afficher(r)
    return r
print("✅ T6 prêt · boîte à outils complète.")
```

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Titres, en-têtes répétés, mots coupés, tableaux et OCR.</div></div>

| Opération | Principe | Pourquoi |
|---|---|---|
| **Repérer les titres** | Ligne écrite nettement plus grand que le corps du texte | Chaque passage connaît sa **section** |
| **Retirer en-têtes et pieds de page** | Ligne répétée, en haut ou en bas, sur au moins la moitié des pages | « Page 3 » pollue chaque passage |
| **Recoller les mots coupés** | « statis-⏎tique » → « statistique » | La recherche trouve le mot entier |
| **Linéariser les tableaux** | Chaque ligne devient « Libellé — colonne : valeur ; … » | Un chiffre isolé n'a pas de sens |
| **OCR** (facultatif) | Reconnaissance du texte des pages scannées | Rendre visibles les numérisations |

La page ci-dessous montre ce que l'extracteur a reconnu : **vert** pour les titres, **gris** pour les paragraphes, **ocre** pour les tableaux, **rouge barré** pour les en-têtes retirés.

```python
def page_annotee(d, numero, dpi=75):
    doc = pymupdf.open("pdf", d.octets)
    page = doc[numero - 1]
    styles = {"titre": (0.0, 0.66, 0.42), "paragraphe": (0.37, 0.41, 0.39),
              "tableau": (0.83, 0.60, 0.0), "ocr": (0.05, 0.49, 0.53)}
    for e in d.elements:
        if e.page == numero:
            page.draw_rect(pymupdf.Rect(e.zone), color=styles[e.type], width=1.4)
    hauteur = page.rect.height
    for l in lignes_de_page(page):
        if (l["zone"].y1 < hauteur * 0.1 or l["zone"].y0 > hauteur * 0.9) and not any(
                e.page == numero and pymupdf.Rect(e.zone).intersects(l["zone"]) for e in d.elements):
            page.draw_rect(l["zone"], color=(0.72, 0.23, 0.18), width=1.2)
            page.draw_line(l["zone"].bl, l["zone"].tr, color=(0.72, 0.23, 0.18), width=0.8)
    return base64.b64encode(page.get_pixmap(dpi=dpi).tobytes("png")).decode()

d_demo = next((d for d in BIB.documents if any(e.type == "tableau" for e in d.elements)), BIB.documents[0])
tab = next((e for e in d_demo.elements if e.type == "tableau"), None)
page_demo = tab.page if tab else 1
bloc = ""
if tab:
    df_t = pd.DataFrame(tab.tableau["lignes"], columns=tab.tableau["entetes"][:len(tab.tableau["lignes"][0])])
    bloc = (f'<div class="rg-titre" style="margin-top:10px;">Le tableau tel qu\'il est lu…</div>'
            f'{df_t.to_html(index=False, classes="rg-t", border=0)}'
            f'<div class="rg-titre" style="margin-top:10px;">… et tel qu\'il est indexé</div>'
            f'<pre style="background:{C["brume"]};padding:10px;border-radius:8px;white-space:pre-wrap;font-size:12px;'
            f'color:{C["encre"]};">{_html.escape(tab.texte)}</pre>')
plan = "".join(f'<div style="margin-left:{(e.niveau - 1) * 16}px;">{"📘" if e.niveau == 1 else "▸"} {_html.escape(e.texte)} '
               f'<span style="color:{C["ardoise"]};font-size:11px;">p. {e.page}</span></div>'
               for e in d_demo.elements if e.type == "titre")
display(HTML(
    f'{_CSS}<div class="rg rg-carte"><div class="rg-titre">🔍 {_html.escape(d_demo.court)} · page {page_demo} vue par l\'extracteur</div>'
    f'<div style="display:flex;gap:20px;flex-wrap:wrap;align-items:flex-start;">'
    f'<img src="data:image/png;base64,{page_annotee(d_demo, page_demo)}" style="width:320px;border:1px solid {C["sauge"]};border-radius:6px;">'
    f'<div style="flex:1;min-width:280px;font-size:13px;"><div class="rg-titre">🧭 Plan détecté</div>'
    f'<div style="line-height:1.6;">{plan}</div>{bloc}</div></div></div>'))
tableau(pd.DataFrame([{"Document": d.court, "Titres": sum(e.type == "titre" for e in d.elements),
                       "Paragraphes": sum(e.type == "paragraphe" for e in d.elements),
                       "Tableaux": sum(e.type == "tableau" for e in d.elements),
                       "Paragraphes OCR": sum(e.type == "ocr" for e in d.elements),
                       "En-têtes retirés": d.infos["en-têtes retirés"]} for d in BIB.documents]),
        titre="Ce que l'extraction a produit, document par document")
```

<div style="color:#5E6964;font-size:14px;line-height:1.45;">Normalisation, glossaire, deux moteurs, fusion et reclassement.</div></div>

1. La question est **normalisée** (minuscules, accents, racines) et **enrichie** par le glossaire métier : « deuxième trimestre » → « T2 », « sans emploi » → « chômage ».
2. **BM25** trouve les mots exacts, **la recherche sémantique** les reformulations. Leurs classements sont **fusionnés** (RRF).
3. Le **reclasseur** mesure la **couverture de la question** par chaque passage (0 à 1) : c'est le niveau de confiance.
4. **Filtre par document** : on peut limiter la recherche à certains documents.
5. **Diversité** : quand plusieurs documents sont interrogés, un même document ne peut pas occuper toutes les places.

Une question de suivi n'a pas de sens seule : « Et pour les hommes ? » ne contient aucun indicateur. Avant la recherche, l'assistant la **réécrit en question autonome** :

| Méthode | Principe | Quand |
|---|---|---|
| **Par le modèle** | Le modèle reçoit les derniers échanges et réécrit la question | Un modèle est disponible et `MEMOIRE_PAR_MODELE = True` |
| **Par règles** | La question précédente est reprise, et la valeur qui change (région, sexe, trimestre, année, milieu, âge, indicateur) est **remplacée** | Sans modèle, ou en secours |

La reformulation est **toujours affichée** (« 🧠 Compris comme : … »), pour que l'utilisateur puisse la vérifier.

> ⚠️ **La mémoire peut se tromper.** Si l'utilisateur change de sujet avec une question très courte, elle peut être rattachée à tort à la précédente. D'où l'affichage systématique, et la possibilité de désactiver la mémoire dans les interfaces.

```python
if DOCUMENTS == ["exemple"]:
    JEU_TEST = [
        # famille, question, valeurs attendues (None = refus attendu), question précédente pour les suivis
        ("Tableau", "Quel est le taux de chômage au deuxième trimestre ?", ["7,8"], None),
        ("Tableau", "Quelle est la population de la région du Centre ?", ["4 870 500"], None),
        ("Texte", "Quelle est la part de l'emploi informel dans l'emploi total ?", ["71,5"], None),
        ("Texte", "Quelle est la taille moyenne d'un ménage ?", ["5,1"], None),
        ("Reformulée", "Les chercheurs peuvent-ils obtenir les microdonnées ?", ["accès sécurisé"], None),
        ("Plusieurs documents", "Comment le taux de chômage a-t-il évolué entre le quatrième trimestre 2025 et le premier trimestre 2026 ?", ["7,4"], None),
        ("Plusieurs documents", "Quand paraîtra le prochain bulletin de l'emploi ?", ["août 2026"], None),
        ("Suivi", "Et pour les hommes ?", ["8,1"], "Quel est le taux de chômage des femmes dans la région du Littoral ?"),
        ("Suivi", "Et au troisième trimestre ?", ["7,7"], "Quel est le taux de chômage au deuxième trimestre ?"),
        ("Page scannée", "Quand le prochain recensement est-il programmé ?", ["2027"], None),
        ("Hors documents", "Quel est le prix du maïs au kilogramme ?", None, None),
        ("Hors documents", "Quel sera le taux de chômage en 2030 ?", None, None),
    ]
else:
    JEU_TEST = []   # 🧪 À compléter : ("Famille", "Question", ["valeur attendue"] ou None, question précédente ou None)
    encadre("Vous utilisez vos propres PDF : complétez <code>JEU_TEST</code> pour valider l'assistant.", "alerte")

def _contient(texte, valeurs):
    t = sans_accents(texte.lower()).replace("\u00a0", " ").replace("\u202f", " ")
    return any(sans_accents(v.lower()) in t for v in valeurs)

LIGNES_TEST = []
for famille, question, attendu, precedente in JEU_TEST:
    hist = [BIB.repondre(precedente)] if precedente else []
    r = BIB.repondre(question, historique=hist)
    ok = r["refus"] if attendu is None else (not r["refus"] and _contient(r["reponse"], attendu))
    LIGNES_TEST.append({"Famille": famille, "Question": question + (f"<br><i>après : {precedente}</i>" if precedente else ""),
                        "Attendu": "refus" if attendu is None else " ou ".join(attendu),
                        "Réponse": _html.escape(r["reponse"][:130]),
                        "Chiffres": ("—" if not r["verification"] or not r["verification"]["nombres"]
                                     else "✅" if not r["verification"]["absents"] else "⚠️"),
                        "Résultat": "✅" if ok else "❌", "_ok": ok})
if LIGNES_TEST:
    TEST = pd.DataFrame(LIGNES_TEST)
    tableau(TEST.drop(columns="_ok"), titre=f"🧪 Jeu de test : {TEST['_ok'].sum()} / {len(TEST)} correct(s) ({TEST['_ok'].mean():.0%})")
    if DOCUMENTS == ["exemple"] and not UTILISER_OCR:
        encadre("La question sur le prochain recensement porte sur une <b>page scannée</b> : sans OCR, le refus est "
                "le comportement attendu. Activez <code>UTILISER_OCR = True</code> pour la rendre lisible.", "info")
```

| Contrôle | Principe | Ce qu'il détecte |
|---|---|---|

```python
import ipywidgets as widgets

HISTORIQUE = []
SUGGESTIONS = (["Quel est le taux de chômage des femmes dans la région du Littoral ?", "Et pour les hommes ?",
                "Comment le taux de chômage a-t-il évolué entre le quatrième trimestre 2025 et le premier trimestre 2026 ?",
                "Quelle est la taille moyenne d'un ménage ?", "Quel est le prix du maïs ?"]
               if DOCUMENTS == ["exemple"] else
               [f"Que dit le document sur : {e.texte} ?" for d in BIB.documents for e in d.elements
                if e.type == "titre" and e.niveau > 1][:4])

def _entete():
    modele = f"{LLM.nom} · {LLM.modele}" if LLM.disponible else "mode extractif"
    return (f'{_CSS}<div class="rg" style="background:linear-gradient(120deg,{C["foret"]} 0%,{C["vert"]} 100%);'
            f'border-radius:14px;padding:14px 20px;color:#fff;"><div style="color:{C["or_"]};font-size:11px;'
            f'font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE</div>'
            f'<div style="font-size:20px;font-weight:800;margin:2px 0;">💬 {len(BIB.documents)} document(s) · '
            f'{len(BIB.passages)} passages</div><div style="font-size:12.5px;color:#E6F6EE;">modèle : '
            f'{_html.escape(modele)} · recherche : {_html.escape(BIB.semantique.nom)}</div></div>')

def _fil():
    if not HISTORIQUE:
        return (f'{_CSS}<div class="rg rg-carte" style="text-align:center;color:{C["ardoise"]};padding:26px;">'
                'Posez une question, ou cliquez sur une suggestion.</div>')
    bulles = "".join(
        f'<div style="display:flex;justify-content:flex-end;margin:12px 0 6px 0;"><div style="background:{C["vert_fonce"]};'
        f'color:#fff;border-radius:14px 14px 2px 14px;padding:8px 14px;max-width:80%;font-size:14px;">'
        f'{_html.escape(r["question"])}</div></div><div style="display:flex;gap:8px;align-items:flex-start;">'
        f'<div style="font-size:22px;">🤖</div><div style="flex:1;">{BIB.carte_html(r, ui_vignettes.value)}</div></div>'
        for r in reversed(HISTORIQUE))
    return (f'{_CSS}<div class="rg" style="max-height:720px;overflow-y:auto;padding:4px 6px;border:1px solid '
            f'{C["sauge"]};border-radius:12px;background:{C["brume"]};">{bulles}</div>')

ui_question = widgets.Text(placeholder="Posez votre question sur les documents…", layout=widgets.Layout(flex="1", width="auto"))
ui_envoyer = widgets.Button(description="Envoyer", button_style="success", icon="paper-plane", layout=widgets.Layout(width="120px"))
ui_etat = widgets.HTML("")
ui_docs = [widgets.Checkbox(value=True, description=d.court, indent=False, layout=widgets.Layout(width="auto", margin="0 14px 0 0"))
           for d in BIB.documents]
ui_k = widgets.IntSlider(value=K_RESULTATS, min=1, max=8, description="Passages", continuous_update=False)
ui_seuil = widgets.FloatSlider(value=float(BIB.seuil), min=0.0, max=1.0, step=0.01, description="Seuil",
                               continuous_update=False, readout_format=".2f")
ui_garde = widgets.Checkbox(value=True, description="Garde-fous activés", indent=False)
ui_memoire = widgets.Checkbox(value=True, description="Mémoire de conversation", indent=False)
ui_vignettes = widgets.Checkbox(value=True, description="Afficher les pages", indent=False)
ui_effacer = widgets.Button(description="Nouvelle conversation", icon="refresh", layout=widgets.Layout(width="200px"))
ui_exporter = widgets.Button(description="Exporter", icon="download", layout=widgets.Layout(width="120px"))
ui_fil = widgets.HTML(_fil())
ui_utile = widgets.Button(description="Utile", icon="thumbs-up", layout=widgets.Layout(width="100px"))
ui_a_revoir = widgets.Button(description="À revoir", icon="thumbs-down", layout=widgets.Layout(width="110px"))
ui_commentaire = widgets.Text(placeholder="Commentaire facultatif sur la dernière réponse…",
                              layout=widgets.Layout(flex="1", width="auto"))

def donner_avis(utile):
    if not HISTORIQUE:
        ui_etat.value = "Posez d'abord une question."
        return
    enregistrer_avis(HISTORIQUE[-1], utile, ui_commentaire.value.strip(), "notebook")
    ui_commentaire.value = ""
    ui_etat.value = (f'<span style="color:{C["vert_fonce"]};">Merci ! Avis enregistré '
                     f'({len(AVIS)} au total, fichier avis_utilisateurs.csv).</span>')

def envoyer(_=None, question=None):
    q = (question or ui_question.value).strip()
    if not q:
        return
    docs = [d.id for d, c in zip(BIB.documents, ui_docs) if c.value]
    if not docs:
        ui_etat.value = f'<span style="color:{C["brique"]};">Cochez au moins un document.</span>'
        return
    ui_question.value = ""
    ui_envoyer.disabled = True
    ui_etat.value = f'<span style="color:{C["bleu"]};">⏳ Recherche dans {len(docs)} document(s)…</span>'
    try:
        r = BIB.repondre(q, historique=HISTORIQUE if ui_memoire.value else None, docs=docs, k=ui_k.value,
                         seuil=ui_seuil.value, garde_fous=ui_garde.value, memoire=ui_memoire.value)
        HISTORIQUE.append(r)
        ui_etat.value = (f'<span style="color:{C["ardoise"]};font-size:12px;">{len(HISTORIQUE)} échange(s) · '
                         f'dernière réponse en {r["duree_s"]} s' + (f' · {LLM.appels} appel(s) au modèle' if LLM.disponible else "") + "</span>")
    except Exception as erreur:
        ui_etat.value = f'<span style="color:{C["brique"]};">⛔ {_html.escape(str(erreur)[:200])}</span>'
    finally:
        ui_envoyer.disabled = False
        ui_fil.value = _fil()

def nouvelle_conversation(_=None):
    HISTORIQUE.clear()
    ui_fil.value = _fil()
    ui_etat.value = ""

def _serialiser(r):
    return {"question": r["question"], "question_autonome": r["question_autonome"], "reponse": r["reponse"],
            "refus": r["refus"], "confiance": round(r["confiance"], 3), "motif": r["motif"], "mode": r["mode"],
            "garde_fous": r["garde_fous"], "heure": r["heure"],
            "sources": [{"document": p.doc, "section": p.section, "pages": p.pages, "type": p.type,
                         "confiance": round(s, 3), "texte": p.texte} for p, s in r["passages"]]}

def exporter(_=None):
    if not HISTORIQUE:
        ui_etat.value = "Rien à exporter pour l'instant."
        return
    h = time.strftime("%Y%m%d_%H%M%S")
    (DOSSIER_SORTIE / f"conversation_{h}.json").write_text(
        json.dumps({"documents": [d.titre for d in BIB.documents], "echanges": [_serialiser(r) for r in HISTORIQUE]},
                   ensure_ascii=False, indent=2), encoding="utf-8")
    chemin_html = DOSSIER_SORTIE / f"conversation_{h}.html"
    chemin_html.write_text(f'<html><head><meta charset="utf-8"><title>Conversation</title></head>'
                           f'<body style="max-width:900px;margin:auto;">{_entete()}{_fil()}</body></html>', encoding="utf-8")
    ui_etat.value = f"✅ Conversation exportée : <code>{chemin_html.name}</code> (et sa version JSON)"
    if ENV["nom"] == "colab":
        from google.colab import files
        files.download(str(chemin_html))

boutons = []
for s in SUGGESTIONS:
    b = widgets.Button(description=(s[:44] + "…") if len(s) > 45 else s, tooltip=s,
                       layout=widgets.Layout(width="auto", margin="2px 4px 2px 0"))
    b.on_click(lambda _, s=s: envoyer(question=s))
    boutons.append(b)
ui_envoyer.on_click(envoyer)
ui_utile.on_click(lambda _: donner_avis(True))
ui_a_revoir.on_click(lambda _: donner_avis(False))
ui_effacer.on_click(nouvelle_conversation)
ui_exporter.on_click(exporter)
ui_vignettes.observe(lambda _: setattr(ui_fil, "value", _fil()), names="value")
try:
    import warnings
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        ui_question.on_submit(envoyer)
except Exception:
    pass

reglages = widgets.Accordion(children=[widgets.VBox([
    ui_k, ui_seuil, widgets.HBox([ui_garde, ui_memoire, ui_vignettes])])])
reglages.set_title(0, "⚙️ Réglages")
reglages.selected_index = None

INTERFACE = widgets.VBox([
    widgets.HTML(_entete()),
    widgets.HTML('<div style="font-size:12.5px;margin:10px 0 2px 0;"><b>Documents interrogés</b></div>'),
    widgets.HBox(ui_docs, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HTML('<div style="font-size:12.5px;margin:8px 0 2px 0;"><b>Suggestions</b></div>'),
    widgets.HBox(boutons, layout=widgets.Layout(flex_flow="row wrap")),
    widgets.HBox([ui_question, ui_envoyer], layout=widgets.Layout(margin="8px 0 2px 0")),
    ui_etat, reglages,
    widgets.HBox([widgets.HTML('<span style="font-size:12.5px;margin-right:6px;"><b>Dernière réponse :</b></span>'),
                  ui_utile, ui_a_revoir, ui_commentaire]),
    widgets.HBox([ui_effacer, ui_exporter]), ui_fil,
], layout=widgets.Layout(width="100%", max_width="980px"))
display(INTERFACE)
```

L'application reprend toutes les fonctions de l'assistant dans une page web :

- **glisser-déposer** de nouveaux PDF, indexés à la volée ;
- **filtre** des documents interrogés ;
- **conversation avec mémoire** et affichage de la reformulation ;
- **sources** avec vignettes de pages surlignées ;
- **réglages** : nombre de passages, seuil, garde-fous, mémoire.

Sur **Colab** et **Kaggle**, Gradio crée un **lien public temporaire** (`https://….gradio.live`), valable tant que la session est active et pour une durée limitée. En **local**, l'application s'ouvre sur `http://127.0.0.1:7860`.

> ⚠️ **Confidentialité.** Toute personne qui possède le lien public peut interroger vos documents et y déposer des fichiers. **Définissez `MOT_DE_PASSE_WEB`** avant de partager un lien, et n'utilisez pas de lien public pour des documents non publiés.

> ⚠️ **Application partagée.** Tous les utilisateurs d'un même lien partagent la même bibliothèque : un PDF déposé par l'un devient interrogeable par tous. Chaque utilisateur garde en revanche sa propre conversation.

```python
def _installer_gradio():
    try:
        import gradio as gr
        if int(gr.__version__.split(".")[0]) >= 5:
            return gr
        print(f"Gradio {gr.__version__} est trop ancien : mise à jour…")
    except ImportError:
        print("Installation de Gradio (une minute)…")
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-U", "gradio>=5"], check=False)
    importlib.invalidate_caches()
    import gradio as gr
    if int(gr.__version__.split(".")[0]) < 5:
        raise RuntimeError("Redémarrez la session (Exécution → Redémarrer), puis réexécutez le notebook.")
    return gr

def _libelles_docs():
    return [f"{d.id} · {d.court}" for d in BIB.documents]

def _texte_resume():
    return (f"**{len(BIB.documents)} document(s)** · {len(BIB.passages)} passages · seuil {BIB.seuil:.2f} · "
            + (f"modèle **{LLM.nom}**" if LLM.disponible else "mode extractif"))

def creer_application():
    gr = _installer_gradio()
    import inspect
    theme = gr.themes.Soft(primary_hue="emerald", secondary_hue="amber", neutral_hue="slate")
    css = ".gradio-container {max-width: 1250px !important} footer {display: none !important}"
    args_blocks = {"title": "Assistant documentaire RAG"}
    params_blocks = inspect.signature(gr.Blocks.__init__).parameters
    if "theme" in params_blocks:
        args_blocks.update(theme=theme, css=css)
    args_chat = {"height": 470, "label": "Conversation"}
    if "feedback_options" in inspect.signature(gr.Chatbot.__init__).parameters:
        args_chat["feedback_options"] = ["Like", "Dislike"]
    if "type" in inspect.signature(gr.Chatbot.__init__).parameters:
        args_chat["type"] = "messages"

    def repondre_web(question, messages, memoire_etat, docs, k, seuil, garde, memoire):
        messages = list(messages or [])
        if not question or not question.strip():
            return "", messages, memoire_etat, gr.update()
        ids = [int(x.split(" · ")[0]) for x in (docs or [])]
        if not ids:
            messages.append({"role": "assistant", "content": "Cochez au moins un document."})
            return "", messages, memoire_etat, gr.update()
        hist = [{"question_autonome": h["question_autonome"], "reponse": h["reponse"]} for h in (memoire_etat or [])]
        r = BIB.repondre(question.strip(), historique=hist if memoire else None, docs=ids, k=int(k),
                         seuil=float(seuil), garde_fous=bool(garde), memoire=bool(memoire))
        texte = r["reponse"]
        if r["reformulee"]:
            texte = f"_🧠 Compris comme : {r['question_autonome']}_\n\n" + texte
        if r["motif"]:
            texte += f"\n\n_Motif : {r['motif']}_"
        if not r["refus"]:
            texte += "\n\n" + "\n".join(f"**[{i}]** {p.citation}" for i, (p, _) in enumerate(r["passages"], 1))
        messages += [{"role": "user", "content": question.strip()}, {"role": "assistant", "content": texte}]
        memoire_etat = (memoire_etat or []) + [{"question": r["question"], "question_autonome": r["question_autonome"],
                                                 "reponse": r["reponse"], "citations": r["citations"]}]
        return "", messages, memoire_etat, f"{_CSS}{BIB.carte_html(r)}"

    def noter(donnees: gr.LikeData, memoire_etat):
        index = donnees.index[0] if isinstance(donnees.index, (list, tuple)) else donnees.index
        echanges = memoire_etat or []
        position = int(index) // 2
        if not echanges or position >= len(echanges):
            return "Avis non enregistré : échange introuvable."
        utile = donnees.liked if isinstance(donnees.liked, bool) else str(donnees.liked).lower() in ("true", "like")
        enregistrer_avis(echanges[position], utile, canal="web")
        return f"Merci ! Avis enregistré ({len(AVIS)} au total)."

    def ajouter_pdf(fichiers, docs):
        if not fichiers:
            return gr.update(), _texte_resume(), ""
        chemins = [f if isinstance(f, str) else getattr(f, "name", str(f)) for f in fichiers]
        nouveaux, erreurs = BIB.charger(chemins)
        BIB.indexer()
        choix = _libelles_docs()
        coches = list(docs or []) + [f"{d.id} · {d.court}" for d in nouveaux]
        message = f"✅ {len(nouveaux)} document(s) ajouté(s)." + "".join(f"\n⚠️ {n} : {e}" for n, e in erreurs)
        return gr.update(choices=choix, value=[c for c in choix if c in coches]), _texte_resume(), message

    with gr.Blocks(**args_blocks) as app:
        gr.HTML(f'<div style="background:linear-gradient(120deg,#00553A 0%,#00A86A 100%);border-radius:14px;'
                f'padding:16px 22px;color:#fff;font-family:\'Segoe UI\',Calibri,sans-serif;">'
                f'<div style="color:#F5C242;font-size:11px;font-weight:700;letter-spacing:2.5px;">ASSISTANT DOCUMENTAIRE · RAG</div>'
                f'<div style="font-size:24px;font-weight:800;">💬 Interrogez vos rapports PDF</div>'
                f'<div style="font-size:13px;color:#E6F6EE;">Réponses sourcées, pages surlignées, refus quand '
                f'l\'information n\'est pas dans les documents.</div></div>')
        memoire_etat = gr.State([])
        with gr.Row():
            with gr.Column(scale=1, min_width=300):
                resume = gr.Markdown(_texte_resume())
                docs = gr.CheckboxGroup(choices=_libelles_docs(), value=_libelles_docs(), label="Documents interrogés")
                depot = gr.File(label="Ajouter des PDF", file_count="multiple", file_types=[".pdf"])
                bouton_ajout = gr.Button("📥 Ajouter et indexer")
                message_ajout = gr.Markdown("")
                with gr.Accordion("⚙️ Réglages", open=False):
                    k = gr.Slider(1, 8, value=K_RESULTATS, step=1, label="Passages transmis")
                    seuil = gr.Slider(0.0, 1.0, value=float(BIB.seuil), step=0.01, label="Seuil de confiance")
                    garde = gr.Checkbox(value=True, label="Garde-fous activés")
                    memoire = gr.Checkbox(value=True, label="Mémoire de conversation")
            with gr.Column(scale=2):
                chat = gr.Chatbot(**args_chat)
                with gr.Row():
                    question = gr.Textbox(placeholder="Posez votre question sur les documents…", show_label=False,
                                          scale=5, autofocus=True)
                    envoyer_web = gr.Button("Envoyer", variant="primary", scale=1)
                gr.Examples(examples=[[s] for s in SUGGESTIONS], inputs=[question], label="Suggestions")
                with gr.Row():
                    nouvelle = gr.Button("🔄 Nouvelle conversation", size="sm")
                    message_avis = gr.Markdown("_Utilisez 👍 / 👎 sous chaque réponse pour donner votre avis._")
                with gr.Accordion("📄 Sources de la dernière réponse", open=True):
                    sources = gr.HTML("<i>Les sources apparaîtront ici.</i>")
        entrees = [question, chat, memoire_etat, docs, k, seuil, garde, memoire]
        sorties = [question, chat, memoire_etat, sources]
        envoyer_web.click(repondre_web, entrees, sorties)
        question.submit(repondre_web, entrees, sorties)
        nouvelle.click(lambda: ([], [], "<i>Les sources apparaîtront ici.</i>"), None, [chat, memoire_etat, sources])
        bouton_ajout.click(ajouter_pdf, [depot, docs], [docs, resume, message_ajout])
        if hasattr(chat, "like"):
            chat.like(noter, [memoire_etat], [message_avis])
    app._args_launch = {} if "theme" in params_blocks else {"theme": theme, "css": css}
    return app

def lancer_application(app):
    import inspect
    auth = (UTILISATEUR_WEB, MOT_DE_PASSE_WEB) if MOT_DE_PASSE_WEB else None
    options = dict(share=PARTAGER_LIEN, auth=auth, prevent_thread_lock=True, quiet=True, show_error=True,
                   inbrowser=(ENV["nom"] == "local"), inline=(ENV["nom"] == "colab"), **app._args_launch)
    parametres = inspect.signature(app.launch).parameters
    options = {k: v for k, v in options.items() if k in parametres}
    try:
        app.launch(**options)
    except Exception as erreur:
        if options.get("share"):
            print(f"⚠️ Lien public indisponible ({type(erreur).__name__}) : lancement en local.")
            options["share"] = False
            app.launch(**options)
        else:
            raise
    lien_public = getattr(app, "share_url", None)
    lien_local = getattr(app, "local_url", None) or "http://127.0.0.1:7860"
    lignes = [f'<div style="font-size:14px;margin:4px 0;">🔗 <b>Lien public :</b> <a href="{lien_public}" target="_blank">{lien_public}</a></div>'
              if lien_public else "",
              f'<div style="font-size:14px;margin:4px 0;">💻 <b>Adresse locale :</b> <code>{lien_local}</code></div>',
              f'<div style="font-size:13px;margin:4px 0;">🔒 Accès : ' + (f'identifiant <code>{UTILISATEUR_WEB}</code> et mot de passe défini'
              if auth else f'<span style="color:{C["brique"]};"><b>sans mot de passe</b></span>') + '</div>']
    display(HTML(f'{_CSS}<div class="rg rg-carte" style="border-left:6px solid {C["vert"]};">'
                 f'<div class="rg-titre">🌐 Application web lancée</div>{"".join(lignes)}'
                 f'<div class="rg-note">Pour l\'arrêter : exécutez <code>APP.close()</code>. '
                 f'Elle s\'arrête aussi à la fin de la session.</div></div>'))
    if lien_public and not auth:
        encadre("Le lien est public et <b>sans mot de passe</b> : définissez <code>MOT_DE_PASSE_WEB</code> "
                "avant de le partager.", "alerte")

if LANCER_INTERFACE_WEB:
    APP = creer_application()
    lancer_application(APP)
else:
    encadre("Application web non lancée (<code>LANCER_INTERFACE_WEB = False</code>).", "info")
```

### Les limites de cet assistant

- **Les mises en page complexes** (colonnes multiples, tableaux sans bordures, notes de bas de page) sont imparfaitement extraites. Vérifiez l'étape B2 sur vos documents.
- **Les graphiques ne sont pas lus**, seuls leurs titres et légendes le sont.
- **Le mode extractif** cite des phrases ou des lignes de tableau : il ne sait ni **calculer** ni **classer** (« quelle région a le chômage le plus élevé ? »). Ces questions demandent un modèle de langage.
- **La mémoire par règles** reconnaît les changements de région, de sexe, de trimestre, d'année, de milieu, d'âge et d'indicateur. Au-delà, un modèle est nécessaire.
- **La vérification des chiffres est littérale** : un chiffre recalculé (une différence, une moyenne) sera signalé comme absent, même s'il est juste.
- **La vérification finale reste humaine** : l'assistant indique la page, un statisticien vérifie avant toute diffusion.

### ✅ Point de contrôle

<details><summary><b>1. Pourquoi afficher « Compris comme… » ?</b></summary>
Parce que la mémoire peut se tromper. L'utilisateur voit la question réellement recherchée et peut reformuler si elle ne correspond pas à son intention.
</details>

<details><summary><b>2. Pourquoi limiter le nombre de passages d'un même document ?</b></summary>
Pour qu'une question comparative reçoive des extraits de plusieurs documents, au lieu de trois passages du même rapport.
</details>

<details><summary><b>3. Pourquoi un mot de passe sur l'application web ?</b></summary>
Un lien public est accessible à quiconque le possède : sans mot de passe, n'importe qui peut interroger vos documents ou en déposer de nouveaux.
</details>

<details><summary><b>4. Une réponse fausse avec le bon passage : que corriger ?</b></summary>
La <b>génération</b> : la consigne ou le modèle. Avec un passage hors sujet : la <b>récupération</b> (extraction, glossaire, seuil).
</details>

### 🛠️ En cas de problème

| Symptôme | Cause probable | Solution |
|---|---|---|
| L'interface C1 ne s'affiche pas | Widgets non pris en charge | Utilisez `demander("…")` ou l'application web |
| « Lien public indisponible » | Internet désactivé ou réseau filtré | Kaggle : *Settings → Internet : On* ; sinon, utilisez l'adresse locale |
| « Redémarrez la session » après l'installation de Gradio | Ancienne version déjà chargée | *Exécution → Redémarrer*, puis *Tout exécuter* |
| Un PDF déposé n'apparaît pas | Fichier protégé ou scanné | Voir le message sous le bouton ; activez l'OCR si besoin |
| « Chiffre absent des sources » sur une bonne réponse | Chiffre calculé ou reformulé par le modèle | Vérifiez le calcul à partir des sources citées |
| « Compris comme » incorrect | Mémoire par règles limitée | Posez une question complète, ou activez un modèle (Groq, Ollama) |
| Trop de refus | Seuil trop élevé | Abaissez le seuil dans les réglages ; enrichissez le glossaire |
| « Quota momentanément atteint » | Limite par minute de Groq | Le notebook attend et réessaie ; pour un usage intensif, utilisez Ollama |
| Ollama très lent | Pas de GPU | Activez un GPU T4, ou utilisez `qwen2.5:1.5b` |
